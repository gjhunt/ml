# NNs: Generalization, Regularization, and Modeling - Review Questions {#sec-nnarch-questions}


### Question 1

Give one precise way in which each of the following can increase model flexibility: depth, width, and amount of training. 

<details>
<summary>Answer</summary>

Increasing depth adds more composed transformations, so the network can build more complex hierarchical representations. Increasing width adds more units in a layer, so the layer can represent more features or directions at once. Increasing the amount of training can let the optimizer fit the empirical risk more closely.

Greater flexibility can increase the opportunity to overfit, but test performance also depends on the data, architecture, regularization, optimization bias, and validation-based model selection. Over-parameterized models may still generalize well, so flexibility alone does not determine test error.

</details>


### Question 2

A dense layer maps $s_i:\mathbb R^{d_{i-1}}\to\mathbb R^{d_i}$ using

$$
h^{(i)}=\phi_i(W^{(i)}h^{(i-1)}+b^{(i)}).
$$

How many trainable parameters does this layer have including the weights $W^{(i)}$ and bias term $b^{(i)}$. 

<details>
<summary>Answer</summary>

The weight matrix has shape

$$
W^{(i)}\in\mathbb R^{d_i\times d_{i-1}},
$$

so it contains $d_i d_{i-1}$ scalar weights. The bias vector has shape

$$
b^{(i)}\in\mathbb R^{d_i},
$$

so it contains $d_i$ scalar biases. Therefore the dense layer has

$$
d_i d_{i-1}+d_i
$$

trainable parameters.

</details>


### Question 3

Compare this with a linear regression model with 20 predictors and an intercept.

<summary>Answer</summary>

The hidden layer has

$$

$$
1\cdot50+1=51
$$

parameters. The full network therefore has

$$
1050+51=1101

parameters.

A linear regression model with 20 predictors and an intercept has $20+1=21$ parameters. The neural network has many more parameters because it learns hidden features before the final scalar output.

</details>


### Question 4

A weight-decay-style update in the lecture is

$$
\theta^{(t+1)}=(1-\eta\lambda)\theta^{(t)}-\eta\nabla_\theta\hat R(\theta^{(t)}).
$$

Which penalized objective gives exactly this update under ordinary gradient descent: $\hat R(\theta)+\lambda\|\theta\|_2^2$ or $\hat R(\theta)+\frac{\lambda}{2}\|\theta\|_2^2$? Justify your answer.

<details>
<summary>Answer</summary>

The objective

$$
\hat R(\theta)+\frac{\lambda}{2}\|\theta\|_2^2
$$

has gradient

$$
\nabla_\theta\hat R(\theta)+\lambda\theta.
$$

Gradient descent gives

$$
\theta^{(t+1)}=\theta^{(t)}-
\eta\{\nabla_\theta\hat R(\theta^{(t)})+\lambda\theta^{(t)}\}
=(1-\eta\lambda)\theta^{(t)}-
\eta\nabla_\theta\hat R(\theta^{(t)}).
$$

If the objective were $\hat R(\theta)+\lambda\|\theta\|_2^2$, the penalty gradient would be $2\lambda\theta$, giving a factor $1-2\eta\lambda$ instead.

</details>


### Question 5

For inverted dropout, the lecture defines

$$
m_i\sim\operatorname{Bernoulli}(1-p),\qquad
\tilde h_i=\frac{m_i h_i}{1-p}.
$$

Verify that $\mathbb E[\tilde h_i]=h_i$.

<details>
<summary>Answer</summary>

Since $m_i\sim\operatorname{Bernoulli}(1-p)$,

$$
\mathbb E[m_i]=1-p.
$$

Treating $h_i$ as fixed during the dropout draw,

$$
\mathbb E[\tilde h_i]
=\mathbb E\left[\frac{m_i h_i}{1-p}\right]
=\frac{h_i}{1-p}\mathbb E[m_i]
=\frac{h_i}{1-p}(1-p)
=h_i.
$$

Thus inverted dropout preserves the activation's expected scale during training.

</details>


### Question 6

Take $h_i=3$ and dropout probability $p=0.25$. Under inverted dropout, what values can $\tilde h_i$ take, and with what probabilities? Compute its mean and variance.

<details>
<summary>Answer</summary>

The unit is dropped with probability $p=0.25$, so $\tilde h_i=0$ with probability $0.25$. It is kept with probability $1-p=0.75$, and then

$$
\tilde h_i=\frac{3}{0.75}=4.
$$

The mean is

$$
0(0.25)+4(0.75)=3.
$$

The variance can be computed directly:

$$
0.25(0-3)^2+0.75(4-3)^2
=0.25\cdot9+0.75\cdot1=3.
$$

</details>


### Question 7

Classify each training pattern as more suggestive of underfitting, overfitting, or reasonable fit. Use the lecture's emphasis on training and validation performance.

1. Training loss high, validation loss high.
2. Training loss low, validation loss much higher and worsening.
3. Training loss low, validation loss low and stable.

<details>
<summary>Answer</summary>

1. High training and validation loss suggests underfitting: the model is not fitting even the training data well.
2. Low training loss but much higher and worsening validation loss suggests overfitting: continued training is fitting training-set idiosyncrasies rather than improving generalization.
3. Low training loss and low stable validation loss suggests a reasonable fit, at least relative to the available validation data.

The diagnosis is based on the gap and trend between training and validation performance, not training performance alone.

</details>


### Question 8

In early stopping, why should the validation set be used to decide when to stop, while the test set should be reserved until after model selection is complete?

<details>
<summary>Answer</summary>

The validation set is part of the model-selection procedure: it helps choose when to stop training and, more generally, tune hyperparameters. If the test set were used for that choice, then information from the test set would influence the fitted model or training procedure.

The test set is meant to estimate performance on new data after all model-selection decisions are finished. Keeping it separate gives a less biased final estimate of generalization performance.

</details>


### Question 9

Suppose each feature $x_j$ is standardized using a mean $\bar x_j$ and standard deviation $s_j$:

$$
x_j\mapsto \frac{x_j-\bar x_j}{s_j}.
$$

A student computes $\bar x_j$ and $s_j$ using the full dataset before splitting the data into training, validation, and test sets.

Explain why this is a form of data leakage. Then state the correct procedure for computing and applying $\bar x_j$ and $s_j$.

<details>
<summary>Answer</summary>

This is data leakage because the preprocessing step uses information from observations that are supposed to be held out for validation or testing. Even though the labels are not used, the feature distributions from the validation and test sets influence the values of $\bar x_j$ and $s_j$. Those values are then used to transform the training data before the model is fit.

The correct procedure is:

1. Split the data into training, validation, and test sets.
2. Compute $\bar x_j$ and $s_j$ using only the training set.
3. Treat those values as fixed preprocessing parameters.
4. Apply the same transformation to the training, validation, and test observations:

$$
x_j\mapsto \frac{x_j-\bar x_j}{s_j}.
$$

The validation and test sets should be transformed using the training-set mean and standard deviation, not their own means and standard deviations.

</details>


### Question 10

In SGD with momentum, suppose

$$
v_t=\rho v_{t-1}+g_t,
\qquad v_0=0,
$$

and $g_t=g$ is the same vector for every step. Derive $v_t$ in closed form.

<details>
<summary>Answer</summary>

Expanding the recursion gives

$$
v_1=g,
$$

$$
v_2=\rho g+g=(1+\rho)g,
$$

and in general

$$
v_t=(1+\rho+\rho^2+\cdots+\rho^{t-1})g.
$$

For $\rho\ne1$,

$$
v_t=\frac{1-\rho^t}{1-\rho}g.
$$

If $0\le \rho<1$, then $v_t\to g/(1-\rho)$ as $t\to\infty$.

</details>


### Question 11

Compute one RMSProp update in one dimension. Let $\theta^{(t)}=2$, $g_t=4$, $s_{t-1}=9$, $\rho=0.9$, $\eta=0.01$, and ignore $\epsilon$ for simplicity. 

<details>
<summary>Answer</summary>

First compute the squared-gradient average:

$$
s_t=0.9(9)+0.1(4^2)=8.1+1.6=9.7.
$$

Then

$$
\sqrt{s_t}=\sqrt{9.7}\approx3.114.
$$

The update is

$$
\theta^{(t+1)}=2-0.01\frac{4}{3.114}
\approx2-0.01285=1.98715.
$$

</details>


### Question 12

Using the lecture's simplified Adam update, take $m_{t-1}=0$, $s_{t-1}=0$, $g_t=2$, $\beta_1=0.9$, $\beta_2=0.99$, $\eta=0.001$, and ignore $\epsilon$. Compute the updated used in the Adam update to the parameters. 

<details>
<summary>Answer</summary>

The first moment is

$$
m_t=0.9(0)+(1-0.9)(2)=0.2.
$$

The squared-gradient average is

$$
s_t=0.99(0)+(1-0.99)(2^2)=0.04.
$$

Thus

$$
\sqrt{s_t}=0.2.
$$

The update size is (ignoring $\epsilon$)

$$
\eta\frac{m_t}{\sqrt{s_t}}
=0.001\frac{0.2}{0.2}=0.001.
$$

</details>

### Question 14

Give a precise distinction between an over-parameterized model and an overfit model.

<details>
<summary>Answer</summary>

A model is over-parameterized when it has many parameters, possibly enough to fit the training data extremely well. This is a statement about model size and capacity relative to the data.

A model is overfit when it generalizes poorly: it performs much better on training data than on new data. This is a statement about the gap between training performance and out-of-sample performance.


</details>


### Question 15

Name three reasons why an over-parameterized neural network can still generalize well.

<details>
<summary>Answer</summary>

First, architecture imposes structure: the model is not an arbitrary function.

Second, explicit regularization can help: the example uses dropout and an $L_2$ kernel regularizer on a dense layer.

Third, optimization can have implicit regularizing effects: gradient-based methods such as Adam do not choose an arbitrary parameter setting but follow a particular training trajectory determined by initialization, mini-batches, gradients, and update rules.

</details>


### Question 16

A student summarizes double descent as: "Once a model is past the interpolation threshold, making it larger always improves test error." Explain why this is not what the lecture claims.

<details>
<summary>Answer</summary>

The lecture says that in some modern machine-learning settings, test error may decrease again after the interpolation threshold. It explicitly does not claim that larger models are always better.

Double descent is a possible pattern, not a universal monotonic guarantee. Test error still depends on the data-generating process, architecture, optimization, regularization, noise, and the way model complexity is measured. The safer conclusion is that parameter count alone is not enough to determine whether a model will overfit.

</details>
