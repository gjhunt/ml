### Question 1

Let $X$ be continuously distributed on $[0,1]$, and let $Y\sim\operatorname{Bernoulli}(1/2)$ be independent of $X$. Suppose the training data are

$$
D={(X_1,Y_1),\ldots,(X_n,Y_n)},
$$

where the $X_i$'s are distinct. Consider the memorizing classifier

$$
\hat f(x)=
\begin{cases}
Y_i, & \text{if } x=X_i \text{ for some training input } X_i,\\
0, & \text{otherwise.}
\end{cases}
$$

Using 0-1 loss,

$$
\ell(\hat f(x),y)=\mathbf 1\{\hat f(x)\neq y\},
$$

argue that $\hat f$ has zero training error but population risk $1/2$.

<details>
<summary>Answer</summary>

The classifier has zero training error because it memorizes the observed labels:

$$
\hat f(X_i)=Y_i
$$

for every training example $i=1,\ldots,n$. Therefore,

$$
\frac{1}{n}\sum_{i=1}^n \mathbf 1\{\hat f(X_i)\neq Y_i\}=0.
$$

For a new draw $(X,Y)$, the input $X$ is almost surely not equal to any of the finitely many training inputs $X_1,\ldots,X_n$, since $X$ has a continuous distribution. Thus $\hat f(X)=0$ almost surely on a new test point. Because $Y\sim\operatorname{Bernoulli}(1/2)$ and is independent of $X$,

$$
R(\hat f)=
\mathbb E[\mathbf 1\{\hat f(X)\neq Y\}]=
P(Y=1)=
\frac{1}{2}.
$$

So the classifier fits the training data perfectly but performs no better than predicting the constant label $0$ on new data. This illustrates that zero empirical risk does not necessarily imply low population risk.

</details>

### Question 3

Consider a three-class classification problem. Scores and actions are separated by $\hat f(x)=a(\hat s(x))$. For three class-scores $t=(2,5,1)$ and associated classes $\mathcal C=\{1,2,3\}$, compute the classification action $a(t)=\arg\max_k t_k$. Then give a score vector with the same action but different score values.

<details>
<summary>Answer</summary>

The largest component of $(2,5,1)$ is the second, so $a(t)=2$. A different score vector such as $(0,10,-7)$ has the same classification because its second component is also largest. This shows that the action can discard numerical information in the score.

### Question 4

Cross-entropy is defined from class scores using $p_y(x)=\exp(\hat s_y(x))/\sum_{k=1}^K\exp(\hat s_k(x))$. Verify that $\sum_{j=1}^K p_j(x)=1$ and compute the loss when all $K$ scores are equal.

<!-- Source: loss functions; softmax probabilities and cross-entropy -->

<details>
<summary>Answer</summary>

Since $p_j=e^{s_j}/\sum_k e^{s_k}$, summing over $j$ gives $\sum_j p_j=(\sum_j e^{s_j})/(\sum_k e^{s_k})=1$. If all scores are equal to $c$, then for any $y$, $p_y=e^c/(Ke^c)=1/K$, so the cross-entropy loss is $-\log(1/K)=\log K$.

</details>

### Question 5

For zero-one loss $\ell(y,s(x))=\mathbf 1\{y\ne a(s(x))\}$, rewrite the population risk $R(s)$ as a probability of an event.

<!-- Source: loss functions; zero-one loss and population risk -->

<details>
<summary>Answer</summary>

The population risk is $R(s)=\mathbb E[\mathbf 1\{Y\ne a(s(X))\}]$. The expectation of an indicator is the probability of its event, so $R(s)=P(Y\ne a(s(X)))$. Thus zero-one risk is the misclassification probability.

</details>

### Question 6

Assume $(x_1,y_1),\ldots,(x_N,y_N)\overset{\mathrm{i.i.d.}}\sim P$. For a fixed score $s$, prove that $\hat R_N(s)$ has expectation $R(s)$.


<details>
<summary>Answer</summary>

Let $Z_n=\ell(y_n,s(x_n))$. The i.i.d. assumption makes the $Z_n$ identically distributed with $\mathbb E Z_n=R(s)$. Therefore $\mathbb E\hat R_N(s)=\mathbb E[N^{-1}\sum_n Z_n]=N^{-1}\sum_n R(s)=R(s)$.

</details>

### Question 7

Consider the training data

$$
D=\{(0,0),(1,0)\}.
$$

Let the loss be squared loss, $\ell(s(x),y)=(s(x)-y)^2$. Construct two score functions $s_1$ and $s_2$ that agree on the training inputs $x=0$ and $x=1$, but have different population risks under the distribution where $X$ is uniform on ${0,1,2}$ and $Y=0$ always.

Use this example to explain why empirical risk alone may not identify out-of-sample behavior.

<details>
<summary>Answer</summary>

Define

$$
s_1(x)=0 \quad \text{for all } x\in\{0,1,2\},
$$

and

$$
s_2(0)=0,\qquad s_2(1)=0,\qquad s_2(2)=1.
$$

The two score functions agree on the training inputs $0$ and $1$. Therefore, they have the same empirical risk on $D$. In fact, both have empirical risk $0$, since both predict $0$ on the two training examples and $Y=0$.

However, their population risks differ. Since $Y=0$ always,

$$
R(s_1)=\mathbb{E}[(s_1(X)-Y)^2]=0,
$$

while

$$
R(s_2)=\mathbb{E}[(s_2(X)-Y)^2]
= \frac{1}{3}(0^2+0^2+1^2)
= \frac{1}{3}.
$$

Thus, the two functions are indistinguishable from the training data but behave differently at the unseen input $x=2$. This shows that empirical risk alone only constrains predictions on observed training inputs. It does not, by itself, determine how a model behaves out of sample.

</details>


### Question 8

Compare population risk minimization and empirical risk minimization using the same loss $\ell$ and class $\mathcal S$. State precisely which object is unknown in one problem and random in the other.

<!-- Source: population risk versus empirical risk; ERM definition -->

<details>
<summary>Answer</summary>

Population risk minimization chooses $s^*\in\arg\min_{s\in\mathcal S}\mathbb E_P\ell(Y,s(X))$, but $P$ is unknown. ERM chooses $\hat s\in\arg\min_{s\in\mathcal S}N^{-1}\sum_n\ell(y_n,s(x_n))$. The empirical objective is computable but random because it depends on the random sample.

</details>

### Question 9

Let $\mathcal S_1\subseteq\mathcal S_2$. Prove that the minimum empirical risk over $\mathcal S_2$ is no larger than the minimum empirical risk over $\mathcal S_1$.

<details>
<summary>Answer</summary>

Because every candidate in $\mathcal S_1$ is also in $\mathcal S_2$, minimizing over $\mathcal S_2$ has at least all candidates available to $\mathcal S_1$. Therefore $\min_{s\in\mathcal S_2}\hat R(s)\le \min_{s\in\mathcal S_1}\hat R(s)$. 

</details>

### Question 11

Higher loss is worse, and losses are typically nonnegative. Construct a loss $\tilde\ell(y,s)=\ell(y,s)+5$ from any nonnegative loss $\ell$. Does ERM over a fixed class change? Prove your answer.

<!-- Source: loss functions; transformations that preserve ERM minimizers -->

<details>
<summary>Answer</summary>

The empirical risk for $\tilde\ell$ is $\hat R_{\tilde\ell}(s)=N^{-1}\sum_n(\ell_n(s)+5)=\hat R_\ell(s)+5$. Adding the same constant to every candidate's objective preserves all orderings, so the ERM minimizers over a fixed class are unchanged.

</details>

### Question 12

A transformation squares the average empirical risk: $Q(s)=\hat R(s)^2$. If $\hat R(s)\ge0$ for every $s$, does minimizing $Q$ give the same minimizers as minimizing $\hat R$?

<details>
<summary>Answer</summary>

Yes. On $[0,\infty)$, the map $u\mapsto u^2$ is strictly increasing, so $\hat R(s_1)\le\hat R(s_2)$ if and only if $Q(s_1)\le Q(s_2)$. 

</details>

### Question 13

For a binary classification problem with score $s(x)\in\mathbb R$ and action $a_c(s)=\mathbf 1\{s\ge c\}$, show how changing $c$ can change zero-one loss. Give a numerical example with one observation.

<!-- Source: score/action separation; action threshold affects loss -->

<details>
<summary>Answer</summary>

Take one observation with $y=1$ and score $s(x)=0.7$. If $c=0.5$, then $a_c(s)=1$ and zero-one loss is $0$. If $c=0.8$, then $a_c(s)=0$ and zero-one loss is $1$. The score is unchanged; only the action rule changed.

</details>

### Question 14

In the statistical setup, $x_n=(X_{n1},\ldots,X_{nD})\in\mathbb R^D$. For a score $s:\mathbb R^D\to\mathbb R^K$, state the dimensions of $s(x_n)$ and the dimension of the action output in $K$-class classification.

</details>

### Question 15

For multi-class scores $s:\mathbb R^D\to\mathbb R^K$, add the same constant $c$ to every component: $s'_k(x)=s_k(x)+c$. Does the argmax action change? Does the cross-entropy loss from the lecture change? Prove both answers.

<!-- Source: classification scores; argmax and cross-entropy invariance to common shifts -->

<details>
<summary>Answer</summary>

The argmax does not change because adding the same constant preserves the ordering of components. For cross-entropy, $p'_y=e^{s_y+c}/\sum_k e^{s_k+c}=e^c e^{s_y}/(e^c\sum_k e^{s_k})=p_y$, so $-\log p'_y=-\log p_y$. Both the action and this loss are invariant to common score shifts.

</details>
