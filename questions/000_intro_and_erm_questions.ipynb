### Question 1

Using the lecture's treatment of supervised data $(x_n,y_n)$ and generalization, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Training loss is computed on the observed sample, but learning targets new draws from $P_{X,Y}$. A rule can fit noise in the sample, so low empirical risk is only a proxy for low population risk.

</details>

### Question 2

Using the lecture's treatment of regression versus classification, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Regression has $y\in\mathbb R$ and real-valued predictions; classification has $y\in\{c_1,\ldots,c_K\}$. The loss should match the output space: squared error for real errors, classification or likelihood losses for categories.

</details>

### Question 3

Using the lecture's treatment of unsupervised learning with only $x_1,\ldots,x_N$, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Without labels there is no supervised target. The mathematical task is exploratory: find structure such as clusters, dependence, or low-dimensional summaries in the marginal distribution of $X$.

</details>

### Question 4

Using the lecture's treatment of score and action separation, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

A score $s(x)$ can be real-valued while the action maps it to a decision, such as $a(s)=1\{s>0\}$. Changing the action threshold can change classifications without refitting the score.

</details>

### Question 5

Using the lecture's treatment of population risk, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Population risk is $R(s)=\mathbb E[\ell(Y,s(X))]$. It averages loss over the data-generating distribution, so it is the ideal out-of-sample criterion.

</details>

### Question 6

Using the lecture's treatment of empirical risk, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Empirical risk is $\hat R(s)=N^{-1}\sum_n \ell(y_n,s(x_n))$. It replaces an unknown expectation by a sample average and is therefore computable.

</details>

### Question 7

Using the lecture's treatment of positive rescaling of an ERM objective, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

If $c>0$, then $\arg\min_s c\hat R(s)=\arg\min_s\hat R(s)$ because multiplication by $c$ preserves all order comparisons.

</details>

### Question 8

Using the lecture's treatment of zero-one loss, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Zero-one loss estimates the probability of an incorrect action, but it is discontinuous and usually nonconvex in model parameters, making direct optimization difficult.

</details>

### Question 9

Using the lecture's treatment of the i.i.d. assumption, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For fixed $s$, the losses are i.i.d. with mean $R(s)$. Laws of large numbers justify $\hat R(s)$ as an approximation to $R(s)$.

</details>

### Question 10

Using the lecture's treatment of models as approximations, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

The lecture’s statistical viewpoint treats data as noisy and incomplete. A model is useful if it predicts or summarizes well, not because it is literally true.

</details>

### Question 11

Using the lecture's treatment of choosing a model class, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

ERM is meaningful only after restricting the candidate functions. Too large a class can interpolate arbitrary training labels and generalize poorly.

</details>

### Question 12

Using the lecture's treatment of identity action in regression, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For squared-error regression the action is often $a(s)=s$, so $\hat f(x)=\hat s(x)$. The score and prediction coincide.

</details>

### Question 13

Using the lecture's treatment of finite-sample ambiguity, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Two functions may tie on empirical risk but differ off sample. This illustrates why evaluation, regularization, and assumptions are needed.

</details>

### Question 14

Using the lecture's treatment of classification with $K$ classes, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

The predictor maps into a finite set $\mathcal C=\{c_1,\ldots,c_K\}$. Multi-class classification is the case $K>2$.

</details>

### Question 15

Using the lecture's treatment of loss functions as decision criteria, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

The learned rule optimizes the chosen loss. If the loss does not encode the scientific or decision goal, ERM can solve the wrong problem perfectly.

</details>

### Question 16

Using the lecture's treatment of unseen data, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

The phrase “new data” means another draw from the same distribution $P$, not merely another row already used for fitting.

</details>

### Question 17

Using the lecture's treatment of supervised versus unsupervised objectives, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Supervised learning estimates a relation between $X$ and $Y$; unsupervised learning studies structure in $X$ alone.

</details>

### Question 18

Using the lecture's treatment of probabilistic uncertainty, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Randomness can represent measurement noise, unobserved variables, or inherent variability; risk averages over these sources.

</details>

### Question 19

Using the lecture's treatment of the method/evaluation split, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Methods construct $\hat f$ from data; evaluation asks how good $\hat f$ is. The two interact but are logically distinct.

</details>

### Question 20

Using the lecture's treatment of overfitting as an ERM failure mode, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

A highly flexible ERM can drive empirical risk down by matching accidental sample patterns, increasing population risk.

</details>
