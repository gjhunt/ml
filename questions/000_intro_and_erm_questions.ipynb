### Question 1

Supervised learning observes pairs $(x_n,y_n)$ and aims to generalize. Construct a binary classification example where a rule has zero training error but population zero-one risk $1/2$.

<!-- Source: supervised learning; generalization versus fitting observed data -->

<details>
<summary>Answer</summary>

Let $X$ be continuous on $[0,1]$ and let $Y\sim\operatorname{Bernoulli}(1/2)$ be independent of $X$. For distinct training inputs, define a memorizing classifier that returns the observed label at each training $x_n$ and returns $0$ elsewhere. It has zero training error. For a new draw, $X$ almost surely is not one of the finitely many training inputs, and $Y$ is independent fair noise, so the error probability is $P(Y=1)=1/2$.

</details>

### Question 3

Only inputs $x_1,\ldots,x_N$ are observed in an unsupervised problem. Give one mathematically precise objective for clustering these inputs and explain why it is not a supervised prediction risk.

<!-- Source: unsupervised learning; clustering as structure discovery -->

<details>
<summary>Answer</summary>

One objective is $\min_{z_1,\ldots,z_N,\mu_1,\ldots,\mu_K}\sum_{n=1}^N\|x_n-\mu_{z_n}\|^2$, the k-means within-cluster sum of squares. It is not supervised prediction risk because no response $Y$ is observed and the objective does not compare predictions to labels $y_n$; it summarizes geometry in the marginal sample of $X$.

</details>

### Question 4

Scores and actions are separated by $\hat f(x)=a(\hat s(x))$. For scores $t=(2,5,1)$ and classes $\mathcal C=\{1,2,3\}$, compute the classification action $a(t)=\arg\max_k t_k$. Then give a score vector with the same action but different score values.

<!-- Source: score/action separation; argmax action for classification -->

<details>
<summary>Answer</summary>

The largest component of $(2,5,1)$ is the second, so $a(t)=2$. A different score vector such as $(0,10,-7)$ has the same action because its second component is also largest. This shows that the action can discard numerical information in the score.

</details>

### Question 5

A regression model uses identity action $a(u)=u$. Prove that, in this case, knowing the score function $\hat s$ is exactly the same as knowing the prediction function $\hat f$.

<!-- Source: score/action separation; identity action in regression -->

<details>
<summary>Answer</summary>

If $a$ is the identity, then $\hat f(x)=a(\hat s(x))=\hat s(x)$ for every input $x$. Thus the two functions have the same domain and same value at every point. Conversely, knowing $\hat f$ gives $\hat s=\hat f$.

</details>

### Question 6

Cross-entropy is defined from class scores by $p_y(x)=\exp(\hat s_y(x))/\sum_{k=1}^K\exp(\hat s_k(x))$. Verify that $\sum_{j=1}^K p_j(x)=1$ and compute the loss when all $K$ scores are equal.

<!-- Source: loss functions; softmax probabilities and cross-entropy -->

<details>
<summary>Answer</summary>

Since $p_j=e^{s_j}/\sum_k e^{s_k}$, summing over $j$ gives $\sum_j p_j=(\sum_j e^{s_j})/(\sum_k e^{s_k})=1$. If all scores are equal to $c$, then $p_y=e^c/(Ke^c)=1/K$, so the cross-entropy loss is $-\log(1/K)=\log K$.

</details>

### Question 7

For zero-one loss $\ell(y,s(x))=\mathbf 1\{y\ne a(s(x))\}$, rewrite the population risk $R(s)$ as a probability of an event.

<!-- Source: loss functions; zero-one loss and population risk -->

<details>
<summary>Answer</summary>

The population risk is $R(s)=\mathbb E[\mathbf 1\{Y\ne a(s(X))\}]$. The expectation of an indicator is the probability of its event, so $R(s)=P(Y\ne a(s(X)))$. Thus zero-one risk is the misclassification probability.

</details>

### Question 8

For squared-error regression with score $s$, write $R(s)$ explicitly as an integral or expectation under $P_{X,Y}$. Then state what quantity would be minimized if $P_{X,Y}$ were known.

<!-- Source: population risk; squared-error regression -->

<details>
<summary>Answer</summary>

The risk is $R(s)=\mathbb E_{(X,Y)\sim P}[(Y-s(X))^2]$, equivalently $\int (y-s(x))^2\,dP_{X,Y}(x,y)$. If $P_{X,Y}$ were known and the candidate class were unrestricted, one would choose a score minimizing this expected squared loss.

</details>

### Question 9

Assume $(x_1,y_1),\ldots,(x_N,y_N)\overset{\mathrm{i.i.d.}}\sim P$. For a fixed score $s$, prove that $\hat R_N(s)$ is an unbiased estimator of $R(s)$ whenever the loss has finite expectation.

<!-- Source: statistical setup; empirical risk as sample average -->

<details>
<summary>Answer</summary>

Let $Z_n=\ell(y_n,s(x_n))$. The i.i.d. assumption makes the $Z_n$ identically distributed with $\mathbb E Z_n=R(s)$. Therefore $\mathbb E\hat R_N(s)=\mathbb E[N^{-1}\sum_n Z_n]=N^{-1}\sum_n R(s)=R(s)$.

</details>

### Question 10

Identify the exact step in the previous unbiasedness argument that may fail if the data are not identically distributed.

<!-- Source: statistical setup; role of identical distribution -->

<details>
<summary>Answer</summary>

The step $\mathbb E Z_n=R(s)$ for every $n$ can fail. If observation $n$ is drawn from $P_n$ rather than a common $P$, then $\mathbb E Z_n=\mathbb E_{P_n}\ell(Y,s(X))$, so the empirical average estimates the average of risks under the $P_n$'s rather than the single target risk under $P$.

</details>

### Question 11

Construct two score functions that agree on the training inputs but can have different population risks. Use this to make precise why empirical risk alone may not identify out-of-sample behavior.

<!-- Source: empirical risk; finite sample ambiguity -->

<details>
<summary>Answer</summary>

Let the input space be $[0,1]$ and suppose the training inputs are a finite set $T$. Define $s_1(x)=0$ for all $x$, and define $s_2(x)=0$ on $T$ but $s_2(x)=1$ off $T$. They have identical empirical risk for any loss depending only on training values. Under a distribution putting positive mass off $T$, their population risks can differ, for example under squared loss with $Y=0$.

</details>

### Question 12

The lecture asks, 'which $s$ do we optimize over? All of them?' Give a concrete reason why optimizing zero-one empirical risk over all functions can be ill-posed for generalization.

<!-- Source: ERM; need to choose a candidate class $\mathcal S$ -->

<details>
<summary>Answer</summary>

If all functions are allowed and training inputs are distinct, one can define $s$ so that $a(s(x_n))=y_n$ for every training point, giving zero empirical classification error. This says little about new inputs, because the function can be arbitrary away from the observed points. A restricted class $\mathcal S$ supplies the inductive bias needed for generalization.

</details>

### Question 13

Compare population risk minimization and empirical risk minimization using the same loss $\ell$ and class $\mathcal S$. State precisely which object is unknown in one problem and random in the other.

<!-- Source: population risk versus empirical risk; ERM definition -->

<details>
<summary>Answer</summary>

Population risk minimization chooses $s^*\in\arg\min_{s\in\mathcal S}\mathbb E_P\ell(Y,s(X))$, but $P$ is unknown. ERM chooses $\hat s\in\arg\min_{s\in\mathcal S}N^{-1}\sum_n\ell(y_n,s(x_n))$. The empirical objective is computable but random because it depends on the random sample.

</details>

### Question 14

Let $\mathcal S_1\subseteq\mathcal S_2$. Prove that the minimum empirical risk over $\mathcal S_2$ is no larger than the minimum empirical risk over $\mathcal S_1$.

<!-- Source: ERM; model class size and training fit -->

<details>
<summary>Answer</summary>

Because every candidate in $\mathcal S_1$ is also in $\mathcal S_2$, minimizing over $\mathcal S_2$ has at least all candidates available to $\mathcal S_1$. Therefore $\inf_{s\in\mathcal S_2}\hat R(s)\le \inf_{s\in\mathcal S_1}\hat R(s)$. This is a training-risk statement and does not by itself guarantee lower population risk.

</details>

### Question 15

Give a classification score vector $t\in\mathbb R^K$ for which the action $\arg\max_k t_k$ is not uniquely defined. State one missing convention that would make $a(t)$ a function.

<!-- Source: score/action separation; argmax tie convention -->

<details>
<summary>Answer</summary>

For $K=3$, take $t=(1,1,0)$. Both classes 1 and 2 maximize the score, so $\arg\max$ is a set unless a tie-breaking rule is specified. A convention such as 'choose the smallest maximizing index' makes $a(t)$ a well-defined function.

</details>

### Question 16

Higher loss is worse, and losses are typically nonnegative. Construct a loss $\tilde\ell(y,s)=\ell(y,s)+5$ from any nonnegative loss $\ell$. Does ERM over a fixed class change? Prove your answer.

<!-- Source: loss functions; transformations that preserve ERM minimizers -->

<details>
<summary>Answer</summary>

The empirical risk for $\tilde\ell$ is $\hat R_{\tilde\ell}(s)=N^{-1}\sum_n(\ell_n(s)+5)=\hat R_\ell(s)+5$. Adding the same constant to every candidate's objective preserves all orderings, so the ERM minimizers over a fixed class are unchanged.

</details>

### Question 17

A different transformation squares the average empirical risk: $Q(s)=\hat R(s)^2$. If $\hat R(s)\ge0$ for every $s$, does minimizing $Q$ give the same minimizers as minimizing $\hat R$? What condition from the lecture makes the answer simple?

<!-- Source: loss functions; nonnegativity and monotone transformations -->

<details>
<summary>Answer</summary>

Yes. On $[0,\infty)$, the map $u\mapsto u^2$ is strictly increasing, so $\hat R(s_1)\le\hat R(s_2)$ if and only if $Q(s_1)\le Q(s_2)$. The lecture's typical nonnegativity condition on losses ensures empirical risks are nonnegative.

</details>

### Question 18

### Question 19

Data, loss, candidate class, and optimizer are all inputs to an ERM algorithm. Give two learning procedures that use the same data and loss but can return different scores because $\mathcal S$ differs.

<!-- Source: ERM algorithm inputs; role of candidate class -->

<details>

With squared loss on the same data, one procedure might use linear scores $s_w(x)=w^\top x$, while another uses quadratic-feature scores $s_w(x)=w_0+w_1x+w_2x^2$. The loss and data are the same, but the candidate classes differ, so the empirical-risk minimizers and resulting scores can differ.

### Question 20

Now give two procedures that use the same data, loss, and class $\mathcal S$ but may differ because the optimizer only approximately minimizes $\hat R$. What mathematical property of the returned score is no longer guaranteed?

<!-- Source: ERM algorithm inputs; optimizer versus exact minimizer -->

<details>
<summary>Answer</summary>

One procedure could find a global minimizer of $\hat R$ while another stops after a finite number of gradient steps. With the approximate optimizer, the returned score need not satisfy $\hat R(\hat s)=\inf_{s\in\mathcal S}\hat R(s)$. It is an algorithmic output rather than an exact ERM solution.

</details>

### Question 21

In the statistical setup, $x_n=(X_{n1},\ldots,X_{nD})\in\mathbb R^D$. For a score $s:\mathbb R^D\to\mathbb R^K$, state the dimensions of $s(x_n)$ and of the action output in $K$-class classification.

<!-- Source: statistical setup; feature dimension and score dimension -->

<details>
<summary>Answer</summary>

For each input $x_n\in\mathbb R^D$, the score $s(x_n)$ is a vector in $\mathbb R^K$, with one component per class. The action $a(s(x_n))$ is a class label in $\mathcal C=\{c_1,\ldots,c_K\}$, not a vector in $\mathbb R^K$.

</details>

### Question 22

Diagnose the false claim: 'Because unsupervised learning has no correct answer, there is no mathematical objective one can optimize.' Give a counterexample objective using only $x_1,\ldots,x_N$.

<!-- Source: unsupervised learning; exploratory goals and mathematical objectives -->

<details>
<summary>Answer</summary>

The claim is false. For example, clustering can optimize $\sum_n\|x_n-\mu_{z_n}\|^2$ over assignments $z_n$ and centers $\mu_k$, using only the inputs. What is absent is a supervised target $y_n$ and a loss comparing predictions to true labels, not mathematical objectives altogether.

</details>

### Question 23

For squared-error regression, identify the population-risk minimizer among all measurable functions by completing the conditional-expectation calculation: minimize $\mathbb E[(Y-g(X))^2\mid X=x]$ over a scalar $g(x)$.

<!-- Source: population risk; squared loss and conditional mean as natural gap-filling result -->

<details>
<summary>Answer</summary>

For fixed $x$, write $m(x)=\mathbb E[Y\mid X=x]$. Then $\mathbb E[(Y-a)^2\mid X=x]=\mathbb E[(Y-m(x))^2\mid X=x]+(m(x)-a)^2$, because the cross term has conditional expectation zero. The minimizer is $a=m(x)$. Thus the unrestricted squared-error population-risk minimizer is $g(x)=\mathbb E[Y\mid X=x]$.

</details>

### Question 24

For absolute-error regression, the lecture notes $|y-\hat f(x)|$ as another loss. In the simple distribution $P(Y=0)=0.6$, $P(Y=10)=0.4$ with no features, compute a constant prediction minimizing squared loss and one minimizing absolute loss.

<!-- Source: loss functions; squared loss versus absolute loss -->

<details>
<summary>Answer</summary>

For squared loss, the best constant is the mean: $0(0.6)+10(0.4)=4$. For absolute loss, any median minimizes expected absolute deviation; here $0$ is a median because $P(Y\le0)=0.6\ge1/2$ and $P(Y\ge0)=1$. The two losses choose different constants.

</details>

### Question 26

Models are treated as approximations rather than exact representations. Formalize this by giving a distribution where the best linear score has positive population squared risk.

<!-- Source: statistical perspective; models as approximations and model misspecification -->

<details>
<summary>Answer</summary>

Let $X\sim\operatorname{Uniform}[-1,1]$ and $Y=X^2$. If the candidate class is linear scores $s_{a,b}(x)=a+bx$, the best linear approximation cannot equal $x^2$ for all $x$ because a quadratic is not affine. </details>

### Question 27

Assume a finite candidate class $\mathcal S=\{s_1,\ldots,s_M\}$. Write ERM as an explicit finite comparison of $M$ numbers. What tie issue must be resolved to make $\hat s$ a single function?

<!-- Source: ERM over a candidate class; argmin as a function versus a set -->

<details>
<summary>Answer</summary>

Compute $\hat R(s_j)=N^{-1}\sum_n\ell(y_n,s_j(x_n))$ for each $j=1,\ldots,M$ and choose an index in $\arg\min_j\hat R(s_j)$. If several $s_j$ achieve the same smallest empirical risk, $\arg\min$ is set-valued. A tie-breaking convention is needed to make $\hat s$ a single function.

</details>

### Question 28

A proof sketch says: '$\hat R_N(s)$ is close to $R(s)$ for each fixed $s$, therefore the empirical minimizer over any huge class has small population risk.' Identify the logical gap.

<!-- Source: generalization questions raised by ERM; pointwise versus uniform control -->

<details>
<summary>Answer</summary>

Pointwise closeness for each fixed $s$ does not imply simultaneous closeness over a large data-dependent class. The empirical minimizer is chosen after seeing the data, so it may exploit random fluctuations. One needs stronger control, such as uniform convergence over $\mathcal S$ or complexity restrictions, to justify the conclusion.

</details>

### Question 29

For multi-class scores $s:\mathbb R^D\to\mathbb R^K$, add the same constant $c$ to every component: $s'_k(x)=s_k(x)+c$. Does the argmax action change? Does the cross-entropy loss from the lecture change? Prove both answers.

<!-- Source: classification scores; argmax and cross-entropy invariance to common shifts -->

<details>
<summary>Answer</summary>

The argmax does not change because adding the same constant preserves the ordering of components. For cross-entropy, $p'_y=e^{s_y+c}/\sum_k e^{s_k+c}=e^c e^{s_y}/(e^c\sum_k e^{s_k})=p_y$, so $-\log p'_y=-\log p_y$. Both the action and this loss are invariant to common score shifts.

</details>

### Question 30

Connect the lecture's two guiding questions, methods and evaluation, to ERM. Which parts of $\hat s=\mathrm{Algorithm}(\mathcal D,\ell,\mathcal S,\mathrm{Optimizer})$ belong to construction, and which mathematical object belongs to evaluation?

<!-- Source: methods versus evaluation; ERM algorithm inputs -->

<details>
<summary>Answer</summary>

Construction is the algorithmic map from data, loss, class, and optimizer to a fitted score $\hat s$. Evaluation asks how good that output is, ideally through population risk $R(\hat s)=\mathbb E_P\ell(Y,\hat s(X))$ or an estimate of it. Thus ERM describes a method, while risk or performance estimation addresses evaluation.

</details>
