### Question 1

LASSO stands for least absolute shrinkage and selection operator. Why?
<details>
<summary>Answer</summary>

The penalty
$$
\lambda\|w\|_1=\lambda\sum_j |w_j|
$$
causes shrinkage because it makes large coefficient magnitudes costly, pulling estimates toward zero.

Selection occurs because the $\ell_1$ penalty can set some coefficients exactly equal to zero. A coefficient with $\hat w_j=0$ effectively removes predictor $j$ from the fitted linear model. This *selects* out the corresponding variable. 

</details>


### Question 2

For $w=(3,-1,0,2)$, compute $\|w\|_1$, $\|w\|_2^2$, and the number of "selected" variables.

<details>
<summary>Answer</summary>

The $\ell_1$ norm is
$$
\|w\|_1=|3|+|-1|+|0|+|2|=6.
$$
The squared $\ell_2$ norm is
$$
\|w\|_2^2=3^2+(-1)^2+0^2+2^2=14.
$$
There are three nonzero coefficients, so the number of selected variables is $3$.

</details>


### Question 3

What is the main optimization hurdle for LASSO as compared to ridge?

<details>
<summary>Answer</summary>

Ridge uses the smooth squared $\ell_2$ penalty $\|w\|_2^2$, so the objective is differentiable and leads to normal equations with a closed-form matrix solution.

LASSO uses the $\ell_1$ penalty $\|w\|_1=\sum_j |w_j|$. The absolute value is not differentiable at zero, so the objective is convex but nonsmooth. Because of this nonsmoothness, LASSO generally does not have a simple closed-form matrix formula like ridge.

</details>


### Question 4

State the subdifferential of $|u|$ at $u>0$, $u=0$, and $u<0$.

<details>
<summary>Answer</summary>

The subdifferential is
$$
\partial |u|=\begin{cases}
\{1\},&u>0,\\
[-1,1],&u=0,\\
\{-1\},&u<0.
\end{cases}
$$
At zero, the subdifferential is an entire interval rather than a single slope.

</details>


### Question 5

In coordinate descent, the partial residual for coordinate $j$ is
$$
r_j=y-\sum_{k\ne j}x_kw_k.
$$
Show that if $\hat y=Xw=\sum_k x_kw_k$, then
$$
r_j=y-\hat y+x_jw_j.
$$

<details>
<summary>Answer</summary>

Since
$$
\hat y=\sum_k x_kw_k=x_jw_j+\sum_{k\ne j}x_kw_k,
$$
we have
$$
\sum_{k\ne j}x_kw_k=\hat y-x_jw_j.
$$
Substitute into the definition of $r_j$:
$$
r_j=y-(\hat y-x_jw_j)=y-\hat y+x_jw_j.
$$

</details>


### Question 6

In the coordinate-descent derivation, $c_j=(1/N)x_j^\top r_j$. Give an interpretation of $c_j$.

<details>
<summary>Answer</summary>

The value
$$
c_j=\frac1N x_j^\top r_j
$$
measures the empirical alignment, or correlation up to scaling, between feature $j$ and the remaining residual. If $|c_j|$ is large, feature $j$ explains residual variation not already explained by the other variables.

</details>


### Question 7

Starting from
$$
\frac{1}{2N}\|r_j-x_jw_j\|_2^2+
\lambda|w_j|,
$$
expand the terms that depend on $w_j$ and derive
$$
g(w_j)=\frac12a_jw_j^2-c_jw_j+
\lambda|w_j|,
$$
where $a_j=(1/N)x_j^\top x_j$ and $c_j=(1/N)x_j^\top r_j$. You can drop any term that doesn't depend on $w_j$. 

<details>
<summary>Answer</summary>

Expand the squared term:
$$
\frac{1}{2N}(r_j-x_jw_j)^\top(r_j-x_jw_j)
=
\frac{1}{2N}\left(r_j^\top r_j-2w_jx_j^\top r_j+w_j^2x_j^\top x_j\right).
$$
The term $(1/(2N))r_j^\top r_j$ does not depend on $w_j$, so it can be dropped for minimization. The remaining terms are
$$
-\frac1N w_jx_j^\top r_j+
\frac12\frac1N x_j^\top x_j w_j^2+
\lambda|w_j|.
$$
Using $a_j=(1/N)x_j^\top x_j$ and $c_j=(1/N)x_j^\top r_j$ gives
$$
g(w_j)=\frac12a_jw_j^2-c_jw_j+
\lambda|w_j|.
$$

</details>


### Question 8

For the coordinate objective
$$
g(w_j)=\frac12a_jw_j^2-c_jw_j+
\lambda|w_j|,
$$
derive the subdifferential.

<details>
<summary>Answer</summary>

The differentiable part is
$$
\frac12a_jw_j^2-c_jw_j,
$$
whose derivative is
$$
a_jw_j-c_j.
$$
The nonsmooth part is $\lambda|w_j|$, whose subdifferential is
$$
\lambda\partial|w_j|.
$$
Adding the derivative of the smooth part and the subdifferential of the nonsmooth part gives
$$
\partial g(w_j)=a_jw_j-c_j+
\lambda\partial|w_j|.
$$

</details>


### Question 9

At $w_j=0$, show that
$$
\partial g(0)=[-c_j-\lambda,
-c_j+
\lambda].
$$
Then state the condition on $c_j$ under which $w_j=0$ satisfies the optimality condition.

<details>
<summary>Answer</summary>

At $w_j=0$,
$$
a_jw_j-c_j=-c_j.
$$
Also,
$$
\partial|0|=[-1,1].
$$
Therefore
$$
\partial g(0)=-c_j+
\lambda[-1,1]=[-c_j-
\lambda,-c_j+
\lambda].
$$
The optimality condition is $0\in\partial g(0)$. This holds exactly when
$$
-c_j-
\lambda\le0\le -c_j+
\lambda,
$$
which is equivalent to
$$
|c_j|\le\lambda.
$$

</details>


### Question 10

Derive the positive-coordinate case of the LASSO update: if $c_j>\lambda$, show that the minimizer satisfies

$$
w_j=\frac{c_j-\lambda}{a_j}.
$$

<details>
<summary>Answer</summary>

The coordinate-wise objective has subdifferential

$$
\partial g(w_j)=a_jw_j-c_j+\lambda \partial |w_j|.
$$

A value of $w_j$ is optimal exactly when

$$
0 \in \partial g(w_j).
$$

Assume $c_j>\lambda$. First check whether $w_j=0$ can be optimal. At $w_j=0$,

$$
\partial g(0)=[-c_j-\lambda,,-c_j+\lambda].
$$

Since $c_j>\lambda$, the upper endpoint satisfies $-c_j+\lambda<0$. Therefore this whole interval is negative, so $0\notin\partial g(0)$. Thus $w_j=0$ is not optimal.

Now check whether a negative value can be optimal. If $w_j<0$, then $\partial |w_j|={-1}$, so

$$
\partial g(w_j)={a_jw_j-c_j-\lambda}.
$$

Since $a_j>0$ and $w_j<0$, we have $a_jw_j<0$. Also, $c_j>\lambda>0$. Therefore $a_jw_j-c_j-\lambda<0$, so the subdifferential cannot contain zero for any $w_j<0$.

So the minimizer must be positive. For $w_j>0$, $\partial |w_j|={1}$, so

$$
\partial g(w_j)={a_jw_j-c_j+\lambda}.
$$

To make zero belong to this set, solve

$$
a_jw_j-c_j+\lambda=0.
$$

This gives

$$
w_j=\frac{c_j-\lambda}{a_j}.
$$

Because $c_j>\lambda$ and $a_j>0$, this value is positive. Therefore it is consistent with the positive case and satisfies the subgradient optimality condition. Since $g$ is convex, it is the minimizer.

</details>


### Question 11

Compute $S(c,\lambda)$ for $\lambda=0.5$ and
$$
c\in\{-1.2,-0.5,-0.1,0.7,2.0\}.
$$

<details>
<summary>Answer</summary>

Using soft-thresholding:

- $S(-1.2,0.5)=-1.2+0.5=-0.7$.
- $S(-0.5,0.5)=0$ because $|c|\le\lambda$.
- $S(-0.1,0.5)=0$ because $|c|\le\lambda$.
- $S(0.7,0.5)=0.7-0.5=0.2$.
- $S(2.0,0.5)=2.0-0.5=1.5$.

</details>

### Question 13

Compare ridge and LASSO in the orthonormal-predictor case. If $z_j$ is the OLS coefficient, what does ridge do to $z_j$, and what does LASSO do to $z_j$?

<details>
<summary>Answer</summary>

In the orthonormal case, ridge rescales every coefficient by the same factor:
$$
\hat w^{ridge}_j=\frac{1}{1+
\lambda}z_j
$$
under the lecture's ridge convention.

LASSO applies soft-thresholding:
$$
\hat w^{lasso}_j=S(z_j,\lambda)=\operatorname{sign}(z_j)(|z_j|-
\lambda)_+.
$$
Thus ridge continuously shrinks all coefficients, while LASSO sets coefficients to zero when $|z_j|\le\lambda$ and shrinks the rest by subtracting $\lambda$ from their magnitude.

</details>


### Question 14

For orthonormal predictors, let $z=(2.0,0.4,-1.5)$ and $\lambda=1$. Compute the LASSO solution and the ridge solution using the shrinkage factor $1/(1+
\lambda)$.

<details>
<summary>Answer</summary>

For LASSO,
$$
S(2.0,1)=1.0,
\qquad
S(0.4,1)=0,
\qquad
S(-1.5,1)=-0.5.
$$
So
$$
\hat w^{lasso}=(1.0,0,-0.5).
$$
For ridge, the shrinkage factor is
$$
\frac{1}{1+

$$
Thus
$$
\hat w^{ridge}=(1.0,0.2,-0.75).
$$

</details>


### Question 15

In two dimensions, describe the set
$$
\{(w_1,w_2): |w_1|+|w_2|\le 1\}.
$$
List its four vertices.

<details>
<summary>Answer</summary>

The set is the two-dimensional $\ell_1$ unit ball, which is a diamond. Its boundary consists of the lines
$$
|w_1|+|w_2|=1.
$$
The four vertices are
$$
(1,0),
\quad
(0,1),
\quad
(-1,0),
\quad
(0,-1).
$$
These vertices lie on the coordinate axes.

</details>


### Question 16

Use the geometry from the lecture to explain why LASSO can produce exact zeros while ridge usually does not.

<details>
<summary>Answer</summary>

In the constrained view, ridge uses a round $\ell_2$ ball, while LASSO uses a diamond-shaped $\ell_1$ ball. The LASSO diamond has corners on the coordinate axes.

The solution is where the smallest attainable loss contour touches the constraint region. Because the $\ell_1$ ball has corners on the axes, tangency often occurs at a point where one or more coordinates are exactly zero. The ridge ball is smooth and round, so tangency is less likely to occur exactly on an axis; ridge usually shrinks coefficients without setting them exactly to zero.

</details>


### Question 17

For coefficient vectors $u=(1,1)$ and $v=(2,0)$, compare their $\ell_1$ and squared $\ell_2$ penalties. Use this to explain why LASSO may be indifferent between spreading and concentrating coefficients in some correlated-feature settings.

<details>
<summary>Answer</summary>

For $u=(1,1)$,
$$
\|u\|_1=2,
\qquad
\|u\|_2^2=1^2+1^2=2.
$$
For $v=(2,0)$,
$$
\|v\|_1=2,
\qquad
\|v\|_2^2=2^2+0^2=4.
$$
The two vectors have the same $\ell_1$ norm but different squared $\ell_2$ norms. This shows that an $\ell_1$ penalty may not strongly prefer spreading weight across correlated variables, while ridge's squared $\ell_2$ penalty favors spreading because it penalizes concentration more heavily.

</details>


### Question 18

Describe the qualitative relationship between $\lambda$ and selected model complexity.

<details>
<summary>Answer</summary>

As $\lambda$ increases, the number of nonzero coefficients decreases. Larger $\lambda$ imposes stronger $\ell_1$ regularization, so more coefficients fail the soft-thresholding condition and become zero.

Thus selected model complexity, measured by the number of nonzero coefficients, decreases as $\lambda$ increases.

</details>


### Question 19

Why should predictors be standardized before applying LASSO?

<details>
<summary>Answer</summary>

LASSO penalizes coefficient magnitudes through
$$
\lambda\sum_j |w_j|.
$$
Coefficient magnitude depends on feature units. If one predictor is measured in thousands, a small coefficient can have a large predictive effect. If another predictor is measured in tiny units, the same predictive effect may require a large coefficient.

Without standardization, LASSO may penalize the small-scale variable more heavily simply because its coefficient must be numerically larger. Standardizing puts predictors on comparable scales so the penalty treats them more fairly.

</details>
