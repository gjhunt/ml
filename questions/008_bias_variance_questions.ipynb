# Bias-Variance Tradeoff - Review Questions {#sec-bias-var-questions}

This notebook contains review questions for `lectures/008_bias_variance.ipynb`. Answers are hidden in collapsible sections.


### Question 1

In the lecture's setup, $\hat f$ is random. Identify the source of this randomness and explain why $f^*(x)$ is not random.

<details>
<summary>Answer</summary>

The fitted predictor $\hat f= \hat f_{\mathcal D}$ is random because it is a function of the random training dataset
$$
\mathcal D=\{(X_i,Y_i)\}_{i=1}^n.
$$
Different draws of $\mathcal D$ can produce different fitted functions and therefore different values $\hat f_{\mathcal D}(x)$.

At a fixed input $x$, $f^*(x)$ is the value of the true regression function at that input. The function $f^*$ is treated as part of the data-generating distribution, not as something re-estimated from each sample. Thus $f^*(x)$ is fixed in the pointwise decomposition.

</details>


### Question 2

Starting from the model $Y=f^*(X)+\varepsilon$, show that
$$
(Y-\hat f(x))^2 = (f^*(x)-\hat f(x))^2+2\varepsilon(f^*(x)-\hat f(x))+\varepsilon^2.
$$

<details>
<summary>Answer</summary>

At a fixed test input $X=x$, the model gives
$$
Y=f^*(x)+\varepsilon.
$$
Therefore
$$
(Y-\hat f(x))^2=(f^*(x)+\varepsilon-\hat f(x))^2.
$$
Rearranging inside the square,
$$
(f^*(x)-\hat f(x)+\varepsilon)^2
=(f^*(x)-\hat f(x))^2+2\varepsilon(f^*(x)-\hat f(x))+\varepsilon^2.
$$
These are the prediction-error term, the cross term, and the noise-squared term.

</details>


### Question 3

Justify the step
$$
\mathbb E[\varepsilon(f^*(x)-\hat f(x))]=0.
$$

<details>
<summary>Answer</summary>

At test time, $\varepsilon$ is the noise in the new response at $X=x$, while $\hat f(x)$ is determined by the training dataset $\mathcal D$. The needed assumptions are that
$$
\mathbb E[\varepsilon\mid X=x]=0
$$
and that the test noise is independent of the training data, hence independent of $\hat f(x)$.

Since $f^*(x)$ is fixed and $\hat f(x)$ is independent of $\varepsilon$,
$$
\mathbb E[\varepsilon(f^*(x)-\hat f(x))]
=\mathbb E[\varepsilon]\mathbb E[f^*(x)-\hat f(x)]=0.
$$

</details>


### Question 4

Distinguish the variance term $\mathrm{Var}_{\mathcal D}(\hat f(x))$ from the noise variance $\sigma^2$. What random quantity varies in each case?

<details>
<summary>Answer</summary>

The variance term
$$
\mathrm{Var}_{\mathcal D}(\hat f(x))
$$
measures how the fitted prediction at $x$ changes across different random training datasets $\mathcal D$.

The noise variance
$$
\sigma^2=\mathrm{Var}(\varepsilon\mid X=x)
$$
measures the randomness in the new response $Y$ around the true signal $f^*(x)$, even if the signal were known perfectly. The first is model instability; the second is irreducible output noise.

</details>


### Question 5

Why is the noise term called irreducible error? Show that even if $\hat f(x)=f^*(x)$ then
$$
E[(Y-\hat f(x))^2 \mid X = x] = \sigma^2.
$$

<details>
<summary>Answer</summary>

If $\hat f(x)=f^*(x)$ exactly, then squared prediction error at $x$ is
$$
(Y-\hat f(x))^2=(f^*(x)+\varepsilon-f^*(x))^2=\varepsilon^2.
$$
Taking expectation gives
$$
\mathbb E[\varepsilon^2\mid X=x]=\sigma^2.
$$
Thus even a perfect predictor of the true signal cannot remove the randomness in $Y$ due to $\varepsilon$. That is why $\sigma^2$ is irreducible.

</details>


### Question 6

A learning procedure has high bias but low variance at a point $x$. Describe what repeated fitted models would look like at $x$ relative to $f^*(x)$.

<details>
<summary>Answer</summary>

Across repeated datasets, the fitted values $\hat f_{\mathcal D}(x)$ would be tightly clustered, because variance is low. However, the cluster would be centered far from $f^*(x)$, because bias is high.

So the procedure is stable but systematically wrong at that point.

</details>


### Question 7

At a fixed $x$, suppose
$$
\mathbb E_{\mathcal D}[\hat f(x)]=2.5,
\qquad
f^*(x)=3.0,
\qquad
\mathrm{Var}_{\mathcal D}(\hat f(x))=0.16,
\qquad
\sigma^2=0.09.
$$
Compute the squared bias and $R_f(x)$.

<details>
<summary>Answer</summary>

The bias is
$$
2.5-3.0=-0.5,
$$
so the squared bias is
$$
(-0.5)^2=0.25.
$$
The risk is
$$
R_f(x)=0.25+0.16+0.09=0.50.
$$

</details>


### Question 8

The lecture says flexible models are often low-bias and high-variance. State this claim in the context of KNN by comparing small $k$ and large $k$.

<details>
<summary>Answer</summary>

In KNN regression, small $k$ uses very local averages. This makes the fitted function flexible, so it can adapt to local structure in $f^*$ and often has lower bias. But it also depends strongly on a small number of training points, so it has higher variance across datasets.

Large $k$ averages many points. This stabilizes predictions and lowers variance, but it can smooth over local structure and increase bias.

</details>


### Question 9

Increasing sample size usually decreases variance while bias typically stays about the same. Explain why this statement assumes the model class or fitting rule is not fundamentally changed as $n$ increases.

<details>
<summary>Answer</summary>

Variance measures sensitivity to the random training dataset. With more data, a fixed fitting rule usually becomes less sensitive to any particular sample, so variance tends to decrease.

Bias is determined largely by the systematic limitations of the model class or procedure. If the same kind of model is used as $n$ grows, those approximation limitations may remain similar. But if the model class or tuning rule changes with $n$, bias can change too. Thus the classical statement assumes the modeling procedure is held mostly fixed.

</details>


### Question 10

Suppose $\mathrm{Bias}_f(x)^2=0.36$, $\mathrm{Var}_f(x)=0.04$, and $\sigma^2=0.09$. If a richer model reduces squared bias to $0.09$ but increases variance to $0.20$, which model has lower pointwise risk at $x$?

<details>
<summary>Answer</summary>

The original model has risk
$$
0.36+0.04+0.09=0.49.
$$
The richer model has risk
$$
0.09+0.20+0.09=0.38.
$$
The richer model has lower pointwise risk at $x$, despite its higher variance, because the reduction in squared bias is larger than the increase in variance.

</details>


### Question 11

A student says: "Because variance is bad, we should always choose the lowest-variance model." Diagnose the mistake using the decomposition from the lecture.

<details>
<summary>Answer</summary>

The mistake is ignoring squared bias. The pointwise risk is
$$
R_f(x)=\mathrm{Bias}_f(x)^2+\mathrm{Var}_f(x)+\sigma^2.
$$
A very low-variance model can still have high risk if its average prediction is far from $f^*(x)$. The goal is not to minimize variance alone, but to balance squared bias and variance, since the noise term is not controlled by the model.

</details>


### Question 12

Suppose the learning procedure always predicts
$$
\hat f(x)=0
$$
for every training dataset, while
$$
f^*(x)=2.
$$
Show that this procedure has zero variance but nonzero bias. 

<details>
<summary>Answer</summary>
    
We then have:
$$
\mathrm{Var}_{\mathcal D}(\hat f(x))=0
$$
because the prediction never changes across datasets. But
$$
\mathrm{Bias}_f(x)=\mathbb E[\hat f(x)]-f^*(x)=0-2=-2,
$$
so the squared bias is $4$.

</details>


### Question 13

Does the bias-variance decomposition work exactly for the 0-1 loss? What about absolute error loss?

<details>
<summary>Answer</summary>

No. The derivation relies on expanding a square:
$$
(Y-\hat f(x))^2.
$$
This expansion produces terms that can be identified as squared bias, variance, and noise, with cross terms that vanish under the stated assumptions.

For other losses, such as absolute error or 0-1 loss, there is no analogous simple quadratic expansion into these exact three terms. The bias-variance idea may still be useful qualitatively, but the displayed formula is specific to squared-error regression.

</details>


### Question 14

A model has pointwise risk $R_f(x)=1.7$, squared bias $0.8$, and variance $0.6$. Under the lecture's decomposition, what is $\sigma^2$?

<details>
<summary>Answer</summary>

Using
$$
R_f(x)=\mathrm{Bias}_f(x)^2+\mathrm{Var}_f(x)+\sigma^2,
$$
we get
$$
\sigma^2=1.7-0.8-0.6=0.3.
$$

</details>


### Question 15

At a point $x$, three procedures have the following squared bias and variance, with common $\sigma^2=0.1$:

| procedure | squared bias | variance |
| --- | ---: | ---: |
| A | 0.50 | 0.05 |
| B | 0.20 | 0.20 |
| C | 0.05 | 0.60 |

Compute the risk for each procedure and identify the best one at this $x$.

<details>
<summary>Answer</summary>

Procedure A has risk
$$
0.50+0.05+0.10=0.65.
$$
Procedure B has risk
$$
0.20+0.20+0.10=0.50.
$$
Procedure C has risk
$$
0.05+0.60+0.10=0.75.
$$
Procedure B is best at this $x$ because it has the smallest total risk.

</details>


### Question 16

A false proof says: "The cross term vanishes because $\mathbb E[\hat f(x)-f^*(x)]=0$." Identify why this proof is wrong in general.

<details>
<summary>Answer</summary>

The statement
$$
\mathbb E[\hat f(x)-f^*(x)]=0
$$
means the estimator is unbiased at $x$. The lecture does not assume unbiasedness; in fact, bias may be nonzero and is one of the terms being measured.

The first cross term involving $\varepsilon$ vanishes because the test noise has mean zero and is independent of $\hat f(x)$. The second cross term in the bias-variance decomposition vanishes because $\hat f(x)-\mathbb E[\hat f(x)]$ is centered, not because $\hat f(x)-f^*(x)$ is centered.

</details>


### Question 17

Suppose the irreducible noise variance is $\sigma^2=1$. At a fixed test point $x_0$, two prediction procedures have the following bias and variance:

| Procedure | Bias$^2$ | Variance |
| --------- | -------: | -------: |
| A         |      $0$ |      $4$ |
| B         |      $3$ |      $1$ |

Using the bias-variance decomposition, show that the two procedures have the same overall risk at $x_0$, but different bias-variance behavior.

<details>
<summary>Answer</summary>

For Procedure A, the risk is

$$
0+4+1=5.
$$

So Procedure A has total risk $5$.

For Procedure B, the risk is

$$
3+1+1=5.
$$

So Procedure B also has total risk $5$.

Therefore, the two procedures have the same overall risk at $x_0$, even though the source of the error is different.

</details>
