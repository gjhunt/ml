# Bias–Variance Tradeoff — Review Questions {#sec-bias-var-questions}



### Question 1

In the lecture's setup, $\hat f_{\mathcal D}$ is random. Identify the source of this randomness and explain why $f^*(x)$ is fixed.

<details>
<summary>Answer</summary>

The fitted predictor $\hat f_{\mathcal D}=A(\mathcal D)$ is random because it is a function of the random training dataset
$$
\mathcal D=\{(x_n,y_n)\}_{n=1}^N.
$$
Different draws of $\mathcal D$ can produce different fitted functions and therefore different values $\hat f_{\mathcal D}(x)$.

At a fixed input $x$, $f^*(x)=\mathbb E[Y\mid X=x]$ is determined by the population distribution. It is not re-estimated for each training sample, so it is fixed in the pointwise decomposition.

</details>


### Question 2

Starting from the model $Y=f^*(X)+\varepsilon$, show that
$$
(Y-\hat f_{\mathcal D}(x))^2 = (f^*(x)-\hat f_{\mathcal D}(x))^2+2\varepsilon(f^*(x)-\hat f_{\mathcal D}(x))+\varepsilon^2.
$$

<details>
<summary>Answer</summary>

At a fixed test input $X=x$, the model gives
$$
Y=f^*(x)+\varepsilon.
$$
Therefore
$$
(Y-\hat f_{\mathcal D}(x))^2=(f^*(x)+\varepsilon-\hat f_{\mathcal D}(x))^2.
$$
Rearranging inside the square,
$$
(f^*(x)-\hat f_{\mathcal D}(x)+\varepsilon)^2
=(f^*(x)-\hat f_{\mathcal D}(x))^2+2\varepsilon(f^*(x)-\hat f_{\mathcal D}(x))+\varepsilon^2.
$$
These are the prediction-error term, the cross term, and the noise-squared term.

</details>


### Question 3

Justify the step
$$
\mathbb E[\varepsilon(f^*(x)-\hat f_{\mathcal D}(x))]=0.
$$

<details>
<summary>Answer</summary>

At test time, $\varepsilon$ is the noise in the new response at $X=x$, while $\hat f_{\mathcal D}(x)$ is determined by the training dataset $\mathcal D$. We assume
$$
\mathbb E[\varepsilon\mid X=x]=0
$$
and that the new test observation is independent of the training data. Using conditional expectation,

$$
\begin{aligned}
&\mathbb E_{\mathcal D,\varepsilon\mid x}[\varepsilon(f^*(x)-\hat f_{\mathcal D}(x))]\\
&=\mathbb E_{\mathcal D}\left[(f^*(x)-\hat f_{\mathcal D}(x))
\mathbb E[\varepsilon\mid X=x,\mathcal D]\right]=0.
\end{aligned}
$$

</details>


### Question 4

Distinguish the variance term $\operatorname{Var}_{\mathcal D}(\hat f_{\mathcal D}(x))$ from the noise variance $\sigma^2$. What random quantity varies in each case?

<details>
<summary>Answer</summary>

The variance term
$$
\operatorname{Var}_{\mathcal D}(\hat f_{\mathcal D}(x))
$$
measures how the fitted prediction at $x$ changes across different random training datasets $\mathcal D$.

The noise variance
$$
\sigma^2=\operatorname{Var}(\varepsilon\mid X=x)
$$
measures the randomness in the new response $Y$ around the true signal $f^*(x)$, even if the signal were known perfectly. The first is model instability; the second is irreducible output noise.

</details>


### Question 5

Why is the noise term called irreducible error? Show that even if $\hat f_{\mathcal D}(x)=f^*(x)$, then
$$
\mathbb E[(Y-\hat f_{\mathcal D}(x))^2 \mid X=x] = \sigma^2.
$$

<details>
<summary>Answer</summary>

If $\hat f_{\mathcal D}(x)=f^*(x)$ exactly, then squared prediction error at $x$ is
$$
(Y-\hat f_{\mathcal D}(x))^2=(f^*(x)+\varepsilon-f^*(x))^2=\varepsilon^2.
$$
Taking expectation gives
$$
\mathbb E[\varepsilon^2\mid X=x]=\sigma^2.
$$
Thus even a perfect predictor of the true signal cannot remove the randomness in $Y$ due to $\varepsilon$. That is why $\sigma^2$ is irreducible.

</details>


### Question 6

A learning procedure has high bias but low variance at a point $x$. Describe what repeated fitted models would look like at $x$ relative to $f^*(x)$.

<details>
<summary>Answer</summary>

Across repeated datasets, the fitted values $\hat f_{\mathcal D}(x)$ would be tightly clustered, because variance is low. However, the cluster would be centered far from $f^*(x)$, because bias is high.

So the procedure is stable but systematically wrong at that point.

</details>


### Question 7

At a fixed $x$, suppose
$$
\mathbb E_{\mathcal D}[\hat f_{\mathcal D}(x)]=2.5,
\qquad
f^*(x)=3.0,
\qquad
\operatorname{Var}_{\mathcal D}(\hat f_{\mathcal D}(x))=0.16,
\qquad
\sigma^2=0.09.
$$
Compute the squared bias and $R(x)$.

<details>
<summary>Answer</summary>

The bias is
$$
2.5-3.0=-0.5,
$$
so the squared bias is
$$
(-0.5)^2=0.25.
$$
The risk is
$$
R(x)=0.25+0.16+0.09=0.50.
$$

</details>


### Question 8

The lecture says flexible models are often low-bias and high-variance. State this claim for nearest-neighbor regression by comparing small $m$ and large $m$.

<details>
<summary>Answer</summary>

With small $m$, nearest-neighbor regression uses very local averages. This makes the fitted function flexible, so it can adapt to local structure in $f^*$ and often has lower bias. But it also depends strongly on a small number of training points, so it has higher variance across datasets.

Large $m$ averages many points. This stabilizes predictions and lowers variance, but it can smooth over local structure and increase bias.

</details>


### Question 9

Why does increasing the sample size $N$ often reduce variance? Explain why bias can nevertheless change as $N$ increases.

<details>
<summary>Answer</summary>

Variance measures sensitivity to the random training dataset. With more data, a fitted predictor often becomes less sensitive to any particular observation, so variance tends to decrease.

Bias may remain similar when the model class and fitting behavior remain similar, but this is not guaranteed. The effective behavior of a procedure can change with $N$. For example, in nearest-neighbor regression, more observations can make a neighborhood around $x$ more local and reduce bias.

</details>


### Question 10

Suppose $\operatorname{Bias}(x)^2=0.36$, $\operatorname{Var}(x)=0.04$, and $\sigma^2=0.09$. If a richer model reduces squared bias to $0.09$ but increases variance to $0.20$, which model has lower pointwise risk at $x$?

<details>
<summary>Answer</summary>

The original model has risk
$$
0.36+0.04+0.09=0.49.
$$
The richer model has risk
$$
0.09+0.20+0.09=0.38.
$$
The richer model has lower pointwise risk at $x$, despite its higher variance, because the reduction in squared bias is larger than the increase in variance.

</details>


### Question 11

A student says: "Because variance is bad, we should always choose the lowest-variance model." Diagnose the mistake using the decomposition from the lecture.

<details>
<summary>Answer</summary>

The mistake is ignoring squared bias. The pointwise risk is
$$
R(x)=\operatorname{Bias}(x)^2+\operatorname{Var}(x)+\sigma^2.
$$
A very low-variance procedure can still have high risk if its average prediction is far from $f^*(x)$. The goal is to balance squared bias and variance, since the noise term is not controlled by the fitting procedure.

</details>


### Question 12

Suppose the learning procedure always predicts
$$
\hat f_{\mathcal D}(x)=0
$$
for every training dataset, while
$$
f^*(x)=2.
$$
Show that this procedure has zero variance but nonzero bias. 

<details>
<summary>Answer</summary>
    
We then have:
$$
\operatorname{Var}_{\mathcal D}(\hat f_{\mathcal D}(x))=0
$$
because the prediction never changes across datasets. But
$$
\operatorname{Bias}(x)=\mathbb E_{\mathcal D}[\hat f_{\mathcal D}(x)]-f^*(x)=0-2=-2,
$$
so the squared bias is $4$.

</details>


### Question 13

Does the lecture's displayed identity—squared bias plus variance plus noise—hold unchanged for 0-1 loss or absolute-error loss?

<details>
<summary>Answer</summary>

No. The derivation relies on expanding a square:
$$
(Y-\hat f_{\mathcal D}(x))^2.
$$
This expansion produces terms that can be identified as squared bias, variance, and noise, with cross terms that vanish under the stated assumptions.

For absolute error or 0-1 loss, there is no analogous simple quadratic expansion into these exact three terms. Other decompositions can be defined, and the bias–variance idea remains useful qualitatively, but this displayed identity is specific to squared-error regression.

</details>


### Question 14

At a point $x$, three procedures have the following squared bias and variance, with common $\sigma^2=0.1$:

| procedure | squared bias | variance |
| --- | ---: | ---: |
| A | 0.50 | 0.05 |
| B | 0.20 | 0.20 |
| C | 0.05 | 0.60 |

Compute the risk for each procedure and identify the best one at this $x$.

<details>
<summary>Answer</summary>

Procedure A has risk
$$
0.50+0.05+0.10=0.65.
$$
Procedure B has risk
$$
0.20+0.20+0.10=0.50.
$$
Procedure C has risk
$$
0.05+0.60+0.10=0.75.
$$
Procedure B is best at this $x$ because it has the smallest total risk.

</details>


### Question 15

A false proof says: "The cross term vanishes because $\mathbb E_{\mathcal D}[\hat f_{\mathcal D}(x)-f^*(x)]=0$." Identify why this proof is wrong in general.

<details>
<summary>Answer</summary>

The statement
$$
\mathbb E_{\mathcal D}[\hat f_{\mathcal D}(x)-f^*(x)]=0
$$
means the estimator is unbiased at $x$. The lecture does not assume unbiasedness; in fact, bias may be nonzero and is one of the terms being measured.

The first cross term involving $\varepsilon$ vanishes because the test noise has conditional mean zero and the test observation is independent of $\mathcal D$. The second cross term vanishes because $\hat f_{\mathcal D}(x)-\mathbb E_{\mathcal D}[\hat f_{\mathcal D}(x)]$ is centered, not because $\hat f_{\mathcal D}(x)-f^*(x)$ is centered.

</details>


### Question 16

Suppose the irreducible noise variance is $\sigma^2=1$. At a fixed test point $x_0$, two prediction procedures have the following bias and variance:

| Procedure | Bias$^2$ | Variance |
| --------- | -------: | -------: |
| A         |      $0$ |      $4$ |
| B         |      $3$ |      $1$ |

Using the bias–variance decomposition, show that the two procedures have the same pointwise risk at $x_0$, but different bias–variance behavior.

<details>
<summary>Answer</summary>

For Procedure A, the risk is

$$
0+4+1=5.
$$

So Procedure A has total risk $5$.

For Procedure B, the risk is

$$
3+1+1=5.
$$

So Procedure B also has total risk $5$.

Therefore, the two procedures have the same pointwise risk at $x_0$, even though the source of the error is different.

</details>
