# Principal Component Analysis (PCA) - Review Questions {#sec-pca-questions}


### Question 1

List the dimensions of the four objects $W$, $P_W$, $Z=XW$, and $XP_W$ when $X\in\mathbb R^{N\times D}$ and $W\in\mathbb R^{D\times q}$ has full column rank. State which of $Z$ and $XP_W$ is written in principal-component coordinates and which is written in the original variable coordinates.

<details>
<summary>Answer</summary>

The matrix $W$ has shape $D\times q$ and its columns form a basis for the chosen $q$-dimensional subspace. The projection matrix

$$
P_W=W(W^TW)^{-1}W^T
$$

has shape $D\times D$ and projects row vectors in $\mathbb R^D$ onto $\operatorname{Col}(W)$.

The score matrix

$$
Z=XW
$$

has shape $N\times q$ and gives the data in the new $q$-dimensional coordinates. The projected data

$$
XP_W
$$

has shape $N\times D$ and gives the projected observations written back in the original $D$ coordinates.

</details>


### Question 2

Verify the lecture's orthonormal-basis shortcut. If the columns of $W\in\mathbb R^{D\times q}$ are orthonormal, compute $W^TW$ and prove that $P_W=WW^T$ is both symmetric $P_W^T = P_W$ and idempotent $P_W^2 = P_W$.

<details>
<summary>Answer</summary>

Orthonormal columns mean $w_i^T w_j=0$ for $i\ne j$ and $w_i^T w_i=1$, so

$$
W^TW=I_q.
$$

Therefore

$$
P_W=W(W^TW)^{-1}W^T=W I_q W^T=WW^T.
$$

It is symmetric because

$$
(WW^T)^T=WW^T.
$$

It is idempotent because

$$
(WW^T)(WW^T)=W(W^TW)W^T=W I_q W^T=WW^T.
$$

Symmetry and idempotence are the characteristic algebraic properties of an orthogonal projection matrix.

</details>


### Question 3

Write the $n$th observation's $i$th principal component score (i.e. $Z_{ni}$) in terms of the row vector $x_n\in\mathbb R^D$ and the loading vector $w_i$. 

<details>
<summary>Answer</summary>

The $i$th score for observation $n$ is the $n$th entry of $Z_i=Xw_i$, so

$$
z_{ni}=x_n w_i.
$$

Writing $x_n=(X_{n1},\ldots,X_{nD})$ and $w_i=(w_{1i},\ldots,w_{Di})^T$ gives

$$
z_{ni}=\sum_{j=1}^D X_{nj}w_{ji}.
$$

At the variable level,

$$
Z_i=Xw_i=w_{1i}X_1+w_{2i}X_2+\cdots+w_{Di}X_D.
$$

Thus each principal component score variable is a linear combination of the original columns of $X$.

</details>


### Question 4

Why does the PCA optimization constrain $\|w\|_2=1$? Show that without this constraint 
$$
\operatorname{Var}(Xw)
$$
can be made arbitrarily large. (Hint: consider $w=c w_0$ for a unit vector $w_0$ and real number $c$.)

<details>
<summary>Answer</summary>

For centered data,

$$
\operatorname{Var}(Xw)=w^T S_X w.
$$

If $w$ has positive projected variance, then for any scalar $c$,

$$
\operatorname{Var}(X(cw))=(cw)^T S_X(cw)=c^2 w^T S_Xw.
$$

Letting $|c|\to\infty$ makes the variance arbitrarily large without changing the direction of $w$. The unit-length constraint removes this artificial scaling freedom, so the optimization chooses a direction rather than a vector length.

</details>


### Question 5

Starting from the sample variance formula for the centered score variable $z=Xw$, derive

$$
\operatorname{Var}(Xw)=w^T S_X w,
\qquad
S_X=\frac{1}{N-1}X^T X.
$$

<details>
<summary>Answer</summary>

Since $X$ is centered, $z=Xw$ is also centered. Its sample variance is

$$
\operatorname{Var}(z)=\frac{1}{N-1}z^Tz.
$$

Substitute $z=Xw$:

$$
\operatorname{Var}(Xw)
=\frac{1}{N-1}(Xw)^T(Xw)
=\frac{1}{N-1}w^T X^T X w.
$$

Using

$$
S_X=\frac{1}{N-1}X^TX,
$$

we get

$$
\operatorname{Var}(Xw)=w^T S_Xw.
$$

</details>


### Question 6

For two centered score variables $Xu$ and $Xv$, derive a formula for their sample covariance. 

<details>
<summary>Answer</summary>

The sample covariance is

$$
\operatorname{Cov}(Xu,Xv)
=\frac{1}{N-1}(Xu)^T(Xv).
$$

Rearranging gives

$$
\operatorname{Cov}(Xu,Xv)
=\frac{1}{N-1}u^T X^T X v
=u^T S_X v.
$$


</details>


### Question 7

Give noncentered vectors $x,y\in\mathbb R^3$ with sample covariance zero but raw dot product $x^Ty\ne 0$.

<details>
<summary>Answer</summary>

Take

$$
x=(1,2,3)^T,
\qquad
 y=(2,1,2)^T.
$$

Their means are $\bar x=2$ and $\bar y=5/3$, so the centered vectors are

$$
x-\bar x\mathbf 1=(-1,0,1)^T,
$$

and

$$
y-\bar y\mathbf 1=(1/3,-2/3,1/3)^T.
$$

Their centered dot product is

$$
(-1)(1/3)+0(-2/3)+1(1/3)=0,
$$

so their sample covariance is zero. But the raw dot product is

$$
x^Ty=1\cdot2+2\cdot1+3\cdot2=10\ne0.
$$

So uncorrelated means orthogonal only after centering.

</details>


### Question 8

Use the covariance formula to prove that every eigenvalue of $S_X$ is nonnegative.

<details>
<summary>Answer</summary>

Let $v\ne0$ be an eigenvector of $S_X$ with eigenvalue $\lambda$. Then

$$
S_Xv=\lambda v.
$$

Multiplying by $v^T$ gives

$$
v^T S_X v=\lambda v^Tv=\lambda\|v\|_2^2.
$$

But

$$
v^T S_Xv=\operatorname{Var}(Xv)\ge0.
$$

Since $\|v\|_2^2>0$, it follows that

$$
\lambda\ge0.
$$

</details>


### Question 9

Complete the main algebra in the simplified Rayleigh quotient theorem: if $A$ is symmetric with orthonormal eigenvectors $v_1,\ldots,v_D$ and $w=\sum_i c_i v_i$, derive the formulas for $\|w\|_2^2$ and $w^TAw$ used in the proof.

<details>
<summary>Answer</summary>

Because the eigenvectors are orthonormal,

$$
\|w\|_2^2=w^Tw
=\left(\sum_i c_i v_i\right)^T\left(\sum_j c_j v_j\right)
=\sum_i\sum_j c_i c_j v_i^T v_j
=\sum_i c_i^2.
$$

Also, $Av_i=\lambda_i v_i$, so

$$
Aw=\sum_i c_i\lambda_i v_i.
$$

Therefore

$$
w^TAw
=\left(\sum_i c_i v_i\right)^T\left(\sum_j c_j\lambda_j v_j\right)
=\sum_i\lambda_i c_i^2.
$$

If $\|w\|_2=1$, then $\sum_i c_i^2=1$, so $w^TAw$ is a weighted average of the eigenvalues with weights $c_i^2$.

</details>


### Question 10

In the restricted Rayleigh quotient theorem, $w$ must satisfy $w\perp v_1,\ldots,v_{k-1}$. Translate this condition into restrictions on the coefficients $c_i$ in $w=\sum_i c_i v_i$, and explain why the maximum value is $\lambda_k$.

<details>
<summary>Answer</summary>

Since the eigenvectors are orthonormal,

$$
w\perp v_j
\quad\Longleftrightarrow\quad
w^T v_j=0
\quad\Longleftrightarrow\quad
c_j=0.
$$

Thus the constraints imply

$$
c_1=\cdots=c_{k-1}=0.
$$

For a unit vector satisfying these restrictions,

$$
w^TAw=\lambda_k c_k^2+\lambda_{k+1}c_{k+1}^2+\cdots+\lambda_Dc_D^2.
$$

The remaining weights sum to one, and the largest remaining eigenvalue is $\lambda_k$. Therefore

$$
w^TAw\le \lambda_k,
$$

with equality when $c_k=1$ and all other remaining coefficients are zero, i.e. $w=v_k$ up to sign.

</details>


### Question 11

The lecture turns the condition $\operatorname{Cov}(Xw,Xw_1)=0$ into $w\perp v_1$. How?

<details>
<summary>Answer</summary>

Since $w_1=v_1$ and $S_Xv_1=\lambda_1v_1$,

$$
\operatorname{Cov}(Xw,Xw_1)=w^T S_X v_1
=w^T(\lambda_1v_1)
=\lambda_1 w^Tv_1.
$$

If $\lambda_1>0$, this covariance is zero exactly when

$$
w^Tv_1=0.
$$

</details>


### Question 12

Derive the covariance matrix of the first $q$ PCA scores and show why it is diagonal when $W_q=[v_1\;v_2\;\cdots\;v_q]$.

<details>
<summary>Answer</summary>

The score matrix is

$$
Z_q=XW_q.
$$

Because $X$ is centered, the sample covariance matrix of the scores is

$$
S_{Z_q}=\frac{1}{N-1}Z_q^T Z_q.
$$

Substitute $Z_q=XW_q$:

$$
S_{Z_q}
=\frac{1}{N-1}W_q^T X^TX W_q
=W_q^T S_X W_q.
$$

The $(i,j)$ entry is

$$
v_i^T S_X v_j=v_i^T(\lambda_j v_j)=\lambda_j v_i^T v_j.
$$

This equals $\lambda_i$ when $i=j$ and $0$ when $i\ne j$. Therefore

$$
S_{Z_q}=\operatorname{diag}(\lambda_1,\ldots,\lambda_q).
$$

</details>


### Question 13

Check the lecture's claim that the PCA score variables are centered whenever the columns of $X$ are centered.

<details>
<summary>Answer</summary>

Let $\mathbf 1\in\mathbb R^N$ be the vector of all ones. The columns of $X$ are centered exactly when

$$
\mathbf 1^T X=0^T.
$$

For $Z_q=XW_q$,

$$
\mathbf 1^T Z_q
=\mathbf 1^T XW_q
=(\mathbf 1^T X)W_q
=0^T W_q
=0^T.
$$

So every column of $Z_q$ has sample mean zero.

</details>


### Question 14

Take the lecture's definition of total variance across the original centered variables and prove

$$
\sum_{j=1}^D \widehat{\operatorname{Var}}(X_j)=\operatorname{tr}(S_X)=\sum_{j=1}^D\lambda_j.
$$

<details>
<summary>Answer</summary>

The diagonal entries of $S_X$ are the sample variances of the original variables:

$$
(S_X)_{jj}=\widehat{\operatorname{Var}}(X_j).
$$

Therefore

$$
\sum_{j=1}^D \widehat{\operatorname{Var}}(X_j)=\sum_{j=1}^D (S_X)_{jj}=\operatorname{tr}(S_X).
$$

For any square matrix, the trace equals the sum of its eigenvalues. 

$$
\operatorname{tr}(S_X)=\sum_{j=1}^D\lambda_j.
$$

</details>


### Question 15

Using the centered data matrix with rows $(1,0)$, $(-1,0)$, $(0,2)$, and $(0,-2)$, compute $S_X$, the two PCA directions, their eigenvalues, and the proportion of variance captured by the first PC.

<details>
<summary>Answer</summary>

Here $N=4$ and

$$
X^TX=\begin{bmatrix}2&0\\0&8\end{bmatrix}.
$$

Thus

$$
S_X=\frac{1}{3}\begin{bmatrix}2&0\\0&8\end{bmatrix}
=\begin{bmatrix}2/3&0\\0&8/3\end{bmatrix}.
$$

The largest eigenvalue is

$$
\lambda_1=8/3
$$

with direction $v_1=(0,1)^T$ up to sign. The second eigenvalue is

$$
\lambda_2=2/3
$$

with direction $v_2=(1,0)^T$ up to sign.

The first PC captures

$$
\frac{8/3}{8/3+2/3}=\frac{8}{10}=0.8
$$

of the total variance.

</details>


### Question 16

A covariance matrix has tied eigenvalues, for example

$$
S=\operatorname{diag}(2,2,1).
$$

Describe exactly what is nonunique about the PCA directions. What remains well-defined if $q=2$?

<details>
<summary>Answer</summary>

The first two eigenvalues are tied, so every unit vector in

$$
\operatorname{span}(e_1,e_2)
$$

is a leading eigenvector with eigenvalue $2$. If $q=1$, the first PC direction is not unique: any line through the origin inside that two-dimensional eigenspace is optimal.

If $q=2$, the individual basis vectors are still not unique, because any orthonormal basis of $\operatorname{span}(e_1,e_2)$ works. However, the two-dimensional subspace itself is well-defined, so the projection onto that subspace is unique.

</details>


### Question 17

Changing a loading vector from $v_i$ to $-v_i$ is allowed in PCA. State what happens to the score $z_i=Xv_i$ and its variance.

<details>
<summary>Answer</summary>

If $v_i$ is replaced by $-v_i$, then the score becomes

$$
X(-v_i)=-Xv_i=-z_i.
$$

The variance is unchanged because

$$
\operatorname{Var}(-z_i)=\operatorname{Var}(z_i).
$$

Thus the sign of a loading vector is arbitrary. Only relative signs within a loading vector have interpretive meaning after a sign convention has been chosen.

</details>


### Question 18

If every column has been centered and scaled to sample variance $1$, what are the diagonal entries and trace of $S_X$? Why does this matter when variables use different units?

<details>
<summary>Answer</summary>

After sample standardization, each variable has sample variance $1$, so

$$
(S_X)_{jj}=1
$$

for every $j$. The covariance matrix is then the sample correlation matrix, and

$$
\operatorname{tr}(S_X)=D.
$$

This matters because variables measured in large numerical units can otherwise dominate the covariance matrix simply due to scale. Standardizing puts all variables on a common variance scale before PCA chooses high-variance directions.

</details>
