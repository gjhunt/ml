# Training Neural Networks with Backpropagation - Review Questions {#sec-backprop-questions}


### Question 1

State what remains the same and what changes when the lecture moves from linear/logistic regression to neural-network training. Your answer should mention the empirical risk, the score function, and the optimization problem.

<details>
<summary>Answer</summary>

The ERM structure is unchanged: choose a score function, choose a loss, and minimize

$$
\hat R(\theta)=\frac1N\sum_{n=1}^N \ell(y_n,s_\theta(x_n)).
$$

What changes is the score class. Instead of a linear score, the model uses a layered composition

$$
s_\theta(x)=s_L(s_{L-1}(\cdots s_1(x)\cdots)),
$$

with many parameters across layers. The resulting optimization problem is usually nonconvex and generally has no closed-form solution, so gradient-based iterative methods are used.

</details>


### Question 2

For the one-hidden-unit model

$$
a_n=wx_n+b,\qquad h_n=\sigma(a_n),\qquad s_n=vh_n+c,
$$

with empirical risk

$$
\hat{R}(\theta)=\frac{1}{N}\sum_{n=1}^N \frac{1}{2}(y_n-s_n)^2
$$

derive the gradients

$$
\frac{\partial \hat{R}}{\partial v},\qquad
\frac{\partial \hat{R}}{\partial c},\qquad
\frac{\partial \hat{R}}{\partial w},\qquad
\frac{\partial \hat{R}}{\partial b}
$$

using the chain rule.

Hint: Work first with one observation,

$$
\ell_n=\frac{1}{2}(y_n-s_n)^2.
$$

Use the paths

$$
\ell_n \to s_n \to v,\qquad
\ell_n \to s_n \to c,\qquad
\ell_n \to s_n \to h_n \to a_n \to w,\qquad
\ell_n \to s_n \to h_n \to a_n \to b.
$$

Compute the intermediate derivatives

$$
\frac{\partial \ell_n}{\partial s_n},\quad
\frac{\partial s_n}{\partial v},\quad
\frac{\partial s_n}{\partial c},\quad
\frac{\partial s_n}{\partial h_n},\quad
\frac{\partial h_n}{\partial a_n},\quad
\frac{\partial a_n}{\partial w},\quad
\frac{\partial a_n}{\partial b},
$$

then average over $n=1,...,N$.

<details>
<summary>Answer</summary>

Let

$$
e_n=s_n-y_n.
$$

Since

$$
\frac{\partial}{\partial s_n}\left(\frac{1}{2}(y_n-s_n)^2\right)
=s_n-y_n=e_n,
$$

we have

$$
\frac{\partial \hat{R}}{\partial v}
=\frac{1}{N}\sum_{n=1}^N e_n h_n,
\qquad
\frac{\partial \hat{R}}{\partial c}
=\frac{1}{N}\sum_{n=1}^N e_n.
$$

Also,

$$
\frac{\partial s_n}{\partial h_n}=v,
\qquad
\frac{\partial h_n}{\partial a_n}=h_n(1-h_n),
\qquad
\frac{\partial a_n}{\partial w}=x_n,
\qquad
\frac{\partial a_n}{\partial b}=1.
$$

Therefore

$$
\frac{\partial \hat{R}}{\partial w}
=\frac{1}{N}\sum_{n=1}^N e_n v h_n(1-h_n)x_n,
\qquad
\frac{\partial \hat{R}}{\partial b}
=\frac{1}{N}\sum_{n=1}^N e_n v h_n(1-h_n).
$$

</details>


### Question 3

Recompute the forward pass for

$$
x=2,\quad y=1,\quad w=0.5,\quad b=0.1,\quad v=1.5,\quad c=-0.2,

using $a=wx+b$, $h=\sigma(a)$, $s=vh+c$, and $L=\frac12(y-s)^2$.

Thus

The score is

$$
s=(1.5)(0.7503)-0.2\approx 0.9254.
$$

Finally,

$$
L=\frac12(1-0.9254)^2\approx 0.0028.
$$

### Question 5

After the backward pass in Question 5, gradient descent with learning rate $\eta>0$ updates $w$ by

$$
w^{(t+1)}=w^{(t)}-\eta \frac{\partial L}{\partial w}.
$$

Does this step increase or decrease $w$ in the numerical example? 

<details>
<summary>Answer</summary>

Here

$$
\frac{\partial L}{\partial w}\approx -0.0419<0.
$$

Therefore

$$
w^{(t+1)}=w^{(t)}-\eta(-0.0419)=w^{(t)}+0.0419\eta.
$$

For any $\eta>0$, the update increases $w$. This is consistent with moving opposite the gradient: a negative derivative means increasing $w$ locally reduces the loss.

</details>


### Question 6

Classify the update rule by its batch size. For a dataset of size $N$, identify which case is full-batch gradient descent, which is strict stochastic gradient descent, and which is mini-batch gradient descent:

$$
|B_t|=N,\qquad |B_t|=1,\qquad 1<|B_t|<N.
$$

<details>
<summary>Answer</summary>

If $|B_t|=N$, the update uses every training observation, so it is full-batch gradient descent.

If $|B_t|=1$, the update uses one randomly selected observation, so it is stochastic gradient descent in the strict sense.

If $1<|B_t|<N$, the update uses a subset larger than one but smaller than the whole dataset, so it is mini-batch gradient descent. As the lecture notes, neural-network practice often uses "SGD" broadly to include this mini-batch case.

</details>


### Question 7

Assume a mini-batch of size $B$ is sampled uniformly with replacement from the training indices ${1,\dots,N}$. Let $k_1,\dots,k_B$ denote the sampled indices, and define the mini-batch gradient as

$$
\hat g_B=\frac{1}{B}\sum_{j=1}^B \nabla_\theta \ell(y_{k_j},s_\theta(x_{k_j})).
$$

Show that the expected value of this mini-batch gradient equals the full empirical-risk gradient:

$$
\mathbb E[\hat g_B]=\nabla_\theta \hat R(\theta).
$$

<details>
<summary>Answer</summary>

For each training observation, define its individual gradient contribution as

$$
g_i=\nabla_\theta \ell(y_i,s_\theta(x_i)).
$$

Since $k_1,\dots,k_B$ are sampled uniformly with replacement from ${1,\dots,N}$, each sampled index $k_j$ has the same distribution. Therefore, for each draw,

$$
\mathbb E[g_{k_j}]=\frac{1}{N}\sum_{i=1}^N g_i.
$$

The mini-batch gradient can be written as

$$
\hat g_B=\frac{1}{B}\sum_{j=1}^B g_{k_j}.
$$

By linearity of expectation,

$$
\mathbb E[\hat g_B]=\frac{1}{B}\sum_{j=1}^B \mathbb E[g_{k_j}].
$$

Substituting the expression for each $\mathbb E[g_{k_j}]$ gives

$$
\mathbb E[\hat g_B]=\frac{1}{B}\sum_{j=1}^B \left(\frac{1}{N}\sum_{i=1}^N g_i\right).
$$

Since the same quantity is added $B$ times,

$$
\mathbb E[\hat g_B]=\frac{1}{N}\sum_{i=1}^N g_i.
$$

Now,

$$
\hat R(\theta)=\frac{1}{N}\sum_{i=1}^N \ell(y_i,s_\theta(x_i)).
$$

Taking the gradient gives

$$
\nabla_\theta \hat R(\theta)=\frac{1}{N}\sum_{i=1}^N \nabla_\theta \ell(y_i,s_\theta(x_i)).
$$

Using the definition of $g_i$, this becomes

$$
\nabla_\theta \hat R(\theta)=\frac{1}{N}\sum_{i=1}^N g_i.
$$

Therefore,

$$
\mathbb E[\hat g_B]=\nabla_\theta \hat R(\theta).
$$

</details>


### Question 8

Suppose there are $N=1250$ training observations and the mini-batch size is $64$. How many updates are in one epoch if the algorithm uses every observation once and includes a final smaller mini-batch? 
<details>
<summary>Answer</summary>

The exact number of updates is

$$
\left\lceil \frac{1250}{64}\right\rceil=20.
$$

There are $19$ full mini-batches of size $64$, which use $19\cdot64=1216$ observations. This leaves $1250-1216=34$ observations for one final smaller mini-batch.

So one epoch has $20$ updates. The rough calculation $1250/64\approx19.5$ suggests about $20$ updates, and the exact count is $20$ when the final partial mini-batch is included.

</details>


### Question 9

In the chain

$$
x\to q\to r\to t\to u\to L,
$$

show explicitly how reusing $\partial L/\partial r$ avoids recomputing the full chain-rule expression for $\partial L/\partial q$.

<details>
<summary>Answer</summary>

The full chain-rule expression is

$$
\frac{\partial L}{\partial q}=
\frac{\partial L}{\partial u}
\frac{\partial u}{\partial t}
\frac{\partial t}{\partial r}
\frac{\partial r}{\partial q}.
$$

But the first three factors are exactly

$$
\frac{\partial L}{\partial r}=
\frac{\partial L}{\partial u}
\frac{\partial u}{\partial t}
\frac{\partial t}{\partial r}.
$$

Therefore

$$
\frac{\partial L}{\partial q}=
\frac{\partial L}{\partial r}\frac{\partial r}{\partial q}.
$$

Backpropagation stores or passes the already-computed derivative $\partial L/\partial r$ and multiplies by only the next local derivative.

</details>


### Question 10

At a single node $r=f(q)$, suppose the upstream gradient is

$$
\frac{\partial L}{\partial r}=5,
$$

and $f(q)=q^3$ at $q=2$. Compute the downstream gradient $\partial L/\partial q$.

<details>
<summary>Answer</summary>

The local derivative is

$$
\frac{\partial r}{\partial q}=f'(q)=3q^2.
$$

At $q=2$,

$$
\frac{\partial r}{\partial q}=3(2)^2=12.
$$

The downstream gradient is upstream times local derivative:

$$
\frac{\partial L}{\partial q}=
\frac{\partial L}{\partial r}\frac{\partial r}{\partial q}
=5\cdot12=60.
$$

</details>


### Question 11

Combine the local derivatives in the scalar graph to obtain a single formula for $\partial J/\partial \alpha$ for one observation. Use

$$
J=\frac{1}{4}(u-t)^4,\qquad u=\gamma r+\delta,\qquad r=\tanh(q),\qquad q=\alpha z+\beta.
$$

Note: $\tanh'(q)=1-\tanh^2(q).$

<details>
<summary>Answer</summary>

The path from $\alpha$ to $J$ is

$$
\alpha \to q \to r \to u \to J.
$$

The chain rule gives

$$
\frac{\partial J}{\partial \alpha}=\frac{\partial J}{\partial u}\frac{\partial u}{\partial r}\frac{\partial r}{\partial q}\frac{\partial q}{\partial \alpha}.
$$

The local derivatives are

$$
\frac{\partial J}{\partial u}=(u-t)^3,\qquad \frac{\partial u}{\partial r}=\gamma,\qquad \frac{\partial r}{\partial q}=\tanh'(q),\qquad \frac{\partial q}{\partial \alpha}=z.
$$

Therefore,

$$
\frac{\partial J}{\partial \alpha}=(u-t)^3\gamma\tanh'(q)z.
$$

Since $\tanh'(q)=1-\tanh^2(q)$ and $r=\tanh(q)$, this can also be written as

$$
\frac{\partial J}{\partial \alpha}=(u-t)^3\gamma(1-r^2)z.
$$

</details>


### Question 12

For a mini-batch $B$ of scalar observations, let $M=|B|$ and define

$$
J_B=\frac{1}{M}\sum_{n\in B} L_n,
$$

where

$$
L_n=\frac{1}{4}(u_n-t_n)^4,\qquad u_n=\gamma r_n+\delta,\qquad r_n=\tanh(q_n),\qquad q_n=\alpha z_n+\beta.
$$

Write the gradients of $J_B$ with respect to $\alpha,\beta,\gamma,\delta$.

<details>
<summary>Answer</summary>

Since

$$
J_B=\frac{1}{M}\sum_{n\in B} L_n,
$$

each parameter gradient is the average of the corresponding single-observation gradients. First compute

$$
\frac{\partial L_n}{\partial u_n}=(u_n-t_n)^3.
$$

For compactness, define

$$
d_n=(u_n-t_n)^3.
$$

For $\gamma$, the chain rule gives

$$
\frac{\partial L_n}{\partial \gamma}=\frac{\partial L_n}{\partial u_n}\frac{\partial u_n}{\partial \gamma}=d_n r_n.
$$

Therefore,

$$
\frac{\partial J_B}{\partial \gamma}=\frac{1}{M}\sum_{n\in B} d_n r_n.
$$

For $\delta$, the chain rule gives

$$
\frac{\partial L_n}{\partial \delta}=\frac{\partial L_n}{\partial u_n}\frac{\partial u_n}{\partial \delta}=d_n.
$$

Therefore,

$$
\frac{\partial J_B}{\partial \delta}=\frac{1}{M}\sum_{n\in B} d_n.
$$

For $\alpha$, the path is $\alpha\to q_n\to r_n\to u_n\to L_n$. The chain rule gives

$$
\frac{\partial L_n}{\partial \alpha}=\frac{\partial L_n}{\partial u_n}\frac{\partial u_n}{\partial r_n}\frac{\partial r_n}{\partial q_n}\frac{\partial q_n}{\partial \alpha}=d_n\gamma\tanh'(q_n)z_n.
$$

Therefore,

$$
\frac{\partial J_B}{\partial \alpha}=\frac{1}{M}\sum_{n\in B} d_n\gamma\tanh'(q_n)z_n.
$$

For $\beta$, the path is $\beta\to q_n\to r_n\to u_n\to L_n$. The chain rule gives

$$
\frac{\partial L_n}{\partial \beta}=\frac{\partial L_n}{\partial u_n}\frac{\partial u_n}{\partial r_n}\frac{\partial r_n}{\partial q_n}\frac{\partial q_n}{\partial \beta}=d_n\gamma\tanh'(q_n).
$$

Therefore,

$$
\frac{\partial J_B}{\partial \beta}=\frac{1}{M}\sum_{n\in B} d_n\gamma\tanh'(q_n).
$$

Since $\tanh'(q_n)=1-\tanh^2(q_n)$ and $r_n=\tanh(q_n)$, the last two gradients can also be written as

$$
\frac{\partial J_B}{\partial \alpha}=\frac{1}{M}\sum_{n\in B} d_n\gamma(1-r_n^2)z_n,
$$

and

$$
\frac{\partial J_B}{\partial \beta}=\frac{1}{M}\sum_{n\in B} d_n\gamma(1-r_n^2).
$$

</details>


### Question 13

For the mini-batch objective

$$
J_B=\frac{1}{M}\sum_{n\in B}L_n,\qquad L_n=\frac{1}{4}(u_n-t_n)^4,
$$

with

$$
u_n=\gamma r_n+\delta,\qquad r_n=\tanh(q_n),\qquad q_n=\alpha z_n+\beta,
$$

why do the gradients for the shared parameters $\alpha,\beta,\gamma,\delta$ sum contributions over the mini-batch, while quantities such as $\partial J_B/\partial r_n$ and $\partial J_B/\partial q_n$ are kept separately for each observation before continuing backward?

<details>
<summary>Answer</summary>

The parameters $\alpha,\beta,\gamma,\delta$ are shared across all observations in the mini-batch. The same value of $\gamma$, for example, appears in every expression

$$
u_n=\gamma r_n+\delta.
$$

Therefore, the total derivative of $J_B$ with respect to $\gamma$ must add the contributions from every observation:

$$
\frac{\partial J_B}{\partial \gamma}=\frac{1}{M}\sum_{n\in B}\frac{\partial L_n}{\partial u_n}\frac{\partial u_n}{\partial \gamma}.
$$

Using $d_n=(u_n-t_n)^3$ and $\partial u_n/\partial \gamma=r_n$, this becomes

$$
\frac{\partial J_B}{\partial \gamma}=\frac{1}{M}\sum_{n\in B}d_n r_n.
$$

The same idea applies to the other shared parameters. For example,

$$
\frac{\partial J_B}{\partial \delta}=\frac{1}{M}\sum_{n\in B}d_n,
$$

and

$$
\frac{\partial J_B}{\partial \alpha}=\frac{1}{M}\sum_{n\in B}d_n\gamma\tanh'(q_n)z_n,\qquad
\frac{\partial J_B}{\partial \beta}=\frac{1}{M}\sum_{n\in B}d_n\gamma\tanh'(q_n).
$$

By contrast, $r_n$ and $q_n$ are not shared parameters. They are intermediate values for a specific observation $n$. Before the backward pass can continue from $r_n$ to $q_n$, it needs one derivative for each observation:

$$
\frac{\partial J_B}{\partial r_n}=\frac{1}{M}d_n\gamma.
$$

Then, for each observation separately,

$$
\frac{\partial J_B}{\partial q_n}=\frac{\partial J_B}{\partial r_n}\frac{\partial r_n}{\partial q_n}=\frac{1}{M}d_n\gamma\tanh'(q_n).
$$

So the rule is: gradients for shared parameters are summed over the mini-batch, while gradients for per-observation intermediate quantities remain indexed by $n$ until they are used by the next local backward rule.

</details>


### Question 14

When a graph branches, gradient contributions add. Verify this in the concrete graph

$$
r=z^2,\qquad t=z^3,\qquad L=r+t.
$$

Compute $\partial L/\partial z$ by adding path contributions, and compare with direct differentiation.

<details>
<summary>Answer</summary>

There are two downstream paths from $z$ to $L$. The contribution through $r$ is

$$
\frac{\partial L}{\partial r}\frac{\partial r}{\partial z}=1\cdot 2z=2z.
$$

The contribution through $t$ is

$$
\frac{\partial L}{\partial t}\frac{\partial t}{\partial z}=1\cdot 3z^2=3z^2.
$$

Adding gives

$$
\frac{\partial L}{\partial z}=2z+3z^2.
$$

Directly, $L=z^2+z^3$, so

$$
\frac{dL}{dz}=2z+3z^2.
$$

The two calculations agree.

</details>


### Question 15

A dense layer has

$$
Z=XW+b,
$$

where $X\in\mathbb R^{B\times d}$, $W\in\mathbb R^{d\times m}$, $b\in\mathbb R^m$, and the upstream gradient is

$$
G=\frac{\partial L}{\partial Z}\in\mathbb R^{B\times m}.
$$

State the shapes of

$$
\frac{\partial L}{\partial X},\qquad
\frac{\partial L}{\partial W},\qquad
\frac{\partial L}{\partial b},
$$

look the dense-layer backward formulas from the lecture, and check that the right-hand side of each formula has the correct shape.

<details>
<summary>Answer</summary>

Each gradient has the same shape as the quantity it differentiates:

$$
\frac{\partial L}{\partial X}\in\mathbb R^{B\times d},\qquad
\frac{\partial L}{\partial W}\in\mathbb R^{d\times m},\qquad
\frac{\partial L}{\partial b}\in\mathbb R^m.
$$

The dense-layer backward formulas are

$$
\frac{\partial L}{\partial X}=GW^T,\qquad
\frac{\partial L}{\partial W}=X^TG,\qquad
\frac{\partial L}{\partial b}=\sum_{i=1}^B G_i,
$$

where $G_i$ is row $i$ of $G$.

Now check the shapes. Since $G\in\mathbb R^{B\times m}$ and $W^T\in\mathbb R^{m\times d}$,

$$
GW^T\in\mathbb R^{B\times d},
$$

which matches the shape of $X$.

Since $X^T\in\mathbb R^{d\times B}$ and $G\in\mathbb R^{B\times m}$,

$$
X^TG\in\mathbb R^{d\times m},
$$

which matches the shape of $W$.

Finally, summing $G\in\mathbb R^{B\times m}$ over the $B$ batch rows gives a vector in $\mathbb R^m$:

$$
\sum_{i=1}^B G_i\in\mathbb R^m,
$$

which matches the shape of $b$.

</details>

### Question 17

Consider the quadratic objective

$$
R(\theta)=\frac{1}{2}\theta^2.
$$

For this objective, the gradient is

$$
R'(\theta)=\theta.
$$

Gradient descent with learning rate $\eta$ therefore gives the update

$$
\theta^{(t+1)}=\theta^{(t)}-\eta \theta^{(t)}.
$$

1. What is the formula for $\theta^{(t)}$ in terms of $\theta^{(0)}$?
2. For which values of $\eta$ does $\theta^{(t)}$ converge to $0$?


<details>
<summary>Answer</summary>

Starting from

$$
\theta^{(t+1)}=\theta^{(t)}-\eta\theta^{(t)},
$$

factor out $\theta^{(t)}$:

$$
\theta^{(t+1)}=(1-\eta)\theta^{(t)}.
$$

So in the form $\theta^{(t+1)}=a\theta^{(t)}$, we have

$$
a=1-\eta.
$$

Applying the update repeatedly gives

$$
\theta^{(t)}=(1-\eta)^t\theta^{(0)}.
$$

This converges to $0$ exactly when the multiplier $1-\eta$ has absolute value less than $1$:

$$
|1-\eta|<1.
$$

Solving this inequality gives

$$
-1<1-\eta<1.
$$

Subtracting $1$ from all three parts gives

$$
-2<-\eta<0.
$$

Multiplying by $-1$ reverses the inequalities:

$$
0<\eta<2.
$$

Therefore, gradient descent converges to $0$ for learning rates satisfying

$$
0<\eta<2.
$$

If $\eta>2$, then $|1-\eta|>1$, so the iterates grow in magnitude. This is the precise sense in which a learning rate that is too large can make gradient descent unstable.

</details>


### Question 18

Changing the mini-batch objective from an average

$$
\frac1{|B_t|}\sum_{n\in B_t}\ell_n(\theta)
$$

to a sum

$$
\sum_{n\in B_t}\ell_n(\theta)
$$

would change the gradient used in the update. What exactly changes, and why does the averaging convention make learning-rate comparisons across batch sizes cleaner?

<details>
<summary>Answer</summary>

The gradient of the sum is $|B_t|$ times the gradient of the average:

$$
\nabla_\theta\sum_{n\in B_t}\ell_n(\theta)
=|B_t|\nabla_\theta\left(\frac1{|B_t|}\sum_{n\in B_t}\ell_n(\theta)\right).
$$

With the same learning rate, using the summed loss would therefore make the step size scale linearly with the batch size. Averaging keeps the gradient on roughly the same scale when $|B_t|$ changes, so the learning rate has a more comparable meaning across different mini-batch sizes.

</details>
