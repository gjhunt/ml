### Question 1

The lecture defines $s_w(x)=w^\top x$ and uses the identity action $a(s)=s$. For $x\in\mathbb R^D$ and $y\in\mathbb R$, write the empirical risk for squared loss and explain why minimizing over scores $s_w$ is equivalent to minimizing over parameters $w$.

<!-- Source: model class, identity action, squared loss, empirical risk -->

<details>
<summary>Answer</summary>

With squared loss, $\hat R(w)=N^{-1}\sum_{n=1}^N (y_n-w^\top x_n)^2$. The map $w\mapsto s_w$ determines the score function in the chosen model class; when this parameterization is identifiable, minimizing over scores in $\mathcal S=\{s_w\}$ is the same as minimizing over $w$. Even if two $w$'s produce the same scores on the training data, the objective depends only on the fitted values $Xw$, so the ERM problem can still be written as a minimization over $w$.

</details>

### Question 2

Starting with $X\in\mathbb R^{N\times D}$ whose $n$th row is $x_n^\top$, expand $\|y-Xw\|_2^2$ componentwise and verify that it equals $\sum_{n=1}^N (y_n-w^\top x_n)^2$.

<!-- Source: matrix-vector form of empirical risk; design matrix notation -->

<details>
<summary>Answer</summary>

The $n$th component of $Xw$ is $x_n^\top w=w^\top x_n$. Hence the $n$th component of $y-Xw$ is $y_n-w^\top x_n$. Squaring and summing components gives $\|y-Xw\|_2^2=\sum_{n=1}^N (y_n-w^\top x_n)^2$, matching the sum of squared errors.

</details>

### Question 3

Compute $\nabla_w L(w)$ for $L(w)=\|y-Xw\|_2^2$. Then state the normal equations and the rank condition under which they have the unique solution $\hat w=(X^\top X)^{-1}X^\top y$.

<!-- Source: least-squares derivative, normal equations, invertibility condition -->

<details>
<summary>Answer</summary>

Expand $L(w)=y^\top y-2w^\top X^\top y+w^\top X^\top Xw$. Since $X^\top X$ is symmetric, $\nabla_wL(w)=-2X^\top y+2X^\top Xw$. Setting the gradient to zero gives $X^\top Xw=X^\top y$. The solution is unique exactly when $X^\top X$ is invertible, equivalently when the columns of $X$ are linearly independent.

</details>

### Question 4

The lecture says the factor $1/N$ may be dropped from $\hat R(w)$. Prove that $\arg\min_w N^{-1}\|y-Xw\|^2$ and $\arg\min_w \|y-Xw\|^2$ are the same set.

<!-- Source: dropping the $1/N$ factor in ERM -->

<details>
<summary>Answer</summary>

Multiplication by the positive constant $1/N$ preserves order: for any $u,v$, $N^{-1}L(u)\le N^{-1}L(v)$ if and only if $L(u)\le L(v)$. Therefore the parameter values minimizing the scaled and unscaled objectives are identical.

</details>

### Question 5

Give a two-observation, one-feature example with $X=\begin{pmatrix}1\2\end{pmatrix}$ and $y=\begin{pmatrix}1\0\end{pmatrix}$. Compute $\hat w$ and the fitted vector $\hat y=X\hat w$ by hand.

<!-- Source: one-dimensional projection example; hand calculation -->

<details>
<summary>Answer</summary>

Here $X^\top X=1^2+2^2=5$ and $X^\top y=1\cdot1+2\cdot0=1$, so $\hat w=1/5$. Thus $\hat y=X\hat w=(1/5,2/5)^\top$. This is the projection of $y$ onto the line spanned by $(1,2)^\top$.

</details>

### Question 6

Let $\hat y=X\hat w$ with $\hat w$ satisfying the normal equations. Prove that the residual vector $r=y-\hat y$ is orthogonal to every column of $X$.

<!-- Source: normal equations and residual orthogonality -->

<details>
<summary>Answer</summary>

The normal equations are $X^\top X\hat w=X^\top y$. Rearranging gives $X^\top(y-X\hat w)=0$, i.e. $X^\top r=0$. The $j$th entry of $X^\top r$ is the inner product between column $j$ of $X$ and $r$, so $r$ is orthogonal to every design column.

</details>

### Question 7

Assume the first column of $X$ is $\mathbf 1=(1,\ldots,1)^\top$. Use residual orthogonality to prove that the least-squares residuals sum to zero. Construct a one-observation example showing the conclusion can fail if the intercept column is omitted.

<!-- Source: intercept column; residuals sum to zero -->

<details>
<summary>Answer</summary>

With an intercept, $\mathbf 1$ is a column of $X$, so $0=\mathbf 1^\top r=\sum_n r_n$. Without an intercept, take $X=(0)$ and $y=(1)$. Every fitted value is $0\cdot w=0$, so the residual is $1$ and the residual sum is not zero.

</details>

### Question 8

Define $P_X=X(X^\top X)^{-1}X^\top$ when $X$ has full column rank. Verify directly that $P_X^\top=P_X$ and $P_X^2=P_X$.

<!-- Source: projection matrix $P_X$; symmetry and idempotence -->

<details>
<summary>Answer</summary>

Since $X^\top X$ is symmetric, so is $(X^\top X)^{-1}$, hence $P_X^\top=P_X$. Also $P_X^2=X(X^\top X)^{-1}X^\top X(X^\top X)^{-1}X^\top=X(X^\top X)^{-1}X^\top=P_X$. Thus $P_X$ is an orthogonal projection matrix.

</details>

### Question 9

Use the projection matrix to prove the Pythagorean identity $\|y\|^2=\|\hat y\|^2+\|y-\hat y\|^2$ when the fitted vector is $\hat y=P_Xy$.

<!-- Source: geometric interpretation; orthogonal projection and residual -->

<details>
<summary>Answer</summary>

The fitted vector $\hat y$ lies in $\operatorname{col}(X)$ and the residual $r=y-\hat y$ is orthogonal to $\operatorname{col}(X)$, hence $r\perp \hat y$. Since $y=\hat y+r$, $\|y\|^2=\|\hat y\|^2+2\hat y^\top r+\|r\|^2=\|\hat y\|^2+\|r\|^2$.

</details>

### Question 10

A student writes $\hat y=(X^\top X)^{-1}X^\top y$. Identify the dimension error and give the correct formulas for $\hat w$ and $\hat y$.

<!-- Source: forming predictions; distinguishing coefficients from fitted values -->

<details>
<summary>Answer</summary>

If $X\in\mathbb R^{N\times D}$, then $(X^\top X)^{-1}X^\top y\in\mathbb R^D$, so it is the coefficient vector $\hat w$, not the $N$-vector of fitted values. The correct formulas are $\hat w=(X^\top X)^{-1}X^\top y$ and $\hat y=X\hat w=X(X^\top X)^{-1}X^\top y$.

</details>

### Question 11

Construct a design matrix with two perfectly collinear columns. Show that $X^\top X$ is singular and exhibit two different coefficient vectors that produce the same fitted values.

<!-- Source: solution stability; exact collinearity example with duplicated variable -->

<details>
<summary>Answer</summary>

Take $X=\begin{pmatrix}1&1\2&2\end{pmatrix}$. The columns are identical, so $X^\top X=\begin{pmatrix}5&5\5&5\end{pmatrix}$ has determinant $0$. For any $c$, $(\beta_1+c,\beta_2-c)^\top$ gives the same fitted vector because $X(\beta_1+c,\beta_2-c)^\top=X(\beta_1,\beta_2)^\top$.

</details>

### Question 12

In the exact-collinearity example $X=[x\ x]$ with $x
e0$, prove that the least-squares loss depends on $\beta_1$ and $\beta_2$ only through their sum $\beta_1+\beta_2$.

<!-- Source: ill-posed simulated data; loss contours under exact collinearity -->

<details>
<summary>Answer</summary>

For $X=[x\ x]$, $X\beta=x\beta_1+x\beta_2=x(\beta_1+\beta_2)$. Therefore $\|y-X\beta\|^2=\|y-x(\beta_1+\beta_2)\|^2$. Coefficients with the same sum have identical fitted values and identical loss, explaining the flat direction in the loss.

</details>

### Question 13

The lecture notes that near-collinearity can make $\hat w$ unstable. Let $X^\top X=Q\Lambda Q^\top$ with eigenvalues $\lambda_j>0$. Express $\hat w$ in the eigenbasis and identify why small $\lambda_j$ amplify perturbations in $X^\top y$.

<!-- Source: solution stability; conditioning of $X^\top X$ -->

<details>
<summary>Answer</summary>

Write $\hat w=Q\Lambda^{-1}Q^\top X^\top y$. In direction $q_j$, the coefficient is scaled by $1/\lambda_j$. If $\lambda_j$ is very small, a small perturbation in the component $q_j^\top X^\top y$ produces a large change in the fitted coefficient along $q_j$.

</details>

### Question 16

Compare the full one-hot encoding without an intercept to the $K-1$ encoding with an intercept. For a single categorical predictor with all categories observed, do they span the same fitted-value subspace? Justify the answer.

<!-- Source: categorical variables; full indicators versus baseline coding -->

<details>
<summary>Answer</summary>

Yes. Full one-hot encoding without an intercept spans all vectors that are constant within each category. Baseline coding with an intercept also assigns one mean to the baseline and one shifted mean to each other category. The parameterization differs, but the set of possible fitted values is the same when all category columns are represented appropriately.

</details>

### Question 17

For data points $x_1,x_2,x_3$ and polynomial features $\phi(x)=(1,x,x^2)$, write the design matrix. State the condition on $x_1,x_2,x_3$ that makes its columns linearly independent.

<!-- Source: polynomial regression and Vandermonde-style design matrix -->

<summary>Answer</summary>

The design matrix is $X=\begin{pmatrix}1&x_1&x_1^2\1&x_2&x_2^2\1&x_3&x_3^2\end{pmatrix}$. Its determinant is $\prod_{i<j}(x_j-x_i)$, so the columns are linearly independent exactly when the three input values are distinct.

### Question 19

Find a nonzero vector $a=(a_1,a_2,a_3)^\top$ showing that the basis functions $(1,x,2x)$ are redundant. What happens to $X^\top X$ for any design matrix built from these three columns?

<!-- Source: basis functions; redundancy and linear dependence -->

<details>
<summary>Answer</summary>

The functions satisfy $0\cdot1+2x-1\cdot(2x)=0$, so $a=(0,1,-1)^\top$ is a nonzero dependence. In any design matrix with columns $1,x,2x$, the third column is twice the second; therefore the columns are linearly dependent and $X^\top X$ is singular.

</details>

### Question 20

The lecture says squared loss penalizes large residuals heavily. Compare residual vectors $r=(10,0,\ldots,0)$ and $s=(1,1,\ldots,1)$ with ten entries. Which has larger squared loss, and what does this illustrate about outliers?

<!-- Source: sensitivity to outliers; squared error loss -->

<details>
<summary>Answer</summary>

For $r$, $\|r\|^2=100$. For $s\in\mathbb R^{10}$, $\|s\|^2=10$. One residual of size $10$ contributes as much as one hundred residuals of size $1$, so a small number of large errors can dominate the least-squares objective.

</details>

### Question 21

Diagnose this false claim: 'If $X^\top X$ is singular, then the fitted vector $X\hat w$ is never unique.' Give a concrete counterexample and verify uniqueness of the fitted vector.

<!-- Source: rank deficiency; fitted values versus coefficient non-identifiability -->

<details>
<summary>Answer</summary>

Let $X=\begin{pmatrix}1&1\2&2\end{pmatrix}$. The coefficient vector is not unique because only $\beta_1+\beta_2$ matters. However, the closest point to $y$ in the column space of $X$ is the unique orthogonal projection of $y$ onto $\operatorname{col}(X)$. Thus all least-squares minimizers have the same fitted vector, even though their coefficients differ.

</details>

### Question 22

Assume $X$ has full column rank. Prove that the Hessian of $L(w)=\|y-Xw\|^2$ is positive definite. What changes when $X$ is rank deficient?

<!-- Source: least-squares Hessian; convexity and uniqueness -->

<details>
<summary>Answer</summary>

The Hessian is $\nabla^2L(w)=2X^\top X$. For any nonzero $v$, $v^\top(2X^\top X)v=2\|Xv\|^2>0$ when $X$ has full column rank, so the Hessian is positive definite. If $X$ is rank deficient, some nonzero $v$ has $Xv=0$, so the Hessian is only positive semidefinite and minimizers may be nonunique.

</details>

### Question 23

In a model with an intercept column, prove that the mean of the fitted values equals the mean of the observed responses: $N^{-1}\sum_n\hat y_n=N^{-1}\sum_n y_n$.

<!-- Source: intercept term; fitted means and residual sum zero -->

<details>
<summary>Answer</summary>

With an intercept, residuals sum to zero: $\sum_n(y_n-\hat y_n)=0$. Rearranging gives $\sum_n y_n=\sum_n\hat y_n$. Dividing by $N$ shows that the fitted and observed means are equal.

</details>

### Question 24

For a new input $x_{new}\in\mathbb R^D$, distinguish the scalar prediction $\hat f(x_{new})$ from the training fitted vector $\hat y$. Write both formulas and their dimensions.

<!-- Source: forming predictions on new inputs versus training data -->

<details>
<summary>Answer</summary>

The new-input prediction is the scalar $\hat f(x_{new})=\hat w^\top x_{new}\in\mathbb R$. The training fitted vector is $\hat y=X\hat w\in\mathbb R^N$, collecting predictions at all observed rows of $X$. They use the same coefficient vector but apply it to different inputs.

</details>

### Question 25

The lecture's real-data example uses one-hot encoding with `drop_first=True`. Translate that software choice into the mathematical statement about the design matrix columns and the intercept.

### Question 26

A feature is duplicated accidentally, so the design matrix is $\tilde X=[X_j\ X_j\ Z]$. Which quantities are identifiable: the two separate duplicated coefficients, their sum, or the fitted values? Justify using $\tilde X\tilde w$.

<!-- Source: solution stability; accidental duplicated variable -->

<details>
<summary>Answer</summary>

If the duplicated coefficients are $a$ and $b$, their contribution is $aX_j+bX_j=(a+b)X_j$. The separate values $a$ and $b$ are not identifiable, but their sum is identifiable to the extent the combined column is identifiable with the other columns. The fitted values can still be unique as the projection onto $\operatorname{col}(\tilde X)$.

</details>

### Question 27

For centered single-predictor data with an intercept, $\sum_n x_n=0$. Derive the least-squares slope formula $\hat w_1=\frac{\sum_n x_n y_n}{\sum_n x_n^2}$.

<!-- Source: intercept and centered predictor; omitted algebra in normal equations -->

<details>
<summary>Answer</summary>

With model $y_n=w_0+w_1x_n+r_n$, the normal equations are $\sum r_n=0$ and $\sum x_nr_n=0$. Since $\sum x_n=0$, the first equation gives $\hat w_0=\bar y$. The second gives $\sum x_n(y_n-\bar y-\hat w_1x_n)=0$. Because $\sum x_n\bar y=0$, $\hat w_1=\sum x_ny_n/\sum x_n^2$.

</details>

### Question 30

State the precise optimization problem whose solution is the projection of $y$ onto $\operatorname{col}(X)$. Then explain why this problem has a unique fitted vector even if the coefficient minimizer is not unique.

<!-- Source: geometric interpretation; projection problem versus coefficient problem -->

<details>
<summary>Answer</summary>

The fitted vector solves $\hat y=\arg\min_{z\in\operatorname{col}(X)}\|y-z\|^2$. A closed subspace of $\mathbb R^N$ has a unique orthogonal projection of $y$, so $\hat y$ is unique. Coefficients need not be unique because different $w$ values can map to the same $z=Xw$ when $X$ has a nontrivial null space.

</details>
