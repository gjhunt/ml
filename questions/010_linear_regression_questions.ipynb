### Question 1

Starting with $X\in\mathbb R^{N\times D}$ whose $n$th row is $x_n^\top$, expand $\|y-Xw\|_2^2$ componentwise and verify that it equals $\sum_{n=1}^N (y_n-w^\top x_n)^2$.

<!-- Source: matrix-vector form of empirical risk; design matrix notation -->

<details>
<summary>Answer</summary>

The $n$th component of $Xw$ is $x_n^\top w=w^\top x_n$. Hence the $n$th component of $y-Xw$ is $y_n-w^\top x_n$. Squaring and summing components gives $\|y-Xw\|_2^2=\sum_{n=1}^N (y_n-w^\top x_n)^2$, matching the sum of squared errors.

</details>

### Question 2

Compute $\nabla_w \hat{R}(w)$ for $\hat{R}(w)=\|y-Xw\|_2^2$. Then state the normal equations and the rank condition under which they have the unique solution $\hat w=(X^\top X)^{-1}X^\top y$.

<!-- Source: least-squares derivative, normal equations, invertibility condition -->

<details>
<summary>Answer</summary>

Expand $\hat{R}(w)=y^\top y-2w^\top X^\top y+w^\top X^\top Xw$. Since $X^\top X$ is symmetric, $\nabla_w \hat{R}(w)=-2X^\top y+2X^\top Xw$. Setting the gradient to zero gives $X^\top Xw=X^\top y$. The solution is unique exactly when $X^\top X$ is invertible, equivalently when the columns of $X$ are linearly independent.

</details>

### Question 3

The lecture says the factor $1/N$ may be dropped from $\hat R(w)$. Argue that $\arg\min_w N^{-1}\|y-Xw\|^2$ and $\arg\min_w \|y-Xw\|^2$ are the same.

<!-- Source: dropping the $1/N$ factor in ERM -->

<details>
<summary>Answer</summary>

Multiplication by the positive constant $1/N$ preserves order: for any $u,v$, $N^{-1}\hat{R}(u)\le N^{-1}\hat{R}(v)$ if and only if $\hat{R}(u)\le \hat{R}(v)$. Therefore the parameter values minimizing the scaled and unscaled objectives are identical.

</details>

### Question 4

Give a two-observation, one-feature example with $X=\begin{pmatrix}1\\2\end{pmatrix}$ and $y=\begin{pmatrix}1\\0\end{pmatrix}$. Compute $\hat w$ and the fitted vector $\hat y=X\hat w$ by hand.

<!-- Source: one-dimensional projection example; hand calculation -->

<details>
<summary>Answer</summary>

Here $X^\top X=1^2+2^2=5$ and $X^\top y=1\cdot1+2\cdot0=1$, so $\hat w=1/5$. Thus $\hat y=X\hat w=(1/5,2/5)^\top$. This is the projection of $y$ onto the line spanned by $(1,2)^\top$.

</details>

### Question 5

Let $\hat y=X\hat w$ with $\hat w$ satisfying the normal equations. Prove that the residual vector $r=y-\hat y$ (for the training data) is orthogonal to every column of $X$.

<!-- Source: normal equations and residual orthogonality -->

<details>
<summary>Answer</summary>

The normal equations are $X^\top X\hat w=X^\top y$. Rearranging gives $X^\top(y-X\hat w)=0$, i.e. $X^\top r=0$. The $j$th entry of $X^\top r$ is the inner product between column $j$ of $X$ and $r$, so $r$ is orthogonal to every design column.

</details>

### Question 7

Let $X \in \mathbb{R}^{n \times d}$ have full column rank, and define

$$
P_X = X(X^\top X)^{-1}X^\top.
$$

Verify directly that $P_X$ is symmetric and idempotent. That is, show that

$$
P_X^\top = P_X
$$

and

$$
P_X^2 = P_X.
$$

In the symmetry calculation, explicitly use the facts that $X^\top X$ is symmetric and that the inverse of an invertible symmetric matrix is also symmetric.

<!-- Source: projection matrix $P_X$; symmetry and idempotence -->

<details>
<summary>Answer</summary>

Let

$$
B = X^\top X.
$$

Since $X$ has full column rank, $B$ is invertible. Also,

$$
B^\top = (X^\top X)^\top = X^\top X = B,
$$

so $B$ is symmetric. Because $B$ is invertible and symmetric, its inverse is also symmetric:

$$
(B^{-1})^\top = (B^\top)^{-1} = B^{-1}.
$$

Now compute the transpose of $P_X$:

$$
P_X^\top=
\left(XB^{-1}X^\top\right)^\top.
$$

Using $(ABC)^\top = C^\top B^\top A^\top$, we get

$$
P_X^\top=
(X^\top)^\top (B^{-1})^\top X^\top.
$$

Since $(X^\top)^\top = X$ and $(B^{-1})^\top = B^{-1}$,

$$
P_X^\top=
XB^{-1}X^\top=
P_X.
$$

Thus $P_X$ is symmetric.

Next, compute $P_X^2$:

$$
P_X^2=
\left(XB^{-1}X^\top\right)
\left(XB^{-1}X^\top\right).
$$

Group the middle terms:

$$
P_X^2=
XB^{-1}(X^\top X)B^{-1}X^\top.
$$

Since $B = X^\top X$, this becomes

$$
P_X^2=
XB^{-1}BB^{-1}X^\top.
$$

Because $B^{-1}B = I$, we have

$$
P_X^2=
XB^{-1}X^\top=
P_X.
$$

Therefore $P_X$ is both symmetric and idempotent. This means it is an orthogonal projection matrix.

</details>


### Question 8

A student writes $\hat y=(X^\top X)^{-1}X^\top y$. Identify the dimension error and give the correct formulas for $\hat w$ and $\hat y$.

<!-- Source: forming predictions; distinguishing coefficients from fitted values -->

<details>
<summary>Answer</summary>

If $X\in\mathbb R^{N\times D}$, then $(X^\top X)^{-1}X^\top y\in\mathbb R^D$, so it is the coefficient vector $\hat w$, not the $N$-vector of fitted values. The correct formulas are $\hat w=(X^\top X)^{-1}X^\top y$ and $\hat y=X\hat w=X(X^\top X)^{-1}X^\top y$.

</details>

### Question 10

Suppose the design matrix has two identical columns:

$$
X =\begin{bmatrix}
\vert & \vert \\
x & x \\
\vert & \vert
\end{bmatrix}=
[x \ \ x],
$$

where $x \in \mathbb{R}^n$ is a nonzero feature vector. Prove that the least-squares loss

$$
L(\beta) = \|y - X\beta\|^2
$$

depends on $\beta_1$ and $\beta_2$ only through their sum $\beta_1+\beta_2$.

<!-- Source: ill-posed simulated data; loss contours under exact collinearity -->

<details>
<summary>Answer</summary>

Since the two columns of $X$ are both equal to $x$, we have

$$
X\beta=
[x \ \ x]
\begin{bmatrix}
\beta_1 \\
\beta_2
\end{bmatrix}=
x\beta_1 + x\beta_2.
$$

Factoring out $x$ gives

$$
X\beta = x(\beta_1+\beta_2).
$$

Therefore the least-squares loss is

$$
L(\beta)=
\|y - X\beta\|^2=
\|y - x(\beta_1+\beta_2)\|^2.
$$

Thus the loss depends on $\beta_1$ and $\beta_2$ only through the single quantity

$$
\beta_1+\beta_2.
$$

In particular, any two coefficient vectors with the same sum produce the same fitted values and therefore the same least-squares loss. For example,

$$
(\beta_1,\beta_2)
\quad \text{and} \quad
(\beta_1 + c, \beta_2 - c)
$$

have the same sum for any $c \in \mathbb{R}$, so they give the same fitted values and the same loss.

</details>

### Question 15

Assume $X$ has full column rank. Find the Hessian of

$$
\hat{R}(w)=\|y-Xw\|^2.
$$

<!-- Source: least-squares Hessian; convexity and uniqueness -->

<details>
<summary>Answer</summary>

First expand the empirical risk:

$$
\hat{R}(w)=\|y-Xw\|^2=(y-Xw)^\top(y-Xw).
$$

Multiplying this out gives

$$
\hat{R}(w)=y^\top y-y^\top Xw-w^\top X^\top y+w^\top X^\top Xw.
$$

The two middle terms are both scalars, and

$$
y^\top Xw=w^\top X^\top y.
$$

Therefore,

$$
\hat{R}(w)=y^\top y-2w^\top X^\top y+w^\top X^\top Xw.
$$

Now differentiate with respect to $w$. The first term, $y^\top y$, is constant in $w$, so its gradient is zero. The second term is linear in $w$, so

$$
\nabla_w(-2w^\top X^\top y)=-2X^\top y.
$$

For the quadratic term, use the identity

$$
\nabla_w(w^\top A w)=(A+A^\top)w.
$$

Here $A=X^\top X$, which is symmetric because

$$
(X^\top X)^\top=X^\top X.
$$

Thus,

$$
\nabla_w(w^\top X^\top Xw)=2X^\top Xw.
$$

Putting these pieces together,

$$
\nabla_w \hat{R}(w)=-2X^\top y+2X^\top Xw.
$$

The Hessian is the derivative of the gradient with respect to $w$. Since $-2X^\top y$ is constant in $w$ and $2X^\top Xw$ is linear in $w$, we get

$$
\nabla_w^2 \hat{R}(w)=2X^\top X.
$$

So the Hessian is

$$
\boxed{\nabla_w^2 \hat{R}(w)=2X^\top X.}
$$

So $X^\top X$ is **positive definite**, which basically means the empirical risk is strictly convex, so it has a unique minimizer.

</details>


### Question 16

In a model with an intercept column, prove that the mean of the fitted values equals the mean of the observed responses: $\frac{1}{N}\sum_n\hat y_n=\frac{1}{N}\sum_n y_n$.

<!-- Source: intercept term; fitted means and residual sum zero -->

<details>
<summary>Answer</summary>

With an intercept, residuals sum to zero: $\sum_n(y_n-\hat y_n)=0$. Rearranging gives $\sum_n y_n=\sum_n\hat y_n$. Dividing by $N$ shows that the fitted and observed means are equal.

</details>

### Question 17

The lecture warns that linear regression cannot capture nonlinear patterns unless nonlinear features are introduced. Give a three-point dataset in one raw predictor for which no affine function $w_0+w_1x$ fits perfectly, but a quadratic feature model can fit perfectly.

<!-- Source: linearity limitation; polynomial feature engineering -->

<details>
<summary>Answer</summary>

Use $(x,y)=(-1,1),(0,0),(1,1)$. An affine function fitting the first and third points symmetrically would have slope $0$ and intercept $1$, which misses $(0,0)$. The quadratic model $s(x)=x^2$ fits all three points exactly using features $(1,x,x^2)$.
