# Feed-Forward Neural Networks - Review Questions {#sec-nn-questions}

### Question 1

A network has layer sizes
$$
[3,4,3,1].
$$
I.e. it has a 3-vector input and outputs a single number. Using the lecture's formula, compute the number of parameters in each layer and the total number of parameters.

<details>
<summary>Answer</summary>

For a layer mapping width $d_{i-1}$ to width $d_i$, the number of parameters is
$$
d_id_{i-1}+d_i.
$$
For $3\to4$:
$$
4\cdot3+4=16.
$$
For $4\to3$:
$$
3\cdot4+3=15.
$$
For $3\to1$:
$$
1\cdot3+1=4.
$$
The total is
$$
16+15+4=35.
$$

</details>


### Question 3

The lecture contrasts feature engineering and feature learning. What's the difference?

<details>
<summary>Answer</summary>

Feature engineering manually constructs new inputs before fitting a model. For example, one could replace $x=(x_1,x_2)$ by
$$
(x_1,x_2,x_1^2,x_2^2,x_1x_2)
$$
and then fit a linear model in this expanded representation.

Feature learning lets the model learn intermediate features from data. In a neural network, the first hidden layer might compute
$$
h^{(1)}=\phi(W^{(1)}x+b^{(1)}),
$$
so the learned features are the coordinates of $h^{(1)}$ rather than manually specified polynomial terms.

</details>


### Question 2

Suppose layer $i$ has input dimension $d_{i-1}=5$ and output dimension $d_i=7$. State the dimensions of $W^{(i)}$, $b^{(i)}$, $W^{(i)}h+b^{(i)}$, and $s_i(h)$.

<details>
<summary>Answer</summary>

The weight matrix has shape
$$
W^{(i)}\in\mathbb R^{7\times5}.
$$
The bias vector has shape
$$
b^{(i)}\in\mathbb R^7.
$$
For $h\in\mathbb R^5$,
$$
W^{(i)}h\in\mathbb R^7,
$$
so
$$
W^{(i)}h+b^{(i)}\in\mathbb R^7.
$$
Because $\phi_i$ maps $\mathbb R^7$ to $\mathbb R^7$ componentwise,
$$
s_i(h)=\phi_i(W^{(i)}h+b^{(i)})\in\mathbb R^7.
$$

</details>


### Question 3

A one-hidden-layer regression network has input dimension $D=10$, $H=6$ hidden units, and a single output dimension. It uses
$$
h=\phi(Wx+b),\qquad s_\theta(x)=v^\top h+c.
$$
Compute the total number of parameters.

<details>
<summary>Answer</summary>

The hidden layer has
$$
W\in\mathbb R^{6\times10}
$$
with $60$ weights and
$$
b\in\mathbb R^6
$$
with $6$ biases. That is $66$ hidden-layer parameters.

The output layer has
$$
v\in\mathbb R^6
$$
with $6$ weights and scalar bias $c$, giving $7$ output-layer parameters.

The total is
$$
66+7=73.
$$

</details>


### Question 4

Why are nonlinear activation functions essential if we want a multilayer network to represent more than a linear model? 
<details>
<summary>Answer</summary>

If every activation is the identity, then each layer is just an affine transformation. A composition of affine transformations is itself affine. Therefore, even a deep network with only identity activations collapses to a single affine score function. Nonlinear activations such as ReLU, sigmoid, or tanh prevent this collapse. They allow the network to build nonlinear transformations of the input and become more flexible than a linear model.

</details>


### Question 5

Prove that a two-layer network with identity activations is affine. Specifically, let
$$
h=W^{(1)}x+b^{(1)},
\qquad
s(x)=W^{(2)}h+b^{(2)}.
$$
Write $s(x)$ as a single affine function of $x$:
$$
s(x)=Ax+a.
$$

<details>
<summary>Answer</summary>

Substitute the first layer into the second:
$$
s(x)=W^{(2)}(W^{(1)}x+b^{(1)})+b^{(2)}.
$$
Distribute:
$$
s(x)=W^{(2)}W^{(1)}x+W^{(2)}b^{(1)}+b^{(2)}.
$$
This has the form
$$
s(x)=Ax+a,
$$
where
$$
A=W^{(2)}W^{(1)},
\qquad
a=W^{(2)}b^{(1)}+b^{(2)}.
$$
Thus the two-layer identity-activation network is just one affine map.

</details>


### Question 6

What does it mean for a network to be feed-forward? 
<details>
<summary>Answer</summary>

Feed-forward means information moves in one direction through the network. The input $x$ is transformed into the first hidden representation $h^{(1)}$, then into $h^{(2)}$, and so on until the output score $s_\theta(x)$ is produced. There are no feedback loops in this description: later layers do not feed their outputs back into earlier layers. Each layer receives input from the previous layer and passes its output forward.

</details>


### Question 7

For a classification network with a single hidden width $H=8$ and $K=3$ classes, the final affine layer has
$$
s_\theta(x)=Vh+c.
$$
State the dimensions of $V$, $c$, and $s_\theta(x)$.

<details>
<summary>Answer</summary>

The hidden representation satisfies
$$
h\in\mathbb R^8.
$$
To produce $K=3$ logits,
$$
V\in\mathbb R^{3\times8},
\qquad
c\in\mathbb R^3.
$$
Then
$$
Vh+c\in\mathbb R^3.
$$
Thus
$$
s_\theta(x)\in\mathbb R^3.
$$

</details>


### Question 8

The lecture says classification network outputs are often treated as logits. What is a logit in this context, and why are logits not probabilities?

<details>
<summary>Answer</summary>

A logit is a raw class score produced by the final affine layer of the network. For $K$ classes, the logit vector is
$$
s_\theta(x)\in\mathbb R^K.
$$
Logits are not probabilities because their entries can be negative, positive, or zero, and they do not have to sum to one. To obtain probabilities, one applies the softmax function:
$$
p_\theta(x)=\operatorname{softmax}(s_\theta(x)).
$$

</details>


### Question 9

A classifier has logits
$$
s=(0,\log 2,\log 3).
$$
Compute the softmax probability vector.

<details>
<summary>Answer</summary>

Softmax probabilities are
$$
p_k=\frac{e^{s_k}}{\sum_j e^{s_j}}.
$$
Here
$$
e^0=1,
\qquad
e^{\log2}=2,
\qquad
e^{\log3}=3.
$$
The denominator is $1+2+3=6$. Therefore
$$
p=\left(\frac16,\frac26,\frac36\right)=\left(\frac16,\frac13,\frac12\right).
$$

</details>

### Question 11

For one-hidden-layer classification,
$$
h=\phi(Wx+b),\qquad s_\theta(x)=Vh+c,
$$
with $K$ classes and $H$ hidden units. Derive the shapes of $W$, $b$, $V$, and $c$ when $x\in\mathbb R^D$.

<details>
<summary>Answer</summary>

The hidden layer maps $\mathbb R^D$ to $\mathbb R^H$, so
$$
W\in\mathbb R^{H\times D},
\qquad
b\in\mathbb R^H.
$$
Then
$$
h=\phi(Wx+b)\in\mathbb R^H.
$$
The output layer maps $\mathbb R^H$ to $\mathbb R^K$, so
$$
V\in\mathbb R^{K\times H},
\qquad
c\in\mathbb R^K.
$$
Thus
$$
s_\theta(x)=Vh+c\in\mathbb R^K.
$$

</details>


### Question 12

For predicted class probabilities
$$
p=(0.1,0.7,0.2)
$$
and true class $y=2$, compute the categorical cross-entropy loss. Use natural logarithms.

<details>
<summary>Answer</summary>

The correct-class probability is
$$
p_y=p_2=0.7.
$$
The loss is
$$
-\log(p_y)=-\log(0.7)\approx 0.357.
$$

</details>


### Question 13

Suppose a hidden layer uses ReLU activation, $\phi(t)=\max\{0,t\}$ componentwise. For pre-activation vector
$$
z=(-2,0,3),
$$
compute the hidden representation $h=\phi(z)$.

<details>
<summary>Answer</summary>

Apply ReLU to each coordinate:
$$
\phi(-2)=0,
\qquad
\phi(0)=0,
\qquad
\phi(3)=3.
$$
Thus
$$
h=(0,0,3).
$$

</details>


### Question 14

A hidden layer maps $\mathbb R^4$ to $\mathbb R^3$ using an affine map and an activation. How many weights and biases does this layer have?

<summary>Answer</summary>

Here $d_{i-1}=4$ and $d_i=3$. The weight matrix has
$$
3\cdot4=12
$$
entries. The bias vector has
$$
3
$$

$$
12+3=15
$$
parameters.

</details>


### Question 15

For a classification network with logits
$$
s=(2,1,0),
$$
Which class is predicted? Does applying softmax change the predicted class?

<details>
<summary>Answer</summary>

The largest logit is the first entry, $2$, so the predicted class is class $1$ if classes are indexed as $1,2,3$. Applying softmax does not change the predicted class because softmax is monotone in each logit relative to the others: the largest logit gets the largest softmax probability.

</details>


### Question 16

Why does the lecture describe neural networks as learning intermediate representations?
<details>
<summary>Answer</summary>

Each hidden layer produces a new representation of the input, such as
$$
h^{(1)}=s_1(x),
\qquad
h^{(2)}=s_2(h^{(1)}).
$$
These representations are not fixed before training; they depend on learned weights and biases. The final layer then makes predictions from the last learned representation.

Thus the network learns both the features and the final prediction rule, rather than fitting a linear model to a fixed feature representation supplied in advance.

</details>
