### Question 1

### Question 2

Verify the algebraic simplification of the binary cross-entropy loss. For $y\in\{0,1\}$, $s\in\mathbb{R}$, and $p=\sigma(s)$, derive
$$
-\{y\log p+(1-y)\log(1-p)\}=\log(1+e^s)-ys.
$$

<!-- Source: Cross-entropy loss and the compressed simplification to $\log(1+e^s)-ys$. -->

<details>
<summary>Answer</summary>

Use $p=e^s/(1+e^s)$ and $1-p=1/(1+e^s)$. Then
$$
\log p=s-\log(1+e^s),\qquad \log(1-p)=-\log(1+e^s).
$$
Substitution gives
$$
-\{y(s-\log(1+e^s))+(1-y)(-\log(1+e^s))\}
=\log(1+e^s)-ys.
$$

</details>


### Question 3

For the two labels separately, compute $\ell(1,s)$ and $\ell(0,s)$ from
$$
\ell(y,s)=\log(1+e^s)-ys.
$$
State the limits of each loss as $s\to\infty$ and $s\to -\infty$, and identify which limits correspond to confident mistakes.

<!-- Source: Understanding the loss through the cases $y=1$ and $y=0$. -->

<details>
<summary>Answer</summary>

For $y=1$,
$$
\ell(1,s)=\log(1+e^s)-s=\log(1+e^{-s}).
$$
Thus $\ell(1,s)\to0$ as $s\to\infty$, and $\ell(1,s)\to\infty$ as $s\to-\infty$. The second limit is a confident mistake.

For $y=0$,
$$
\ell(0,s)=\log(1+e^s).
$$
Thus $\ell(0,s)\to\infty$ as $s\to\infty$, and $\ell(0,s)\to0$ as $s\to-\infty$. The first limit is a confident mistake.

</details>


### Question 4

Differentiate the single-observation loss
$$
\ell_n(w)=-\{y_n\log p_n+(1-y_n)\log(1-p_n)\},
\qquad p_n=\sigma(x_n^\top w),
$$
with respect to $w$. Give the chain-rule calculation leading to $\nabla_w\ell_n=(p_n-y_n)x_n$.

<!-- Source: Gradient derivation for one summand, where the lecture says the algebra simplifies nicely. -->

<details>
<summary>Answer</summary>

First,
$$
\frac{\partial \ell_n}{\partial p_n}
=-\frac{y_n}{p_n}+\frac{1-y_n}{1-p_n}
=\frac{p_n-y_n}{p_n(1-p_n)}.
$$
Also,
$$
\nabla_w p_n=\sigma'(x_n^\top w)x_n=p_n(1-p_n)x_n.
$$
Therefore
$$
\nabla_w\ell_n
=\frac{p_n-y_n}{p_n(1-p_n)}p_n(1-p_n)x_n
=(p_n-y_n)x_n.
$$

</details>


### Question 5

Translate the componentwise gradient into matrix form. Given $X\in\mathbb{R}^{N\times D}$, $p=\sigma(Xw)$, and $y\in\mathbb{R}^N$, verify the dimensions and derive
$$
\nabla \hat R(w)=\frac{1}{N}X^\top(p-y).
$$

<!-- Source: Vector-form empirical risk and the matrix gradient $X^\top(p-y)/N$. -->

<details>
<summary>Answer</summary>

The componentwise gradient is
$$
\frac1N\sum_{n=1}^N(p_n-y_n)x_n.
$$
Here $p-y\in\mathbb{R}^N$ and $X^\top\in\mathbb{R}^{D\times N}$, so $X^\top(p-y)\in\mathbb{R}^D$. Its $j$-th coordinate is
$$
\sum_{n=1}^N X_{nj}(p_n-y_n),
$$
which is exactly the $j$-th coordinate of the summation.

</details>


### Question 6

Compare the logistic-regression gradient with the least-squares gradient. For least squares with objective $N^{-1}\|Xw-y\|_2^2/2$, the gradient is $N^{-1}X^\top(Xw-y)$. State exactly what changes in logistic regression.

<!-- Source: Comparison between residuals $Xw-y$ and logistic residuals $p-y$. -->

<details>
<summary>Answer</summary>

In logistic regression,
$$
\nabla \hat R(w)=\frac1N X^\top(p-y),\qquad p=\sigma(Xw).
$$
The linear fitted values $Xw$ are replaced by fitted probabilities $p$.

### Question 7

Diagnose the false claim: "Since logistic regression has gradient $X^\top(p-y)/N$, the minimizer is obtained by setting $p=y$."

<!-- Source: No-closed-form discussion and comparison with least-squares normal equations. -->

<details>
<summary>Answer</summary>

The claim confuses the condition for a zero gradient with the stronger condition $p=y$.

For logistic regression, the gradient of the empirical risk is

$$
\nabla_w \widehat{R}(w)=\frac{1}{N}X^\top(p-y),
$$

where

$$
p=\sigma(Xw).
$$

At a minimizer, a first-order condition is

$$
X^\top(p-y)=0.
$$

However, this does **not** generally imply

$$
p-y=0.
$$

It only says that the residual vector $p-y$ is orthogonal to every column of $X$. Equivalently, $p-y$ lies in the orthogonal complement of the column space of $X$.

This is similar to least squares. In linear regression, setting the gradient to zero gives

$$
X^\top(Xw-y)=0,
$$

not necessarily

$$
Xw-y=0.
$$

The model only needs the residual to be orthogonal to the feature directions, not exactly zero for every observation.

There is another issue in logistic regression: each entry of $p$ has the form

$$
p_n=\sigma(x_n^\top w),
$$

so $p_n$ always lies strictly between $0$ and $1$ for finite $w$. If the labels satisfy $y_n\in\{0,1\}$, then setting $p=y$ would require probabilities to be exactly $0$ or exactly $1$, which cannot happen for finite parameter values.

Thus the correct stationarity condition is

$$
X^\top(p-y)=0,
$$

not

$$
p=y.
$$

This is one reason logistic regression does not lead to a simple closed-form normal equation like ordinary least squares

</details>

### Question 10

Connect the one-score and two-score forms of a binary linear classifier. Suppose

$$
s_1(x)=w_1^\top x,
\qquad
s_2(x)=w_2^\top x,
$$

and the classifier predicts the class with the larger score:

$$
\hat{y}(x)=\arg\max_{k\in\{1,2\}} s_k(x).
$$

Show that this decision rule can be written using a single vector $w$.

<!-- Source: Over-complete two-score parameterization and reduction to $w=w_1-w_2$. -->

<details>
<summary>Answer</summary>

The classifier predicts class $1$ exactly when the score for class $1$ is at least as large as the score for class $2$:

$$
s_1(x)\ge s_2(x).
$$

Substituting the definitions of the two scores gives

$$
w_1^\top x\ge w_2^\top x.
$$

Moving all terms to one side,

$$
w_1^\top x-w_2^\top x\ge 0.
$$

Factoring out $x$ gives

$$
(w_1-w_2)^\top x\ge 0.
$$

Therefore, if we define

$$
w=w_1-w_2,
$$

then the same decision rule is

$$
\text{predict class }1 \text{ if } w^\top x\ge 0,
$$

and predict class $2$ otherwise.

Thus the two-score classifier depends only on the difference between the two weight vectors, $w_1-w_2$. The separate vectors $w_1$ and $w_2$ are an over-parameterization for binary classification, since the decision boundary is determined by the single vector

$$
w=w_1-w_2.
$$

</details>


### Question 11

Use monotonicity of the sigmoid to prove the equivalence between predicting class 1 based on (1) the probability-threshold rule and (2) the score-threshold rule, i.e., that the two are equivalent:
$$
p_w(x)\ge \frac12 \quad\Longleftrightarrow\quad w^\top x\ge0.
$$

<!-- Source: Prediction rule $p_w(x)\ge1/2$ and boundary $w^\top x=0$. -->

<details>
<summary>Answer</summary>

Since $p_w(x)=\sigma(w^\top x)$, solve $\sigma(t)\ge1/2$:
$$
\frac{1}{1+e^{-t}}\ge\frac12
\iff 2\ge1+e^{-t}
\iff 1\ge e^{-t}
\iff t\ge0.
$$
Taking $t=w^\top x$ proves the equivalence.

</details>


### Question 12

Describe the decision boundary in $\mathbb{R}^2$ when an intercept is included. If $\tilde x=(1,x_1,x_2)^\top$ and $w=(b,w_1,w_2)^\top$, write the decision boundary equation and solve it for $x_2$ when $w_2\ne0$.

<!-- Source: Linear decision boundary $\hat w^\top x=0$, including the intercept column used in examples. -->

<details>
<summary>Answer</summary>

The boundary is
$$
b+w_1x_1+w_2x_2=0.
$$
If $w_2\ne0$, then
$$
x_2=-\frac{b+w_1x_1}{w_2}
=-\frac{w_1}{w_2}x_1-\frac{b}{w_2}.
$$
This is a line in the original two-dimensional input space.

</details>


### Question 13

### Question 15

Determine how the decision rule changes when the probability threshold is not $1/2$. For fixed $\tau\in(0,1)$, derive an equivalent condition on the score $w^\top x$ for predicting class $1$ under the rule $p_w(x)\ge \tau$.

<!-- Source: Thresholding probabilities and the special $1/2$ threshold used in the lecture. -->

<details>
<summary>Answer</summary>

The logistic regression probability is

$$
p_w(x)=\sigma(w^\top x).
$$

We predict class $1$ when

$$
p_w(x)\ge \tau.
$$

Substituting $p_w(x)=\sigma(w^\top x)$ gives

$$
\sigma(w^\top x)\ge \tau.
$$

Since the sigmoid function is increasing, we can apply $\sigma^{-1}$ to both sides without changing the direction of the inequality:

$$
w^\top x\ge \sigma^{-1}(\tau).
$$

Now solve for the inverse sigmoid. Let

$$
\tau=\sigma(t)=\frac{1}{1+e^{-t}}.
$$

Then

$$
\tau(1+e^{-t})=1.
$$

So

$$
\tau e^{-t}=1-\tau.
$$

Dividing by $\tau$ gives

$$
e^{-t}=\frac{1-\tau}{\tau}.
$$

Taking logs,

$$
-t=\log\left(\frac{1-\tau}{\tau}\right).
$$

Therefore,

$$
t=\log\left(\frac{\tau}{1-\tau}\right).
$$

Thus

$$
\sigma^{-1}(\tau)=\log\left(\frac{\tau}{1-\tau}\right).
$$

So the decision rule is:

$$
\text{predict class }1 \text{ if } w^\top x\ge \log\left(\frac{\tau}{1-\tau}\right).
$$

When $\tau=1/2$, this threshold is

$$
\log\left(\frac{1/2}{1-1/2}\right)=\log(1)=0.
$$

So the usual threshold $\tau=1/2$ gives the familiar rule: predict class $1$ when $w^\top x\ge 0$.

</details>

### Question 16

Connect logistic regression to a probabilistic model. Suppose that, for each observation,

$$
Y_n\mid x_n\sim \operatorname{Bernoulli}(p_n),
\qquad p_n=\sigma(x_n^\top w),
$$

independently across $n$. This means the model assigns probability $p_n$ to the event $Y_n=1$ and probability $1-p_n$ to the event $Y_n=0$.

Write down the joint probability mass function for this model. Then take the negative log of this probability and compare the result to the empirical cross-entropy objective.

<!-- Source: The lecture's cautious probabilistic interpretation of $p_w(x)$ and cross-entropy loss. -->

<details>
<summary>Answer</summary>

For one observation, the Bernoulli model assigns probability

$$
p_n^{y_n}(1-p_n)^{1-y_n}
$$

to the observed label $y_n$.

Since the observations are assumed independent, the probability assigned to the full observed label vector is the product

$$
\prod_{n=1}^N p_n^{y_n}(1-p_n)^{1-y_n}.
$$

A common way to turn this probability into a loss is to take the negative log. This gives

$$
-\log\left(\prod_{n=1}^N p_n^{y_n}(1-p_n)^{1-y_n}\right).
$$

Using the fact that the log of a product is the sum of the logs, we get

$$
-\sum_{n=1}^N \log\left(p_n^{y_n}(1-p_n)^{1-y_n}\right).
$$

Using the fact that $\log(a^b)=b\log(a)$, this becomes

$$
-\sum_{n=1}^N \left[y_n\log p_n+(1-y_n)\log(1-p_n)\right].
$$

The empirical cross-entropy risk is this same quantity divided by $N$:

$$
\widehat{R}(w)=-\frac{1}{N}\sum_{n=1}^N \left[y_n\log p_n+(1-y_n)\log(1-p_n)\right].
$$

Thus, minimizing empirical cross-entropy is the same as choosing the parameters that make the observed labels as probable as possible under the Bernoulli model. The factor $1/N$ changes the scale of the objective but not the minimizer.

</details>
