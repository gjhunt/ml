### Question 1

In binary logistic regression, answer the following about the sigmoid map: what is the relevant probability, loss, gradient, or decision-rule formula, and why does it matter?

<details>
<summary>Answer</summary>

The key object is the Bernoulli model $p_w(x)=\sigma(w^Tx)$. For the sigmoid map, the negative log-likelihood gives a convex score-based loss with gradient $X^T(p-y)$; decisions are obtained later by thresholding probabilities or scores.
