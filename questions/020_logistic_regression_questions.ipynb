### Question 1

Starting from
$$
\sigma(t)=\frac{1}{1+e^{-t}},
$$
compute $\sigma'(t)$ and express the answer only in terms of $\sigma(t)$. Then determine where the derivative is largest and give its largest value.

<!-- Source: Sigmoid definition and the lecture claim $\sigma'(t)=\sigma(t)(1-\sigma(t))$. -->

<details>
<summary>Answer</summary>

Differentiate directly:
$$
\sigma'(t)=\frac{e^{-t}}{(1+e^{-t})^2}.
$$
Since $\sigma(t)=1/(1+e^{-t})$ and $1-\sigma(t)=e^{-t}/(1+e^{-t})$,
$$
\sigma'(t)=\sigma(t)(1-\sigma(t)).
$$
Writing $u=\sigma(t)\in(0,1)$, the derivative is $u(1-u)$, maximized at $u=1/2$. The largest value is $1/4$, attained at $t=0$.

</details>


### Question 2

Verify the algebraic simplification of the binary cross-entropy loss. For $y\in\{0,1\}$, $s\in\mathbb{R}$, and $p=\sigma(s)$, derive
$$
-\{y\log p+(1-y)\log(1-p)\}=\log(1+e^s)-ys.
$$

<!-- Source: Cross-entropy loss and the compressed simplification to $\log(1+e^s)-ys$. -->

<details>
<summary>Answer</summary>

Use $p=e^s/(1+e^s)$ and $1-p=1/(1+e^s)$. Then
$$
\log p=s-\log(1+e^s),\qquad \log(1-p)=-\log(1+e^s).
$$
Substitution gives
$$
-\{y(s-\log(1+e^s))+(1-y)(-\log(1+e^s))\}
=\log(1+e^s)-ys.
$$

</details>


### Question 3

For the two labels separately, compute $\ell(1,s)$ and $\ell(0,s)$ from
$$
\ell(y,s)=\log(1+e^s)-ys.
$$
State the limits of each loss as $s\to\infty$ and $s\to -\infty$, and identify which limits correspond to confident mistakes.

<!-- Source: Understanding the loss through the cases $y=1$ and $y=0$. -->

<details>
<summary>Answer</summary>

For $y=1$,
$$
\ell(1,s)=\log(1+e^s)-s=\log(1+e^{-s}).
$$
Thus $\ell(1,s)\to0$ as $s\to\infty$, and $\ell(1,s)\to\infty$ as $s\to-\infty$. The second limit is a confident mistake.

For $y=0$,
$$
\ell(0,s)=\log(1+e^s).
$$
Thus $\ell(0,s)\to\infty$ as $s\to\infty$, and $\ell(0,s)\to0$ as $s\to-\infty$. The first limit is a confident mistake.

</details>


### Question 4

Differentiate the single-observation loss
$$
\ell_n(w)=-\{y_n\log p_n+(1-y_n)\log(1-p_n)\},
\qquad p_n=\sigma(x_n^\top w),
$$
with respect to $w$. Give the chain-rule calculation leading to $\nabla_w\ell_n=(p_n-y_n)x_n$.

<!-- Source: Gradient derivation for one summand, where the lecture says the algebra simplifies nicely. -->

<details>
<summary>Answer</summary>

First,
$$
\frac{\partial \ell_n}{\partial p_n}
=-\frac{y_n}{p_n}+\frac{1-y_n}{1-p_n}
=\frac{p_n-y_n}{p_n(1-p_n)}.
$$
Also,
$$
\nabla_w p_n=\sigma'(x_n^\top w)x_n=p_n(1-p_n)x_n.
$$
Therefore
$$
\nabla_w\ell_n
=\frac{p_n-y_n}{p_n(1-p_n)}p_n(1-p_n)x_n
=(p_n-y_n)x_n.
$$

</details>


### Question 5

Translate the componentwise gradient into matrix form. Given $X\in\mathbb{R}^{N\times D}$, $p=\sigma(Xw)$, and $y\in\mathbb{R}^N$, verify the dimensions and derive
$$
\nabla \hat R(w)=\frac{1}{N}X^\top(p-y).
$$

<!-- Source: Vector-form empirical risk and the matrix gradient $X^\top(p-y)/N$. -->

<details>
<summary>Answer</summary>

The componentwise gradient is
$$
\frac1N\sum_{n=1}^N(p_n-y_n)x_n.
$$
Here $p-y\in\mathbb{R}^N$ and $X^\top\in\mathbb{R}^{D\times N}$, so $X^\top(p-y)\in\mathbb{R}^D$. Its $j$-th coordinate is
$$
\sum_{n=1}^N X_{nj}(p_n-y_n),
$$
which is exactly the $j$-th coordinate of the summation.

</details>


### Question 6

Fill in the Hessian derivation. Starting from
$$
\nabla \hat R(w)=\frac{1}{N}X^\top(p-y),\qquad p=\sigma(Xw),
$$
show that
$$
\nabla^2\hat R(w)=\frac{1}{N}X^\top W X,
\quad
W=\operatorname{diag}(p_1(1-p_1),\ldots,p_N(1-p_N)).
$$

<!-- Source: Hessian formula and the diagonal weight matrix $W$. -->

<details>
<summary>Answer</summary>

Let $z=Xw$. The Jacobian of $p=\sigma(z)$ with respect to $z$ is $W$, because the sigmoid is applied elementwise. Since $z=Xw$, the Jacobian of $p$ with respect to $w$ is $WX$. Differentiating
$$
\frac1N X^\top(p-y)
$$
therefore gives
$$
\nabla^2\hat R(w)=\frac1N X^\top W X.
$$

</details>


### Question 7

Classify the Hessian. For arbitrary $X$ and $w$, prove that
$$
H(w)=\frac{1}{N}X^\top W X
$$
is positive semidefinite. Then state a sufficient condition under which it is positive definite.

<!-- Source: Convexity implication of the Hessian form $X^\top W X/N$. -->

<details>
<summary>Answer</summary>

For any $v\in\mathbb{R}^D$,
$$
v^\top H(w)v=\frac1N(Xv)^\top W(Xv)
=\frac1N\sum_{n=1}^N p_n(1-p_n)(x_n^\top v)^2\ge0.
$$
Thus $H(w)$ is positive semidefinite. Since $p_n\in(0,1)$, all diagonal weights are positive; $H(w)$ is positive definite if $X$ has full column rank.

</details>


### Question 8

Identify the missing indexing correction in the lecture's first ERM display. Rewrite the empirical risk using observation-specific variables $(x_n,y_n)$ and the correct summation range.

<!-- Source: The ERM display has compressed/incorrect indexing such as $\sum_{i=n}^N$ and unindexed $x,y$. -->

<details>
<summary>Answer</summary>

The corrected empirical risk is
$$
\hat R(w)=\frac1N\sum_{n=1}^N
-\Big(y_n\log\sigma(w^\top x_n)+(1-y_n)\log(1-\sigma(w^\top x_n))\Big).
$$
Equivalently, with $p_n=\sigma(x_n^\top w)$,
$$
\hat R(w)=-\frac1N\sum_{n=1}^N\{y_n\log p_n+(1-y_n)\log(1-p_n)\}.
$$

</details>


### Question 9

Why does multiplying the empirical risk by $N$ not change the optimizer? Give a precise statement involving $\arg\min$, and mention one condition needed for the statement to make sense.

<!-- Source: Lecture remark that the factor $1/N$ does not affect the minimizer. -->

<details>
<summary>Answer</summary>

If $a>0$, then
$$
\arg\min_w f(w)=\arg\min_w af(w),
$$
because $f(u)\le f(v)$ if and only if $af(u)\le af(v)$. Here $a=N>0$. The two objectives must be defined on the same feasible set; if a minimizer exists for one, the same points minimize the other.

</details>


### Question 10

Compare the logistic-regression gradient with the least-squares gradient. For least squares with objective $N^{-1}\|Xw-y\|_2^2/2$, the gradient is $N^{-1}X^\top(Xw-y)$. State exactly what changes in logistic regression and why the replacement is not merely cosmetic.

<!-- Source: Comparison between residuals $Xw-y$ and logistic residuals $p-y$. -->

<details>
<summary>Answer</summary>

In logistic regression,
$$
\nabla \hat R(w)=\frac1N X^\top(p-y),\qquad p=\sigma(Xw).
$$
The linear fitted values $Xw$ are replaced by fitted probabilities $p$. This is not cosmetic because $p$ depends nonlinearly on $w$, so the objective is not quadratic and
$$
X^\top(\sigma(Xw)-y)=0
$$
does not yield normal equations solvable by linear algebra.

</details>


### Question 11

Diagnose the false claim: "Since logistic regression has gradient $X^\top(p-y)/N$, the minimizer is obtained by setting $p=y$, hence $w=X^{-1}y$ when $X$ is square." Identify two distinct mathematical errors in this argument.

<!-- Source: No-closed-form discussion and comparison with least-squares normal equations. -->

<details>
<summary>Answer</summary>

First, $X^\top(p-y)=0$ does not generally imply $p=y$; it only says $p-y$ is orthogonal to the columns of $X$. Second, even $p=y$ would mean $\sigma(Xw)=y$, not $Xw=y$. Exact binary probabilities $0$ or $1$ require infinite logits, so $w=X^{-1}y$ is not a valid logistic-regression solution.

</details>

### Question 14

Find the Newton direction in one dimension. For one data point with $x=1$, $y=1$, and current parameter $w=0$, compute $p$, the gradient, the Hessian, and the Newton update $w_{\mathrm{new}}=w-H(w)^{-1}\nabla\hat R(w)$.

<!-- Source: Newton update and Hessian formula in a simple special case. -->

<details>
<summary>Answer</summary>

Here $p=\sigma(0)=1/2$. With $N=1$,
$$
\nabla\hat R(w)=(p-y)x=(1/2-1)=-1/2.
$$
The Hessian is
$$
H(w)=p(1-p)x^2=(1/2)(1/2)=1/4.
$$
Thus
$$
w_{\mathrm{new}}=0-(1/4)^{-1}(-1/2)=2.
$$

</details>


### Question 15

Relate Newton's method to a quadratic approximation. Starting from
$$
q(\Delta)=\hat R(w)+\nabla\hat R(w)^\top\Delta+\frac12\Delta^\top H(w)\Delta,
$$
with $H(w)$ invertible, minimize $q$ over $\Delta$ and obtain the Newton step.

<!-- Source: Taylor approximation derivation of Newton's method. -->

<details>
<summary>Answer</summary>

Differentiate with respect to $\Delta$:
$$
\nabla_\Delta q(\Delta)=\nabla\hat R(w)+H(w)\Delta.
$$
Setting this equal to zero gives
$$
H(w)\Delta=-\nabla\hat R(w),
\qquad
\Delta^\star=-H(w)^{-1}\nabla\hat R(w).
$$
Therefore
$$
w\leftarrow w+\Delta^\star=w-H(w)^{-1}\nabla\hat R(w).
$$

</details>


### Question 16

State where invertibility enters Newton's method for logistic regression. Give a concrete example of a design matrix $X$ for which $X^\top W X$ is singular for every $w$, even though all diagonal entries of $W$ are positive.

<!-- Source: Newton formula requiring $(X^\top W X)^{-1}$. -->

<details>
<summary>Answer</summary>

Invertibility is needed in
$$
(X^\top W X)^{-1}X^\top(p-y).
$$
For example,
$$
X=\begin{pmatrix}1&2\\2&4\end{pmatrix}
$$
has rank $1$. For any positive diagonal $W$,
$$
\operatorname{rank}(X^\top W X)\le \operatorname{rank}(X)=1<2,
$$
so $X^\top W X$ is singular.

</details>


### Question 17

Connect the one-score and two-score binary classifiers. If
$$
s_1(x)=w_1^\top x,
\qquad s_2(x)=w_2^\top x,
$$
and prediction is by $\arg\max\{s_1(x),s_2(x)\}$, express the same decision rule using one vector $w$. Also describe one transformation of $(w_1,w_2)$ that leaves every prediction unchanged.

<!-- Source: Over-complete two-score parameterization and reduction to $w=w_1-w_2$. -->

<details>
<summary>Answer</summary>

Predict class $1$ exactly when
$$
s_1(x)\ge s_2(x)
\iff (w_1-w_2)^\top x\ge0.
$$
Thus $w=w_1-w_2$. The transformation
$$
(w_1,w_2)\mapsto(w_1+a,w_2+a)
$$
for any vector $a$ leaves every score difference, and hence every prediction, unchanged.

</details>


### Question 18

Use monotonicity of the sigmoid to prove the equivalence between the probability-threshold rule and the score-threshold rule:
$$
p_w(x)\ge \frac12 \quad\Longleftrightarrow\quad w^\top x\ge0.
$$

<!-- Source: Prediction rule $p_w(x)\ge1/2$ and boundary $w^\top x=0$. -->

<details>
<summary>Answer</summary>

Since $p_w(x)=\sigma(w^\top x)$, solve $\sigma(t)\ge1/2$:
$$
\frac{1}{1+e^{-t}}\ge\frac12

### Question 19

### Question 21

Give a precise counterexample to the claim: "Logistic regression always has a finite empirical-risk minimizer." Use a dataset with one-dimensional features and binary labels, and show why the infimum is approached but not attained.

<!-- Source: Perfect linear separability and the claim that no finite minimum is achieved. -->

<details>
<summary>Answer</summary>

Use $(x_1,y_1)=(1,1)$ and $(x_2,y_2)=(-1,0)$. For $w=c>0$, the scores are $c$ and $-c$. The losses are
$$
\ell(1,c)=\log(1+e^{-c}),
\qquad
\ell(0,-c)=\log(1+e^{-c}),
$$
so the empirical risk tends to $0$ as $c\to\infty$. No finite $w$ attains risk $0$, since logistic probabilities are strictly between $0$ and $1$.

</details>


### Question 22

Locate exactly where strict separability is used in the scaling argument. Assume $y_i=1\Rightarrow w^\top x_i>0$ and $y_i=0\Rightarrow w^\top x_i<0$. For $c\to\infty$, prove that each individual loss $\ell(y_i,cw^\top x_i)\to0$.

<!-- Source: Separable-data derivation with scaling $w\mapsto cw$. -->

<details>
<summary>Answer</summary>

If $y_i=1$, set $a_i=w^\top x_i>0$. Then
$$
\ell(1,ca_i)=\log(1+e^{ca_i})-ca_i=\log(1+e^{-ca_i})\to0.
$$
If $y_i=0$, set $a_i=w^\top x_i<0$. Then
$$
\ell(0,ca_i)=\log(1+e^{ca_i})\to\log(1+0)=0.
$$
Strict inequalities force the scaled scores to $+\infty$ or $-\infty$.

</details>

### Question 26

Determine how the decision rule changes when the threshold is not $1/2$. For fixed $\tau\in(0,1)$, derive an equivalent condition on the score $w^\top x$ for predicting class $1$ under the rule $p_w(x)\ge\tau$.

<!-- Source: Thresholding probabilities and the special $1/2$ threshold used in the lecture. -->

Since $\sigma$ is increasing,
$$
\sigma(w^\top x)\ge\tau
\iff

Solving $\tau=1/(1+e^{-t})$ gives

t=\log\frac{\tau}{1-	au}.

w^\top x\ge\log\frac{\tau}{1-	au}.

### Question 27

Interpret the sign of a single gradient contribution. For an observation with $y_n=1$, explain precisely why $(p_n-y_n)x_n$ points in a direction that gradient descent uses to increase $x_n^\top w$. Repeat the sign analysis for $y_n=0$.

<!-- Source: Gradient contribution $(p_n-y_n)x_n$ and its classification interpretation. -->

<details>
<summary>Answer</summary>

For $y_n=1$, $p_n-y_n=p_n-1<0$, so the contribution to the gradient is a negative multiple of $x_n$. Gradient descent subtracts the gradient, moving $w$ in the positive $x_n$ direction and increasing $x_n^\top w$.

For $y_n=0$, $p_n-y_n=p_n>0$, so gradient descent moves $w$ in the negative $x_n$ direction and decreases $x_n^\top w$.

</details>


### Question 28

Show why squared error composed with the sigmoid can be non-convex. For one observation with $y=1$ and score $s$, define
$$
g(s)=(1-\sigma(s))^2.
$$
Find an interval of $s$-values on which $g''(s)<0$, and conclude that $g$ is not convex on all of $\mathbb{R}$.

<!-- Source: Lecture claim that squared error with the sigmoid leads to a non-convex optimization problem. -->

<details>
<summary>Answer</summary>

Let $p=\sigma(s)$. Then
$$
g'(s)=-2p(1-p)^2.
$$
Using $p'=p(1-p)$,
$$
g''(s)=-2p(1-p)^2(1-3p).
$$
If $p<1/3$, equivalently $s<\log(1/2)=-\log2$, then $g''(s)<0$. Therefore $g$ is not convex on all of $\mathbb{R}$.

</details>


### Question 29

Find the missing probabilistic interpretation. If one assumes
$$
Y_n\mid x_n\sim \operatorname{Bernoulli}(p_n),
\qquad p_n=\sigma(x_n^\top w),
$$
independently across $n$, derive the negative log-likelihood and compare it to the empirical cross-entropy objective.

<!-- Source: The lecture's cautious probabilistic interpretation of $p_w(x)$ and cross-entropy loss. -->

<details>
<summary>Answer</summary>

The likelihood is
$$
L(w)=\prod_{n=1}^N p_n^{y_n}(1-p_n)^{1-y_n}.
$$
Therefore
$$
-\log L(w)=-\sum_{n=1}^N\{y_n\log p_n+(1-y_n)\log(1-p_n)\}.
$$
The empirical cross-entropy risk is this same quantity divided by $N$, so minimizing it is equivalent to maximum likelihood under the Bernoulli model.

</details>


### Question 30

Check a claimed weighted least-squares form of the Newton step. With $W=\operatorname{diag}(p_n(1-p_n))$, show that the Newton update solves
$$
(X^\top W X)w^{(t+1)}=X^\top W z^{(t)}
$$
for a suitable working response $z^{(t)}$. Derive $z^{(t)}$.

<!-- Source: Fun fact connecting Newton's method to IRLS/weighted least squares. -->

<details>
<summary>Answer</summary>

Ignoring the common factor $1/N$, Newton's update is
$$
w^{(t+1)}=w^{(t)}-(X^\top W X)^{-1}X^\top(p-y).
$$
Multiplying by $X^\top W X$,
$$
X^\top W Xw^{(t+1)}=X^\top W Xw^{(t)}-X^\top(p-y).
$$
Since $X^\top W Xw^{(t)}=X^\top W(Xw^{(t)})$,
$$
X^\top W Xw^{(t+1)}=X^\top W\{Xw^{(t)}+W^{-1}(y-p)\}.
$$
Thus
$$
z^{(t)}=Xw^{(t)}+W^{-1}(y-p).
$$

</details>
