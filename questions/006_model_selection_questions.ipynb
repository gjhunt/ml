# Model Selection - Review Questions {#sec-model-sel-questions}



### Question 1

For a fixed learning algorithm $A$, we train
$$
\hat f = A(D_{train})
$$
and evaluate performance on an independent evaluation set $D_{eval}$. What changes when we move from a fixed algorithm $A$ to fitting over a family $\{A_\lambda:\lambda\in\Lambda\}$ and choosing the best candidate?

<details>
<summary>Answer</summary>

For a fixed algorithm $A$, the evaluation set is independent of the fitted model $\hat f=A(D_{train})$, so the evaluation error estimates performance on new data.

In model selection, each fixed candidate $\hat f_\lambda$ can still be evaluated on $D_{val}$ because $\hat f_\lambda$ was trained only on $D_{train}$. The problem arises when we choose
$$
\hat\lambda=\arg\min_{\lambda\in\Lambda}e_\lambda.
$$

Now $\hat\lambda$ depends on $D_{val}$. Therefore the selected model, $\hat f_{\hat\lambda}$, also depends on $D_{val}$. The validation set has become part of the model-building procedure, not just an independent evaluation set.

As a result, the selected validation error $e_{\hat\lambda}$ is typically optimistically biased. It is the minimum of several noisy estimates, so it tends to make the selected model look better than it really is. To estimate final performance honestly, we need a separate test set that was not used for either fitting or selecting $\lambda$.

</details>

### Question 2

For three candidate polynomial degrees ($\lambda$), the validation MSEs $e_\lambda$ are

| $\lambda$ | 1 | 3 | 5 |
| ---: | ---: | ---: | ---: |
| $e_\lambda$ | 4.8 | 2.1 | 2.7 |

Using the lecture's loss-minimization rule, compute $\hat\lambda$. 

<details>
<summary>Answer</summary>

The selected degree is
$$
\hat\lambda=\arg\min_{\lambda\in\{1,3,5\}}e_\lambda=3,
$$
because $2.1$ is the smallest validation MSE.


</details>


### Question 3

For a fixed $\lambda$, why can $\hat{\mathrm{Perf}}(D_{val},\hat f_\lambda)$ still be a valid estimate of that candidate's performance, even though the selected validation error $\hat{\mathrm{Perf}}(D_{val},\hat f_{\hat\lambda})$ is typically optimistic?

<details>
<summary>Answer</summary>

For a fixed $\lambda$, the model $\hat f_\lambda=A_\lambda(D_{train})$ is trained without using $D_{val}$. If $D_{val}$ is an independent sample from the target distribution, then $D_{val}\perp \hat f_\lambda$, so validation performance estimates fresh-data performance for that fixed candidate.

After selection, however, $\hat\lambda$ is chosen because it performed well on $D_{val}$. The reported selected validation error is no longer the error of a candidate fixed in advance; it is the minimum over several noisy validation estimates. This selection step creates dependence between $D_{val}$ and the final chosen model, making the selected validation error optimistic.

</details>


### Question 4

The lecture says the difference between tuning weights $w$ inside an algorithm and choosing a hyperparameter $\lambda$ is less important than it may seem. Why? Give a reason in terms of the final model-building procedure.

<details>
<summary>Answer</summary>

Both operations are data-dependent choices that affect the final fitted model. Tuning weights $w$ uses training data to choose parameters within a model class. Choosing $\lambda$ uses validation or cross-validation data to choose among model classes, complexities, or training rules. In both cases, the output is a function of data.

The relevant object is therefore the whole pipeline $\mathcal P(D)$, not just an isolated algorithm call. If the pipeline uses data to choose $w$, $\lambda$, preprocessing choices, or model classes, those choices are part of the learning procedure that produced $\hat f$.

</details>


### Question 5

In a three-way split, assign each dataset to its role: $D_{train}$, $D_{val}$, and $D_{test}$. For each one, state whether it is allowed to influence the fitted candidate models, the selected $\hat\lambda$, and the final reported performance estimate.

<details>
<summary>Answer</summary>

$D_{train}$ is used to fit candidate models $\hat f_\lambda=A_\lambda(D_{train})$. It influences fitted candidate models, but it should not be used as the final performance estimate.

$D_{val}$ is used to choose $\hat\lambda$ by comparing candidate validation errors. It influences the selected model through $\hat\lambda$, so it is part of model building and should not be used as the final unbiased evaluation set.

$D_{test}$ is held out until the end. It should not influence candidate fitting or $\hat\lambda$. Its role is to provide the final reported estimate $\hat{\mathrm{Perf}}(D_{test},\hat f)$.

</details>


### Question 5

A colleague trains several models, chooses the one with the smallest test error, and then reports that same test error as the final performance estimate. Diagnose a potential problem with this logic. 

<details>
<summary>Answer</summary>

The flaw is that the test set has been used for model selection. Once test performance determines which model is chosen, the test set is no longer a pure final evaluation set; it has become validation data. The final selected model depends on the test set through the choice of model.

</details>


### Question 6

Why should we evaluate on $D_{test}$ only once, at the end? What would happen if the analyst repeatedly used $D_{test}$ to make decisions.

<details>
<summary>Answer</summary>

The test set is useful for final evaluation only if it remains independent of the final model-building procedure. If the analyst repeatedly checks test performance and changes models, hyperparameters, preprocessing, or features in response, then the final model depends on the test set. It no longer provides an independent estimate of fresh-data performance, and the final reported test error can become optimistically biased.

</details>


### Question 7

True or False: Since the validation set $D_{val}$ was not used to fit any individual candidate model $\hat f_\lambda$, it is independent of the final selected model $\hat f_{\hat\lambda}$.

<details>
<summary>Answer</summary>

False.

It is true that, for each fixed $\lambda$, the candidate model
$$
\hat f_\lambda=A_\lambda(D_{train})
$$
was fit without using $D_{val}$.

The false step is concluding that $D_{val}$ is therefore independent of the final selected model. The validation set is used to choose
$$
\hat\lambda=\arg\min_{\lambda\in\Lambda}\hat{\mathrm{Perf}}(D_{val},\hat f_\lambda).
$$

So $\hat\lambda$ depends on $D_{val}$. Since the final selected model is $\hat f_{\hat\lambda}$, the final model can also depend on $D_{val}$, even though no individual candidate was directly fit using $D_{val}$.

</details>

### Question 8

The lecture says selection bias is worse when $\Lambda$ is large or searched aggressively. Why?

<details>
<summary>Answer</summary>

Each validation score $e_\lambda$ is a noisy estimate of the corresponding candidate's true fresh-data performance. When the procedure chooses $\hat\lambda$ by minimizing $e_\lambda$, it tends to pick a candidate whose validation estimate is low partly because of genuine performance and partly because of favorable noise. As $\Lambda$ grows or the search becomes more aggressive, there are more noisy estimates from which to choose. The minimum of many noisy estimates is more likely to be unusually low, so the selected validation error becomes more optimistic.

</details>


### Question 9

For two candidates $\lambda=1,2$, suppose their true performances are equal to $1$, but their validation errors are
$$
e_1=1+\varepsilon_1,\qquad e_2=1+\varepsilon_2,
$$
where $\varepsilon_1$ and $\varepsilon_2$ are independent and take values $-0.2$ and $0.2$ with probability $1/2$ each. Compute $\mathbb E[\min(e_1,e_2)]$ and compare it with the true performance $1$.

<details>
<summary>Answer</summary>

The possible pairs of errors are equally likely:

- $(-0.2,-0.2)$ gives $\min(e_1,e_2)=0.8$.
- $(-0.2,0.2)$ gives $\min(e_1,e_2)=0.8$.
- $(0.2,-0.2)$ gives $\min(e_1,e_2)=0.8$.
- $(0.2,0.2)$ gives $\min(e_1,e_2)=1.2$.

Therefore
$$
\mathbb E[\min(e_1,e_2)]
=\frac14(0.8+0.8+0.8+1.2)=0.9.
$$
Even though both candidates have true performance $1$ and error distributed symmetrically around zero, the expected selected validation error is $0.9$, which is optimistically low.

</details>


### Question 10

Suppose we are doing polynomial model selection. For each degree $d$, let $\hat f_d$ be the polynomial regression model of degree $d$ fit on the training set $D_{train}$. Let

$$
e_d=\hat{\mathrm{MSE}}(D_{val},\hat f_d)
$$

be the validation mean squared error for degree $d$.

Now suppose we choose not only the polynomial degree, but also how large a degree-search space to allow. For a maximum allowed degree $K$, define

$$
\hat d_K=\arg\min_{0\le d\le K} e_d.
$$

The selected validation error for this search space is

$$
e_{\hat d_K}=\min_{0\le d\le K} e_d.
$$

Show that the selected validation error $e_{\hat d_K}$ is nonincreasing as $K$ increases. In other words, explain why allowing yourself to search over degrees $0,\dots,K+1$ cannot produce a larger selected validation error than searching only over degrees $0,\dots,K$.

<details>
<summary>Answer</summary>

When $K$ increases to $K+1$, the search set expands from

$$
\{0,\dots,K\}
$$

to

$$
\{0,\dots,K,K+1\}.
$$

The old selected degree $\hat d_K$ is still available in the larger search set. Therefore, the minimum over the larger set cannot be greater than the minimum over the smaller set:

$$
\min_{0\le d\le K+1} e_d
\le
\min_{0\le d\le K} e_d.
$$

Equivalently,

$$
e_{\hat d_{K+1}}\le e_{\hat d_K}.
$$

So the selected validation error is nonincreasing as $K$ increases.

This does not mean that true test performance must improve as $K$ increases. It only means that the best validation error found inside a larger search space cannot get worse, because the larger search space contains all the old options plus one additional degree.

</details>

### Question 11

Does the monotonicity in Question 19 imply that test error of the selected model must also be nonincreasing in $K$? 

<details>
<summary>Answer</summary>

No. The monotonicity applies to the selected validation error because the procedure explicitly minimizes validation error over an expanding set. Test error is not used in the selection rule. As $K$ increases, the selected degree may change to one with lower validation error but higher test error. 

</details>


### Question 12

Nested cross-validation uses two loops: an inner loop and an outer loop. What is the purpose of each loop? 

<details>
<summary>Answer</summary>

The inner loop is used for model selection. Within each outer training set, the inner cross-validation loop compares candidate models or hyperparameter values and chooses the one with the best inner validation performance.

The outer loop is used for performance estimation. In each outer split, the entire model-selection procedure is run using only the outer training data. The selected model is then evaluated on the outer test fold, which was not used for either fitting candidates or choosing hyperparameters.

Thus, the inner loop answers:

> Which model or hyperparameter setting should we choose?

The outer loop answers:

> How well does the full selection-and-fitting procedure perform on new data?

</details>



### Question 13

In nested CV with $N=120$ observations, $k_{outer}=4$, and $k_{inner}=3$, assume folds are equal size. For one outer fold, how many observations are in $D_{test}^{(j)}$ and $D_{train\text{-}val}^{(j)}$? Inside that outer training set, how many observations are used to train each inner-fold candidate model?

<details>
<summary>Answer</summary>

With $k_{outer}=4$, each outer test fold has
$$
120/4=30
$$
observations. The corresponding outer training-validation set has
$$
120-30=90
$$
observations.

Inside that set, $k_{inner}=3$ means each inner validation fold has $90/3=30$ observations, so each inner candidate model is trained on
$$
90-30=60
$$
observations.

</details>


### Question 14

A data scientist claims: "Nested CV estimates the performance of the exact final model I will train using all $N$ observations in the data set." Is this true?

<details>
<summary>Answer</summary>

The claim is not quite correct. Nested CV estimates the performance of the full model-selection procedure, when that procedure is applied to training samples of the outer-training size. Each outer fold runs the entire inner selection procedure on $D\setminus D_j$ and evaluates the resulting model on $D_j$. The outer average is therefore evidence about the procedure's expected fresh-data performance, not the exact conditional performance of the one final model later trained on all $N$ observations.

</details>


### Question 15

Why is it reasonable, after choosing $\hat\lambda$ using $D_{val}$, to retrain the final model on $D_{train}\cup D_{val}$ before evaluating on $D_{test}$?

<details>
<summary>Answer</summary>

After $\hat\lambda$ has been chosen, $D_{val}$ has already served its role in model building. Retraining on $D_{train}\cup D_{val}$ uses more non-test data to fit the final model, which can improve the final model because it has more training observations. This does not invalidate the final test evaluation as long as $D_{test}$ remains untouched by training and selection. 

</details>


### Question 16

An analyst has a dataset and wants to train and evaluate a predictive model. Before splitting the data into training, validation, and test sets, the analyst selects the top 20 features by computing each feature's correlation with the response using all observations. The analyst then splits the data, trains several candidate models using the selected features, chooses among them using validation error, and reports the final test error.

Does data leakage occur in this pipeline? 

<details>
<summary>Answer</summary>

The leakage occurs during feature selection.

The test set improperly influences the preprocessing step, because the top 20 features were selected using all observations, including observations from $D_{test}$. Feature selection is part of the model-building pipeline, so information from the test set has entered the construction of the final model. As a result, the final model depends on $D_{test}$ even though the model was not directly fit on $D_{test}$. The test set is no longer independent of the final model, so the reported test error can be optimistically biased.

</details>


### Question 17

Compare a validation set and a test set in one sentence each, using the criterion "is this data allowed to affect the final model?" Then explain why the names matter less than the role the data actually plays.

<details>
<summary>Answer</summary>

A validation set is allowed to affect the final model through choices such as $\hat\lambda$, model class, preprocessing, or feature selection. A test set is not allowed to affect the final model; it is reserved for final performance estimation. The names matter less than the role because a dataset called "test" becomes validation data if it is used to choose among models. Conversely, a dataset functions as a test set only if it remains unused until the final evaluation.

</details>
