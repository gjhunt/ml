### Question 1

For $f:\mathbb R^D\to\mathbb R$, define $\nabla f(x)$ through $df=\nabla f(x)^Tdx$ and state the shape of the gradient.

<!-- Source: Matrix calculus lecture; derivative identity 1 -->

<details>
<summary>Answer</summary>

For $f:\mathbb R^D\to\mathbb R$, the gradient is the vector satisfying $df=\nabla f(x)^T dx$. It represents all first-order directional derivatives.

</details>

### Question 2

Differentiate $f(x)=a^Tx+b$ by perturbing $x$ to $x+dx$ and collecting the linear term.

<!-- Source: Matrix calculus lecture; derivative identity 2 -->

<details>
<summary>Answer</summary>

If $f(x)=a^Tx+b$, then $df=a^Tdx$, so $\nabla f=a$.

</details>

### Question 3

The formula $\nabla_x(x^TAx)=2Ax$ is sometimes quoted. Determine the correct formula when $A$ is not symmetric and identify the extra assumption needed for $2Ax$.

<!-- Source: Matrix calculus lecture; derivative identity 3 -->

<details>
<summary>Answer</summary>

For $f(x)=x^TAx$, $df=dx^TAx+x^TAdx=((A+A^T)x)^Tdx$, hence $\nabla f=(A+A^T)x$.

</details>

### Question 4

With $A=A^T$, simplify the differential of $x^TAx$ to obtain the gradient.

<!-- Source: Matrix calculus lecture; derivative identity 4 -->

<details>
<summary>Answer</summary>

When $A=A^T$, the quadratic-form gradient reduces to $2Ax$.

</details>

### Question 5

For $g:\mathbb R^D\to\mathbb R^K$, write the entries and dimensions of $J_g(x)$, then verify that $dg\approx J_g(x)dx$ has the correct shape.

<!-- Source: Matrix calculus lecture; derivative identity 5 -->

<details>
<summary>Answer</summary>

For $g:\mathbb R^D\to\mathbb R^K$, $J_{ij}=\partial g_i/\partial x_j$ and $J\in\mathbb R^{K\times D}$. It maps $dx$ to $dg$ to first order.

</details>

### Question 8

### Question 10

State the Hessian matrix for $f:\mathbb R^D\to\mathbb R$ and give the second-order reason $H\succeq0$ signals local convexity.

<!-- Source: Matrix calculus lecture; derivative identity 10 -->

<details>
<summary>Answer</summary>

The Hessian of scalar $f$ has entries $\partial^2 f/\partial x_i\partial x_j$. Positive semidefiniteness is the local second-order signature of convexity.

</details>

### Question 11

A Jacobian appears transposed in two references. Use numerator versus denominator layout to explain how both conventions can be internally consistent.

<!-- Source: Matrix calculus lecture; derivative identity 11 -->

<details>
<summary>Answer</summary>

Numerator and denominator layouts transpose Jacobians relative to each other. Fixing a convention prevents chain-rule shape errors.

</details>

### Question 12

Compute the Jacobian of the affine map $g(x)=Ax+b$ and specify the dimensions of $A$, $x$, and $b$.

<!-- Source: Matrix calculus lecture; derivative identity 12 -->

<details>
<summary>Answer</summary>

For $g(x)=Ax+b$, each output row has derivative equal to the corresponding row of $A$, so $J_g=A$.

</details>

### Question 13

For $g(t)=(g_1(t),\ldots,g_K(t))$, identify the derivative object and its shape under the lecture's Jacobian convention.

<!-- Source: Matrix calculus lecture; derivative identity 13 -->

<details>
<summary>Answer</summary>

For $g(t)$, the derivative is the vector $(g_1 prime(t),\ldots,g_K prime(t))$, i.e. a $K\times1$ Jacobian.

</details>

### Question 14

Apply the product rule to $M(X)=X^TX$ and write $dM$ in terms of $dX$.

<!-- Source: Matrix calculus lecture; derivative identity 14 -->

<details>
<summary>Answer</summary>

For $X^TX$, $d(X^TX)=dX^T X+X^T dX$ by the product rule.

</details>

### Question 15

For $r(x)=Ax-b$, derive $\nabla_x\|r(x)\|_2^2$ using the chain rule through $r$.

<!-- Source: Matrix calculus lecture; derivative identity 15 -->

<details>
<summary>Answer</summary>

For $f(x)=\|Ax-b\|^2$, $\nabla f=2A^T(Ax-b)$; with the factor $1/2$, the 2 disappears.

</details>

### Question 16

Compute the Hessian of $\frac12\|Ax-b\|_2^2$ and prove it is positive semidefinite.

<!-- Source: Matrix calculus lecture; derivative identity 16 -->

<details>
<summary>Answer</summary>

For $f(x)=\frac12\|Ax-b\|^2$, $\nabla^2 f=A^TA\succeq0$.

</details>

### Question 17

If $g_i(x)=\phi(x_i)$ for a differentiable scalar $\phi$, determine every entry of $J_g(x)$.

<!-- Source: Matrix calculus lecture; derivative identity 17 -->

<details>
<summary>Answer</summary>

If $g_i(x)=\phi(x_i)$, then $J_g=\operatorname{diag}(\phi prime(x_i))$.

</details>

### Question 18

Use the definition of directional derivative to justify that a constant scalar function has zero gradient.

<!-- Source: Matrix calculus lecture; derivative identity 18 -->

<details>
<summary>Answer</summary>

A constant has zero first-order change in every direction, so its derivative object is zero.

</details>

### Question 19

Give a concrete shape-check that would catch writing $J_g\nabla f$ instead of $J_g^T\nabla f$ in the scalar-output chain rule.

<!-- Source: Matrix calculus lecture; derivative identity 19 -->

<details>

The equations $df=\nabla f^Tdx$ and $dg=Jdx$ force dimensions and reveal missing transposes.

Why does the differential identity $df=\langle \nabla_Xf,dX\rangle_F$ prevent ambiguity when optimizing over matrix parameters?

Losses often depend on vectors or matrices of parameters, so differential notation keeps optimization derivations concise.
