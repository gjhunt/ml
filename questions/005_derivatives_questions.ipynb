### Question 1

Using the lecture's treatment of the gradient of a scalar-output vector-input function, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For $f:\mathbb R^D\to\mathbb R$, the gradient is the vector satisfying $df=\nabla f(x)^T dx$. It represents all first-order directional derivatives.

</details>

### Question 2

Using the lecture's treatment of linear functions, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

If $f(x)=a^Tx+b$, then $df=a^Tdx$, so $\nabla f=a$.

</details>

### Question 3

Using the lecture's treatment of quadratic forms, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For $f(x)=x^TAx$, $df=dx^TAx+x^TAdx=((A+A^T)x)^Tdx$, hence $\nabla f=(A+A^T)x$.

</details>

### Question 4

Using the lecture's treatment of symmetric quadratic forms, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

When $A=A^T$, the quadratic-form gradient reduces to $2Ax$.

</details>

### Question 5

Using the lecture's treatment of Jacobians, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For $g:\mathbb R^D\to\mathbb R^K$, $J_{ij}=\partial g_i/\partial x_j$ and $J\in\mathbb R^{K\times D}$. It maps $dx$ to $dg$ to first order.

</details>

### Question 6

Using the lecture's treatment of the vector chain rule, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

If $h(x)=f(g(x))$, then $\nabla_x h=J_g(x)^T
abla f(g(x))$. This follows from $dh=\nabla f^T J_g dx$.

</details>

### Question 7

### Question 8

Using the lecture's treatment of linear trace functions, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

If $f(X)=\operatorname{tr}(A^TX)$, then $df=\operatorname{tr}(A^TdX)$, so $\nabla_X f=A$.

</details>

### Question 9

Using the lecture's treatment of the Frobenius norm, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For $f(X)=\frac12\|X-A\|_F^2$, $df=\langle X-A,dXangle_F$, so $\nabla_X f=X-A$.

</details>

### Question 10

Using the lecture's treatment of Hessians, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

The Hessian of scalar $f$ has entries $\partial^2 f/\partial x_i\partial x_j$. Positive semidefiniteness is the local second-order signature of convexity.

</details>

### Question 11

Using the lecture's treatment of layout conventions, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

Numerator and denominator layouts transpose Jacobians relative to each other. Fixing a convention prevents chain-rule shape errors.

</details>

### Question 12

Using the lecture's treatment of affine maps, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For $g(x)=Ax+b$, each output row has derivative equal to the corresponding row of $A$, so $J_g=A$.

</details>

### Question 13

Using the lecture's treatment of scalar-input vector-output derivatives, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For $g(t)$, the derivative is the vector $(g_1 prime(t),\ldots,g_K prime(t))$, i.e. a $K\times1$ Jacobian.

</details>

### Question 14

Using the lecture's treatment of matrix-valued product rules, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For $X^TX$, $d(X^TX)=dX^T X+X^T dX$ by the product rule.

</details>

### Question 15

Using the lecture's treatment of least-squares derivatives, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For $f(x)=\|Ax-b\|^2$, $\nabla f=2A^T(Ax-b)$; with the factor $1/2$, the 2 disappears.

</details>

### Question 16

Using the lecture's treatment of least-squares Hessians, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For $f(x)=\frac12\|Ax-b\|^2$, $\nabla^2 f=A^TA\succeq0$.

</details>

### Question 17

Using the lecture's treatment of elementwise transformations, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

If $g_i(x)=\phi(x_i)$, then $J_g=\operatorname{diag}(\phi prime(x_i))$.

</details>

### Question 18

Using the lecture's treatment of constant functions, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

A constant has zero first-order change in every direction, so its derivative object is zero.

</details>

### Question 19

Using the lecture's treatment of shape checking, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

The equations $df=\nabla f^Tdx$ and $dg=Jdx$ force dimensions and reveal missing transposes.

</details>
