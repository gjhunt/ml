### Question 5

For $g:\mathbb R^D\to\mathbb R^K$, define the Jacobian entries and explain how $J_g$ maps $dx$ to $dg$.

<details>
<summary>Answer</summary>

For $g:\mathbb R^D\to\mathbb R^K$, $J_{ij}=\partial g_i/\partial x_j$ and $J\in\mathbb R^{K\times D}$. It maps $dx$ to $dg$ to first order.

</details>

### Question 10

What is the Hessian of a scalar function, and what does positive semidefiniteness of the Hessian indicate?

<details>
<summary>Answer</summary>

The Hessian of scalar $f$ has entries $\partial^2 f/\partial x_i\partial x_j$. Positive semidefiniteness is the local second-order signature of convexity.

</details>

### Question 11

Why can numerator-layout and denominator-layout conventions lead to transpose errors in chain-rule calculations?

<details>
<summary>Answer</summary>

Numerator and denominator layouts transpose Jacobians relative to each other. Fixing a convention prevents chain-rule shape errors.

</details>

### Question 12

Compute the Jacobian of $g(x)=Ax+b$ and check its dimensions.

<details>
<summary>Answer</summary>

For $g(x)=Ax+b$, each output row has derivative equal to the corresponding row of $A$, so $J_g=A$.

</details>

### Question 13

For a vector-valued function $g(t)$ of one scalar variable, what derivative object is natural?

<details>
<summary>Answer</summary>

For $g(t)$, the derivative is the vector $(g_1 prime(t),\ldots,g_K prime(t))$, i.e. a $K\times1$ Jacobian.

</details>

### Question 17

<details>

### Question 18

Why is the derivative of a constant scalar function equal to zero?

<details>
<summary>Answer</summary>

A constant has zero first-order change in every direction, so its derivative object is zero.

</details>

### Question 19

How can checking shapes catch missing transposes in matrix calculus?

<details>
<summary>Answer</summary>

The equations $df=\nabla f^Tdx$ and $dg=Jdx$ force dimensions and reveal missing transposes.

</details>

### Question 20

Why is differential notation useful for machine-learning optimization problems involving vectors and matrices?

<details>
<summary>Answer</summary>

Losses often depend on vectors or matrices of parameters, so differential notation keeps optimization derivations concise.

</details>
