### Question 1

For $w\in\mathbb{R}^d$ and a constant matrix $A\in\mathbb{R}^{d\times d}$, compute both $\nabla_w(w^\top A w)$ and $\nabla_w^2(w^\top A w)$ without assuming that $A$ is symmetric. State the shape of each derivative object.

<!-- Source: quadratic vector identity $\nabla_w(w^\top A w)=(A+A^\top)w$ and Hessian discussion -->

<details>
<summary>Answer</summary>

Since
$$w^\top A w=\sum_{i=1}^d\sum_{j=1}^d w_iA_{ij}w_j,$$
we have
$$\frac{\partial}{\partial w_k}(w^\top A w)=\sum_j A_{kj}w_j+\sum_i w_iA_{ik}=((A+A^\top)w)_k.$$
Thus $\nabla_w(w^\top A w)=(A+A^\top)w\in\mathbb{R}^d$. Differentiating the gradient gives
$$\nabla_w^2(w^\top A w)=A+A^\top\in\mathbb{R}^{d\times d}.$$

</details>

### Question 2

The lecture writes $\nabla_w\,\frac12 w^\top A w=Aw$ when $A$ is symmetric. Identify the exact assumption needed, and give a concrete $2\times2$ counterexample showing the formula can fail without it.

<!-- Source: symmetric-matrix assumption in the half-quadratic formula -->

<details>
<summary>Answer</summary>

The needed assumption is $A=A^\top$. In general,
$$\nabla_w\frac12 w^\top A w=\frac12(A+A^\top)w.$$
Take
$$A=\begin{bmatrix}0&1\\0&0\end{bmatrix},\qquad w=\begin{bmatrix}u\\v\end{bmatrix}.$$
Then $\frac12w^\top Aw=\frac12uv$, so the gradient is $\frac12(v,u)^\top$. But $Aw=(v,0)^\top$, which is not equal in general. For example, at $w=(1,1)^\top$ the two vectors are $(1/2,1/2)^\top$ and $(1,0)^\top$.

</details>

### Question 3

Fill in the coordinate argument behind $\nabla_w(a^\top w)=a$. Use $a,w\in\mathbb{R}^d$ and explicitly compute the $k$th partial derivative.

<!-- Source: linear vector derivative identity $\nabla_w(a^\top w)=a$ -->

<details>
<summary>Answer</summary>

Since
$$a^\top w=\sum_{i=1}^d a_iw_i,$$
the $k$th partial derivative is
$$\frac{\partial}{\partial w_k}(a^\top w)=\sum_{i=1}^d a_i\frac{\partial w_i}{\partial w_k}=a_k.$$
Therefore the gradient vector is $(a_1,\ldots,a_d)^\top=a$.

</details>

### Question 5

A scalar-valued function $L:\mathbb{R}^n\to\mathbb{R}$ may be differentiated either as a Jacobian or as a gradient. State the shapes of $J_L(x)$ and $\nabla_xL$, then give the precise relationship between them under the lecture's numerator-layout convention.

<!-- Source: notational tension between scalar-output Jacobians and gradients -->

<details>
<summary>Answer</summary>

Under numerator layout, the Jacobian has output coordinates as rows and input coordinates as columns. Since $L$ has one output coordinate,
$$J_L(x)\in\mathbb{R}^{1\times n}.$$
The gradient is arranged to match the vector input, so $\nabla_xL\in\mathbb{R}^n$, viewed as a column vector. The relationship is
$$J_L(x)=(\nabla_xL)^\top.$$

</details>

### Question 6

Diagnose the shape error in the claim: if $z=f(w)\in\mathbb{R}^m$, $w\in\mathbb{R}^d$, and $L=g(z)$ is scalar-valued, then $\nabla_wL=J_f(w)\nabla_zL$. State the corrected formula and verify its shape.

<!-- Source: scalar-output chain rule through a vector intermediate -->

<details>
<summary>Answer</summary>

The product $J_f(w)\nabla_zL$ is not generally defined with the lecture's shapes: $J_f(w)\in\mathbb{R}^{m\times d}$ and $\nabla_zL\in\mathbb{R}^m$. The inner dimensions $d$ and $m$ do not match unless accidentally $d=m$.

The correct formula is
$$\nabla_wL=J_f(w)^\top\nabla_zL.$$
Here $J_f(w)^\top\in\mathbb{R}^{d\times m}$ and $\nabla_zL\in\mathbb{R}^m$, so the result lies in $\mathbb{R}^d$, matching the shape of $w$.

</details>

### Question 7

Recover the scalar-output chain rule from the vector-output chain rule. Let $x\in\mathbb{R}^n$, $z=f(x)\in\mathbb{R}^m$, and $L=g(z)\in\mathbb{R}$. Starting from $J_{g\circ f}(x)=J_g(z)J_f(x)$, derive $\nabla_xL=J_f(x)^\top\nabla_zL$.

<!-- Source: connection between vector-output and scalar-output chain rules -->

<details>
<summary>Answer</summary>

Because $g:\mathbb{R}^m\to\mathbb{R}$, its Jacobian is a row vector:
$$J_g(z)=(\nabla_zL)^\top\in\mathbb{R}^{1\times m}.$$
The vector-output chain rule gives
$$J_{g\circ f}(x)=J_g(z)J_f(x)=(\nabla_zL)^\top J_f(x).$$
But $J_{g\circ f}(x)=(\nabla_xL)^\top$. Transposing both sides yields
$$\nabla_xL=J_f(x)^\top\nabla_zL.$$

</details>

### Question 8

For $f:\mathbb{R}^2\to\mathbb{R}^3$ defined by
$$f(x_1,x_2)=\begin{bmatrix}x_1+2x_2\\ x_1x_2\\ x_2^2\end{bmatrix},$$
compute $J_f(x)$ using the lecture's row-output, column-input convention.

<!-- Source: definition of the Jacobian for vector-output functions -->

<details>
<summary>Answer</summary>

The rows correspond to output coordinates and the columns to input coordinates:
$$J_f(x)=\begin{bmatrix}
\partial f_1/\partial x_1 & \partial f_1/\partial x_2\\
\partial f_2/\partial x_1 & \partial f_2/\partial x_2\\
\partial f_3/\partial x_1 & \partial f_3/\partial x_2
\end{bmatrix}
=\begin{bmatrix}
1&2\\
x_2&x_1\\
0&2x_2
\end{bmatrix}.$$
Its shape is $3\times2$, equal to output dimension times input dimension.

</details>

### Question 9

For the rotation matrix
$$R(\theta)=\begin{bmatrix}\cos\theta&-\sin\theta\\ \sin\theta&\cos\theta\end{bmatrix},$$
compute $dR/d\theta$ and verify that $dR/d\theta=R(\theta)S$ for a constant matrix $S$. Identify $S$.

<!-- Source: scalar-input matrix-output rotation example -->

<details>
<summary>Answer</summary>

Entrywise differentiation gives
$$\frac{dR}{d\theta}=\begin{bmatrix}-\sin\theta&-\cos\theta\\ \cos\theta&-\sin\theta\end{bmatrix}.$$
Let
$$S=\begin{bmatrix}0&-1\\1&0\end{bmatrix}.$$
Then

</details>

### Question 10

Differentiate the identity $R(\theta)^\top R(\theta)=I$ for the lecture's rotation matrix. What does the result imply about $R(\theta)^\top R'(\theta)$?

<!-- Source: rotation example and matrix-valued differentiation -->

<details>
<summary>Answer</summary>

Differentiating both sides gives
$$R'(\theta)^\top R(\theta)+R(\theta)^\top R'(\theta)=0.$$
Let $B=R(\theta)^\top R'(\theta)$. Then $B^\top=R'(\theta)^\top R(\theta)$, so the equation says
$$B^\top+B=0.$$
Thus $R(\theta)^\top R'(\theta)$ is skew-symmetric. For this rotation, it equals $S=\begin{bmatrix}0&-1\\1&0\end{bmatrix}$.

</details>

### Question 11

Verify $\nabla_W\operatorname{tr}(A^\top W)=A$ by expanding the trace in entries. Assume $A,W\in\mathbb{R}^{m\times n}$ and $A$ is constant with respect to $W$.

<!-- Source: matrix linear trace derivative identity -->

<details>
<summary>Answer</summary>

The trace expands as
$$\operatorname{tr}(A^\top W)=\sum_{i=1}^m\sum_{j=1}^n A_{ij}W_{ij}.$$
Therefore
$$\frac{\partial}{\partial W_{kl}}\operatorname{tr}(A^\top W)=A_{kl}.$$
Since this holds for every entry, the matrix gradient has entries $(\nabla_WL)_{kl}=A_{kl}$, so $\nabla_W\operatorname{tr}(A^\top W)=A$.

</details>

### Question 13

Connect the Frobenius norm identity to the trace identity by proving
$$\|W\|_F^2=\operatorname{tr}(W^\top W)$$
for $W\in\mathbb{R}^{m\times n}$, then use entries to compute $\nabla_W\|W\|_F^2$.

<!-- Source: Frobenius norm formula and its gradient -->

<details>
<summary>Answer</summary>

The diagonal entry $(W^\top W)_{jj}$ is $\sum_{i=1}^m W_{ij}^2$. Hence
$$\operatorname{tr}(W^\top W)=\sum_{j=1}^n\sum_{i=1}^m W_{ij}^2=\|W\|_F^2.$$
Differentiating entrywise,
$$\frac{\partial}{\partial W_{ij}}\|W\|_F^2=2W_{ij}.$$
Therefore $\nabla_W\|W\|_F^2=2W$.

</details>

### Question 14

For $W\in\mathbb{R}^{m\times n}$ and fixed $B\in\mathbb{R}^{m\times n}$, compute
$$\nabla_W\left(\frac12\|W-B\|_F^2\right).$$
Give the entrywise reasoning.

<!-- Source: Frobenius norm derivative as matrix version of $\frac12\|w\|^2$ -->

<details>
<summary>Answer</summary>

Expand the objective as
$$L(W)=\frac12\sum_{i=1}^m\sum_{j=1}^n (W_{ij}-B_{ij})^2.$$
For each entry,
$$\frac{\partial L}{\partial W_{ij}}=W_{ij}-B_{ij}.$$
Thus
$$\nabla_W\left(\frac12\|W-B\|_F^2\right)=W-B.$$

</details>

### Question 17

For $f(x)=Ax+b$ with $A\in\mathbb{R}^{m\times n}$ and $b\in\mathbb{R}^m$, verify from coordinates that $J_f(x)=A$. State exactly where the constant shift disappears.

<!-- Source: Jacobian of linear and affine maps -->

<details>
<summary>Answer</summary>

The $i$th coordinate is
$$f_i(x)=\sum_{j=1}^n A_{ij}x_j+b_i.$$
Thus
$$\frac{\partial f_i}{\partial x_k}=A_{ik}+\frac{\partial b_i}{\partial x_k}=A_{ik},$$
because $b_i$ is constant with respect to every input coordinate. Therefore the $(i,k)$ entry of $J_f(x)$ is $A_{ik}$, so $J_f(x)=A$.

</details>

### Question 19

For $L(w)=w^\top w$ at a point $w\in\mathbb{R}^d$, compare the exact change $L(w+h)-L(w)$ with the second-order approximation
$$L(w)+(\nabla L(w))^\top h+\frac12h^\top H_L(w)h.$$
Is the approximation exact?

<!-- Source: Hessian and second-order approximation formula -->

<details>
<summary>Answer</summary>

Here $\nabla L(w)=2w$ and $H_L(w)=2I$. The second-order expression for $L(w+h)$ is
$$L(w)+2w^\top h+\frac12h^\top(2I)h=w^\top w+2w^\top h+h^\top h.$$

$$(w+h)^\top(w+h)=w^\top w+2w^\top h+h^\top h.$$
Thus the second-order approximation is exact for this quadratic function.

### Question 20

Locate the missing smoothness assumption behind treating the Hessian as a symmetric matrix. Give a function $L:\mathbb{R}^2\to\mathbb{R}$ for which mixed partials at the origin are not equal, and compute the two mixed partials there.

<!-- Source: implicit assumptions behind Hessian symmetry and second partial derivatives -->

<details>
<summary>Answer</summary>

A standard sufficient assumption is that the second partial derivatives are continuous in a neighborhood of the point. Without such regularity, mixed partials can differ.

Define
$$L(x,y)=\begin{cases}\dfrac{xy(x^2-y^2)}{x^2+y^2},&(x,y)
e(0,0),\0,&(x,y)=(0,0).\end{cases}$$
First, $L(x,0)=L(0,y)=0$. Also
$$\frac{\partial L}{\partial x}(0,y)=\lim_{h\to0}\frac{L(h,y)-0}{h}=-y,$$
so $L_{xy}(0,0)=\lim_{y\to0}\frac{L_x(0,y)-L_x(0,0)}{y}=-1$.
Similarly,
$$\frac{\partial L}{\partial y}(x,0)=\lim_{h\to0}\frac{L(x,h)-0}{h}=x,$$
so $L_{yx}(0,0)=1$. The mixed partials differ.

</details>

### Question 22

Identify the flaw in this false extension of the elementwise rule: for a matrix-valued $F(t)$, $\frac{d}{dt}F(t)^2=2F(t)F'(t)$. Give a $2\times2$ counterexample where the proposed formula is wrong.

<!-- Source: limits of scalar-looking rules for matrix products versus elementwise rules -->

<details>
<summary>Answer</summary>

The flaw is that matrix multiplication is generally noncommutative. The product rule gives
$$\frac{d}{dt}(F(t)^2)=F'(t)F(t)+F(t)F'(t),$$
not necessarily $2F(t)F'(t)$.

Take
$$F(t)=\begin{bmatrix}t&t\\0&1\end{bmatrix},\qquad F'(t)=\begin{bmatrix}1&1\\0&0\end{bmatrix}.$$
Then
$$F'F+FF'=\begin{bmatrix}2t&2t+1\\0&0\end{bmatrix},$$
while
$$2FF'=\begin{bmatrix}2t&2t\\0&0\end{bmatrix}.$$
They differ in the $(1,2)$ entry.

</details>

### Question 27

For $g:\mathbb{R}^m\to\mathbb{R}^p$ and $f:\mathbb{R}^n\to\mathbb{R}^m$, compute the $(i,k)$ entry of $J_{g\circ f}(x)$ in summation notation and match it to the matrix product $J_g(f(x))J_f(x)$.

<!-- Source: coordinate form of the vector-output chain rule -->

<details>
<summary>Answer</summary>

Let $z=f(x)$. The $i$th output coordinate is $g_i(z)$. By the multivariable chain rule,
$$\frac{\partial (g\circ f)_i}{\partial x_k}(x)=\sum_{j=1}^m \frac{\partial g_i}{\partial z_j}(z)\frac{\partial f_j}{\partial x_k}(x).$$
This is exactly the $(i,k)$ entry of the product $J_g(z)J_f(x)$, since $J_g(z)_{ij}=\partial g_i/\partial z_j$ and $J_f(x)_{jk}=\partial f_j/\partial x_k$.

</details>

### Question 28

A student claims that $\nabla_W\operatorname{tr}(AW)=A$ for all $A,W\in\mathbb{R}^{m\times m}$. Decide whether this is correct under the lecture's gradient convention. If not, give the corrected formula.

### Question 30

Sort the following derivative rules into scalar-output gradient rules, vector-output Jacobian rules, and scalar-input matrix-output rules: $\nabla_W\|W\|_F^2=2W$, $J_{Ax+b}=A$, $\frac{d}{dt}(t^2A)=2tA$, $\nabla_w(w^\top Aw)=(A+A^\top)w$, and $J_{g\circ f}=J_gJ_f$. Give one sentence explaining the classification criterion.

<!-- Source: summary distinction among derivative objects and basic identities -->

<details>
<summary>Answer</summary>

Scalar-output gradient rules: $\nabla_W\|W\|_F^2=2W$ and $\nabla_w(w^\top Aw)=(A+A^\top)w$.

Vector-output Jacobian rules: $J_{Ax+b}=A$ and $J_{g\circ f}=J_gJ_f$.

Scalar-input matrix-output rule: $\frac{d}{dt}(t^2A)=2tA$.

The criterion is the shape of the output and input: scalar outputs use gradients shaped like the variable, vector outputs use output-by-input Jacobians, and a matrix depending on one scalar is differentiated entrywise into another matrix of the same output shape.

</details>
