# K-means - Review Questions {#sec-kmeans-questions}

### Question 1

In the lecture's notation, identify the roles and dimensions of $X$, $x_n$, $c_n$, and $\mu_k$. Which of these objects is absent from the supervised-learning loss $\ell(y,s(x))$ setup, and why does that matter for clustering?

<details>
<summary>Answer</summary>

The data matrix is

$$
X\in\mathbb R^{N\times D},
$$

with row $x_n^T$ representing observation $n$, so $x_n\in\mathbb R^D$. The cluster assignment

$$
c_n\in\{1,\ldots,K\}
$$

names the cluster assigned to observation $n$. The centroid

$$
\mu_k\in\mathbb R^D
$$

summarizes cluster $k$.

The supervised target $y$ is absent. Because there is no response variable, k-means must define its own criterion for useful structure: a partition whose points are close to their assigned centroids.

</details>


### Question 2

Translate the two displayed forms of the k-means objective into each other:

$$
\sum_{n=1}^N\|x_n-\mu_{c_n}\|_2^2
\qquad\text{and}\qquad
\sum_{k=1}^K\sum_{n:c_n=k}\|x_n-\mu_k\|_2^2.
$$

Why are they the same quantity?

<details>
<summary>Answer</summary>

The first expression sums once over every observation. For observation $n$, the assigned centroid is $\mu_{c_n}$.

The second expression groups the same terms by cluster. If $c_n=k$, then $\mu_{c_n}=\mu_k$, so the contribution of observation $n$ appears in the inner sum for that specific $k$.

Every observation has exactly one assignment $c_n\in\{1,\ldots,K\}$, so each squared distance appears exactly once in both expressions. Therefore the two objectives are identical reorganizations of the same sum.

</details>


### Question 3

With fixed centroids $\mu_1,\ldots,\mu_K$, justify the nearest-centroid assignment rule

$$
c_n=\arg\min_{k\in\{1,
\ldots,K\}}\|x_n-\mu_k\|_2^2.
$$

Your answer should make clear why the choices for different observations can be done separately. 

<details>
<summary>Answer</summary>

With centroids fixed, the objective is

$$
\sum_{n=1}^N\|x_n-\mu_{c_n}\|_2^2.
$$

The choice of $c_n$ appears only in the $n$th term. There is no constraint coupling $c_n$ to $c_m$ for $m\ne n$. Therefore minimizing the full sum over assignments is equivalent to minimizing each term separately.

For each observation $x_n$, the smallest possible contribution is achieved by choosing the centroid with the smallest squared Euclidean distance.

</details>


### Question 4

For fixed nonempty cluster assignments, derive the centroid update

$$
\mu_k=\frac{1}{N_k}\sum_{n:c_n=k}x_n.
$$

Hint: calculate the gradient. 

<details>
<summary>Answer</summary>

For cluster $k$, minimize

$$
f_k(\mu)=\sum_{n:c_n=k}\|x_n-\mu\|_2^2.
$$

The gradient is

$$
\nabla_\mu f_k(\mu)=2\sum_{n:c_n=k}(\mu-x_n).
$$

Setting it equal to zero gives

$$
2\left(N_k\mu-\sum_{n:c_n=k}x_n\right)=0,
$$

so

$$
\mu_k=\frac{1}{N_k}\sum_{n:c_n=k}x_n.
$$

</details>


### Question 5

Work through one Lloyd step for one-dimensional data $x=0,2,9,11$ with $K=2$ and initial centroids $\mu_1=0$, $\mu_2=10$. Assign points, compute the objective before the centroid update, then compute the updated centroids and the new objective with the same assignments.

<details>
<summary>Answer</summary>

Nearest-centroid assignments are

$$
0,2\mapsto \mu_1=0,
\qquad
9,11\mapsto \mu_2=10.
$$

The objective before updating centroids is

$$
(0-0)^2+(2-0)^2+(9-10)^2+(11-10)^2=0+4+1+1=6.
$$

The updated centroids are the cluster means:

$$
\mu_1'=\frac{0+2}{2}=1,
\qquad
\mu_2'=\frac{9+11}{2}=10.
$$

With the same assignments, the new objective is

$$
(0-1)^2+(2-1)^2+(9-10)^2+(11-10)^2=1+1+1+1=4.
$$

The centroid update decreased the objective.

</details>


### Question 6

Calculate the within-cluster sum of squares for the two-dimensional points

$$
(0,0),\ (2,0),\ (0,2),\ (8,8)
$$

under the partition $A=\{(0,0),(2,0),(0,2)\}$ and $B=\{(8,8)\}$, using each cluster's mean as its centroid.

<details>
<summary>Answer</summary>

The centroid of cluster $A$ is

$$
\mu_A=\left(\frac{0+2+0}{3},\frac{0+0+2}{3}\right)=\left(\frac23,\frac23\right).
$$

The cluster $A$ squared distances are

$$
\left\|(0,0)-\left(\frac23,\frac23\right)\right\|^2=\frac89,
$$

$$
\left\|(2,0)-\left(\frac23,\frac23\right)\right\|^2=\frac{20}{9},
$$

and

$$
\left\|(0,2)-\left(\frac23,\frac23\right)\right\|^2=\frac{20}{9}.
$$

Cluster $B$ has one point, so its centroid is $(8,8)$ and its contribution is $0$. Therefore

$$
\operatorname{WCSS}=\frac89+\frac{20}{9}+\frac{20}{9}=\frac{48}{9}=\frac{16}{3}.
$$

</details>


### Question 7

A proposed clustering of the one-dimensional points $0,4,5,9$ with $K=2$ is

$$
A=\{0,9\},\qquad B=\{4,5\}.
$$

Compare its within-cluster sum of squares with the partition

$$
A'=\{0,4\},\qquad B'=\{5,9\}.
$$

Which partition is better for the k-means objective?

<details>
<summary>Answer</summary>

For $A=\{0,9\}$, the centroid is $4.5$ and the contribution is

$$
(0-4.5)^2+(9-4.5)^2=20.25+20.25=40.5.
$$

For $B=\{4,5\}$, the centroid is $4.5$ and the contribution is

$$
(4-4.5)^2+(5-4.5)^2=0.25+0.25=0.5.
$$

So this partition has WCSS $41$.

For $A'=\{0,4\}$, the centroid is $2$ and the contribution is $4+4=8$. For $B'=\{5,9\}$, the centroid is $7$ and the contribution is $4+4=8$. Thus this partition has WCSS $16$.

The second partition is better because $16<41$.

</details>


### Question 8

The lecture states that the full k-means problem is hard generally a very hard problem to solve exactly. Why?

<details>
<summary>Answer</summary>

If centroids are fixed, each assignment can be chosen independently by nearest centroid. If assignments are fixed, each centroid is just the mean of its assigned points.

The full problem is harder because the assignments are discrete and interact with the centroids. There are many possible partitions of $N$ observations into $K$ clusters, and each partition gives a different set of optimal centroids and a different objective value. Lloyd's algorithm improves the current partition greedily, but it does not search all partitions.

</details>


### Question 9

Show that one full Lloyd iteration cannot increase the k-means objective. Separate the argument into the assignment step and the centroid-update step.

<details>
<summary>Answer</summary>

In the assignment step, the centroids are fixed. Assigning each point to its nearest centroid minimizes that point's contribution to the objective, so the total objective cannot increase.

In the centroid-update step, the assignments are fixed. Replacing each centroid by the mean of its assigned points minimizes the within-cluster squared distance for that cluster, so this step also cannot increase the objective.

A full Lloyd iteration is the composition of two nonincreasing steps. Therefore the objective after the iteration is less than or equal to the objective before it.

</details>


### Question 10

Cluster labels are arbitrary. Formalize this statement: if $c_1,\ldots,c_N$ and centroids $\mu_1,
\ldots,\mu_K$ define a clustering, what happens to the objective after applying a permutation to the cluster names?

<details>
<summary>Answer</summary>

For a permutation $pi$ define new labels and centroids by

$$
c_n'=\pi(c_n),
\qquad
\mu_{\pi(k)}'=\mu_k.
$$

Then each observation is paired with the same physical centroid as before, only under a different cluster name. Thus

$$
\|x_n-\mu_{c_n}\|^2=\|x_n-\mu'_{c_n'}\|^2
$$

for every $n$. Summing over $n$ gives the same objective value.

Therefore the numerical labels do not matter; only the partition of observations into groups matters.

</details>


### Question 11

Argue that the minimum k-means objective cannot increase when $K$ is increased from $K$ to $K+1$.

<details>
<summary>Answer</summary>

Take an optimal $K$-cluster solution. In the $(K+1)$-cluster problem, we can keep the same assignments and centroids and leave the extra centroid unused. This gives the same objective value, so the optimum with $K+1$ clusters is no larger.

If one requires every cluster to be nonempty and $K<N$, split one existing cluster by assigning one of its points to the new centroid placed exactly at that point. That point then contributes zero to its new cluster, while the remaining clusters can keep the previous centroids or improve them. The objective still cannot increase.

</details>


### Question 12

In the extreme case $K=N$, show that the k-means objective can be made equal to zero. What does this imply about choosing $K$ by minimizing inertia alone?

<details>
<summary>Answer</summary>

Assign each observation to its own cluster and set

$$
\mu_n=x_n
$$

for that singleton cluster. Then every squared distance is

$$
\|x_n-\mu_{c_n}\|^2=\|x_n-x_n\|^2=0.
$$

Therefore the total objective is zero.

This means choosing $K$ by inertia alone would favor the largest possible $K$, even when that partition is not useful. That is why the lecture introduces heuristics such as elbow plots and silhouette scores.

</details>


### Question 13

K-means uses squared Euclidean distance:

$$
\|x_n-\mu_k\|_2^2=\sum_{j=1}^D(x_{nj}-\mu_{kj})^2.
$$

If the first variable is replaced by $100X_1$, by what factor does its squared contribution to distance change? Connect this calculation to the lecture's scaling warning.

<details>
<summary>Answer</summary>

The first-coordinate contribution changes from

$$
(x_{n1}-\mu_{k1})^2
$$

to

$$
(100x_{n1}-100\mu_{k1})^2=10000(x_{n1}-\mu_{k1})^2.
$$

So that coordinate's squared contribution is multiplied by $10000$.

This explains the lecture's warning: if one variable is measured on a much larger scale, it can dominate Euclidean distances and therefore dominate the k-means assignments.

</details>


### Question 14

Compare these three data geometries and classify where k-means is most appropriate: equal-size separated spherical blobs, two interlocking moons, and one small dense cluster next to one large diffuse cluster. Give the reason for each classification.

<details>
<summary>Answer</summary>

Equal-size separated spherical blobs are the best match. K-means represents each cluster by a centroid and assigns points by Euclidean distance, so compact spherical groups with similar spread fit its geometry.

Two interlocking moons are a poor match because the clusters are non-convex. Nearest-centroid regions are Voronoi cells, which cannot follow crescent-shaped groups well.

A small dense cluster next to a large diffuse cluster can also be a poor match because k-means tends to prefer clusters with comparable within-cluster squared spread. Very different sizes or densities can pull centroids and boundaries away from the intended grouping.

</details>


### Question 15

Diagnose the following claim: "Since inertia always decreases as $K$ increases, the best model is the one with the largest $K$ we are willing to fit."

<details>
<summary>Answer</summary>

The claim confuses optimization of the training objective with useful clustering. Inertia decreases with $K$ because extra centroids make the training objective easier to minimize. In the extreme case $K=N$, inertia can be zero by assigning each point to its own cluster.

That does not mean the resulting partition is meaningful. Choosing $K$ requires an additional modeling judgment or diagnostic, such as an elbow plot, silhouette score, domain knowledge, or external labels used only for evaluation.

</details>
