### Question 1

For k-means, write the objective or Lloyd update associated with clustering from $X$ alone and explain why that step decreases or preserves the objective.

<!-- Source: K-means lecture; section related to: For k-means, write the objective or Lloyd update associated with clustering from X alone a -->

<details>
<summary>Answer</summary>

For clustering from $X$ alone, k-means minimizes within-cluster squared distances to centroids. Lloyd’s algorithm alternates nearest-centroid assignment and centroid recomputation; each step does not increase the objective, but the final solution may be only local.

</details>
