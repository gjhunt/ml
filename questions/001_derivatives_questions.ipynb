# Matrix Calculus for Machine Learning - Review Questions {#sec-deriv-questions}

These questions practice gradients, shape checking, linear and quadratic expressions, the chain rule, and derivatives with respect to matrix parameters. Answers are hidden in collapsible sections.


### Question 1

Let $L:\mathbb{R}^2\to\mathbb{R}$ be
$$
L(w_1,w_2)=3w_1^2+2w_1w_2-w_2.
$$
Compute $\nabla_wL$ and state its shape.

<details>
<summary>Answer</summary>

The partial derivatives are
$$
\frac{\partial L}{\partial w_1}=6w_1+2w_2,
\qquad
\frac{\partial L}{\partial w_2}=2w_1-1.
$$
Therefore
$$
\nabla_wL=
\begin{pmatrix}
6w_1+2w_2\\
2w_1-1
\end{pmatrix}
\in\mathbb{R}^2,
$$
which has the same shape as $w$.

</details>


### Question 2

For fixed $a,x\in\mathbb{R}^D$ and variable $w\in\mathbb{R}^D$, compute the gradients of $a^\top w$ and $w^\top x$. Justify one identity by expanding it in coordinates.

<details>
<summary>Answer</summary>

The identities are
$$
\nabla_w(a^\top w)=a,
\qquad
\nabla_w(w^\top x)=x.
$$
For example,
$$
a^\top w=\sum_{j=1}^D a_jw_j,
$$
whose $k$th partial derivative is $a_k$. Stacking the partial derivatives gives $a$.

</details>


### Question 3

For fixed $A\in\mathbb{R}^{D\times D}$, use coordinates to derive
$$
\nabla_w(w^\top Aw)=(A+A^\top)w.
$$
Then state the simplified formula when $A$ is symmetric.

<details>
<summary>Answer</summary>

Expanding gives
$$
w^\top Aw=\sum_i\sum_jw_iA_{ij}w_j.
$$
Differentiating with respect to $w_k$ picks up terms with $i=k$ and terms with $j=k$:
$$
\frac{\partial}{\partial w_k}(w^\top Aw)
=\sum_jA_{kj}w_j+\sum_iA_{ik}w_i
=(Aw)_k+(A^\top w)_k.
$$
Hence the gradient is $(A+A^\top)w$. If $A=A^\top$, it simplifies to $2Aw$.

</details>


### Question 4

Diagnose the missing assumption in
$$
\nabla_w\left(\frac12w^\top Aw\right)=Aw.
$$
Give a concrete $2\times2$ counterexample showing why the assumption is needed.

<details>
<summary>Answer</summary>

The identity requires $A$ to be symmetric. In general,
$$
\nabla_w\left(\frac12w^\top Aw\right)
=\frac12(A+A^\top)w.
$$
Take
$$
A=\begin{pmatrix}0&1\\0&0\end{pmatrix},
\qquad
w=\begin{pmatrix}u\\v\end{pmatrix}.
$$
Then $\frac12w^\top Aw=uv/2$, whose gradient is $(v/2,u/2)^\top$, while $Aw=(v,0)^\top$. These are not generally equal.

</details>


### Question 5

Let $w,a\in\mathbb{R}^D$, $X\in\mathbb{R}^{N\times D}$, and $y\in\mathbb{R}^N$. State the shape of each expression:

1. $Xw$
2. $y-Xw$
3. $\|y-Xw\|^2$
4. $X^\top(y-Xw)$

Then explain why $X(y-Xw)$ is not generally a valid matrix product.

<details>
<summary>Answer</summary>

The shapes are:

1. $Xw\in\mathbb{R}^N$.
2. $y-Xw\in\mathbb{R}^N$.
3. $\|y-Xw\|^2$ is a scalar.
4. $X^\top(y-Xw)\in\mathbb{R}^D$ because $X^\top$ is $D\times N$.

The product $X(y-Xw)$ attempts to multiply an $N\times D$ matrix by an $N$-vector. Its inner dimensions are $D$ and $N$, which need not agree.

</details>


### Question 6

Let $X\in\mathbb{R}^{N\times D}$ and $y\in\mathbb{R}^N$ be fixed, and define
$$
L(w)=\frac12w^\top X^\top Xw-w^\top X^\top y.
$$
Verify that $X^\top X$ is symmetric and compute $\nabla_wL$. Check the shape of your answer.

<details>
<summary>Answer</summary>

Since
$$
(X^\top X)^\top=X^\top X,
$$
the matrix $X^\top X$ is symmetric. Therefore
$$
\nabla_w\left(\frac12w^\top X^\top Xw\right)=X^\top Xw.
$$
The second term is linear in $w$, with gradient $-X^\top y$. Hence
$$
\nabla_wL=X^\top Xw-X^\top y=X^\top(Xw-y).
$$
This is a $D$-vector, matching the shape of $w$.

</details>


### Question 7

Apply the scalar-intermediate chain rule to
$$
L(w)=\frac12(a^\top w-b)^2,
$$
where $a,w\in\mathbb{R}^D$ and $b\in\mathbb{R}$. Identify a scalar intermediate $s(w)$ and an outer function $\ell(s)$ such that $L(w)=\ell(s(w))$, then derive $\nabla_wL(w)$.

<details>
<summary>Answer</summary>

Let
$$
s(w)=a^\top w-b,
\qquad
\ell(s)=\frac12s^2.
$$
Then $L(w)=\ell(s(w))$. Since
$$
\frac{d\ell}{ds}=s,
\qquad
\nabla_ws(w)=a,
$$
the chain rule gives
$$
\nabla_wL(w)=\frac{d\ell}{ds}\nabla_ws(w)
=(a^\top w-b)a.
$$

</details>


### Question 8

Starting from
$$
\sigma(s)=\frac{1}{1+e^{-s}},
$$
compute $d\sigma/ds$ and express the result only in terms of $p=\sigma(s)$.

<details>
<summary>Answer</summary>

Direct differentiation gives
$$
\frac{d\sigma}{ds}=\frac{e^{-s}}{(1+e^{-s})^2}.
$$
Since
$$
p=\frac1{1+e^{-s}},
\qquad
1-p=\frac{e^{-s}}{1+e^{-s}},
$$
we obtain
$$
\frac{d\sigma}{ds}=p(1-p).
$$

</details>


### Question 9

For fixed $x$ and $y\in\{0,1\}$, define
$$
s(w)=x^\top w,
\qquad
p(w)=\sigma(s(w)),
$$
and let
$$
\ell_y(p)=-\left[y\log p+(1-y)\log(1-p)\right],
\qquad
L(w)=\ell_y(p(w)).
$$
Use the chain rule through $p$ and $s$ to derive $\nabla_wL(w)=(p(w)-y)x$. Display the three derivative factors before simplifying.

<details>
<summary>Answer</summary>

Writing $s=s(w)$ and $p=p(w)$, the three factors are
$$
\frac{d\ell_y}{dp}
=-\frac{y}{p}+\frac{1-y}{1-p}
=\frac{p-y}{p(1-p)},
$$
$$
\frac{dp}{ds}=p(1-p),
\qquad
\nabla_ws(w)=x.
$$
Therefore
$$
\nabla_wL(w)
=\frac{d\ell_y}{dp}
\frac{dp}{ds}\nabla_ws(w)
=(p-y)x.
$$

</details>


### Question 10

Let
$$
z_1(w)=w_1+w_2,
\qquad
z_2(w)=w_1w_2,
$$
and define
$$
\ell(z)=\frac12(z_1^2+z_2^2),
\qquad
L(w)=\ell(z(w)).
$$
Use
$$
\frac{\partial L}{\partial w_j}
=\sum_{k=1}^2
\frac{\partial \ell}{\partial z_k}
\frac{\partial z_k}{\partial w_j}
$$
to compute $\nabla_wL$ at $w=(1,2)^\top$.

<details>
<summary>Answer</summary>

At $w=(1,2)^\top$, $(z_1,z_2)=(3,2)$ and
$$
\frac{\partial\ell}{\partial z_1}=z_1,
\qquad
\frac{\partial\ell}{\partial z_2}=z_2.
$$
Thus
$$
\frac{\partial L}{\partial w_1}
=z_1(1)+z_2(w_2)=3+4=7,
$$
and
$$
\frac{\partial L}{\partial w_2}
=z_1(1)+z_2(w_1)=3+2=5.
$$
Therefore
$$
\nabla_wL=
\begin{pmatrix}7\\5\end{pmatrix}.
$$

</details>


### Question 11

Let $W\in\mathbb{R}^{2\times2}$ be variable, let
$$
B=\begin{pmatrix}1&1\\2&3\end{pmatrix},
$$
and define
$$
L(W)=\frac12\sum_{d=1}^2\sum_{k=1}^2(W_{dk}-B_{dk})^2.
$$
Derive $\nabla_WL$. Then evaluate both $L(W)$ and $\nabla_WL$ at
$$
W=\begin{pmatrix}2&-1\\0&4\end{pmatrix}.
$$

<details>
<summary>Answer</summary>

For each entry,
$$
\frac{\partial L}{\partial W_{dk}}=W_{dk}-B_{dk},
$$
so $\nabla_WL=W-B$. At the specified matrix,
$$
W-B=
\begin{pmatrix}
1&-2\\
-2&1
\end{pmatrix}.
$$
Therefore
$$
L(W)=\frac12(1^2+(-2)^2+(-2)^2+1^2)=5
$$
and
$$
\nabla_WL=
\begin{pmatrix}
1&-2\\
-2&1
\end{pmatrix}.
$$

</details>


### Question 12

Write
$$
W=\begin{pmatrix}u&v\\r&s\end{pmatrix}
$$
and define
$$
L(W)=uv+(r-s)^2.
$$
Compute the four partial derivatives and arrange them in the $2\times2$ matrix $\nabla_WL$.

<details>
<summary>Answer</summary>

The partial derivatives are
$$
\frac{\partial L}{\partial u}=v,
\qquad
\frac{\partial L}{\partial v}=u,
$$
$$
\frac{\partial L}{\partial r}=2(r-s),
\qquad
\frac{\partial L}{\partial s}=-2(r-s).
$$
Placing each derivative in the position of its corresponding entry gives
$$
\nabla_WL=
\begin{pmatrix}
v&u\\
2(r-s)&-2(r-s)
\end{pmatrix}.
$$

</details>
