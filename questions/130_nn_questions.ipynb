### Question 1

For neural networks, explain how neurons as affine maps plus activations changes the represented function compared with a purely linear model.

<!-- Source: Neural Networks lecture; section related to: For neural networks, explain how neurons as affine maps plus activations changes the repre -->

<details>
<summary>Answer</summary>

For neurons as affine maps plus activations, the lecture views a network as compositions of affine maps and nonlinear activations. Without nonlinear activations the composition collapses to one affine map; with them, the model can learn nonlinear features but the ERM is nonconvex.

</details>
