### Question 1

Using the lecture's treatment of boosted regression trees, give a concise mathematical explanation, derivation, or counterexample as appropriate.

<details>
<summary>Answer</summary>

For boosted regression trees, boosting builds an additive score $s_m=s_{m-1}+\nu h_m$. For squared error, $h_m$ fits residuals; more generally it approximates a negative functional gradient, with learning rate and tree complexity acting as regularization.

</details>
