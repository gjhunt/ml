# K-Nearest Neighbors - Review Questions {#sec-knn-questions}



### Question 1

The lecture begins with the risk
$$
R(s)=\mathbb E_{(X,Y)\sim P}[\ell(Y,s(X))].
$$
What is the target $s^*$ trying to find?

<details>
<summary>Answer</summary>

The target is
$$
s^*=\arg\min_s R(s),
$$
the score function with smallest expected loss under the true data-generating distribution $P$. 

</details>


### Question 2

Starting from
$$
R(s)=\mathbb E_{(X,Y)}[\ell(Y,s(X))],
$$
use the law of total expectation to derive the conditional-risk form
$$
R(s)=\mathbb E_X[L_X(s(X))],
\qquad
L_x(t)=\mathbb E[\ell(Y,t)\mid X=x].
$$

<details>
<summary>Answer</summary>

By the law of total expectation,
$$
R(s)=\mathbb E_X\left[\mathbb E[\ell(Y,s(X))\mid X]\right].
$$
After conditioning on $X=x$, the value $s(X)$ becomes $s(x)$, so the inner expectation is
$$
\mathbb E[\ell(Y,s(x))\mid X=x].
$$
Defining
$$
L_x(t)=\mathbb E[\ell(Y,t)\mid X=x]
$$
gives
$$
R(s)=\mathbb E_X[L_X(s(X))].
$$

</details>


### Question 3

The lecture says unrestricted risk minimization can be done pointwise. Why are we free to choosen each value of $s(x)$ separately for each value of $x$?

<details>
<summary>Answer</summary>

The key assumption is that the function class is unrestricted so that the values $s(x)$ can be chosen independently at different inputs. If we have no restrictions on $s$ then we can do this. After all, what is a function other than an assignment of $s(x)$ to each $x$? If we can choose any function then we can choose each $s(x)$ completely separately for each $x$. 

</details>


### Question 4

We showed that 
$$
s^*(x)=\operatorname*{argmin}_{t}\mathbb{E}[\ell(Y,t)\mid X=x].
$$
For squared-loss regression, this is
$$
s^*(x)=\arg\min_t\mathbb E[(Y-t)^2\mid X=x].
$$
Show that this is equivalent to:
$$
s^*(x)=\mathbb{E}[Y\mid X=x].
$$

<details>
<summary>Answer</summary>

Expand the conditional objective:
$$
\mathbb E[(Y-t)^2\mid X=x]
=\mathbb E[Y^2\mid X=x]-2t\mathbb E[Y\mid X=x]+t^2.
$$
Differentiate with respect to $t$:
$$
-2\mathbb E[Y\mid X=x]+2t.
$$
Setting this equal to zero gives
$$
t=\mathbb E[Y\mid X=x].
$$
The second derivative is $2>0$, so this critical point is the unique minimizer. 

</details>


### Question 5

At a particular input value $x$, suppose the conditional distribution of $Y\mid X=x$ assigns probability $0.25$ to $0$ and probability $0.75$ to $4$. Under squared loss, compute the optimal prediction $s^*(x)$.

<details>
<summary>Answer</summary>

Under squared loss,
$$
s^*(x)=\mathbb E[Y\mid X=x].
$$
Here
$$
\mathbb E[Y\mid X=x]=0(0.25)+4(0.75)=3.
$$
So the optimal prediction is $s^*(x)=3$.

</details>


### Question 6

Why is the exact-match estimator
$$
\hat s(x)=\frac{1}{|\{n:X_n=x\}|}\sum_{n:X_n=x}Y_n
$$
usually not useful when $X\in\mathbb R^D$ has a continuous distribution? Make a probabilistic argument based on the fact that:
$$
P(X=x)=0.
$$

<details>
<summary>Answer</summary>

If $X$ has a continuous distribution, then for a fixed query point $x$,
$$
P(X=x)=0.
$$
With finitely many training observations, the probability of observing many, or even any, exact matches $X_n=x$ is usually zero (or negligible). So exact conditional averaging is not usable. Moreover, this woud only define predictions are points where we *exactly* have training data, but we want to be able to predict everywhere. 

</details>


### Question 7

Training inputs on the real line are
$$
X_1=0,\quad X_2=1,\quad X_3=3,\quad X_4=4
$$
with responses
$$
Y_1=2,\quad Y_2=4,\quad Y_3=10,\quad Y_4=12.
$$
Using Euclidean distance and query point $x=2.6$, compute the KNN regression prediction for $k=2$.

<details>
<summary>Answer</summary>

The distances from $x=2.6$ are
$$
|0-2.6|=2.6,\quad |1-2.6|=1.6,\quad |3-2.6|=0.4,\quad |4-2.6|=1.4.
$$
The two nearest inputs are $3$ and $4$, with responses $10$ and $12$. Therefore
$$
\hat s(2.6)=\frac{10+12}{2}=11.
$$

</details>


### Question 8

For the same data as Question 7, compute the KNN regression prediction at $x=2.6$ for $k=4$. Compare it with the $k=2$ prediction and state what changed.

<details>
<summary>Answer</summary>

For $k=4$, all four responses are averaged:
$$
\hat s(2.6)=\frac{2+4+10+12}{4}=7.
$$
The $k=2$ prediction was $11$. Increasing $k$ added farther-away points with smaller responses, so the prediction moved away from the local average near $x=2.6$ and toward a more global average.

</details>


### Question 9

In $k$-nearest neighbors regression, how does the choice of $k$ affect the flexibility of the fitted curve?

<details>
<summary>Answer</summary>

Smaller values of $k$ make the fitted curve more flexible. Each prediction is based on only a few nearby points, so the curve can respond strongly to local changes in the training data.

Larger values of $k$ make the fitted curve less flexible. Each prediction averages over more points, which smooths out local variation and produces a more stable curve.

So, decreasing $k$ increases flexibility, while increasing $k$ decreases flexibility.

</details>


### Question 10

Compare KNN regression and linear regression as estimators. Under squared loss, both methods target the conditional mean function
$$
m(x)=\mathbb E[Y\mid X=x].
$$
State the main difference in how the two methods estimate it.

<details>
<summary>Answer</summary>

Linear regression estimates this target using a global parametric form, usually $m(x)\approx w^\top x$. This imposes a single linear relationship across the feature space.

KNN estimates the target locally by averaging responses near the query point:
$$
\hat s(x)=\frac1k\sum_{n\in\mathcal N_k(x)}Y_n.
$$
It does not impose a global linear form, but its behavior depends strongly on local neighborhoods and the choice of distance.

</details>


### Question 11

In KNN classification with classes ${1,\dots,C}$, show that minimizing 0-1 loss pointwise is equivalent to choosing the class with largest conditional probability $P(Y=c\mid X=x)$.

<details>
<summary>Answer</summary>

In classification, the prediction is a class label. Suppose that at a fixed input value $x$, we predict class $c$.

Under 0-1 loss, the loss is

$$
\ell(Y,c)=\mathbf{1}\{Y\neq c\}.
$$

This loss is $0$ if the prediction is correct and $1$ if the prediction is incorrect.

The conditional risk at $x$ is the expected loss given $X=x$:

$$
L_x(c)=\mathbb{E}[\mathbf{1}\{Y\neq c\}\mid X=x].
$$

Because the expectation of an indicator is the probability of the event, this becomes

$$
L_x(c)=P(Y\neq c\mid X=x).
$$

So, for a fixed $x$, minimizing the conditional 0-1 risk means choosing the class $c$ that makes the probability of misclassification as small as possible.

Now use the fact that, if we predict class $c$, the only way to be correct is for $Y=c$. Therefore,

$$
P(Y\neq c\mid X=x)=1-P(Y=c\mid X=x).
$$

Thus,

$$
L_x(c)=1-P(Y=c\mid X=x).
$$

Since the $1$ is the same for every possible class $c$, minimizing $L_x(c)$ is the same as maximizing $P(Y=c\mid X=x)$.

Therefore, the pointwise optimal classifier is

$$
f^*(x)=\operatorname*{argmax}_{c} P(Y=c\mid X=x).
$$

This is the Bayes classifier. It says that, at each input $x$, the optimal prediction under 0-1 loss is the most likely class conditional on $X=x$.

</details>


### Question 12

For KNN classification, prove that the score
$$
\hat s_c(x)=\frac1k\sum_{n\in\mathcal N_k(x)}\mathbf 1\{Y_n=c\}
$$
is the fraction of neighbors whose class is $c$.

<details>
<summary>Answer</summary>

Each indicator $\mathbf 1\{Y_n=c\}$ equals $1$ if neighbor $n$ has class $c$ and $0$ otherwise. Summing over $n\in\mathcal N_k(x)$ counts how many of the $k$ neighbors have class $c$. Dividing by $k$ gives the fraction of neighbors in class $c$.


</details>


### Question 13

A KNN classifier uses $k=7$. Among the seven nearest neighbors of a query point, the class labels are
$$
A,A,B,A,C,B,A.
$$
Compute $\hat s_A(x)$, $\hat s_B(x)$, and $\hat s_C(x)$, and give the predicted class.

<details>
<summary>Answer</summary>

The counts are $4$ for class $A$, $2$ for class $B$, and $1$ for class $C$. Therefore
$$
\hat s_A(x)=\frac47,
\qquad
\hat s_B(x)=\frac27,
\qquad
\hat s_C(x)=\frac17.
$$
The largest score is $\hat s_A(x)$, so the predicted class is $A$.

</details>


### Question 14

A student claims: "Scaling cannot change KNN predictions because scaling does not change the labels $Y_n$." Diagnose the flaw in this claim.

<details>
<summary>Answer</summary>

The flaw is that KNN predictions depend not only on labels but also on which points are selected as nearest neighbors. Scaling changes the distances between feature vectors, and therefore it can change the neighborhood set $\mathcal N_k(x)$.

Even with the same labels $Y_n$, a different neighbor set can produce a different average response in regression or a different majority vote in classification.

</details>


### Question 15

For KNN regression, describe the qualitative behavior at the two extremes $k=1$ and $k=N$. Which one is more sensitive to individual training responses?

<details>
<summary>Answer</summary>

At $k=1$, the prediction is the response of the single nearest training point. It is highly sensitive to individual training responses and can vary sharply as the nearest neighbor changes.

At $k=N$, every prediction averages all training responses:
$$
\hat s(x)=\frac1N\sum_{n=1}^N Y_n.
$$
This ignores the query location entirely because every point has the same neighbor set. Thus $k=1$ is most sensitive to individual responses, while $k=N$ ignores $x$ the most.

</details>
