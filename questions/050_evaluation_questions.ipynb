### Question 1

For a fixed regression function $f$ and fresh data $D=\{(X_n,Y_n)\}_{n=1}^M\sim P^M$, the lecture defines
$$
\hat{\mathrm{Perf}}(D,f)=\frac1M\sum_{n=1}^M (Y_n-f(X_n))^2.
$$
Show that 
$$\mathrm{Perf}(f)=\mathbb E_D[\hat{\mathrm{Perf}}(D,f)\mid f] = E_{(X,Y)\sim P}[(Y-f(X))^2]
$$ for a new generic draw $(X,Y)\sim P$. 

<details>
<summary>Answer</summary>

Since $f$ is fixed under the conditional expectation,
$$
\mathrm{Perf}(f)=\mathbb E\left[\frac1M\sum_{n=1}^M (Y_n-f(X_n))^2\right]
=\frac1M\sum_{n=1}^M \mathbb E[(Y_n-f(X_n))^2].
$$
The i.i.d. assumption gives each summand the same distribution as $(Y-f(X))^2$, so
$$
\mathrm{Perf}(f)=\mathbb E_{(X,Y)\sim P}[(Y-f(X))^2].
$$

</details>




### Question 2

For a fixed classifier $f$, let

$$
\hat{\mathrm{Perf}}(D,f)=\frac{1}{M}\sum_{n=1}^M \mathbf{1}\{f(X_n)\ne Y_n\}.
$$

Express $\mathrm{Perf}(f)$ as a probability.

Hint: $\mathbb{P}(A)=\mathbb{E}[\mathbf{1}\{A\}]$.

<details>
<summary>Answer</summary>

The quantity $\hat{\mathrm{Perf}}(D,f)$ is the average classification error on the evaluation dataset $D$. Each indicator $\mathbf{1}\{f(X_n)\ne Y_n\}$ equals 1 if the classifier makes a mistake on observation $n$, and equals 0 otherwise.

Since $f$ is fixed, the only randomness comes from the evaluation data. The true performance is the expected value of this evaluation error on fresh data:

$$
\mathrm{Perf}(f)=\mathbb{E}_D[\hat{\mathrm{Perf}}(D,f)].
$$

Substituting the definition of $\hat{\mathrm{Perf}}(D,f)$ gives

$$
\mathrm{Perf}(f)=\mathbb{E}_D\left[\frac{1}{M}\sum_{n=1}^M \mathbf{1}\{f(X_n)\ne Y_n\}\right].
$$

By linearity of expectation,

$$
\mathrm{Perf}(f)=\frac{1}{M}\sum_{n=1}^M \mathbb{E}\left[\mathbf{1}\{f(X_n)\ne Y_n\}\right].
$$

Because the observations are drawn from the same distribution, each term in the average is the probability that $f$ makes a mistake on a fresh observation. Therefore,

$$
\mathrm{Perf}(f)=\mathbb{E}_{(X,Y)\sim P}[\mathbf{1}\{f(X)\ne Y\}].
$$

Using the identity $\mathbb{P}(A)=\mathbb{E}[\mathbf{1}\{A\}]$, we can write this as

$$
\mathrm{Perf}(f)=\mathbb{P}_{(X,Y)\sim P}(f(X)\ne Y).
$$

So the true performance of a fixed classifier is the probability that it misclassifies a fresh draw from the data-generating distribution.

</details>

### Question 3

The lecture says $D_{eval}\perp D_{train}$ is sufficient for $D_{eval}\perp\hat f$ when $\hat f=A(D_{train})$. Why?

<details>
<summary>Answer</summary>

A standard property of independence says that if $U\perp V$, then $U\perp g(V)$ for every function $g$. Taking $U=D_{eval}$, $V=D_{train}$, and $g=A$ gives $D_{eval}\perp A(D_{train})=\hat f$.

</details>




### Question 4

Construct a concrete edge case, matching the lecture's discussion, where $D_{eval}=D_{train}$ but nevertheless $D_{eval}\perp\hat f$. 

<details>
<summary>Answer</summary>

Let $A(D)$ ignore its input and always return the constant model $\hat f(x)=0$. Then $\hat f$ is deterministic, so it is independent of every random variable, including $D_{eval}=D_{train}$. Hence $D_{eval}\perp\hat f$ even though $D_{eval}$ and $D_{train}$ are identical. This illustrates that independence from the fitted model is the key condition, not literal separation from the training set by itself.

</details>




### Question 5

For least squares regression with noise variance $\sigma^2$, the lecture states the optimism formula
$$
\mathbb E[\mathrm{Perf}(\hat f)-\hat{\mathrm{Perf}}(D_{train},\hat f)]
=\frac{2\sigma^2}{N}d.
$$
Compute the optimism when $N=100$, $d=8$, and $\sigma^2=4$. Interpret the units of the answer.

<details>
<summary>Answer</summary>

Substitution gives
$$
\frac{2\sigma^2}{N}d=\frac{2\cdot 4}{100}\cdot 8=\frac{64}{100}=0.64.
$$
Since the metric is mean squared error, the optimism is measured in squared-response units. It says the expected training MSE is lower than the corresponding fresh-data MSE by $0.64$.

</details>




### Question 6

Use the covariance version of Efron's optimism theorem to explain precisely why no optimism appears when every fitted prediction $\hat y_i$ is independent of its observed response $y_i$. 

<details>
<summary>Answer</summary>

If $\hat y_i\perp y_i$ and both variables have finite second moments, then
$$
\mathrm{Cov}(\hat y_i,y_i)=\mathbb E[\hat y_i y_i]-\mathbb E[\hat y_i]\mathbb E[y_i]=0.
$$
Therefore
$$
\frac{2}{N}\sum_{i=1}^N\mathrm{Cov}(\hat y_i,y_i)=0.
$$
Under the theorem's assumptions, the expected gap between true performance and training performance is then $0$.

</details>




### Question 7

Assume that $y$ and $\hat y$ are vectors in $\mathbb{R}^N$, with

$$
\hat y = Sy,
$$

where $S$ is a fixed $N\times N$ matrix. Also assume

$$
y_i=\mu_i+\varepsilon_i,
$$

where the errors $\varepsilon_i$ are independent, have mean 0, and have variance $\sigma^2$.

Show that

$$
\sum_{i=1}^N \mathrm{Cov}(\hat y_i,y_i)=\sigma^2\mathrm{tr}(S).
$$

<details>
<summary>Answer</summary>

Since $\hat y=Sy$, the $i$th fitted value is

$$
\hat y_i=\sum_{j=1}^N S_{ij}y_j.
$$

Therefore,

$$
\mathrm{Cov}(\hat y_i,y_i)=\mathrm{Cov}\left(\sum_{j=1}^N S_{ij}y_j,y_i\right).
$$

Because covariance is linear in its first argument and the entries of $S$ are fixed constants,

$$
\mathrm{Cov}(\hat y_i,y_i)=\sum_{j=1}^N S_{ij}\mathrm{Cov}(y_j,y_i).
$$

Now use the model $y_i=\mu_i+\varepsilon_i$. The mean terms $\mu_i$ are fixed, so they do not affect covariance. Thus,

$$
\mathrm{Cov}(y_j,y_i)=\mathrm{Cov}(\varepsilon_j,\varepsilon_i).
$$

Because the errors are independent and each has variance $\sigma^2$,

$$
\mathrm{Cov}(y_j,y_i)=
\begin{cases}
\sigma^2, & j=i,\\
0, & j\ne i.
\end{cases}
$$

So in the sum

$$
\sum_{j=1}^N S_{ij}\mathrm{Cov}(y_j,y_i),
$$

all terms are zero except the term with $j=i$. Therefore,

$$
\mathrm{Cov}(\hat y_i,y_i)=S_{ii}\sigma^2.
$$

Finally, summing over $i$ gives

$$
\sum_{i=1}^N \mathrm{Cov}(\hat y_i,y_i)=\sum_{i=1}^N S_{ii}\sigma^2.
$$

Since $\sigma^2$ is constant,

$$
\sum_{i=1}^N \mathrm{Cov}(\hat y_i,y_i)=\sigma^2\sum_{i=1}^N S_{ii}.
$$

By definition, the trace of $S$ is the sum of its diagonal entries:

$$
\mathrm{tr}(S)=\sum_{i=1}^N S_{ii}.
$$

Therefore,

$$
\sum_{i=1}^N \mathrm{Cov}(\hat y_i,y_i)=\sigma^2\mathrm{tr}(S).
$$

</details>

### Question 8

A model-selection procedure chooses polynomial the degree $\hat r$ with the smallest evaluation MSE on a fixed evaluation set, then reports that same minimum as the performance estimate. Identify the independence condition from the lecture that is broken, and describe the expected direction of the bias.

<details>
<summary>Answer</summary>

After selecting $\hat r$ using the evaluation set, the final chosen model or procedure depends on $D_{eval}$. Therefore $D_{eval}\perp\hat f$ no longer holds for the selected model. The reported minimum evaluation error is typically optimistic: among several noisy estimates, selecting the smallest tends to underestimate the fresh-data error of the selected degree.

</details>




### Question 9

Cross-validation estimates algorithm-level performance rather than the performance of one fixed model. Write the target quantity for an algorithm $A$ trained on $n$ observations, and explain why $k$-fold CV more directly targets $n=N(k-1)/k$ than $n=N$.

<details>
<summary>Answer</summary>

The algorithm-level target for training size $n$ is
$$
\mathbb E_{D_{train}\sim P^n}[\mathrm{Perf}(A(D_{train}))].
$$
In each $k$-fold CV fit, the algorithm is trained on $N(k-1)/k$ observations, not all $N$. Therefore the models being evaluated are draws from the distribution of $A(D_{train})$ with training size $N(k-1)/k$. The final model trained on all $N$ observations may perform slightly better, which is why the lecture describes CV as somewhat pessimistic.

</details>




### Question 10

A collaborator asks which of the $k$ fitted models from cross-validation should be deployed. What's the answer?

<details>
<summary>Answer</summary>

None of the fold-specific models $\hat f^{(j)}=A(D\setminus D_j)$ is the deployed model. They are auxiliary models used to estimate the performance of the training procedure. After choosing the procedure $A$ and estimating its performance, one trains
$$
\hat f_{final}=A(D_{all})
$$
on all available data and deploys that model. The CV score is interpreted as evidence about the procedure's expected fresh-data performance, not as the fresh-data performance of one particular fold model.

</details>




### Question 11

Resampling by repeated random splits produces
$$
R_B=\frac1B\sum_{b=1}^B \hat{\mathrm{Perf}}(D_{eval}^{(b)},\hat f^{(b)}).
$$
Assuming the random splits are independent, compute $\mathrm{Var}(R_B)$ in terms of the variance of one split score $S_b = \hat{\mathrm{Perf}}(D_{eval}^{(b)},\hat f^{(b)})$ assuming that the $S_b$ are independent with common variance $\mathrm{Var}(S_b) = \tau^2$.

<details>
<summary>Answer</summary>

If the split scores $S_b=\hat{\mathrm{Perf}}(D_{eval}^{(b)},\hat f^{(b)})$ are independent conditional on the dataset and each has variance $\tau^2$, then
$$
\mathrm{Var}(R_B)=\mathrm{Var}\left(\frac1B\sum_{b=1}^B S_b\right)
=\frac1{B^2}\sum_{b=1}^B\mathrm{Var}(S_b)
=\frac{\tau^2}{B}.
$$
(Note that real split scores are often dependent because they reuse observations.)

</details>




### Question 12

For a binary classifier, suppose the evaluation set gives $TP=40$, $FP=10$, $FN=20$, and $TN=30$. Compute classification error and F1 score. Which one uses $TN$ directly?

<details>
<summary>Answer</summary>

The evaluation size is $M=40+10+20+30=100$. Classification error is
$$
\frac{FP+FN}{M}=\frac{10+20}{100}=0.30.
$$
The F1 score is
$$
\frac{2TP}{2TP+FP+FN}=\frac{80}{80+10+20}=\frac{80}{110}=\frac{8}{11}\approx 0.727.
$$

</details>




### Question 13

Suppose we evaluate two classifiers, A and B, on the same dataset of 100 observations. The dataset has 10 positive cases and 90 negative cases.

The confusion matrices are:

| Classifier | TP | FP | FN | TN |
| ---------- | -: | -: | -: | -: |
| A          |  0 |  0 | 10 | 90 |
| B          |  8 | 20 |  2 | 70 |

For each classifier, calculate:

1. The classification error.
2. The F1 score.

Then answer the following:

* Which classifier is better according to classification error?
* Which classifier is better according to F1 score?
* Briefly explain what this example shows about metric choice.

<details>
<summary>Answer</summary>

For classifier A,

$$
\text{Error}_A = \frac{0 + 10}{100} = 0.10
$$

and

$$
F1_A = \frac{2(0)}{2(0) + 0 + 10} = 0.
$$

For classifier B,

$$
\text{Error}_B = \frac{20 + 2}{100} = 0.22
$$

and

$$
F1_B = \frac{2(8)}{2(8) + 20 + 2}
= \frac{16}{38}
\approx 0.421.
$$

Classifier A is better by classification error because (0.10 < 0.22).

Classifier B is better by F1 score because (0.421 > 0).

So the two metrics rank the classifiers in opposite orders. This shows that the choice of evaluation metric can change which model appears better.

</details>


### Question 14

A dataset is randomly split into $D_{train}$ and $D_{eval}$, but preprocessing parameters such as centering and scaling are fit using all observations before the split. Does the lecture's independence requirement automatically hold? Give a precise answer.

<details>
<summary>Answer</summary>

No. If preprocessing parameters are estimated from all observations and then used by the training algorithm, the fitted model can depend on information from $D_{eval}$. Then $D_{eval}\perp\hat f$ need not hold, even if the raw rows were randomly split. To preserve the fresh-data interpretation, preprocessing parameters that affect the model must be fit using only $D_{train}$ within each split or fold, then applied to the corresponding evaluation data.

</details>




### Question 15

A false proof claims: "Because each cross-validation evaluation fold $D_j$ was not used to train $\hat f^{(j)}$, the $k$ fold scores are independent random variables." Identify the false step and give the correct dependence issue.

<details>
<summary>Answer</summary>

The false step is confusing independence of a fold from its own fitted model with independence among the fold scores. The training sets $D\setminus D_j$ overlap heavily across folds, and the fold scores are all functions of the same original dataset $D$. Therefore the scores are generally dependent. Cross-validation can still be useful, but the simple variance formula for an average of independent fold scores is not automatically valid.

</details>




### Question 16

The following statement is missing an important assumption:

"If $D_{eval}$ is independent of $\hat f$, then $\hat{\mathrm{Perf}}(D_{eval},\hat f)$ estimates $\mathrm{Perf}(\hat f)$."

What assumption is missing?

<details>
<summary>Answer</summary>

The missing assumption is that the evaluation data must come from the same distribution used to define fresh-data performance. In other words, we need

$$
D_{eval}\sim P^M.
$$

Independence alone only says that the evaluation data were not used to train the model. That helps prevent data reuse or overfitting to the evaluation set. But it does not guarantee that the evaluation set represents the target population we care about.

An important idea is **training/testing dataset drift**. A test set can be independent of the fitted model and still give the wrong answer if it comes from a different distribution than the fresh data we care about. In that case, the evaluation error estimates performance on the test distribution, not performance on the target distribution $P$

</details>

### Question 17

In the lecture's cartoon plot of training error and testing error versus model complexity, divide the horizontal axis into three qualitative regions: low complexity, intermediate complexity, and high complexity. For each region, state whether the main problem is underfitting, reasonable generalization, or overfitting, and describe the typical relationship between training error and testing error.

<details>
<summary>Answer</summary>

At low complexity, the model class is too restricted to capture the signal, so both training error and testing error are relatively high. This is underfitting: the model is not fitting either the observed data or new data very well.

At intermediate complexity, the model has enough flexibility to capture much of the real signal without adapting too strongly to sample-specific noise. Training error is lower than in the low-complexity region, and testing error is typically near its minimum. The train-test gap may exist, but it is not yet large.

At high complexity, training error keeps decreasing, sometimes nearly to zero, but testing error increases. This is overfitting: the model is fitting noise or accidental patterns in $D_{train}$ that do not repeat in fresh data.

</details>


### Question 18

A student looks at the complexity plot and says: "If increasing model complexity always lowers training error, then the most complex model must be the best model." Diagnose the mistake using the lecture's distinction between performance on $D_{train}$ and performance on new independent data.

<details>
<summary>Answer</summary>

The mistake is treating training error as the target quantity. Training error measures how well $\hat f$ fits the same data used to construct it, so it is affected by the dependence between $D_{train}$ and $\hat f$. It can keep improving as complexity increases because the model has more ways to adapt to this particular dataset.

The lecture's target is performance on new independent data, represented by $\mathrm{Perf}(\hat f)$ or by evaluation error on $D_{eval}$ satisfying $D_{eval}\sim P^M$ and $D_{eval}\perp \hat f$. A highly complex model may lower training error by fitting sample-specific quirks, but those quirks do not generalize. Therefore the most complex model can have worse testing error even though it has better training error.

</details>


### Question 20

Two polynomial degrees have nearly the same evaluation error on an independent evaluation set. Degree 4 has noticeably higher training error, while degree 14 has much lower training error. Based on the lecture's discussion of complexity and overfitting, what would be a reasonable reason to prefer degree 4? 

<details>
<summary>Answer</summary>

If the independent evaluation errors are essentially tied, then the observed advantage of degree 14 is only on the training data. That advantage is less trustworthy because training error is optimistic: the degree-14 model has more capacity to adapt to noise and sample-specific quirks in $D_{train}$.

A reasonable reason to prefer degree 4 is that it achieves comparable estimated fresh-data performance with less dependence on the particular training sample. The argument is not that lower complexity is automatically better; it is that, when independent evaluation performance is similar, the extra training-fit improvement from the more complex model is weak evidence of better generalization and stronger evidence of possible overfitting.

</details>


### Question 21

The lecture says complex models are more sensitive to the data. Translate this sensitivity statement into the language of Efron's optimism discussion: what tends to happen to the covariance terms $\mathrm{Cov}(\hat y_i,y_i)$ as the fitted model adapts more strongly to $D_{train}$? How does that relate to the widening gap between training and testing error in the complexity diagram?

<details>
<summary>Answer</summary>

When a fitted model adapts strongly to $D_{train}$, each fitted value $\hat y_i$ tends to depend more strongly on the observed response $y_i$. In Efron's optimism formula, that stronger dependence appears as larger positive covariance terms $\mathrm{Cov}(\hat y_i,y_i)$.

Larger covariance terms imply larger optimism: the training error is expected to be farther below the fresh-data error. This is the same phenomenon shown in the high-complexity part of the cartoon diagram, where the training curve continues downward while the testing curve turns upward and the train-test gap widens.

</details>
