# Random Forests and Bagging - Review Questions {#sec-rf-questions}

### Question 1

A single deep tree usually has low bias but high variance. State precisely what bagging changes about the model-building process and why this targets variance rather than the functional form of an individual tree.

<details>
<summary>Answer</summary>

Bagging draws many bootstrap samples from the training data, fits one base tree $\hat s_b$ to each bootstrap sample, and averages their predictions. The individual fitted models are still trees, so the base functional form is not replaced by a different model class.

The main change is that the final predictor is an average of many randomized predictors. If the individual trees are unstable, their predictions vary substantially across bootstrap samples. Averaging reduces the part of that variation that is not shared across trees, so bagging primarily targets variance.

</details>


### Question 2

Let $Z_1,\ldots,Z_B$ have the same mean $\mu$, and define $\bar Z_B=B^{-1}\sum_{b=1}^B Z_b$. Fill in the linearity-of-expectation step proving $\mathbb E[\bar Z_B]=\mu$.

<details>
<summary>Answer</summary>

By linearity of expectation,

$$
\mathbb E[\bar Z_B]
=\mathbb E\left[\frac{1}{B}\sum_{b=1}^B Z_b\right]
=\frac{1}{B}\sum_{b=1}^B \mathbb E[Z_b].
$$

Since every $Z_b$ has mean $\mu$,

$$
\mathbb E[\bar Z_B]
=\frac{1}{B}\sum_{b=1}^B \mu
=\frac{B\mu}{B}
=\mu.
$$

So averaging does not change the mean of the base random variables.

</details>


### Question 3

Assume $Z_1,\ldots,Z_B$ are uncorrelated and each has variance $\sigma^2$. Derive

$$
\operatorname{Var}(\bar Z_B)=\frac{\sigma^2}{B}.
$$

<details>
<summary>Answer</summary>

Since

$$
\bar Z_B=\frac{1}{B}\sum_{b=1}^B Z_b,
$$

we have

$$
\operatorname{Var}(\bar Z_B)
=\frac{1}{B^2}\operatorname{Var}\left(\sum_{b=1}^B Z_b\right).
$$

For uncorrelated variables, the variance of the sum is the sum of variances:

$$
\operatorname{Var}\left(\sum_{b=1}^B Z_b\right)=\sum_{b=1}^B\sigma^2=B\sigma^2.
$$

Thus

$$
\operatorname{Var}(\bar Z_B)=\frac{1}{B^2}B\sigma^2=\frac{\sigma^2}{B}.
$$

</details>


### Question 4

For equicorrelated variables with $\operatorname{Var}(Z_b)=\sigma^2$ and $\operatorname{Corr}(Z_b,Z_{b'})=\rho$ for $b\ne b'$, derive

$$
\operatorname{Var}(\bar Z_B)=\rho\sigma^2+\frac{1-\rho}{B}\sigma^2.
$$

Make explicit where the factor $B(B-1)$ comes from.

<details>
<summary>Answer</summary>

First,

$$
\operatorname{Cov}(Z_b,Z_{b'})=\rho\sigma^2
\quad(b\ne b').
$$

Then

$$
\operatorname{Var}\left(\sum_{b=1}^B Z_b\right)
=\sum_{b=1}^B\operatorname{Var}(Z_b)
+\sum_{b\ne b'}\operatorname{Cov}(Z_b,Z_{b'}).
$$

There are $B$ variance terms and $B(B-1)$ ordered off-diagonal covariance terms. Therefore

$$
\operatorname{Var}\left(\sum_{b=1}^B Z_b\right)
=B\sigma^2+B(B-1)\rho\sigma^2.
$$

Dividing by $B^2$ gives

$$
\operatorname{Var}(\bar Z_B)
=\frac{1}{B^2}\{B\sigma^2+B(B-1)\rho\sigma^2\}
=\frac{\sigma^2}{B}+\frac{B-1}{B}\rho\sigma^2.
$$

Since

$$
\frac{\sigma^2}{B}+\frac{B-1}{B}\rho\sigma^2
=\rho\sigma^2+\frac{1-\rho}{B}\sigma^2,
$$

the formula follows.

</details>


### Question 5

Compute the variance of an average when $\sigma^2=16$, $B=25$, and $\rho=0.10$. Also compute the limiting variance as $B\to\infty$.

<details>
<summary>Answer</summary>

Using the lecture's formula,

$$
\operatorname{Var}(\bar Z_B)=\rho\sigma^2+\frac{1-\rho}{B}\sigma^2.
$$

Substitute $\sigma^2=16$, $B=25$, and $\rho=0.10$:

$$
0.10(16)+\frac{0.90}{25}(16)=1.6+0.576=2.176.
$$

As $B\to\infty$, the second term vanishes, so the limiting variance is

$$
\rho\sigma^2=0.10(16)=1.6.
$$

</details>


### Question 6

A forest design A has $\rho=0.20$ and $\sigma^2=9$. Design B has $\rho=0.08$ and $\sigma^2=25$. For very large $B$, which design has smaller limiting variance, and what does this show about making trees less regularized?

<details>
<summary>Answer</summary>

For very large $B$, the limiting variance is approximately

$$
\rho\sigma^2.
$$

Design A has limiting variance

$$
0.20\cdot 9=1.8.
$$

Design B has limiting variance

$$
0.08\cdot 25=2.0.
$$

So A has the smaller limiting variance, even though B has lower correlation. This illustrates the lecture's point: reducing correlation is useful, but not if the individual tree variance $\sigma^2$ increases enough to offset the lower $\rho$.

</details>


### Question 7

A bootstrap sample has size $N$ and is drawn with replacement from $N$ training observations. Derive the probability that a fixed observation is out of bag for one bootstrap sample, and compute the approximate value for large $N$.

<details>
<summary>Answer</summary>

On one draw, the probability the fixed observation is not selected is

$$
1-\frac{1}{N}.
$$

A bootstrap sample consists of $N$ independent draws with replacement, so the probability it is never selected is

$$
\left(1-\frac{1}{N}\right)^N.
$$

As $N$ grows,

$$
\left(1-\frac{1}{N}\right)^N\to e^{-1}\approx0.368.
$$

Thus each bootstrap tree leaves out about $36.8\%$ of the training observations.

</details>


### Question 8

Define the out-of-bag set $O_b$ for a particular tree $b$ and let the index set $B_n=\{b:n\in O_b\}$. Why does the OOB prediction for $x_n$ act like an internal validation prediction?

<details>
<summary>Answer</summary>

For tree $b$, $O_b$ is the set of training indices left out of the bootstrap sample used to fit that tree. For a fixed training observation $n$,

$$
B_n=\{b:n\in O_b\}
$$

is the set of trees for which observation $n$ was out of bag.

The OOB score is

$$
\hat s_{\text{OOB},n}(x_n)=\frac{1}{|B_n|}\sum_{b\in B_n}\hat s_b(x_n).
$$

Each tree in this average was fit without using observation $n$. Therefore the OOB prediction evaluates $x_n$ using only models that did not train on $x_n$, which makes it analogous to a validation prediction.

</details>


### Question 9

For regression, three bagged trees predict $2.4$, $1.8$, and $3.0$ at the same input $x$. Compute $\hat s_{\mathrm{bag}}(x)$.

<details>
<summary>Answer</summary>

The bagged regression score is the average of the tree scores:

$$
\hat s_{\mathrm{bag}}(x)=\frac{1}{B}\sum_{b=1}^B\hat s_b(x).
$$

With the three predictions,

$$
\hat s_{\mathrm{bag}}(x)=\frac{2.4+1.8+3.0}{3}=\frac{7.2}{3}=2.4.
$$

For regression, the final prediction is the same number because the action is the identity.

</details>


### Question 10

Three binary classification trees produce soft class-probability vectors

$$
(0.51,0.49),\qquad (0.51,0.49),\qquad (0.01,0.99).
$$

Compute the hard-vote prediction and the soft-vote prediction. 

<details>
<summary>Answer</summary>

For hard voting, the first two trees predict class $1$ and the third predicts class $2$. The vote proportions are

$$
\left(\frac{2}{3},\frac{1}{3}\right),
$$

so hard voting predicts class $1$.

For soft voting, average the probability vectors:

$$
\frac{1}{3}\{(0.51,0.49)+(0.51,0.49)+(0.01,0.99)\}
=(0.3433\ldots,0.6566\ldots).
$$

Soft voting predicts class $2$.

The example illustrates that hard voting ignores confidence information, while soft voting uses the probability estimates and can be strongly affected by them.

</details>


### Question 11

If a dataset has $D=100$ predictors and a random forest uses $M=\sqrt D$ candidate features at each split, how many features are considered per split? What fraction of the features is this?

<details>
<summary>Answer</summary>

Since

$$
M=\sqrt{100}=10,
$$

the split considers $10$ features.

The fraction of the full feature set is

$$
\frac{10}{100}=0.10.
$$

So each split searches over about $10\%$ of the predictors.

</details>


### Question 12

Suppose there are only a few predictors, and nearly all signal is carried by one strong predictor. Why might a random forest perform worse than bagging ordinary trees?

<details>
<summary>Answer</summary>

Bagged trees with all features available can repeatedly use that predictor when it gives the best split.

If a random forest uses very small $M$, many splits will not be allowed to consider the strong predictor. The trees become weaker because important splits are unavailable. The lower correlation may not compensate for this increased bias, so the random forest can perform worse than bagging.

</details>


### Question 13

The lecture warns that for impurity-based importance scores, correlated features can split importance. Why?

<details>
<summary>Answer</summary>

If two predictors are strongly correlated, either one may produce similar split improvements. Different trees or splits may choose one or the other, so the total importance can be divided between them. Each correlated variable may look less important individually than the shared predictive signal really is.

</details>


### Question 14

Why are random forests usually not sensitive to monotone rescaling of individual features? 
<details>
<summary>Answer</summary>

Monotone rescaling preserves the ordering of observations along the feature, so the same left/right partition can be represented with a transformed threshold. Since random forests are built from such splits, standardization is usually not required for basic random forests.

</details>


### Question 15

Explain why random forests are usually poor at extrapolating outside the range of the training data.

<details>
<summary>Answer</summary>

A tree predicts a constant value inside each terminal leaf. A random forest averages many such tree predictions, so its prediction is still an average of constants determined by training-data leaf values.

Outside the training range, a new point is routed into terminal leaves according to threshold rules, but the leaf predictions are still based on training outcomes observed inside the training data. The model has no mechanism like a linear slope that continues a trend beyond the observed range. Therefore random forests tend to flatten rather than extrapolate trends.

</details>


### Question 16

A student says: "A random forest with more trees must eventually overfit because it has more parameters." Is this true?

<details>
<summary>Answer</summary>

No. The number of trees $B$ is different from the flexibility of individual trees. Increasing $B$ averages more randomized trees and mainly reduces Monte Carlo noise in the ensemble prediction. It does not make each individual tree deeper.

Overfitting is more directly controlled by tree-size parameters such as maximum depth, minimum samples per leaf, and minimum samples needed to split. More trees increase computation, but they do not cause overfitting in the same way that making each tree more flexible can.

</details>


### Question 17

Summarize the main tuning parameters from the lecture: $B$, $M$, and tree-size controls. For each one, state the primary tradeoff it controls.

<details>
<summary>Answer</summary>

$B$ is the number of trees. Larger $B$ reduces Monte Carlo noise and stabilizes predictions, but costs more time and memory.

$M$ is the number of candidate features considered at each split. Smaller $M$ tends to reduce correlation among trees, but it can make individual splits weaker and increase bias if important variables are often unavailable.

Tree-size controls include maximum depth, minimum samples per leaf, and minimum samples needed to split. Larger trees tend to have lower bias but higher variance before averaging; stronger restrictions make individual trees smoother or less flexible.

</details>
