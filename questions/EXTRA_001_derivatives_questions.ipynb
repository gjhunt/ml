# Matrix Calculus for Machine Learning - Review Questions {#sec-deriv-questions}



### Question 1

For $w\in\mathbb{R}^d$ and a constant matrix $A\in\mathbb{R}^{d\times d}$, show that $\nabla_w(w^\top A w) = (A+A^\top)w$. 

<!-- Source: quadratic vector identity $\nabla_w(w^\top A w)=(A+A^\top)w$ and Hessian discussion -->

<details>
<summary>Answer</summary>

We want to compute the gradient of the scalar-valued function

$$
f(w) = w^\top A w,
$$

where $w \in \mathbb{R}^d$ and $A \in \mathbb{R}^{d \times d}$ is constant.

First, write the quadratic form in coordinates. Since

$$
w^\top A w=
\sum_{i=1}^d \sum_{j=1}^d w_i A_{ij} w_j,
$$

we can compute the $k$th component of the gradient by differentiating with respect to $w_k$:

$$
\frac{\partial}{\partial w_k}(w^\top A w)=
\frac{\partial}{\partial w_k}
\left(
\sum_{i=1}^d \sum_{j=1}^d w_i A_{ij} w_j
\right).
$$

The variable $w_k$ can appear in two places: as the left factor $w_i$ when $i=k$, and as the right factor $w_j$ when $j=k$. Therefore,

$$
\frac{\partial}{\partial w_k}(w^\top A w)=
\sum_{j=1}^d A_{kj} w_j
+
\sum_{i=1}^d w_i A_{ik}.
$$

The first term is the $k$th entry of $Aw$:

$$
\sum_{j=1}^d A_{kj} w_j = (Aw)_k.
$$

The second term can be rewritten as the $k$th entry of $A^\top w$, since

$$
(A^\top w)_k=
\sum_{i=1}^d (A^\top)_{ki} w_i=
\sum_{i=1}^d A_{ik} w_i.
$$

Thus,

$$
\frac{\partial}{\partial w_k}(w^\top A w)=
(Aw)_k + (A^\top w)_k=
((A + A^\top)w)_k.
$$

Since this holds for each coordinate $k=1,\ldots,d$, we conclude that

$$
\nabla_w(w^\top A w) = (A + A^\top)w.
$$


</details>

### Question 2

The lecture writes $\nabla_w\,\frac12 w^\top A w=Aw$ when $A$ is symmetric. Give a concrete $2\times2$ counterexample showing the formula can fail when $A$ is not symmetric. 

<!-- Source: symmetric-matrix assumption in the half-quadratic formula -->

<details>
<summary>Answer</summary>

The needed assumption is $A=A^\top$. In general,
$$\nabla_w\frac12 w^\top A w=\frac12(A+A^\top)w.$$
Take
$$A=\begin{bmatrix}0&1\\0&0\end{bmatrix},\qquad w=\begin{bmatrix}u\\v\end{bmatrix}.$$
Then $\frac12w^\top Aw=\frac12uv$, so the gradient is $\frac12(v,u)^\top$. But $Aw=(v,0)^\top$, which is not equal in general. 

</details>

### Question 3

Fill in the coordinate argument behind $\nabla_w(a^\top w)=a$. Use $a,w\in\mathbb{R}^d$ and explicitly compute the $k$th partial derivative.

<!-- Source: linear vector derivative identity $\nabla_w(a^\top w)=a$ -->

<details>
<summary>Answer</summary>

Since
$$a^\top w=\sum_{i=1}^d a_iw_i,$$
the $k$th partial derivative is
$$\frac{\partial}{\partial w_k}(a^\top w)=\sum_{i=1}^d a_i\frac{\partial w_i}{\partial w_k}=a_k.$$
Therefore the gradient vector is $(a_1,\ldots,a_d)^\top=a$.

</details>

### Question 5

A scalar-valued function $L:\mathbb{R}^n\to\mathbb{R}$ may be differentiated either as a Jacobian or as a gradient. State the shapes of $J_L(x)$ and $\nabla_xL$, then give the precise relationship between them under the lecture's numerator-layout convention.

<!-- Source: notational tension between scalar-output Jacobians and gradients -->

<details>
<summary>Answer</summary>

Under numerator layout, the Jacobian has output coordinates as rows and input coordinates as columns. Since $L$ has one output coordinate,
$$J_L(x)\in\mathbb{R}^{1\times n}.$$
The gradient is arranged to match the vector input, so $\nabla_xL\in\mathbb{R}^n$, viewed as a column vector. The relationship is
$$J_L(x)=(\nabla_xL)^\top.$$

</details>

### Question 6

Diagnose the shape error in the claim: if $z=f(w)\in\mathbb{R}^m$, $w\in\mathbb{R}^d$, and $L=g(z)$ is scalar-valued, then $\nabla_wL=J_f(w)\nabla_zL$. State the corrected formula and verify its shape.

<!-- Source: scalar-output chain rule through a vector intermediate -->

<details>
<summary>Answer</summary>

The product $J_f(w)\nabla_zL$ is not generally defined with the lecture's shapes: $J_f(w)\in\mathbb{R}^{m\times d}$ and $\nabla_zL\in\mathbb{R}^m$. The inner dimensions $d$ and $m$ do not match unless $d=m$.

The correct formula is
$$\nabla_wL=J_f(w)^\top\nabla_zL.$$
Here $J_f(w)^\top\in\mathbb{R}^{d\times m}$ and $\nabla_zL\in\mathbb{R}^m$, so the result lies in $\mathbb{R}^d$, matching the shape of $w$.

</details>

### Question 7

For $f:\mathbb{R}^2\to\mathbb{R}^3$ defined by
$$f(x_1,x_2)=\begin{bmatrix}x_1+2x_2\\ x_1x_2\\ x_2^2\end{bmatrix},$$
compute $J_f(x)$ using the lecture's row-output, column-input convention.

<!-- Source: definition of the Jacobian for vector-output functions -->

<details>
<summary>Answer</summary>

The rows correspond to output coordinates and the columns to input coordinates:
$$J_f(x)=\begin{bmatrix}
\partial f_1/\partial x_1 & \partial f_1/\partial x_2\\
\partial f_2/\partial x_1 & \partial f_2/\partial x_2\\
\partial f_3/\partial x_1 & \partial f_3/\partial x_2
\end{bmatrix}
=\begin{bmatrix}
1&2\\
x_2&x_1\\
0&2x_2
\end{bmatrix}.$$
Its shape is $3\times2$, equal to output dimension times input dimension.

</details>

### Question 9

Verify $\nabla_W\operatorname{tr}(A^\top W)=A$ by expanding the trace in entries. Assume $A,W\in\mathbb{R}^{m\times n}$ and $A$ is constant with respect to $W$.

<!-- Source: matrix linear trace derivative identity -->

<details>
<summary>Answer</summary>

The trace expands as
$$\operatorname{tr}(A^\top W)=\sum_{i=1}^m\sum_{j=1}^n A_{ij}W_{ij}.$$
Therefore
$$\frac{\partial}{\partial W_{kl}}\operatorname{tr}(A^\top W)=A_{kl}.$$
Since this holds for every entry, the matrix gradient has entries $(\nabla_WL)_{kl}=A_{kl}$, so $\nabla_W\operatorname{tr}(A^\top W)=A$.

</details>

### Question 10

Connect the Frobenius norm identity to the trace identity by proving
$$\|W\|_F^2=\operatorname{tr}(W^\top W)$$
for $W\in\mathbb{R}^{m\times n}$, then use entries to compute $\nabla_W\|W\|_F^2$.

<!-- Source: Frobenius norm formula and its gradient -->

<details>
<summary>Answer</summary>

Recall that the Frobenius norm of a matrix $W \in \mathbb{R}^{m \times n}$ is defined by

$$
\|W\|_F^2=
\sum_{i=1}^m \sum_{j=1}^n W_{ij}^2.
$$

We want to connect this to the trace expression $\operatorname{tr}(W^\top W)$.

First, note that $W^\top W \in \mathbb{R}^{n \times n}$. Its $(j,j)$ diagonal entry is

$$
(W^\top W)_{jj}=
\sum_{i=1}^m (W^\top)_{ji} W_{ij}.
$$

Since $(W^\top)_{ji} = W_{ij}$, this becomes

$$
(W^\top W)_{jj}=
\sum_{i=1}^m W_{ij}^2.
$$

The trace of a matrix is the sum of its diagonal entries, so

$$
\operatorname{tr}(W^\top W)=
\sum_{j=1}^n (W^\top W)_{jj}.
$$

Substituting the expression for the diagonal entries gives

$$
\operatorname{tr}(W^\top W)=
\sum_{j=1}^n \sum_{i=1}^m W_{ij}^2.
$$

This is just the sum of the squares of all entries of $W$, so

$$
\operatorname{tr}(W^\top W)=
\|W\|_F^2.
$$

Now compute the gradient entry by entry. Since

$$
\|W\|_F^2=
\sum_{i=1}^m \sum_{j=1}^n W_{ij}^2,
$$

the partial derivative with respect to a single entry $W_{ij}$ is

$$
\frac{\partial}{\partial W_{ij}} \|W\|_F^2=
2W_{ij}.
$$

Thus the matrix of partial derivatives has $(i,j)$ entry equal to $2W_{ij}$. Therefore,

$$
\nabla_W \|W\|_F^2 = 2W.
$$

Equivalently, using the trace form,

$$
\nabla_W \operatorname{tr}(W^\top W) = 2W.
$$

</details>


### Question 11

For $W\in\mathbb{R}^{m\times n}$ and fixed $B\in\mathbb{R}^{m\times n}$, compute
$$\nabla_W\left(\frac12\|W-B\|_F^2\right).$$
Give the entrywise reasoning.

<!-- Source: Frobenius norm derivative as matrix version of $\frac12\|w\|^2$ -->

<details>
<summary>Answer</summary>

Expand the objective as
$$L(W)=\frac12\sum_{i=1}^m\sum_{j=1}^n (W_{ij}-B_{ij})^2.$$
For each entry,
$$\frac{\partial L}{\partial W_{ij}}=W_{ij}-B_{ij}.$$
Thus
$$\nabla_W\left(\frac12\|W-B\|_F^2\right)=W-B.$$

</details>

### Question 12

Use the linear approximation
$$f(x+dx)\approx f(x)+J_f(x)dx$$
to justify the lecture's choice that $J_f(x)$ should have shape $m\times n$ for $f:\mathbb{R}^n\to\mathbb{R}^m$.

<!-- Source: numerator-layout convention and the clean approximation formula -->

<details>
<summary>Answer</summary>

The increment $dx$ has shape $n\times1$, and $f(x+dx)-f(x)$ has shape $m\times1$. For the product $J_f(x)dx$ to have shape $m\times1$, the Jacobian must have shape $m\times n$. This is exactly output dimension times input dimension, matching numerator layout.

</details>

### Question 13

For $f(x)=Ax+b$ with $A\in\mathbb{R}^{m\times n}$ and $b\in\mathbb{R}^m$, verify from coordinates that $J_f(x)=A$. 

<!-- Source: Jacobian of linear and affine maps -->

<details>
<summary>Answer</summary>

The $i$th coordinate is
$$f_i(x)=\sum_{j=1}^n A_{ij}x_j+b_i.$$
Thus
$$\frac{\partial f_i}{\partial x_k}=A_{ik}+\frac{\partial b_i}{\partial x_k}=A_{ik},$$
because $b_i$ is constant with respect to every input coordinate. Therefore the $(i,k)$ entry of $J_f(x)$ is $A_{ik}$, so $J_f(x)=A$.

</details>

### Question 14

For the elementwise rule in the lecture, let
$$F(t)=\begin{bmatrix}t&t^2\\-t&1\end{bmatrix},\qquad \phi(u)=u^3.$$
Compute $\frac{d}{dt}\phi(F(t))$ using $\phi'(F(t))\odot F'(t)$.

<!-- Source: entrywise matrix-output chain rule with Hadamard product -->

<details>
<summary>Answer</summary>

We have
$$F'(t)=\begin{bmatrix}1&2t\\-1&0\end{bmatrix},\qquad \phi'(F(t))=3F(t)\odot F(t)=\begin{bmatrix}3t^2&3t^4\\3t^2&3\end{bmatrix}.$$
Thus
$$\frac{d}{dt}\phi(F(t))=\phi'(F(t))\odot F'(t)=\begin{bmatrix}3t^2&6t^5\\-3t^2&0\end{bmatrix}.$$
For example, the $(1,2)$ entry is $(t^2)^3=t^6$, whose derivative is $6t^5$, matching the formula.

</details>

### Question 15

Evaluate the chain rule for a concrete scalar objective. Let $f(w)=Bw$ with $B\in\mathbb{R}^{m\times d}$ and let $L=\frac12\|z\|^2$ where $z=f(w)$. Compute $\nabla_wL$ using the lecture's formula.

<!-- Source: chain rule through vector intermediate and squared norm gradient -->

<details>
<summary>Answer</summary>

For $z=Bw$, the Jacobian is $J_f(w)=B$. Since $L=\frac12z^\top z$, we have $\nabla_zL=z$. The chain rule gives
$$\nabla_wL=J_f(w)^\top\nabla_zL=B^\top z=B^\top Bw.$$
The result lies in $\mathbb{R}^d$, matching $w$.

</details>

### Question 16

For $T\in\mathbb{R}^{2\times2\times2}$, define
$$L(T)=\sum_{i=1}^2\sum_{j=1}^2\sum_{k=1}^2 T_{ijk}^2.$$
Compute the entry $(\nabla_TL)_{ijk}$ and state the shape of $\nabla_TL$.

<!-- Source: tensor-input scalar-output same-shape gradient rule -->

<details>
<summary>Answer</summary>

Each entry differentiates independently:
$$\frac{\partial L}{\partial T_{ijk}}=2T_{ijk}.$$
Therefore
$$(\nabla_TL)_{ijk}=2T_{ijk},\qquad \nabla_TL=2T.$$
The gradient has the same shape as $T$, namely $2\times2\times2$.

</details>

### Question 17

Which of the following expressions are well-shaped under the lecture's conventions? (i) $J_fdx$ for $f:\mathbb{R}^n\to\mathbb{R}^m$, (ii) $J_f^\top dx$, (iii) $J_f^\top v$ for $v\in\mathbb{R}^m$, (iv) $v^\top J_f$ for $v\in\mathbb{R}^m$.

<details>
<summary>Answer</summary>

Here $J_f\in\mathbb{R}^{m\times n}$ and $dx\in\mathbb{R}^n$.

(i) $J_fdx$ is valid and has shape $m$.

(ii) $J_f^\top dx$ is invalid in general because $J_f^\top\in\mathbb{R}^{n\times m}$ expects an $m$-vector, not an $n$-vector.

(iii) $J_f^\top v$ is valid and has shape $n$.

(iv) $v^\top J_f$ is valid as a row vector of shape $1\times n$.

</details>
