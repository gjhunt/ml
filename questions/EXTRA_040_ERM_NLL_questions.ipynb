### Question 1

Suppose
$$
p_\theta(x,y)=p_\theta(y\mid x)p(x).
$$
Argue that: 
$$
\arg\max_\theta \prod_{n=1}^N p_\theta(x_n,y_n)
=\arg\max_\theta \prod_{n=1}^N p_\theta(y_n\mid x_n).
$$
Explain why the argument fails if the marginal distribution $p(x)$ is instead depends on $\theta$: $p_\theta(x)$.

<details>
<summary>Answer</summary>

We have
$$
\prod_{n=1}^N p_\theta(x_n,y_n)
=\prod_{n=1}^N p_\theta(y_n\mid x_n)p(x_n).
$$
Since $\prod_n p(x_n)$ does not depend on $\theta$,
$$
\arg\max_\theta \prod_{n=1}^N p_\theta(x_n,y_n)
=\arg\max_\theta \prod_{n=1}^N p_\theta(y_n\mid x_n).
$$
If the marginal were $p_\theta(x)$, then $\prod_n p_\theta(x_n)$ could affect which $\theta$ maximizes the likelihood, so it could not generally be discarded.

</details>

### Question 2

A regression model assumes Laplace conditional noise:
$$
y\mid x\sim \operatorname{Laplace}(w^\top x,b),
$$
with fixed scale $b>0$ and density
$$
p_w(y\mid x)=\frac{1}{2b}\exp\left(-\frac{|y-w^\top x|}{b}\right).
$$
Derive the NLL and identify the corresponding ERM loss for estimating $w$.

<details>
<summary>Answer</summary>

The negative log-likelihood is
$$
-\log p_w(y\mid x)=\log(2b)+\frac{|y-w^\top x|}{b}.
$$
For fixed $b$, the term $\log(2b)$ and the positive scale factor $1/b$ do not affect the optimizer over $w$. Therefore the corresponding ERM loss is absolute error,
$$
\ell(y,w^\top x)=|y-w^\top x|.
$$

</details>

### Question 3

Suppose a count model assumes

$$
y\mid x\sim \operatorname{Poisson}(\lambda_\theta(x)),
\qquad
\lambda_\theta(x)=\exp(s_\theta(x)),
$$

where $y\in\{0,1,2,\ldots\}$. Derive the negative log-likelihood for one observation. Then identify which term can be ignored when optimizing over $\theta$.

<details>
<summary>Answer</summary>

The Poisson probability is

$$
p_\theta(y\mid x)=\frac{e^{-\lambda_\theta(x)}\lambda_\theta(x)^y}{y!}.
$$

Taking the negative log gives

$$
-\log p_\theta(y\mid x)=
\lambda_\theta(x)-y\log\lambda_\theta(x)+\log(y!).
$$

Since $\lambda_\theta(x)=e^{s_\theta(x)}$, we have $\log\lambda_\theta(x)=s_\theta(x)$. Therefore,

$$
-\log p_\theta(y\mid x)=
e^{s_\theta(x)}-ys_\theta(x)+\log(y!).
$$

The term $\log(y!)$ depends only on the observed count $y$, not on $\theta$. So for optimizing $\theta$, the relevant loss is

$$
e^{s_\theta(x)}-ys_\theta(x).
$$

</details>

### Question 4

For multiclass classification with labels $y\in\{1,\ldots,K\}$, suppose the model outputs scores $s_1(x),\ldots,s_K(x)$ and probabilities
$$
p_\theta(Y=k\mid x)=\frac{e^{s_k(x)}}{\sum_{j=1}^K e^{s_j(x)}}.
$$
Derive the NLL for an observation with class label $y$. Show that this is equivalent to the multiclass cross-entropy loss.

<details>
<summary>Answer</summary>

The NLL is
$$
-\log p_\theta(Y=y\mid x)
=-\log\frac{e^{s_y(x)}}{\sum_{j=1}^K e^{s_j(x)}}.
$$
Therefore
$$
-\log p_\theta(Y=y\mid x)
=\log\left(\sum_{j=1}^K e^{s_j(x)}\right)-s_y(x).
$$
This is the multiclass cross-entropy loss for a single observation.

</details>

### Question 5

The hinge loss for binary labels $y\in\{-1,1\}$ is

$$
\ell(y,s)=\max\{0,1-ys\}.
$$

Suppose we tried to interpret hinge loss as an exact negative log-likelihood, so that

$$
p(y\mid s)=e^{-\ell(y,s)}.
$$

For this to be a valid binary probability model, the two probabilities $p(1\mid s)$ and $p(-1\mid s)$ would need to add up to $1$ for every finite score $s$.

Check what happens when $s\geq 1$. Compute $e^{-\ell(1,s)}$ and $e^{-\ell(-1,s)}$, then add them together. Does this give a valid probability distribution?

<details>
<summary>Answer</summary>

When $y=1$ and $s\geq 1$,

$$
\ell(1,s)=\max\{0,1-s\}=0.
$$

So

$$
e^{-\ell(1,s)}=e^0=1.
$$

When $y=-1$ and $s\geq 1$,

$$
\ell(-1,s)=\max\{0,1+s\}=1+s.
$$

So

$$
e^{-\ell(-1,s)}=e^{-(1+s)}.
$$

Adding the two values gives

$$
e^{-\ell(1,s)}+e^{-\ell(-1,s)}=1+e^{-(1+s)}.
$$

Since $e^{-(1+s)}>0$, this sum is greater than $1$. Therefore, these exponentiated negative hinge losses are not normalized probabilities.

So hinge loss is a useful classification loss, but it is not exactly the NLL of a normalized binary conditional model in this direct form.

</details>

### Question 6

For the discrete distribution
$$
p=(1/2,1/4,1/4),
$$
compute the entropy $H(p)$, leaving logarithms unevaluated.

<details>
<summary>Answer</summary>

Entropy is
$$
H(p)=-\sum_i p_i\log p_i.
$$
Thus
$$
H(p)=-\frac12\log\frac12-\frac14\log\frac14-\frac14\log\frac14.
$$
Equivalently,
$$
H(p)=\frac12\log 2+\frac12\log 4.
$$

</details>

### Question 7

For discrete distributions
$$
p=(1/2,1/4,1/4),
\qquad
q=(1/4,1/4,1/2),
$$
compute the cross-entropy $H(p,q)$ and $\operatorname{KL}(p\|q)$, leaving logarithms unevaluated.

<details>
<summary>Answer</summary>

The cross-entropy is
$$
H(p,q)=-\frac12\log\frac14-\frac14\log\frac14-\frac14\log\frac12.
$$
The KL divergence is
$$
\operatorname{KL}(p\|q)
=\frac12\log\frac{1/2}{1/4}
+\frac14\log\frac{1/4}{1/4}
+\frac14\log\frac{1/4}{1/2}.
$$
So
$$
\operatorname{KL}(p\|q)=\frac12\log 2+\frac14\log 1+\frac14\log\frac12.
$$

</details>

### Question 8

For Bernoulli distributions with true parameter $p=0.7$ and model parameter $q=0.4$, compute the cross-entropy $H(p,q)$, the entropy $H(p)$, and $\operatorname{KL}(p\|q)$. Leave logarithms unevaluated.

<details>
<summary>Answer</summary>

For Bernoulli parameters,
$$
H(p,q)=-p\log q-(1-p)\log(1-q).
$$
Thus
$$
H(0.7,0.4)=-0.7\log(0.4)-0.3\log(0.6),
$$
$$
H(0.7)=-0.7\log(0.7)-0.3\log(0.3),
$$
and
$$
\operatorname{KL}(0.7\|0.4)=0.7\log\frac{0.7}{0.4}+0.3\log\frac{0.3}{0.6}.
$$

</details>

### Question 9

For a fixed Bernoulli true parameter $p\in(0,1)$, minimize
$$
H(p,q)=-p\log q-(1-p)\log(1-q)
$$
over $q\in(0,1)$. 

<details>
<summary>Answer</summary>

Differentiate:
$$
\frac{d}{dq}H(p,q)=-\frac{p}{q}+\frac{1-p}{1-q}.
$$
The zero condition gives $(1-p)q=p(1-q)$, hence $q=p$. 

</details>

### Question 10

For Bernoulli distributions, compute $\operatorname{KL}(0.2\|0.6)$ and $\operatorname{KL}(0.6\|0.2)$. Leave the answer in logarithmic form and state why the two values differ.

<details>
<summary>Answer</summary>

Using
$$
\operatorname{KL}(p\|q)=p\log\frac{p}{q}+(1-p)\log\frac{1-p}{1-q},
$$
we get
$$
\operatorname{KL}(0.2\|0.6)=0.2\log\frac13+0.8\log 2,
$$
and
$$
\operatorname{KL}(0.6\|0.2)=0.6\log 3+0.4\log\frac12.
$$
The weights differ because the expectation is taken under the first distribution in $\operatorname{KL}(p\|q)$.

</details>

### Question 11

For normal distributions, use
$$
\operatorname{KL}(\mathcal N(\mu_p,\sigma_p^2)\|\mathcal N(\mu_q,\sigma_q^2))
=\log\frac{\sigma_q}{\sigma_p}+\frac{\sigma_p^2+(\mu_p-\mu_q)^2}{2\sigma_q^2}-\frac12.
$$
Compute both KL directions for $p=\mathcal N(0,1)$ and $q=\mathcal N(0,4)$.

<details>
<summary>Answer</summary>

For $\operatorname{KL}(p\|q)$, $\sigma_p=1$, $\sigma_q=2$, and the means agree, so
$$
\operatorname{KL}(p\|q)=\log 2+\frac{1}{8}-\frac12=\log 2-\frac38.
$$
For $\operatorname{KL}(q\|p)$, $\sigma_p=2$, $\sigma_q=1$, so
$$
\operatorname{KL}(q\|p)=\log\frac12+2-\frac12=\frac32-\log 2.
$$
The two directions are unequal.

</details>

### Question 12

A false interpretation says: "For Gaussian distributions, KL divergence is zero whenever the means are equal." Refute it with a concrete pair of Gaussian distributions and compute the KL value.

<details>
<summary>Answer</summary>

Take $p=\mathcal N(0,1)$ and $q=\mathcal N(0,4)$. The means are equal, but
$$
\operatorname{KL}(p\|q)=\log 2+\frac18-\frac12=\log 2-\frac38>0.
$$
KL is zero only when the full distributions agree; for univariate Gaussians, both mean and variance must match.

</details>

### Question 13

At fixed $x$, explain the decomposition
$$
H(p(\cdot\mid x),p_\theta(\cdot\mid x))
=H(p(\cdot\mid x))
+
\operatorname{KL}(p(\cdot\mid x)\|p_\theta(\cdot\mid x)).
$$
Which term is irreducible, and which term measures model mismatch?

<details>
<summary>Answer</summary>

The entropy term $H(p(\cdot\mid x))$ is irreducible because it depends only on the true conditional distribution of $Y$ given $X=x$. It is the best possible average log-loss at that input. The KL term measures model mismatch: it is the excess average log-loss from using $p_\theta(\cdot\mid x)$ instead of the true conditional distribution.

</details>
