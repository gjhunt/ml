### Question 1

For scores $z=(z_1,\dots,z_K)\in\mathbb{R}^K$, the lecture defines
$$p_k=\frac{e^{z_k}}{\sum_{j=1}^K e^{z_j}}.$$
Verify directly that $p_k>0$ for every $k$ and $\sum_{k=1}^K p_k=1$. Then value that $p$ takes on when all scores are equal to the same constant $c$.

<details>
<summary>Answer</summary>

Since $e^{z_k}>0$ and the denominator $\sum_j e^{z_j}$ is positive, each $p_k>0$. Also
$$\sum_{k=1}^K p_k=\frac{\sum_k e^{z_k}}{\sum_j e^{z_j}}=1.$$
If $z_1=\cdots=z_K=c$, then
$$p_k=\frac{e^c}{K e^c}=\frac1K,$$
so $p=(1/K,\dots,1/K)$.

</details>


### Question 2

The prediction rule in the lecture uses either largest score or largest softmax probability. Show that
$$\arg\max_k z_k=\arg\max_k \frac{e^{z_k}}{\sum_{j=1}^K e^{z_j}},$$
where $z_k = w_k^\top x$. 

<details>
<summary>Answer</summary>

The denominator $\sum_{j=1}^K e^{z_j}$ is the same positive number for every class, and $u\mapsto e^u$ is strictly increasing. Hence
$$z_a>z_b \iff e^{z_a}>e^{z_b} \iff p_a>p_b.$$
Therefore the maximizing class is identical for scores and softmax probabilities, since they are increasing functions of one another. 

</details>


### Question 3

A single observation has true class $c$ and one-hot vector $t$. Starting from
$$\ell(t,p)=-\sum_{k=1}^K t_k\log p_k,$$
compute $\ell(t,p)$ explicitly in terms of only $p_c$. Then explain why a predicted probability $p_c=1$ gives the smallest possible loss.

<details>
<summary>Answer</summary>

Because $t_c=1$ and $t_k=0$ for $k \neq c$, 
$$\ell(t,p)=-\log p_c.$$
Since $0<p_c\le 1$, we have $\log p_c\le 0$, so $-\log p_c\ge 0$. The minimum possible value is $0$, attained only when $p_c=1$.

</details>


### Question 4

Fill in the algebra that converts:
$$-\log\left(\frac{e^{w_c^\top x}}{\sum_{j=1}^K e^{w_j^\top x}}\right)$$
into
$$\log\left(\sum_{j=1}^K e^{w_j^\top x}\right)-w_c^\top x.$$

<details>
<summary>Answer</summary>

Using $-\log(a/b)=-\log a+\log b$,
$$-\log\left(\frac{e^{w_c^\top x}}{\sum_j e^{w_j^\top x}}\right)
=-\log(e^{w_c^\top x})+\log\left(\sum_j e^{w_j^\top x}\right).$$
Since $\log(e^{w_c^\top x})=w_c^\top x$, the expression is
$$\log\left(\sum_j e^{w_j^\top x}\right)-w_c^\top x.$$

</details>


### Question 5

Check the matrix dimensions in the ERM formulation. With $X\in\mathbb{R}^{N\times D}$, $W\in\mathbb{R}^{D\times K}$, $T\in\mathbb{R}^{N\times K}$, and $P=\operatorname{softmax}(XW)$ rowwise, list the dimensions of $XW$, $P-T$, and $X^\top(P-T)$. Why is the final gradient shape correct?

<details>
<summary>Answer</summary>

We have $XW\in\mathbb{R}^{N\times K}$, so $P\in\mathbb{R}^{N\times K}$ and $P-T\in\mathbb{R}^{N\times K}$. Since $X^\top\in\mathbb{R}^{D\times N}$,
$$X^\top(P-T)\in\mathbb{R}^{D\times K}.$$
This matches the shape of $W$, as required for $\nabla_W\hat R(W)$.

</details>


### Question 6

For one data point with scores $z=W^\top x\in\mathbb{R}^K$ and one-hot label $t$, define
$$L(z)=-\sum_{a=1}^K t_a\log p_a,\qquad p_a=\frac{e^{z_a}}{\sum_j e^{z_j}}.$$
Using the softmax derivative identity $\partial p_a/\partial z_b=p_a(\mathbf 1\{a=b\}-p_b)$, compute $\partial L/\partial z_b$ and simplify it to a compact expression.

<details>
<summary>Answer</summary>

By the chain rule,
$$\frac{\partial L}{\partial z_b}=-\sum_a t_a\frac1{p_a}\frac{\partial p_a}{\partial z_b}.$$
Using $\partial p_a/\partial z_b=p_a(\mathbf 1\{a=b\}-p_b)$ gives
$$\frac{\partial L}{\partial z_b}=-\sum_a t_a(\mathbf 1\{a=b\}-p_b)=-t_b+p_b\sum_a t_a.$$
Since $t$ is one-hot, $\sum_a t_a=1$, so
$$\frac{\partial L}{\partial z_b}=p_b-t_b.$$

</details>


### Question 7

Consider the standard softmax derivative identity

$$
\frac{\partial P_{na}}{\partial Z_{nb}} = P_{na}(\mathbf{1}\{a=b\}-P_{nb}).
$$

Verify this identity separately for the cases $a=b$ and $a\neq b$ by differentiating the softmax quotient.

<details>
<summary>Answer</summary>

Fix a row $n$. To simplify notation, write

$$
S = \sum_j e^{Z_j}, \qquad P_a = \frac{e^{Z_a}}{S}.
$$

First suppose $a=b$. Then both the numerator and denominator depend on $Z_a$. By the quotient rule,

$$
\frac{\partial P_a}{\partial Z_a} = \frac{e^{Z_a}S - e^{Z_a}e^{Z_a}}{S^2}.
$$

Rewriting in terms of probabilities gives

$$
\frac{\partial P_a}{\partial Z_a} = \frac{e^{Z_a}}{S}\left(1-\frac{e^{Z_a}}{S}\right) = P_a(1-P_a).
$$

This matches the identity because when $a=b$,

$$
P_a(\mathbf{1}\{a=b\}-P_b) = P_a(1-P_a).
$$

Now suppose $a\neq b$. Then the numerator $e^{Z_a}$ does not depend on $Z_b$, but the denominator does. Again using the quotient rule,

$$
\frac{\partial P_a}{\partial Z_b} = \frac{0\cdot S - e^{Z_a}e^{Z_b}}{S^2}.
$$

Rewriting in terms of probabilities gives

$$
\frac{\partial P_a}{\partial Z_b} = -\frac{e^{Z_a}}{S}\frac{e^{Z_b}}{S} = -P_aP_b.
$$

This matches the identity because when $a\neq b$,

$$
P_a(\mathbf{1}\{a=b\}-P_b) = P_a(0-P_b) = -P_aP_b.
$$

Therefore, in both cases,

$$
\frac{\partial P_{na}}{\partial Z_{nb}} = P_{na}(\mathbf{1}\{a=b\}-P_{nb}).
$$

</details>




### Question 8

Show precisely why the parameterization is redundant. For any $a\in\mathbb{R}^D$, define $w_k'=w_k+a$ for every class $k$. Prove that $p_k(x;W')=p_k(x;W)$ for every $x$ and $k$.

<details>
<summary>Answer</summary>

The new score is
$$w_k'^\top x=(w_k+a)^\top x=w_k^\top x+a^\top x.$$
Thus every score has the same added constant $a^\top x$. The softmax probability becomes
$$p_k(x;W')=\frac{e^{w_k^\top x+a^\top x}}{\sum_j e^{w_j^\top x+a^\top x}}
=\frac{e^{a^\top x}e^{w_k^\top x}}{e^{a^\top x}\sum_j e^{w_j^\top x}}=p_k(x;W).$$

</details>


### Question 9

For $K=2$, multinomial logistic regression has two class weight vectors, $w_1$ and $w_2$. However, the class probability depends only on their difference. Your goal is to show that the two-class softmax model can be rewritten as a binary logistic regression model with one contrast weight vector.

Starting from

$$
p_1(x)=\frac{e^{w_1^\top x}}{e^{w_1^\top x}+e^{w_2^\top x}},
$$

rewrite $p_1(x)$ in the form

$$
p_1(x)=\sigma(v^\top x),
$$

where $\sigma(u)=1/(1+e^{-u})$. Identify the contrast vector $v$ in terms of $w_1$ and $w_2$.

<details>
<summary>Answer</summary>

Divide the numerator and denominator by $e^{w_1^\top x}$:

$$
p_1(x)=\frac{1}{1+e^{w_2^\top x-w_1^\top x}}.
$$

Since $w_2^\top x-w_1^\top x=-(w_1-w_2)^\top x$, we can write

$$
p_1(x)=\frac{1}{1+e^{-(w_1-w_2)^\top x}}.
$$

Using $\sigma(u)=1/(1+e^{-u})$, this becomes

$$
p_1(x)=\sigma((w_1-w_2)^\top x).
$$

Therefore, the relevant binary logistic regression weight vector is

$$
v=w_1-w_2.
$$

Only the contrast vector $w_1-w_2$ matters for the probability $p_1(x)$. The individual vectors $w_1$ and $w_2$ are not separately identifiable from the model probabilities.

</details>

### Question 13

The lecture calls models with scores $s_k(x)=h(w_k^\top x)$ linear classifiers when $h$ is strictly increasing. Prove that the predicted class is unchanged by replacing $s_k(x)$ with $w_k^\top x$.

<details>
<summary>Answer</summary>

For any two classes $j,k$, strict monotonicity gives
$$h(w_j^\top x)>h(w_k^\top x) \iff w_j^\top x>w_k^\top x,$$
and equality is also preserved because $h$ is a function. Therefore the ordering of all class scores is the same before and after applying $h$, and hence the predicted class is unchanged.

</details>


### Question 14

The lecture says that the logistic regression objective is not quadratic. One way to check this is to look at the second derivative. If a function is quadratic, its second derivative should be constant.

Consider the simplest possible case: one observation, $K=2$, true class $1$, and scores $(\theta,0)$. Then

$$
p_1=\frac{e^\theta}{e^\theta+1},
$$

so the loss is

$$
L(\theta)=-\log p_1=\log(1+e^{-\theta}).
$$

The first derivative is

$$
L'(\theta)=-\frac{1}{1+e^\theta}.
$$

Compute $L''(\theta)$. Is it constant? What does this tell you about whether $L(\theta)$ is quadratic?

<details>
<summary>Answer</summary>

Differentiating

$$
L'(\theta)=-\frac{1}{1+e^\theta}
$$

gives

$$
L''(\theta)=\frac{e^\theta}{(1+e^\theta)^2}.
$$

This depends on $\theta$, so it is not constant. Therefore $L(\theta)$ is not quadratic.

</details>

### Question 15

The gradient $\nabla_W \hat R(W)$ has the same shape as the weight matrix $W$. This means each entry of the gradient tells us how the loss changes with respect to one particular weight.

Suppose $D=2$ and $K=3$, so $W$ has two rows and three columns. List the six scalar partial derivatives contained in $\nabla_W \hat R(W)$, and place each one in the correct position in the $2\times 3$ gradient matrix.

<details>
<summary>Answer</summary>

Write

$$
W=\begin{pmatrix}
W_{11}&W_{12}&W_{13}\\
W_{21}&W_{22}&W_{23}
\end{pmatrix}.
$$

Since the gradient has the same shape as $W$,

$$
\nabla_W\hat R(W)=\begin{pmatrix}
\frac{\partial\hat R}{\partial W_{11}} & \frac{\partial\hat R}{\partial W_{12}} & \frac{\partial\hat R}{\partial W_{13}}\\
\frac{\partial\hat R}{\partial W_{21}} & \frac{\partial\hat R}{\partial W_{22}} & \frac{\partial\hat R}{\partial W_{23}}
\end{pmatrix}.
$$

Column $k$ contains the derivatives with respect to the weight vector for class $k$. Row $d$ contains the derivatives with respect to the coefficient for feature $d$ across all classes.

</details>
