### Question 1

In softmax regression, answer the following about linear class scores: what is the relevant score/probability/loss formula, and what does it imply for prediction or identifiability?

<!-- Source: Multiclass Logistic Regression lecture; section related to: In softmax regression, answer the following about linear class scores: what is the relevan -->

<details>
<summary>Answer</summary>

With class scores $s_k=w_k^Tx$, softmax gives $p_k=e^{s_k}/\sum_j e^{s_j}$. The issue of linear class scores follows from the cross-entropy $-\sum_k y_k\log p_k$, whose gradients compare predicted probabilities with one-hot labels.

</details>
