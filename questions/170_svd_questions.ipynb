### Question 1

For diagonal matrices, state the algebraic relationship involving eigenvectors, singular vectors, or $A=U\Sigma V^T$, and interpret it geometrically.

<details>
<summary>Answer</summary>

For diagonal matrices, SVD writes $A=U\Sigma V^T$ with orthonormal singular vectors and nonnegative singular values. It decomposes a linear map into rotation/reflection, coordinate scaling, and another rotation/reflection, and truncated SVD gives optimal low-rank approximations.

</details>
