# Ridge Regression - Review Questions {#sec-ridge-questions}



### Question 1

The lecture describes ridge regression as an example of regularization. How does this regularization work for ridge, what quantity controls the strength of regularization, and what happens at the two extremes $\lambda=0$ and $\lambda\to\infty$?

<details>
<summary>Answer</summary>

Ridge regularizes the coefficient vector $w$ by adding the penalty
$$
\lambda\|w\|_2^2
$$
to the squared-error training objective. The regularization strength is controlled by $\lambda\ge0$.

At $\lambda=0$, the penalty disappears and the problem reduces to OLS. As $\lambda\to\infty$, any nonzero penalized coefficient becomes extremely costly, so the fitted slopes are driven toward $0$. An unpenalized intercept need not approach zero.

</details>


### Question 2

For $w=(3,-4,12)$, compute $\|w\|_2^2$. If $\lambda=0.5$, what is the ridge penalty value $\lambda\|w\|_2^2$?

<details>
<summary>Answer</summary>

The squared norm is
$$
\|w\|_2^2=3^2+(-4)^2+12^2=9+16+144=169.
$$
Thus the penalty value is
$$
\lambda\|w\|_2^2=0.5(169)=84.5.
$$

</details>


### Question 3

Why does minimizing training error alone fail to control model complexity? How does ridge control model complexity?

<details>
<summary>Answer</summary>

Training error can be reduced by increasing flexibility, but greater flexibility can fit noise and sample-specific patterns. For example, in polynomial regression, increasing degree can keep lowering training error while eventually increasing test error. In nearest-neighbor regression, very small $m$ gives highly local fits that can have high variance.

Ridge addresses the same issue for linear regression by keeping the model class linear but penalizing large coefficients. The penalty discourages overly flexible or unstable coefficient choices even if they improve training error.

</details>


### Question 4

Starting from the unnormalized ridge objective used in the solving section,
$$
J_\lambda(w)=\|y-Xw\|_2^2+\lambda\|w\|_2^2,
$$
expand $J_\lambda(w)$ into quadratic form involving $y^\top y$, $w^\top X^\top y$, $w^\top X^\top Xw$, and $w^\top w$.

<details>
<summary>Answer</summary>

First,
$$
\|y-Xw\|_2^2=(y-Xw)^\top(y-Xw).
$$
Expanding gives
$$
y^\top y-y^\top Xw-w^\top X^\top y+w^\top X^\top Xw.
$$
Since $y^\top Xw$ is a scalar equal to $w^\top X^\top y$, this becomes
$$
y^\top y-2w^\top X^\top y+w^\top X^\top Xw.
$$
Adding the penalty gives
$$
J_\lambda(w)=y^\top y-2w^\top X^\top y+w^\top X^\top Xw+
\lambda w^\top w.
$$

</details>


### Question 5

Differentiate the expanded objective from Question 4 and derive the ridge normal equations.

<details>
<summary>Answer</summary>

Using the standard rules,
$$
\nabla_w(y^\top y)=0,
\qquad
\nabla_w(w^\top X^\top y)=X^\top y,
$$
$$
\nabla_w(w^\top X^\top Xw)=2X^\top Xw,
\qquad
\nabla_w(w^\top w)=2w.
$$
Therefore
$$
\nabla_wJ_\lambda(w)=2X^\top Xw-2X^\top y+2\lambda w.
$$
Setting this equal to zero and dividing by $2$ gives
$$
X^\top Xw-X^\top y+
\lambda w=0.
$$
Thus
$$
(X^\top X+
\lambda I_D)w=X^\top y.
$$

</details>


### Question 6

In the scalar case $D=1$, let $X^\top X=9$, $X^\top y=18$, and $\lambda=3$ under the unnormalized convention. Compute the OLS coefficient and the ridge coefficient.

<details>
<summary>Answer</summary>

The OLS coefficient is
$$
\widehat w_{\mathrm{OLS}}=(X^\top X)^{-1}X^\top y=\frac{18}{9}=2.
$$
The ridge coefficient is
$$
\widehat w_\lambda=(X^\top X+
\lambda)^{-1}X^\top y=\frac{18}{9+3}=\frac{18}{12}=1.5.
$$
Ridge shrinks the coefficient toward zero.

</details>


### Question 7

Why does ridge regression stabilize solutions when features are highly correlated? Give a precise explanation using the lecture's example where $X_{:2}\approx X_{:1}$.

<details>
<summary>Answer</summary>

If $X_{:2}\approx X_{:1}$, then
$$
w_1X_{:1}+w_2X_{:2}\approx (w_1+w_2)X_{:1}.
$$
Many pairs $(w_1,w_2)$ with similar sums can produce nearly the same predictions. OLS may therefore choose large cancelling coefficients, such as $(100,-99)$, with little change in training error.

Ridge penalizes
$$
w_1^2+w_2^2,
$$
so large cancelling coefficients become costly. Among coefficient vectors with similar fit, ridge prefers smaller Euclidean norm, producing a more stable solution.

</details>


### Question 8

The lecture states that ridge replaces $X^\top X$ with $X^\top X+\lambda I_D$, which is invertible when $\lambda>0$. Suppose $X^\top Xv=\sigma^2v$ for a nonzero vector $v$. Show that $v$ is also an eigenvector of $X^\top X+\lambda I_D$, identify its new eigenvalue, and explain why this supports the invertibility claim.

<details>
<summary>Answer</summary>

Compute
$$
(X^\top X+\lambda I_D)v=X^\top Xv+\lambda I_Dv.
$$
Using $X^\top Xv=\sigma^2v$ and $I_Dv=v$,
$$
(X^\top X+\lambda I_D)v=\sigma^2v+\lambda v=(\sigma^2+\lambda)v.
$$
Thus $v$ remains an eigenvector, and the new eigenvalue is $\sigma^2+\lambda$. Because $X^\top X$ has nonnegative eigenvalues, $\sigma^2+\lambda>0$ whenever $\lambda>0$. The ridge matrix therefore has no zero eigenvalues and is invertible.

</details>


### Question 9

For orthonormal predictors satisfying $X^\top X=I_D$, derive the ridge estimator and express it in terms of the OLS estimator.

<details>
<summary>Answer</summary>

The ridge estimator is
$$
\widehat w_\lambda=(X^\top X+
\lambda I_D)^{-1}X^\top y.
$$
Using $X^\top X=I_D$,
$$
\widehat w_\lambda=(I_D+
\lambda I_D)^{-1}X^\top y=((1+
\lambda)I_D)^{-1}X^\top y.
$$
Therefore
$$
\widehat w_\lambda=\frac{1}{1+
\lambda}X^\top y.
$$
Since $\widehat w_{\mathrm{OLS}}=X^\top y$ in this case,
$$
\widehat w_\lambda=\frac{1}{1+
\lambda}\widehat w_{\mathrm{OLS}}.
$$

</details>


### Question 10

In an orthonormal-design problem, suppose $\widehat w_{\mathrm{OLS}}=(4,-2,1)$ and $\lambda=3$. Compute $\widehat w_\lambda$.

<details>
<summary>Answer</summary>

The shrinkage factor is
$$
\frac{1}{1+
\lambda}=\frac14.
$$
Therefore
$$
\widehat w_\lambda=\frac14(4,-2,1)=\left(1,-\frac12,\frac14\right).
$$

</details>


### Question 11

The lecture recommends standardizing predictors before ridge regression. Why is this necessary for the penalty $\lambda\sum_jw_j^2$ to treat predictors comparably?

<details>
<summary>Answer</summary>

Coefficient size depends on feature units. If one predictor is measured in thousands and another in units near one, the same predictive effect can require very different coefficient magnitudes. Since ridge penalizes coefficient magnitudes directly,
$$
\lambda\sum_jw_j^2,
$$
the penalty depends on the measurement scale of each feature.

Standardizing features to comparable scales, usually mean zero and variance one, makes coefficient sizes more comparable. Then the penalty is less distorted by arbitrary units.

</details>


### Question 12

State the penalized and constrained forms of ridge regression. 

<details>
<summary>Answer</summary>

The penalized form is
$$
\min_w \|y-Xw\|_2^2+\lambda\|w\|_2^2.
$$
The constrained form is
$$
\min_w \|y-Xw\|_2^2\quad\text{s.t.}\quad \|w\|_2^2\le t.
$$

</details>


### Question 13

For a diagonal design with
$$
X^\top X=\begin{pmatrix}4&0\\0&1\end{pmatrix},
\qquad
X^\top y=\begin{pmatrix}8\\3\end{pmatrix},
$$
compute the ridge solution for $\lambda=1$.

<details>
<summary>Answer</summary>

The ridge system is
$$
(X^\top X+
\lambda I)w=X^\top y.
$$
With $\lambda=1$,
$$
X^\top X+
I=
\begin{pmatrix}5&0\\0&2\end{pmatrix}.
$$
Therefore
$$
w=\begin{pmatrix}5^{-1}&0\\0&2^{-1}\end{pmatrix}
\begin{pmatrix}8\\3\end{pmatrix}
=\begin{pmatrix}8/5\\3/2\end{pmatrix}.
$$
So
$$
\hat w_\lambda=(1.6,1.5)^\top.
$$

</details>


### Question 14

How does increasing $\lambda$ affect the bias-variance tradeoff in ridge regression? Give the expected direction of training error and test error for very small versus moderate $\lambda$.

<details>
<summary>Answer</summary>

Increasing $\lambda$ shrinks coefficients and reduces model flexibility. This usually increases bias but lowers variance.

For very small $\lambda$, ridge is close to OLS: training error is low, but variance may be high and test error can be poor in noisy, collinear, or high-dimensional settings. For moderate $\lambda$, training error may increase, but test error can decrease because the reduction in variance and instability outweighs the increase in bias. For very large $\lambda$, underfitting can make test error increase again.

</details>


### Question 15

The riboflavin example has $D>N$. Why can OLS have multiple coefficient solutions in this setting, and how does ridge with $\lambda>0$ resolve that ambiguity?

<details>
<summary>Answer</summary>

When $D>N$, the $D$ columns of $X$ cannot all be linearly independent. Consequently, there is a nonzero vector $z$ such that $Xz=0$. If $w$ is an OLS solution, then $w+z$ gives the same fitted values because
$$
X(w+z)=Xw+Xz=Xw.
$$
Thus the training error may not determine a unique coefficient vector.

Ridge adds the penalty $\lambda\|w\|_2^2$, favoring smaller coefficient vectors among solutions with similar fit. For $\lambda>0$, its normal equations use
$$
X^\top X+\lambda I_D,
$$
which is invertible, so ridge has a unique coefficient solution.


</details>
