# Elastic Net and Penalized Logistic Regression - Review Questions {#sec-en-questions}


### Question 1

In the elastic net objective
$$
\frac{1}{2N}\|y-Xw\|_2^2+\lambda\left(\rho\|w\|_1+\frac{1-\rho}{2}\|w\|_2^2\right),
$$
state the role of $\lambda$ and the role of $\rho$. What happens at $\rho=0$ and $\rho=1$?

<details>
<summary>Answer</summary>

The parameter $\lambda\ge0$ controls the overall strength of regularization. Larger $\lambda$ places more weight on the penalty relative to the squared-error fit term.

The parameter $\rho\in[0,1]$ controls the mix between the $\ell_1$ and $\ell_2$ parts. At $\rho=1$, the penalty is purely $\ell_1$, giving LASSO. At $\rho=0$, the penalty is purely squared $\ell_2$, giving ridge up to constant scaling of $\lambda$.

</details>


### Question 2

For $w=(2,-1,0)$, $\lambda=3$, and $\rho=0.4$, compute the elastic net penalty term
$$
\lambda\left(\rho\|w\|_1+\frac{1-\rho}{2}\|w\|_2^2\right).
$$

<details>
<summary>Answer</summary>

First,
$$
\|w\|_1=|2|+|-1|+0=3,
$$
and
$$
\|w\|_2^2=2^2+(-1)^2+0^2=5.
$$
Thus the unscaled mixed penalty is
$$
0.4(3)+\frac{1-0.4}{2}(5)=1.2+0.3(5)=1.2+1.5=2.7.
$$
Multiplying by $\lambda=3$ gives
$$
3(2.7)=8.1.
$$

</details>


### Question 3

Why does the lecture say elastic net is especially useful when predictors are strongly correlated? Compare the expected coefficient behavior of ridge, LASSO, and elastic net in a correlated signal group.

<details>
<summary>Answer</summary>

Ridge tends to spread weight across correlated predictors because its $\ell_2$ penalty stabilizes large coefficient swings but does not create exact zeros. LASSO encourages sparsity, so in a correlated group it may select one representative variable and set the others to zero, which can make selected variables unstable.

Elastic net combines these behaviors. Its $\ell_1$ part can remove irrelevant variables, while its $\ell_2$ part encourages stability and can keep several variables from a correlated signal group instead of forcing a single representative.

</details>


### Question 4

The lecture's coordinate update is
$$
w_j\leftarrow \frac{S(z_j,\lambda\rho)}{a_j+\lambda(1-\rho)}.
$$
Show explicitly that this reduces to the LASSO update when $\rho=1$ and to a ridge-like update when $\rho=0$.

<details>
<summary>Answer</summary>

If $\rho=1$, then the threshold is $\lambda\rho=\lambda$ and the denominator is
$$
a_j+\lambda(1-1)=a_j.
$$
Thus
$$
w_j\leftarrow \frac{S(z_j,\lambda)}{a_j},
$$
which is the LASSO coordinate update.

If $\rho=0$, then the threshold is $0$, so $S(z_j,0)=z_j$, and the denominator is
$$
a_j+\lambda.
$$
Thus
$$
w_j\leftarrow \frac{z_j}{a_j+\lambda},
$$
which is ridge-like shrinkage.

</details>


### Question 5

For intermediate $0<\rho<1$, why does elastic net both threshold and shrink? Use the coordinate update formula in your answer.

<details>
<summary>Answer</summary>

The numerator
$$
S(z_j,\lambda\rho)
$$
soft-thresholds $z_j$. If $|z_j|\le\lambda\rho$, the numerator is zero, so the coefficient is set exactly to zero. This is the LASSO-like thresholding effect.

When $|z_j|>\lambda\rho$, the numerator is nonzero but has reduced magnitude. Then the denominator
$$
a_j+\lambda(1-\rho)
$$
further shrinks the coefficient. This is the ridge-like stabilization effect.

</details>


### Question 6

Assume standardized predictors so $a_j=1$. Let $z_j=2.5$, $\lambda=1$, and $\rho=0.5$. Compute the elastic net coordinate update.

<details>
<summary>Answer</summary>

The threshold is
$$
\lambda\rho=1(0.5)=0.5.
$$
Since $z_j=2.5>0.5$,
$$
S(2.5,0.5)=2.5-0.5=2.0.
$$
The denominator is
$$
a_j+\lambda(1-\rho)=1+1(0.5)=1.5.
$$
Thus
$$
w_j\leftarrow \frac{2.0}{1.5}=\frac43.
$$

</details>


### Question 7

Suppose $z_j=0.3$, $a_j=1$, $\lambda=1$, and $\rho=0.5$. Compute the elastic net coordinate update.

<details>
<summary>Answer</summary>

The threshold is
$$
\lambda\rho=0.5.
$$
Since $|z_j|=0.3\le0.5$,
$$
S(z_j,0.5)=0.
$$
Therefore
$$
w_j\leftarrow \frac{0}{1+1(1-0.5)}=0.
$$
The coordinate signal is not strong enough to overcome the $\ell_1$ part of the penalty, so the coefficient is set to zero.

</details>


### Question 8

The lecture says different books and software use different constants in front of the squared-error and ridge terms. Explain why multiplying the whole penalty by a constant often only changes the scale of $\lambda$ rather than the statistical idea.

<details>
<summary>Answer</summary>

Suppose an objective contains
$$
\lambda c\,\Omega(w)
$$
for some positive constant $c$. If we define a new tuning parameter
$$
\lambda'=c\lambda,
$$
then the penalty becomes
$$
\lambda'\Omega(w).
$$
Thus changing constants can often be absorbed by changing the numerical value of $\lambda$. The path is parameterized differently, but the basic fit-complexity tradeoff is the same.

</details>


### Question 9

Why does tuning elastic net generally require a two-dimensional grid search, while ridge and LASSO each require only a one-dimensional penalty search in the lecture's setup?

<details>
<summary>Answer</summary>

Ridge and LASSO each have one main penalty-strength parameter, $\lambda$, after the penalty type is fixed. Elastic net has both the overall strength $\lambda$ and the mixing parameter $\rho$.

The value of $\lambda$ controls how much regularization is applied, while $\rho$ controls the balance between $\ell_1$ sparsity and $\ell_2$ stabilization. Both affect the fitted model, so cross-validation typically searches over pairs $(\lambda,\rho)$.

</details>


### Question 10

Why is the intercept $w_0$ usually not penalized in penalized logistic regression, according to the same logic used for ridge and LASSO?

<details>
<summary>Answer</summary>

The intercept controls the baseline log-odds, or baseline class probability, before feature effects are applied. It is not a measure of model complexity in the same way that feature coefficients are.

Penalizing $w_0$ would artificially shrink the baseline probability toward a default value, potentially introducing unnecessary bias. Therefore the penalty is usually applied only to $w$, not to $w_0$.

</details>


### Question 11

In `sklearn`, suppose two logistic models differ only in `C`: one has `C=0.01` and one has `C=100`. Which one is more strongly regularized? Which one would you expect to have more nonzero coefficients under an elastic net penalty with fixed `l1_ratio`?

<details>
<summary>Answer</summary>

`C` is inverse regularization strength, so smaller `C` means stronger regularization. Thus `C=0.01` is more strongly regularized than `C=100`.

With fixed `l1_ratio`, the model with `C=100` has weaker regularization, so more coefficients can enter the model. The model with `C=0.01` is expected to have fewer nonzero coefficients.

</details>


### Question 12

Map the `sklearn` parameter `l1_ratio` to the lecture's penalty mix. What do `l1_ratio=0`, `l1_ratio=1`, and `l1_ratio=0.5` mean?

<details>
<summary>Answer</summary>

`l1_ratio` plays the role of the elastic net mixing parameter $\rho$.

- `l1_ratio=0` means $\rho=0$, giving a purely ridge-like logistic penalty.
- `l1_ratio=1` means $\rho=1$, giving a purely LASSO logistic penalty.
- `l1_ratio=0.5` gives an elastic net penalty with both $\ell_1$ and $\ell_2$ components.

</details>


### Question 13

For an elastic net logistic path with fixed `l1_ratio=0.5`, suppose the nonzero counts at $C=0.01,0.1,1,10$ are $2,8,18,30$. Interpret the trend in terms of regularization strength.

<details>
<summary>Answer</summary>

As $C$ increases, regularization weakens because $C$ is inverse regularization strength. With weaker regularization, more coefficients are allowed to enter the model.

The nonzero counts increasing from $2$ to $30$ are therefore consistent with the lecture's statement that larger $C$ allows larger and more numerous coefficients.

</details>
