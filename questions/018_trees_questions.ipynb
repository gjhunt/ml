# Trees - Review Questions {#sec-tree-questions}


### Question 1

In the supervised setup of the lecture, training data are $(x_1,y_1),\ldots,(x_N,y_N)$ with $x_n\in\mathbb R^D$. State how a tree prediction rule differs structurally from the parametric examples $s_w(x)=w^Tx$ and logistic regression with a linear score.

<details>
<summary>Answer</summary>

The parametric examples use one global formula whose parameters apply throughout the input space, such as a single linear score $w^Tx$.

A tree instead recursively partitions the input space into regions $R_1,\ldots,R_M$ using splits of the form $x_j\le t$ versus $x_j>t$. Inside each leaf region, it makes a simple prediction, usually a constant for regression or a class-proportion vector for classification. Thus the fitted score is piecewise constant rather than globally linear.

</details>


### Question 2

A two-dimensional tree first splits on $x_1\le 0.4$, then, within the right child $x_1> 0.4$, splits on $x_2\le 0.7$. List the three resulting leaf regions as subsets of $\mathbb R^2$.

<details>
<summary>Answer</summary>

The first leaf is the left child of the first split:

$$
R_1=\{x\in\mathbb R^2:x_1\le 0.4\}.
$$

The right child is split again, giving

$$
R_2=\{x\in\mathbb R^2:x_1>0.4,\ x_2\le 0.7\},
$$

and

$$
R_3=\{x\in\mathbb R^2:x_1>0.4,\ x_2>0.7\}.
$$

Each region is axis-aligned: in two dimensions these are rectangles or unbounded rectangles. Repeated threshold splits create an axis-aligned rectangular partition of feature space.

</details>


### Question 3

For a regression tree with leaf regions $R_1,\ldots,R_M$, explain why the fitted score can be written as

$$
\hat s(x)=\sum_{m=1}^M \hat c_m\mathbf 1\{x\in R_m\}.
$$

<details>
<summary>Answer</summary>

A regression tree predicts one constant $\hat c_m$ inside leaf $R_m$. The indicator $\mathbf 1\{x\in R_m\}$ is $1$ if $x$ falls in that leaf and $0$ otherwise.

The leaf regions must form a partition of the feature space covered by the tree: they are disjoint and every input reaches exactly one leaf. Therefore exactly one indicator is $1$, and the sum returns exactly the constant attached to that leaf.

</details>


### Question 4

Fill in the proof of the lecture's claim that, under squared-error loss, the best constant prediction in a region $R_m$ is

$$
\hat c_m=\frac{1}{N_m}\sum_{n:x_n\in R_m}y_n.
$$

<details>
<summary>Answer</summary>

For a fixed region $R_m$, minimize

$$
L(c)=\sum_{n:x_n\in R_m}(y_n-c)^2.
$$

Differentiate:

$$
L'(c)=\sum_{n:x_n\in R_m}2(c-y_n)=2\left(N_m c-\sum_{n:x_n\in R_m}y_n\right).
$$

Setting $L'(c)=0$ gives

$$
N_m c=\sum_{n:x_n\in R_m}y_n,
$$

so

$$
\hat c_m=\frac{1}{N_m}\sum_{n:x_n\in R_m}y_n.
$$

Since $L(c)$ is a convex quadratic in $c$, this critical point is the global minimizer.

</details>


### Question 5

For outcomes $2,4,7,9$ in a single regression-tree leaf, compute the leaf prediction and the leaf's contribution to $RSS(T)$.

<details>
<summary>Answer</summary>

The leaf prediction is the mean:

$$
\hat c=\frac{2+4+7+9}{4}=\frac{22}{4}=5.5.
$$

The residual sum of squares contribution is

$$
(2-5.5)^2+(4-5.5)^2+(7-5.5)^2+(9-5.5)^2.
$$

Thus

$$
RSS=(12.25)+(2.25)+(2.25)+(12.25)=29.
$$

</details>


### Question 6

For the one-dimensional regression data

$$
(x,y)=(1,1),(2,2),(4,10),(7,11),(8,12),
$$

suppose a tree requires each leaf to contain at least two training observations. In other words, a split is only allowed if it leaves at least two observations on the left and at least two observations on the right.

List the allowed candidate first-split thresholds and compute which one has the smaller post-split RSS.

<details>
<summary>Answer</summary>

The sorted $x$ values are

$$
1,2,4,7,8.
$$

For a numeric predictor, candidate thresholds are placed halfway between consecutive sorted $x$ values:

$$
1.5,\ 3,\ 5.5,\ 7.5.
$$

However, the minimum leaf size rule removes thresholds that would leave only one observation on either side. The split at $1.5$ would put only $(1,1)$ on the left, and the split at $7.5$ would put only $(8,12)$ on the right. So those are not allowed.

The allowed candidate thresholds are therefore

$$
t=\frac{2+4}{2}=3,
\qquad
t=\frac{4+7}{2}=5.5.
$$

For $t=3$, the split is

$$
{(1,1),(2,2)}
\qquad\text{and}\qquad
{(4,10),(7,11),(8,12)}.
$$

The left outcomes are $1,2$, with mean $1.5$, so the left RSS is

$$
(1-1.5)^2+(2-1.5)^2=0.5.
$$

The right outcomes are $10,11,12$, with mean $11$, so the right RSS is

$$
(10-11)^2+(11-11)^2+(12-11)^2=2.
$$

Thus the total RSS for $t=3$ is

$$
0.5+2=2.5.
$$

For $t=5.5$, the split is

$$
{(1,1),(2,2),(4,10)}
\qquad\text{and}\qquad
{(7,11),(8,12)}.
$$

The left outcomes are $1,2,10$, with mean $13/3$. The left RSS is

$$
\left(1-\frac{13}{3}\right)^2+
\left(2-\frac{13}{3}\right)^2+
\left(10-\frac{13}{3}\right)^2 = 
\frac{146}{3}.
$$

The right outcomes are $11,12$, with mean $11.5$, so the right RSS is

$$
(11-11.5)^2+(12-11.5)^2=0.5.
$$

Thus the total RSS for $t=5.5$ is

$$
\frac{146}{3}+\frac12=
\frac{295}{6}
\approx 49.17.
$$

The better split is $t=3$, because it gives the smaller post-split RSS.

</details>


### Question 7

Why is it enough, for a numeric predictor, to try thresholds halfway between consecutive sorted observed values rather than every $t\in\mathbb R$?

<details>
<summary>Answer</summary>

For a split $x_j\le t$ versus $x_j>t$, the left and right training sets change only when $t$ crosses an observed value of $x_j$ in the node. Between two consecutive sorted observed values, the set of observations on each side is constant.

Since the child-node means and RSS depend only on which observations are assigned to each child, every threshold in the same open interval gives the same training RSS. A midpoint between consecutive observed values is a representative threshold for that interval.

</details>


### Question 8

Why can a "greedy" split can fail to produce the globally best tree, even if each chosen split is locally optimal at the node where it is made.

<details>
<summary>Answer</summary>

A greedy tree chooses the split that gives the largest immediate loss reduction at the current node. That choice constrains all later descendant splits because each later split can only refine one of the child regions created now.

A different first split with a smaller immediate improvement might create child regions that can be split much more effectively later. Since the greedy procedure does not evaluate all future trees before choosing the current split, local optimality of each split does not imply global optimality of the full tree.

</details>


### Question 9

A binary tree has maximum depth $d$, where the root has depth $0$ and each split increases depth by $1$. What is the largest possible number of leaves? 

<details>
<summary>Answer</summary>

A full binary tree with depth $d$ has at most

$$
2^d
$$

leaves, because each level can double the number of active regions.

</details>


### Question 10

In a classification leaf with class **counts** $(3,7,0)$ for classes $1,2,3$, compute the empirical class-proportion vector $\hat p_m$, the score $\hat s(x)$ for any $x$ in that leaf, and the predicted class.

<details>
<summary>Answer</summary>

The leaf contains

$$
N_m=3+7+0=10
$$

observations. The empirical class-proportion vector is

$$
\hat p_m=\left(\frac{3}{10},\frac{7}{10},0\right)^T=(0.3,0.7,0)^T.
$$

For any $x\in R_m$, the classification tree score is

$$
\hat s(x)=\hat p_m.
$$

The predicted class is

$$
\arg\max_k \hat p_{mk}=2.
$$

</details>


### Question 11

For a node with class proportions $(0.2,0.5,0.3)$, compute the misclassification error, Gini impurity, and entropy. Use natural logarithms for entropy.

<details>
<summary>Answer</summary>

The misclassification error is

$$
1-\max(0.2,0.5,0.3)=1-0.5=0.5.
$$

The Gini impurity is

$$
1-(0.2^2+0.5^2+0.3^2)=1-(0.04+0.25+0.09)=0.62.
$$

The entropy is

$$
-[0.2\log(0.2)+0.5\log(0.5)+0.3\log(0.3)].
$$

Numerically this is approximately

$$
-[-0.3219-0.3466-0.3612]=1.0297.
$$

</details>


### Question 12

Show algebraically that the Gini impurity formula

$$
\sum_{k=1}^K \hat p_k(1-\hat p_k)
$$

is equal to

$$
1-\sum_{k=1}^K \hat p_k^2.
$$

<details>
<summary>Answer</summary>

Expand the first expression:

$$
\sum_{k=1}^K \hat p_k(1-\hat p_k)
=\sum_{k=1}^K \hat p_k-\sum_{k=1}^K \hat p_k^2.
$$

Because the $\hat p_k$ are class proportions, they sum to one:

$$
\sum_{k=1}^K \hat p_k=1.
$$

Therefore

$$
\sum_{k=1}^K \hat p_k(1-\hat p_k)=1-\sum_{k=1}^K\hat p_k^2.
$$

</details>


### Question 13

In binary classification, let $p$ be the proportion of class $1$ in a node. Write the misclassification error, Gini impurity, and entropy as functions of $p$. Which values of $p$ make the node pure?

<details>
<summary>Answer</summary>

The class proportions are $(p,1-p)$. The misclassification error is

$$
1-\max(p,1-p)=\min(p,1-p).
$$

The Gini impurity is

$$
p(1-p)+(1-p)p=2p(1-p).
$$

The entropy is

$$
-[p\log p+(1-p)\log(1-p)],
$$

with the usual convention that $0\log 0=0$.

The node is pure when all observations belong to one class, so

$$
p=0\quad\text{or}\quad p=1.
$$

All three impurity measures are zero at those values.

</details>


### Question 14

Show why Gini impurity can be more sensitive than misclassification error. Compare nodes with class proportions $(0.60,0.20,0.20)$ and $(0.60,0.30,0.10)$.

<details>
<summary>Answer</summary>

For $(0.60,0.20,0.20)$, the misclassification error is

$$
1-0.60=0.40,
$$

and the Gini impurity is

$$
1-(0.60^2+0.20^2+0.20^2)=1-(0.36+0.04+0.04)=0.56.
$$

For $(0.60,0.30,0.10)$, the misclassification error is

$$
1-0.60=0.40,
$$

and the Gini impurity is

$$
1-(0.60^2+0.30^2+0.10^2)=1-(0.36+0.09+0.01)=0.54.
$$

The misclassification error is unchanged because the largest class proportion remains $0.60$. The Gini impurity changes from $0.56$ to $0.54$ because it depends on all class proportions. This illustrates the lecture's point that Gini can detect purity changes that misclassification error ignores.

</details>


### Question 15

Why do tree use the weighted child-node impurity

$$
Q(j,t)=\frac{N_1}{N_R}I(R_1)+\frac{N_2}{N_R}I(R_2)
$$

instead of the unweighted average $\{I(R_1)+I(R_2)\}/2$?

<details>
<summary>Answer</summary>

The split criterion should represent the average impurity experienced by observations in the parent node after the split. A child containing many observations should contribute more to that average than a child containing only a few observations.

The weights $N_1/N_R$ and $N_2/N_R$ are the fractions of parent-node observations going to the two children. An unweighted average would let a tiny pure child count as much as a large impure child, overstating the quality of splits that isolate a small number of observations.

</details>


### Question 16

Claim: "A deep tree is always better because it can make more splits." Is this correct?

<details>
<summary>Answer</summary>

More depth increases flexibility, so a deeper tree can reduce training error by fitting finer regions. But lower training error does not guarantee lower test error.

</details>


### Question 17

Explain why ordinary tree splits are not sensitive to monotone rescaling of a numeric feature. E.g., if $z_j=100x_j$, what split in $z_j$ is equivalent to $x_j\le t$?

<details>
<summary>Answer</summary>

The split

$$
x_j\le t
$$

is equivalent to

$$
100x_j\le 100t.
$$

Since $z_j=100x_j$, the equivalent split is

$$
z_j\le 100t.
$$

A monotone rescaling preserves the ordering of observations along that feature, so the same partition can be represented with a transformed threshold. This is why basic trees usually do not require standardization, unlike distance-based methods such as k-means.

</details>
