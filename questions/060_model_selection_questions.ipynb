### Question 1

For train/validation/test splits, which part of the train/validation/test or nested-CV procedure is involved, and why is this separation necessary?

<!-- Source: Model Selection lecture; section related to: For train/validation/test splits, which part of the train/validation/test or nested-CV pro -->

<details>
<summary>Answer</summary>

The lecture treats train/validation/test splits as a selection problem: train candidates on training data, choose hyperparameters on validation data, and reserve test or outer-fold data for an honest final estimate. Reusing the test set would make it part of training-by-selection.

</details>
