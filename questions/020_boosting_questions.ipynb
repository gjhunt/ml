# Boosting Review Questions {#sec-boosting-questions}

### Question 1

Contrast boosting trees with bagging trees. What is different about how the trees are fit?

<details>
<summary>Answer</summary>

Bagging and random forests fit many trees in parallel, using bootstrap samples and possibly randomized feature subsets, and then average or vote over their predictions. Boosting fits trees sequentially: tree $m$ is chosen after the previous score function $\hat s_{m-1}$ has already been built.

For squared-error regression, "where the previous model did poorly" is represented by the residuals

$$
r_{nm}=y_n-\hat s_{m-1}(x_n).
$$

The next weak learner $\hat h_m$ is fit to these residuals, so it tries to correct the current model's errors at the training points.

</details>


### Question 2

Starting from the recursive update

$$
\hat s_m(x)=\hat s_{m-1}(x)+\nu\hat h_m(x),
$$

expand $\hat s_M(x)$ in terms of $\hat s_0$ and the weak learners $\hat h_1,\ldots,\hat h_M$.

<details>
<summary>Answer</summary>

Apply the recursion repeatedly:

$$
\hat s_1(x)=\hat s_0(x)+\nu\hat h_1(x),
$$

$$
\hat s_2(x)=\hat s_1(x)+\nu\hat h_2(x)
=\hat s_0(x)+\nu\hat h_1(x)+\nu\hat h_2(x).
$$

Continuing to $M$ steps gives

$$
\hat s_M(x)=\hat s_0(x)+\sum_{m=1}^M\nu\hat h_m(x).
$$

Thus boosting builds an additive score function from sequential weak learners.

</details>


### Question 3

For regression outcomes $y=(1,3,5)$, compute the initial constant score $\hat s_0(x)=\bar y$. If the first fitted weak learner predicts $\hat h_1(x_1)=-1$, $\hat h_1(x_2)=0$, and $\hat h_1(x_3)=1$, compute the updated training predictions for $\nu=0.5$.

<details>
<summary>Answer</summary>

The initial score is the sample mean:

$$
\hat s_0=\bar y=\frac{1+3+5}{3}=3.
$$

So the initial predictions are $(3,3,3)$. The boosting update is

$$
\hat s_1(x_n)=\hat s_0(x_n)+\nu\hat h_1(x_n).
$$

With $\nu=0.5$, the updates are

$$
0.5(-1,0,1)=(-0.5,0,0.5).
$$

Thus the updated predictions are

$$
(3,3,3)+(-0.5,0,0.5)=(2.5,3,3.5).
$$

</details>


### Question 4

Fill in the derivative calculation showing that residual boosting is gradient boosting for squared-error loss

$$
\ell(y,s)=\frac12(y-s)^2.
$$

<details>
<summary>Answer</summary>

Differentiate with respect to the score value $s$:

$$
\frac{\partial}{\partial s}\frac12(y-s)^2
=\frac12\cdot 2(y-s)(-1)
=s-y.
$$

Therefore the negative derivative is

$$
-\frac{\partial \ell(y,s)}{\partial s}=y-s.
$$

At iteration $m$, substituting $s=\hat s_{m-1}(x_n)$ gives

$$
r_{nm}=y_n-\hat s_{m-1}(x_n),
$$

which is exactly the ordinary residual.

</details>


### Question 5

Suppose the current residual at a training point is $r=4$, and the fitted weak learner predicts exactly this residual at that point, $\hat h_m(x)=4$. Compute the new residual after one update for $\nu=1$, $\nu=0.25$, and $\nu=0.05$.

<details>
<summary>Answer</summary>

The current prediction is increased by $\nu\hat h_m(x)=4\nu$. Since the old residual is $r=y-\hat s_{m-1}(x)=4$, the new residual is

$$
r_{\text{new}}=4-4\nu=4(1-\nu).
$$

For $\nu=1$:

$$
r_{\text{new}}=0.
$$

For $\nu=0.25$:

$$
r_{\text{new}}=4(0.75)=3.
$$

For $\nu=0.05$:

$$
r_{\text{new}}=4(0.95)=3.8.
$$

This illustrates why smaller learning rates make smaller corrections.

</details>


### Question 6

In the finite-sample viewpoint of the lecture, what is the functional gradient of $\hat R(s)$ with respect to $s$, evaluated at the training points? Write it as a vector.

<details>
<summary>Answer</summary>

At the training points, the functional gradient is the gradient of the empirical risk with respect to the score values

$$
s(x_1),\ldots,s(x_N).
$$

Ignoring the common factor $1/N$ used in the empirical average, the lecture writes the relevant vector as

$$
\left(
\frac{\partial \ell(y_1,s(x_1))}{\partial s(x_1)},
\ldots,
\frac{\partial \ell(y_N,s(x_N))}{\partial s(x_N)}
\right).
$$

The negative of the $n$th component indicates the direction in which the score value $s(x_n)$ should move to reduce the loss at that training point. These negative values are the pseudo-residuals.

</details>


### Question 7

Why does gradient boosting fit a weak learner $\hat h_m$ to pseudo-residuals instead of simply updating the score values $s(x_n)$ directly at the training points?

<details>
<summary>Answer</summary>

Directly updating

$$
s_m(x_n)=s_{m-1}(x_n)+\nu r_{nm}
$$

would define new score values only at the training inputs. It would not specify how to predict at a new point $x$ that was not in the training set.

Gradient boosting fits a weak learner $\hat h_m$ so that

$$
\hat h_m(x_n)\approx r_{nm}.
$$

This produces an actual predictive function. The model can then update scores for both training points and new inputs using

$$
\hat s_m(x)=\hat s_{m-1}(x)+\nu\hat h_m(x).
$$

</details>


### Question 8

The generic gradient boosting algorithm fits $\hat h_m$ by solving approximately

$$
\arg\min_{h\in\mathcal H}\sum_{n=1}^N(r_{nm}-h(x_n))^2.
$$

Why is this a regression task even when the original problem is classification?

<details>
<summary>Answer</summary>

The target values in this auxiliary problem are the pseudo-residuals $r_{nm}$, which are numerical values. For binary logistic boosting they are

$$
y_n-p_{nm},
$$

and for multiclass boosting they are values like

$$
t_{nk}-p_{nk}^{(m-1)}.
$$

These are real-valued quantities, not class labels. Therefore each weak learner is fit to predict real-valued pseudo-residuals, so the weak-learner fitting step is a regression task regardless of the original supervised learning task.

</details>


### Question 9

For binary cross-entropy in the form

$$
\ell(y,s)=-ys+\log(1+e^s),
$$

derive

$$
\frac{\partial \ell(y,s)}{\partial s}=p-y,
\qquad p=\sigma(s).
$$

<details>
<summary>Answer</summary>

Differentiate term by term:

$$
\frac{\partial}{\partial s}(-ys)=-y,
$$

and

$$
\frac{\partial}{\partial s}\log(1+e^s)=\frac{e^s}{1+e^s}.
$$

But

$$
\frac{e^s}{1+e^s}=\frac{1}{1+e^{-s}}=\sigma(s)=p.
$$

Therefore

$$
\frac{\partial \ell(y,s)}{\partial s}=p-y.
$$

The negative gradient is $y-p$.

</details>


### Question 10

Derive the lecture's initial constant score for binary logistic boosting:

$$
\hat s_0=\log\left(\frac{\bar y}{1-\bar y}\right),
$$

where $\bar y=N^{-1}\sum_n y_n$.

<details>
<summary>Answer</summary>

The constant-score objective is

$$
L(c)=\sum_{n=1}^N[-y_n c+\log(1+e^c)].
$$

Differentiate:

$$
L'(c)=\sum_{n=1}^N\left[-y_n+\frac{e^c}{1+e^c}\right]
=-\sum_{n=1}^N y_n+N\sigma(c).
$$

Set this equal to zero:

$$
\sigma(c)=\frac1N\sum_{n=1}^N y_n=\bar y.
$$

Solving $\sigma(c)=\bar y$ gives

$$
\frac{e^c}{1+e^c}=\bar y
\quad\Longrightarrow\quad
 e^c=\frac{\bar y}{1-\bar y},
$$

so

$$
c=\log\left(\frac{\bar y}{1-\bar y}\right).
$$

</details>


### Question 11

Compute the initial binary logistic boosting score $\hat s_0$ when the positive-class fraction is $\bar y=0.20$. Is the initial score positive or negative, and why?

<details>
<summary>Answer</summary>

The initial score is the empirical log-odds:

$$
\hat s_0=\log\left(\frac{0.20}{1-0.20}\right)
=\log\left(\frac{0.20}{0.80}\right)
=\log(0.25).
$$

Thus

$$
\hat s_0\approx -1.386.
$$

It is negative because the positive class is rare: its empirical probability is less than $1/2$, so its odds are less than $1$ and the log-odds are negative.

</details>


### Question 12

For binary logistic gradient boosting, compute the pseudo-residuals $r=y-p$ for observations with $(y,p)=(1,0.70)$, $(0,0.20)$, and $(1,0.10)$. Which observation asks for the largest upward movement in logit score?

<details>
<summary>Answer</summary>

The pseudo-residuals are

$$
1-0.70=0.30,
$$

$$
0-0.20=-0.20,
$$

and

$$
1-0.10=0.90.
$$

The largest upward movement in the logit score corresponds to the largest positive pseudo-residual, which is $0.90$ for the observation with $y=1$ and $p=0.10$.

</details>


### Question 13

A binary boosted classifier has current score $s=0$ at a point, a weak learner prediction $\hat h_m(x)=2$, and learning rate $\nu=0.1$. Compute the updated score and the updated probability $\sigma(s)$.

<details>
<summary>Answer</summary>

The update is on the logit scale:

$$
s_{\text{new}}=s+\nu\hat h_m(x)=0+0.1(2)=0.2.
$$

The updated probability is

$$
\sigma(0.2)=\frac{1}{1+e^{-0.2}}\approx0.550.
$$

So the update changes the score to $0.2$ and the probability to about $0.55$.

</details>


### Question 14

For one observation in a three-class problem, suppose the true class is $2$ and the current softmax probabilities are

$$
(p_1,p_2,p_3)=(0.20,0.50,0.30).
$$

Compute the multiclass pseudo-residual vector $(r_1,r_2,r_3)$.

<details>
<summary>Answer</summary>

The one-hot vector for true class $2$ is

$$
(t_1,t_2,t_3)=(0,1,0).
$$

The multiclass pseudo-residuals are

$$
r_k=t_k-p_k.
$$

Therefore

$$
(r_1,r_2,r_3)=(0-0.20,
1-0.50,
0-0.30)=(-0.20,0.50,-0.30).
$$

</details>


### Question 15

Show that the multiclass pseudo-residuals for a single observation always sum to zero:

$$
\sum_{k=1}^K (t_{nk}-p_{nk})=0.
$$

<details>
<summary>Answer</summary>

For one observation, the one-hot vector satisfies

$$
\sum_{k=1}^K t_{nk}=1,
$$

because exactly one class is the observed class. The softmax probabilities also sum to one:

$$
\sum_{k=1}^K p_{nk}=1.
$$

Therefore

$$
\sum_{k=1}^K(t_{nk}-p_{nk})
=\sum_{k=1}^K t_{nk}-\sum_{k=1}^K p_{nk}
=1-1=0.
$$

</details>


### Question 16

In multiclass gradient boosting with $K=4$ classes and $M=75$ boosting iterations, how many regression trees are fit?

<details>
<summary>Answer</summary>

The lecture describes fitting $K$ regression trees per boosting iteration. With $K=4$ and $M=75$, the total number of trees is

$$
K M=4\cdot75=300.
$$

</details>


### Question 17

Explain why training loss can keep decreasing as more boosting iterations are added while test loss may flatten or increase. Use the roles of $M$ and early stopping in your answer.

<details>
<summary>Answer</summary>

Each additional boosting iteration adds another weak learner to the additive score function, increasing model flexibility. The new learner is chosen to improve the current model on the training data, so training loss often continues to decrease.

Test loss measures generalization. After enough iterations, the model may start fitting training-sample noise or idiosyncrasies, so test loss can stop improving or increase. The number of trees $M$ is therefore a regularization parameter, and early stopping chooses $M$ using validation error, cross-validation, or test-like information rather than training error.

</details>


### Question 18

For squared-error loss and a fixed tree leaf $R_{jm}$, prove that the leaf-specific update

$$
\gamma_{jm}=\arg\min_\gamma \sum_{x_n\in R_{jm}}(y_n-\hat s_{m-1}(x_n)-\gamma)^2
$$

is the mean current residual in that leaf.

<details>
<summary>Answer</summary>

Let

$$
r_n=y_n-\hat s_{m-1}(x_n).
$$

Then the objective inside the leaf is

$$
L(\gamma)=\sum_{x_n\in R_{jm}}(r_n-\gamma)^2.
$$

Differentiate:

$$
L'(\gamma)=2\sum_{x_n\in R_{jm}}(\gamma-r_n)
=2\left(N_{jm}\gamma-\sum_{x_n\in R_{jm}}r_n\right).
$$

Setting this equal to zero gives

$$
\gamma_{jm}=\frac{1}{N_{jm}}\sum_{x_n\in R_{jm}}r_n.
$$

So for squared-error loss, the optimal leaf update is the mean residual in that leaf.

</details>
