# Introduction to Statistical Machine Learning

## What is Statistical Machine Learning?

Statistical machine learning is the study of how to build and analyze data-driven methods using probabilistic thinking.

At a high level, the goal is to extract useful information from data. This might involve making predictions, identifying patterns, or summarizing complex relationships. What distinguishes the "statistical" perspective is the explicit role of uncertainty. Data are rarely clean or complete, and any conclusions we draw are inherently approximate.

Rather than treating models as exact representations of reality, we treat them as approximations. Probability and statistics provides a framework for reasoning about how we should build and evaluate machine learning methods. We will use this statistical perspective when it is useful, and eschew it when it is not.

## Types of Learning Problems

### Supervised Learning

We begin with data. For supervised learning we observe a collection of $N$ samples:

$$
(x_1, y_1), (x_2, y_2), \dots, (x_N, y_N)
$$

Each $x_n$ represents an input or feature vector, and $y_n$ represents an associated outcome. The central task is to use this data to construct a function $\hat{f}$ that captures the relationship between inputs and outputs.

**Key:** The goal is not just to fit the observed data, but to learn a rule that generalizes. That is, we want $\hat{f}$ to work well on new, unseen inputs.

In supervised learning, we observe both inputs and outputs. The term "supervised" reflects the fact that the outputs $y_n$ provide guidance. They tell us what the correct answer should be for each input.

The objective is to learn a function $\hat{f}$ such that:

$$
y \approx \hat{f}(x)
$$

In other words, we want to predict the output $y$ from the input $x$.

There are two main types of supervised problems.

**Regression**

- The output $y$ is continuous (real-valued so that $y\in\mathbb{R}$).
- Examples:
  - Predicting stock market performance from economic indicators
  - Predicting adult height from childhood measurements
- In this case $\hat{f}: \mathbb{R} \to \mathbb{R}$

**Classification**

- The output $y$ is discrete or categorical ($y \in \mathcal{C} = \{c_1,\ldots,c_K\}$)
- Examples:
  - Predicting whether a borrower will default (yes/no)
  - Classifying an image into categories such as plant species (so-called **multi-class classification**)
- In this case $\hat{f}: \mathbb{R} \to \mathcal{C}$

### Unsupervised Learning

In unsupervised learning, we do not observe outputs. We are given only inputs:
$$
x_1, x_2, \dots, x_N
$$

There is no explicit prediction task. Instead, the goal is to uncover structure in the data.

Typical objectives include:

- Identifying clusters or groupings
- Discovering relationships between variables
- Summarizing the main patterns or trends

Because there is no "correct answer" provided, unsupervised learning is inherently more exploratory. It often serves as a tool for understanding data rather than making direct predictions. Although it can be used in conjunction with supervised learning. 

## What Does It Mean to "Learn"?

Learning consists of **choosing** a **good** function $\hat{f}$ based on the observed data. This raises two fundamental questions that will guide the course:

1. **Methods:** How do we construct $\hat{f}$?  
   What principles do we use to choose a prediction function $\hat{f}$?

2. **Evaluation:** How do we determine if $\hat{f}$ is good?  
   What criteria do we use to measure performance? This can be used to help choose $\hat{f}$ but also to report its performance.

## Statistical Setup for Supervised Learning

To study supervised learning, we adopt a statistical viewpoint.

We assume the data are drawn i.i.d. from a distribution $P=P_{X,Y}$ so that
$$
(x_1, y_1), \dots, (x_N, y_N) \overset{\text{i.i.d.}}{\sim} P
$$

Each input $x_n$ is a $D$-dimensional vector:
$$
x_n = (X_{n1}, X_{n2}, \dots, X_{nD}) \in \mathbb{R}^D
$$

There is an underlying probabilistic relationship between $x$ and $y$. The randomness may reflect measurement noise, unobserved variables, or inherent variability in the system, depending on your philosophy of probability. 

The goal is to use the observed data to construct a function $\hat{f}$ that predicts $y$ from $x$:
$$
y \approx \hat{y} = \hat{f}(x).
$$

## A Paradigm: Empirical Risk Minimization (ERM)

How should we learn from the data? There are many paradigms that we can use in statistical learning. In general our objective is to construct a function $\hat{f}$ that performs well on **new** data drawn from the distribution $P$. 

### Score and Action Separation

In many presentations, the prediction rule $\hat{f}$ is treated as a single object. In practice, however, it is often useful to separate it into two components.

We write the prediction rule as:
$$
\hat{f}(x) = a(\hat{s}(x)),
$$
where:

- $\hat{s}$ is a **score function**
- $a$ is an **action function**

The score function $\hat{s}(x)$ assigns a numerical score (or a vector of scores) to each input $x$. The action function $a$ then maps this score to a final prediction $\hat{y}$. The **score function** encodes what the model "believes" about the input. The **action function** determines how that belief is converted into a concrete decision. In some ways the score is the *raw* output of the prediction function, before it has been massaged into the right human-readable form.

Typically, the action function $a$ is fixed in advance, and so most of the interesting and difficult parts of machine learning happen in figuring out $\hat{s}$. Consequently, this separation is useful because evaluating prediction function is often more naturally defined in terms of the score, rather than the final prediction. In any case, we lose nothing by doing this. 

In **regression**, this score/action distinction is typically not important. The score itself is the prediction, so we typically will take:
$$
a = \text{identity}, \quad a(x) = x,
$$
in which case $\hat{f}(x) = \hat{s}(x)$ and so the distinction collapses. 

In **classification**, the distinction becomes more meaningful. Suppose we are predicting one of $K$ classes.

Often, the score function takes the form:
$$
\hat{s}: \mathbb{R}^D \to \mathbb{R}^K,
$$
where each component $\hat{s}(x)_k$ represents a score for class $k$. Larger values indicate that the model considers class $k$ more plausible (in an informal sense).

The action function then selects a label:
$$
a: \mathbb{R}^K \to \mathcal{C}, \quad
a(t) = \arg\max_{k\in\mathcal{C}} t_k.
$$
Here $\arg\max$ is just the function that returns the index of the maximizing value. 

Thus, the final prediction is:
$$
\hat{f}(x) = \arg\max_{k\in\mathcal{C}} \hat{s}(x)_k.
$$

This score/action decomposition is not strictly necessary, but it often provides a cleaner way to think about how models produce predictions. Unless otherwise noted, these are the choices for $a$ we will use henceforth. We don't lose anything by doing this, since if we fix $a$ then knowing $\hat{s}$ allows me to recover $\hat{f}$. 

### Loss Functions

A primary thing we need to do when *learning* a prediction function $\hat{f}$ is to evaluate how *good* the predictions actually are. We can do this using a **loss function** $\ell$. To evaluate a prediction $\hat{f}(x)$ against the true outcome $y$, we let
$$
\ell(y, \hat{s}(x)).
$$
quantify the accuracy. Here higher loss $\ell$ is worse. Typically we want this quantification to be some non-negative number. Consequently,

- in **regression** typically $\ell: \mathbb{R} \times \mathbb{R} \to \mathbb{R}^+$,
- in **classification** typically $\ell: \mathcal{C} \times \mathbb{R}^K \to \mathbb{R}^+$.

Examples include: 

- Squared error: $\ell(y, \hat{f}(x)) = (y - \hat{f}(x))^2$ for regression, (sloppily conflating $\hat{s}$ with $\hat{f}$)
- Absolute error: $\ell(y, \hat{f}(x)) = |y - \hat{f}(x)|$ for regression,
- Zero-one loss: $\ell(y,\hat{s}(x)) = \mathbb{1}(y \neq a(\hat{s}(x))) = \mathbb{1}(y \neq \hat{f}(x))$
- Cross-entropy loss: let
  $$
  p_y(x) = \frac{\exp(\hat{s}_y(x))}{\sum_{k=1}^K \exp(\hat{s}_k(x))}
  $$
  and then define
  $$
    \ell(y, \hat{s}(x)) =
    -\log \left( p_y(x) \right).
    $$
  

The loss function provides a way to compare different candidate prediction rules.

### Population Risk

Ideally, we would like to choose a function $\hat{s}$ that minimizes the loss over the true data-generating distribution $P$. That is, for some new, unseen data from our data generating distribution, we'd like $\hat{s}$ to not incur much loss. This leads to the notion of **risk**:
$$
R(s) = \mathbb{E}_{(x,y) \sim P}[\ell(y, s(x))].
$$
This quantity measures (in expectation) how well $s$ performs on average for new samples drawn from $P$.

The optimal scoring rule in this framework is then
$$
s^* = \arg\min_s R(s).
$$

This represents the best possible score function within this framework, assuming we had access to the true distribution $P$.

**Problem:** the distribution $P$ is unknown, so the expectation cannot be computed directly.

### Empirical Risk

Since we do not have access to $P$, we approximate the population risk using the observed data. This leads to the *empirical risk*:
$$
\hat{R}(s) = \hat{R}_N(s) = \frac{1}{N} \sum_{n=1}^N \ell(y_n, s(x_n)).
$$

The empirical risk is simply the average loss over the training sample. Ideally, then, we could do something like: 

$$
\hat{s} = \arg\min_s \hat{R}(s).
$$

**More Problems:** which $s$ do we optimize over? All of them? 

In practice, we have to choose some class of candidate functions $\mathcal{S}$.

**Empirical Risk Minimization (ERM)** is then the principle of choosing $\hat{s}$ that minimizes the empirical risk:
$$
\hat{s} = \arg\min_{s\in\mathcal{S}} \hat{R}(s).
$$

In words, we select the function that best fits the observed data according to the chosen loss function.

**Even more problems:** How do we do this optimization? In practice we need to choose some `Optimizer` (either closed-form or an algorithm) to try to optimize $\hat{R}(s)$ over $\mathcal{S}$. 

Consequently, determing our score function $\hat{s}$ using ERM ends up being a function of

- the data $\mathcal{D} = \{(x_n,y_n)\}_{n=1}^N$
- the loss function $\ell$ which determines our empirical risk
- $\mathcal{S}$, the space of functions
- the `Optimizer` to search over this space,

One could think of it as:

$$
\hat{s} = \text{Algorithm}(\mathcal{D}, \ell, \mathcal{S}, {\tt Optimizer}).
$$

While ERM is a natural and widely used approach, it raises important questions. For example:
- Does minimizing empirical risk lead to good performance on new data?
- Can a model fit the training data too well?
- How does the choice of model class affect the outcome?

These questions, and others, motivate the study of generalization, regularization, and model complexity, which we will develop later.

Furthermore, ERM isn't the only game in town. Although, for an introductory course its a good framework to hang much of what we learn upon.
