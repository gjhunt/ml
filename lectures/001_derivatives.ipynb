# Matrix Calculus for Machine Learning

In single-variable calculus, the basic question is how one quantity changes as another quantity changes. If $f: \mathbb{R} \to \mathbb{R}$, then the **derivative** $\frac{df}{dx}$ measures how the scalar output changes as the scalar input changes. The familiar rules describe common patterns: constants differentiate to zero, linear terms differentiate to their coefficients, and quadratic terms produce a factor of two.

In multivariable calculus, the same idea is extended to functions with several inputs. For a scalar-valued function such as $L(w_1,\ldots,w_d)$, the **gradient** collects the partial derivatives into one vector. **Matrix calculus** systematizes this notation and extends it to expressions involving vectors and matrices. The underlying calculus is not new. The main difference is that every object now has a shape, and the notation has to keep track of those shapes.

## Scalar Objectives and Gradients

Most fitting problems in this course minimize a **scalar** loss. If
$$
L:\mathbb{R}^D\to\mathbb{R},\qquad
w=(w_1,\ldots,w_D)^\top,
$$
then its gradient is the column vector
$$
\nabla_w L=
\begin{pmatrix}
\frac{\partial L}{\partial w_1}\\
\vdots\\
\frac{\partial L}{\partial w_D}
\end{pmatrix}
\in\mathbb{R}^D.
$$

**The rule:** the gradient has the same shape as the input vector. Its $j$th entry measures how the loss changes when $w_j$ changes while the other coordinates are held fixed.

**Example.** If
$$
L(w) = L(w_1,w_2)=w_1^2+3w_1w_2-2w_2,
$$
then
$$
\nabla_wL=
\begin{pmatrix}
2w_1+3w_2\\
3w_1-2
\end{pmatrix}.
$$
Nothing fundamentally new has happened: we computed ordinary partial derivatives and stacked them into a vector.


## Shape Checking

Shapes provide a quick way to catch mistakes. If $w,a\in\mathbb{R}^D$, $X\in\mathbb{R}^{N\times D}$, and $y\in\mathbb{R}^N$, then

| Expression | Shape |
|---|---:|
| $a^\top w$ | scalar |
| $Xw$ | $N$-vector |
| $y-Xw$ | $N$-vector |
| $\|y-Xw\|^2$ | scalar |
| $X^\top(y-Xw)$ | $D$-vector |
| $\nabla_w L$ | $D$-vector |

Consequently, $X^\top(y-Xw)$ has the correct shape to be a gradient with respect to $w$, while $X(y-Xw)$ usually is not even a valid matrix product.

Shape checking cannot prove that a derivative is correct, but it can quickly prove that many incorrect formulas are impossible.


## Linear Expressions

For a constant $c$ and a fixed vector $a\in\mathbb{R}^D$,
$$
\nabla_wc=0,
\qquad
\nabla_w(a^\top w)=a.
$$

To verify the second identity, expand
$$
a^\top w=\sum_{j=1}^D a_jw_j.
$$
The partial derivative with respect to $w_k$ is $a_k$, so the gradient is $a$.

**Statistical-learning example.** For a linear score
$$
s_w(x)=w^\top x,
$$
the input $x$ is fixed when we differentiate with respect to the model parameters. Therefore
$$
\nabla_ws_w(x)=x.
$$

## Squared Norms and Quadratic Forms

Since
$$
w^\top w=\sum_{j=1}^D w_j^2,
$$
we have
$$
\nabla_w(w^\top w)=2w,
\qquad
\nabla_w\left(\frac12w^\top w\right)=w.
$$

More generally, for a fixed $D\times D$ matrix $A$,
$$
\nabla_w(w^\top Aw)=(A+A^\top)w.
$$
If $A$ is symmetric, this becomes
$$
\nabla_w(w^\top Aw)=2Aw,
\qquad
\nabla_w\left(\frac12w^\top Aw\right)=Aw.
$$

The symmetry condition matters. Only the symmetric part $(A+A^\top)/2$ contributes to the scalar quadratic form.

**Example.** If $X\in\mathbb{R}^{N\times D}$, then $X^\top X$ is symmetric because
$$
(X^\top X)^\top=X^\top X.
$$
Hence
$$
\nabla_w\left(\frac12w^\top X^\top Xw\right)=X^\top Xw.
$$


## The Chain Rule for a Scalar Intermediate

Suppose $s:\mathbb{R}^D\to\mathbb{R}$ is a scalar intermediate and $\ell:\mathbb{R}\to\mathbb{R}$ is a scalar function. Define the composite objective $L:\mathbb{R}^D\to\mathbb{R}$ by
$$
L(w)=\ell(s(w)).
$$
Thus the composition is
$$
w\longmapsto s(w)\longmapsto \ell(s(w))=L(w).
$$
Writing $s=s(w)$ for the scalar intermediate value, the ordinary chain rule becomes
$$
\nabla_wL(w)=\frac{d\ell}{ds}\,\nabla_ws(w),
$$
where $d\ell/ds$ is evaluated at $s=s(w)$.

**Example.** Define
$$
s(w)=a^\top w-b,
\qquad
\ell(s)=\frac12s^2.
$$
Then
$$
L(w)=\ell(s(w))=\frac12(a^\top w-b)^2.
$$
Since
$$
\frac{d\ell}{ds}=s,
\qquad
\nabla_ws(w)=a,
$$
the chain rule gives
$$
\nabla_wL(w)=\frac{d\ell}{ds}\nabla_ws(w)
=(a^\top w-b)a.
$$

### Example: 

For fixed $x$ and $y$, let
$$
s(w)=x^\top w,
\qquad
p(w)=\sigma(s(w)),
$$
where $\sigma(s)=1/(1+e^{-s})$. Define the outer loss
$$
\ell_y(p)=-\left[y\log p+(1-y)\log(1-p)\right]
$$
and the complete objective
$$
L(w)=\ell_y(p(w)).
$$

To reduce clutter, write $s=s(w)$ and $p=p(w)$. The needed scalar derivatives are
$$
\frac{d\ell_y}{dp}
=-\frac{y}{p}+\frac{1-y}{1-p}
=\frac{p-y}{p(1-p)},
$$
$$
\frac{dp}{ds}=p(1-p),
\qquad
\nabla_ws(w)=x.
$$
Applying the chain rule through the two scalar intermediates gives
$$
\nabla_wL(w)
=\frac{d\ell_y}{dp}
\frac{dp}{ds}\nabla_ws(w)
=(p-y)x.
$$



## The Chain Rule Through Several Intermediates

Suppose $z:\mathbb{R}^D\to\mathbb{R}^m$ is a vector intermediate and $\ell:\mathbb{R}^m\to\mathbb{R}$ is a scalar function. Define the composite objective by
$$
L(w)=\ell(z(w)).
$$
For each parameter coordinate $w_j$, the chain rule gives
$$
\frac{\partial L}{\partial w_j}
=\sum_{k=1}^m
\frac{\partial \ell}{\partial z_k}
\frac{\partial z_k}{\partial w_j}.
$$
Here the derivatives of $\ell$ are evaluated at $z(w)$.


**Example.** Let

$$
z_1(w)=w_1+w_2,\qquad
z_2(w)=w_1w_2,
$$

and define the outer function and composite objective by

$$
\ell(z)=\frac12(z_1^2+z_2^2),
\qquad
L(w)=\ell(z(w)).
$$

Both $w_1$ and $w_2$ influence the loss through both $z_1$ and $z_2$.

First differentiate the loss with respect to the intermediate variables:

$$
\frac{\partial \ell}{\partial z_1}=z_1,
\qquad
\frac{\partial \ell}{\partial z_2}=z_2.
$$

Next differentiate the intermediate variables with respect to the parameters:

$$
\frac{\partial z_1}{\partial w_1}=1,
\qquad
\frac{\partial z_1}{\partial w_2}=1,
$$

and

$$
\frac{\partial z_2}{\partial w_1}=w_2,
\qquad
\frac{\partial z_2}{\partial w_2}=w_1.
$$

At

$$
w=(1,2)^\top,
$$

the intermediate values are

$$
z_1=1+2=3,
\qquad
z_2=(1)(2)=2.
$$

For $w_1$, the chain rule gives two contributions, one through $z_1$ and one through $z_2$:

$$
\frac{\partial L}{\partial w_1}=
\frac{\partial \ell}{\partial z_1}
\frac{\partial z_1}{\partial w_1}
+
\frac{\partial \ell}{\partial z_2}
\frac{\partial z_2}{\partial w_1}.
$$

Substituting the derivatives and evaluating at the point,

$$
\frac{\partial L}{\partial w_1}=
z_1(1)+z_2(w_2)=
3(1)+2(2)=
7.
$$

Similarly, for $w_2$,

$$
\frac{\partial L}{\partial w_2}=
\frac{\partial \ell}{\partial z_1}
\frac{\partial z_1}{\partial w_2}+
\frac{\partial \ell}{\partial z_2}
\frac{\partial z_2}{\partial w_2},
$$

so

$$
\frac{\partial L}{\partial w_2}=
z_1(1)+z_2(w_1)=
3(1)+2(1)=
5.
$$

Thus the gradient with respect to $w$ is

$$
\nabla_w L=
\begin{pmatrix}
7\\
5
\end{pmatrix}.
$$


## Matrix Parameters Are Bookkeeping

Consider instead an entire matrix $W$ with columns $w_i\in\mathbb{R}^D$:
$$
W=(w_1,\ldots,w_K)\in\mathbb{R}^{D\times K}.
$$
When a scalar loss depends on a matrix, define its gradient entry by entry:
$$
(\nabla_WL)_{dk}=\frac{\partial L}{\partial W_{dk}}.
$$
Thus $\nabla_WL$ has the same $D\times K$ shape as $W$. This introduces no new calculus; it simply arranges the ordinary partial derivatives in the same layout as the parameters.


### Example: Matching a Target Matrix

Suppose $W\in\mathbb{R}^{2\times2}$ is variable and $B\in\mathbb{R}^{2\times2}$ is fixed. Define
$$
L(W)=\frac12\sum_{d=1}^2\sum_{k=1}^2(W_{dk}-B_{dk})^2.
$$
Only one term in the sum depends on a given entry $W_{dk}$, so
$$
\frac{\partial L}{\partial W_{dk}}=W_{dk}-B_{dk}.
$$
Arranging these four partial derivatives in the same positions as the entries of $W$ gives
$$
\nabla_WL=W-B.
$$

For example, if
$$
W=
\begin{pmatrix}
2&-1\\
0&4
\end{pmatrix},
\qquad
B=
\begin{pmatrix}
1&1\\
2&3
\end{pmatrix},
$$
then
$$
\nabla_WL=
\begin{pmatrix}
1&-2\\
-2&1
\end{pmatrix}.
$$
