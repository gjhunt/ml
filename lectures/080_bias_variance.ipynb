# Bias-Variance Tradeoff


Why does a learned model make mistakes? One useful way to think about this is coming from two sources:

**Bias**
  - Error from restricting ourselves to a model class
  - We approximate a complex reality with a simpler function

**Variance**
  - Error from sensitivity to the particular training dataset
  - Different samples → different fitted models

Our key intuition is flexible models are low bias, high variance, where as rigid models are high bias, low variance. They tend to move opposite of each other. This tension is the **bias–variance tradeoff**.

## Statistical Setup

Let's try to make this more rigorous. The easiest way to do this is in a *regression* context with a squared-error loss.

Assume a standard data-generating model:
$$
Y = f^*(X) + \varepsilon
$$

Here $f^*$ is the **true** model, now just the *best* one. 

Where:
- $f^*(X)$: true underlying signal
- $\varepsilon$: noise

pretty mild assumptions assumptions:
- $\mathbb{E}[\varepsilon \mid X] = 0$
- $\mathrm{Var}(\varepsilon \mid X) = \sigma^2$

Assume we have some training data: 

$$
\mathcal{D} = \{(X_i, Y_i)\}_{i=1}^n \overset{iid}{\sim} P
$$

We use this training data to learn a predictor:
$$
\hat f = \hat f_{\mathcal{D}}
$$

Here, $\hat f$ is **random** because it depends on the random dataset $\mathcal{D}$.

Under this model there are **two sources of randomness**:
1. Noise in outputs $\varepsilon$
2. Randomness of the dataset $\mathcal{D}$

Consider evaluating at a new new test point:
$$
(X, Y) \sim P
$$

We let's consider the point-wise risk (for a fixed $x$):
$$
R_f(x) = \mathbb{E}\big[(Y - \hat f(x))^2 \mid X = x\big]
$$

Notice here that we are using **squared error** to measure fit. This is important, and our derivation won't work (exactly) without it. 

We define quantities **at a fixed $x$**. So that the expectation is over the randomness of the dataset $\mathcal{D}$. Define the **bias** as :
$$
\mathrm{Bias}_f(x) = \mathbb{E}_{\mathcal{D}}[\hat f(x)] - f^*(x)
$$

The interpretation is that *bias* measures far the *average model* is from truth. Its a measure of **model misspecification**. 

Define the **variance** as 
$$
\mathrm{Var}_f(x) = \mathrm{Var}_{\mathcal{D}}(\hat f(x))
$$

The interpretation is how much predictions fluctuate across datasets. Variance measures **instability**.

We can show that 

$$R_f(x)=
\mathrm{Bias}_f(x)^2
+
\mathrm{Var}_f(x)
+
\sigma^2
$$

So that pointwise risk is a combination of:

- **Squared Bias**: Error from using the wrong function class
- **Variance**: Error from estimating with finite data
- **$\sigma^2$**: Irreducible error (Even perfect prediction of $f^*$ cannot remove this.)

## Derivation

Let's show that: 
$$
R_f(x) = \mathbb{E}\big[(Y - \hat f(x))^2 \mid X = x\big]
= \mathrm{Bias}_f(x)^2 + \mathrm{Var}_f(x) + \sigma^2
$$


**Step 1:** Substitute the data-generating model

Recall:
$$
Y = f^*(x) + \varepsilon
$$

Plug into the risk:
$$
R(x) = \mathbb{E}\big[(f^*(x) + \varepsilon - \hat f(x))^2\big]
$$

Rearrange:
$$
= \mathbb{E}\big[(f^*(x) - \hat f(x) + \varepsilon)^2\big]
$$


**Step 2:** Expand the square

$$
= \mathbb{E}\big[(f^*(x) - \hat f(x))^2 + 2\varepsilon(f^*(x) - \hat f(x)) + \varepsilon^2\big]
$$

Split expectation:
$$
= \mathbb{E}\big[(f^*(x) - \hat f(x))^2\big]
+ 2\mathbb{E}\big[\varepsilon(f^*(x) - \hat f(x))\big]
+ \mathbb{E}[\varepsilon^2]
$$


**Step 3:** Handle each term

(i) Noise term
$$
\mathbb{E}[\varepsilon^2] = \sigma^2
$$
 (ii) Cross term
$$
\mathbb{E}[\varepsilon(f^*(x) - \hat f(x))] = 0
$$

This happens because 
- $\varepsilon$ has mean zero given $X$
- $\hat f(x)$ depends only on $\mathcal{D}$, independent of $\varepsilon$ at test time

So
$$
\mathbb{E}[\varepsilon(f^*(x) - \hat f(x))] = \mathbb{E}[\varepsilon]\mathbb{E}[(f^*(x) - \hat f(x))] =  0 \cdot \mathbb{E}[(f^*(x) - \hat f(x))] = 0
$$

**Step 4:** Focus on the main term

We now have:
$$
R(x) = \mathbb{E}\big[(f^*(x) - \hat f(x))^2\big] + \sigma^2
$$

Now decompose:
$$
f^*(x) - \hat f(x)=
\big(f^*(x) - \mathbb{E}[\hat f(x)]\big)
+
\big(\mathbb{E}[\hat f(x)] - \hat f(x)\big)
$$


**Step 5:** Expand again

$$
(f^*(x) - \hat f(x))^2=
\big(f^*(x) - \mathbb{E}[\hat f(x)]\big)^2
+
\big(\mathbb{E}[\hat f(x)] - \hat f(x)\big)^2
+\
2\big(f^*(x) - \mathbb{E}[\hat f(x)]\big)\big(\mathbb{E}[\hat f(x)] - \hat f(x)\big)
$$

Taking expectation over $\mathcal{D}$:

- First term is constant:
$$
= \big(f^*(x) - \mathbb{E}[\hat f(x)]\big)^2
$$

since $f^*$ is constant, and the expectation is certainly a constant. 

- Second term:
$$
= \mathrm{Var}_{\mathcal{D}}(\hat f(x))
$$

- Cross term vanishes:
$$
\mathbb{E}\big[\hat f(x) - \mathbb{E}[\hat f(x)]\big] = 0
$$
this happens because
$$
\mathbb{E}\big[\hat f(x) - \mathbb{E}[\hat f(x)]\big]=
\mathbb{E}[\hat f(x)] - \mathbb{E}[\hat f(x)]= 0
$$

So the entire cross term will be zero. 


**Step 6:** Final result

Putting this all together, we get: 

$$
\mathbb{E}\big[(f^*(x) - \hat f(x))^2\big]=
\underbrace{\big(\mathbb{E}[\hat f(x)] - f^*(x)\big)^2}_{\mathrm{Bias}_f(x)^2}
+
\underbrace{\mathrm{Var}(\hat f(x))}_{\mathrm{Var}_f(x)}
$$

So we get our result:
$$
R(x)=
\mathrm{Bias}_f(x)^2
+
\mathrm{Var}_f(x)
+
\sigma^2
$$

This decomposition separates three sources of error:

- **Squred Bias**: error from model misspecification  
- **Variance**: error from dataset sensitivity  
- **Noise**: irreducible randomness in $Y$

## Why do we call it a tradeoff?

Typically, we find that flexibility typically goes with model complexity, and:

- Increasing model flexibility:
  - reduces bias (better approximation of $f^*$)
  - increases variance (more sensitivity to $\mathcal{D}$)

- Decreasing flexibility:
  - increases bias
  - reduces variance

Examples of flexibility include:
- Polynomial degree
- Number of parameters
- KNN neighborhood size

## Role of Sample Size

As the sample size increases:

- **Variance decreases** 
- **Bias typically stays the same**

Recall that variance measures how much $\hat f(x)$ changes across different training datasets. So this is more sensitive with a small sample size. However bias often depends more on the **model class**, less so than the dataset size. (In practice, bias can sometimes decrease slightly with $n$, but classically not as much as variance increases). 

Practical takeaway:

- If variance is high → get more data or regularize
- If bias is high → use a richer model


Here is an example of what these actual quantities look like in a simulation:

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsRegressor

rng = np.random.default_rng()

# True regression function
def f(x):
    return np.sin(2 * np.pi * x)**2

# Noise level
sigma = 0.3

# Grid where we evaluate bias/variance
x_grid = np.linspace(0, 1, 200).reshape(-1, 1)
f_true = f(x_grid.ravel())

n_train = 100
k_values = [int(i) for i in np.linspace(1,n_train,num=n_train)]
n_datasets = 300
```

```python
bias2_list = []
var_list = []
noise_list = []
test_error_list = []

for k in k_values:
    preds = []

    for _ in range(n_datasets):
        # Simulate one training dataset
        X_train = rng.uniform(0, 1, size=n_train)
        y_train = f(X_train) + rng.normal(0, sigma, size=n_train)

        # Fit KNN regression
        model = KNeighborsRegressor(n_neighbors=k)
        model.fit(X_train.reshape(-1, 1), y_train)

        # Predict on common grid
        y_hat = model.predict(x_grid)
        preds.append(y_hat)

    preds = np.array(preds)  # shape: (n_datasets, len(x_grid))

    # Average prediction over datasets
    mean_pred = preds.mean(axis=0)

    # Pointwise bias^2 and variance
    bias2_x = (mean_pred - f_true) ** 2
    var_x = preds.var(axis=0)

    # Global averages over x
    bias2 = bias2_x.mean()
    variance = var_x.mean()
    noise = sigma**2
    test_error = bias2 + variance + noise

    bias2_list.append(bias2)
    var_list.append(variance)
    noise_list.append(noise)
    test_error_list.append(test_error)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(k_values, bias2_list, marker='o', label='Bias^2')
plt.plot(k_values, var_list, marker='o', label='Variance')
plt.plot(k_values, test_error_list, marker='o', label='Bias^2 + Variance + Noise')
plt.axhline(sigma**2, linestyle='--', label='Noise')
plt.xlabel('k in KNN regression')
plt.ylabel('Error')
plt.title('Bias-Variance Tradeoff in KNN Regression')
plt.legend()
plt.show()
```

While technically this very mathematically precise statement about a bias-variance decomposition depends on using a squared-error loss in a regression context. It is still often a good guiding principle for other losses and even classification.
