# Unsupervised Learning 

So far, much of our setup has been supervised. We observe data of the form

$$
(x_1,y_1),\ldots,(x_N,y_N),
$$

and the goal is to learn a rule that predicts $y$ from $x$.

In unsupervised learning, there are no observed outcomes. We only observe

$$
x_1,\ldots,x_N,
$$

where each

$$
x_n \in \mathbb{R}^D.
$$

The goal is not prediction in the usual supervised sense. Instead, the goal is to find useful structure in the inputs themselves.

Typical unsupervised learning tasks include:

- finding clusters or groups of similar observations;
- finding low-dimensional summaries of high-dimensional data;
- finding unusual observations;
- estimating latent structure that may help us understand the data.

We move from trying to summarize $P(x,y)$ to trying to get useful summaries of $P(x)$ alone.
