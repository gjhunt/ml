# Constrained Optimization via Primal–Dual Methods

We consider a general constrained optimization problem. The goal is to minimize a function subject to constraints.

## Primal Problem

Let $x \in \mathbb{R}^n$ be the decision variable.

We define the **primal problem**:

$$
p^* = \inf_x f_0(x)
$$

subject to

$$
f_i(x) \le 0 \quad (i = 1,\dots,m),
\qquad
h_j(x) = 0 \quad (j = 1,\dots,p).
$$

- $f_0(x)$: objective function (what we want to minimize)  
- $f_i(x) \le 0$: inequality constraints (limits or resources)  
- $h_j(x) = 0$: equality constraints (exact requirements)  
- $p^*$: optimal value  

A point $x$ satisfying all constraints is called **primal feasible**.

## Lagrangian

To handle constraints, we introduce multipliers:

- $\lambda_i \ge 0$ for inequalities  
- $\nu_j \in \mathbb{R}$ for equalities  

The **Lagrangian** is

$$
\mathcal{L}(x, \lambda, \nu)
=
f_0(x)
+
\sum_{i=1}^m \lambda_i f_i(x)
+
\sum_{j=1}^p \nu_j h_j(x).
$$

**Interpretation:**  
- Each constraint is turned into a **penalty term**  
- If a constraint is violated, it increases the objective  
- The multipliers control how strongly we penalize inequality violations  

So the Lagrangian converts a constrained problem into an unconstrained one, parameterized by $(\lambda,\nu)$. It is (essentially) a relaxation of the problem.

## Dual Function

For fixed multipliers $(\lambda, \nu)$, define

$$
g(\lambda, \nu)=
\inf_x \mathcal{L}(x, \lambda, \nu).
$$

This is called the **dual function**.

We also define

$$
\hat{x}(\lambda, \nu)
\in
\arg\min_x \mathcal{L}(x, \lambda, \nu).
$$
to be the point that achieves the minimum for some fixed $\lambda$ and $\nu$. 

Important:

- $\hat{x}(\lambda,\nu)$ minimizes the Lagrangian  
- It **ignores primal feasibility**  
- It is just the best point under those penalties  

## Weak Duality

A fundamental fact:

$$
g(\lambda, \nu) \le f_0(x)
$$

for any feasible $x$ and any $\lambda \ge 0$.

Taking the best feasible $x$ gives

$$
g(\lambda, \nu) \le p^*
$$

The dual always gives a **lower bound** on the best possible primal value.


## Dual Problem

The **dual problem** is

$$
d^* =
\sup_{\lambda \ge 0, \nu} g(\lambda, \nu).
$$

We are choosing penalties to make the Lagrangian lower-bound approximation as "tight" as possible. Let $(\lambda^*, \nu^*)$ be an optimal solution to the dual problem. By weak duality $g(\lambda^*, \nu^*) \le p^*$. 

Now define

$$
\hat{x}
=
\hat{x}(\lambda^*, \nu^*)
\in
\arg\min_x \mathcal{L}(x, \lambda^*, \nu^*).
$$

We already defined this notation, we're just applying it to the dual optimal values now. 

At this point we know

$$
\mathcal{L}(\hat{x}, \lambda^*, \nu^*) = g(\lambda^*, \nu^*)
$$

but we do **not** yet know:

- whether $\hat{x}$ is feasible  
- whether $\hat{x}$ is optimal  

##  KKT Conditions

A triple $(x, \lambda, \nu)$ satisfies the **KKT conditions** if it satisfies:

### Primal feasibility
$$
f_i(x) \le 0,
\qquad
h_j(x) = 0
$$

### Dual feasibility
$$
\lambda_i \ge 0
$$

### Stationarity
$$
\nabla_x \mathcal{L}(x, \lambda, \nu) = 0
$$

### Complementary slackness
$$
\lambda_i f_i(x) = 0
$$

**Interpretation:**

- Stationarity → no local improvement possible  
- Feasibility → constraints satisfied  
- Complementary slackness → only “active” constraints matter  

## Strategy

We already have:

- $(\lambda^*, \nu^*)$ from the dual  
- $\hat{x}$ from minimizing the Lagrangian at these values.

So:

- stationarity holds (by construction) of finding $\hat{x}$
- dual feasibility holds (from dual problem)  

To prove optimality, we must check:

- primal feasibility
- complementary slackness

for $\hat{x}$.

## Duality Gap Identity

For any $x$ and $(\lambda, \nu)$,

$$
f_0(x) - g(\lambda, \nu)
=
-\sum_{i=1}^m \lambda_i f_i(x)
-
\sum_{j=1}^p \nu_j h_j(x)
+
\left[
\mathcal{L}(x, \lambda, \nu) - g(\lambda, \nu)
\right]
$$

**Interpretation:**  
The gap splits into:

- constraint violations  
- failure to minimize the Lagrangian  

Let's evaluate at $\hat{x}, \lambda^*, \nu^*$.

Each term becomes zero:

- Since $\hat{x}$ minimizes the Lagrangian:
$$
\mathcal{L}(\hat{x}, \lambda^*, \nu^*) - g(\lambda^*, \nu^*) = 0
$$

- If feasibility holds:
$$
h_j(\hat{x}) = 0
$$

- If complementary slackness holds:
$$
\lambda_i^* f_i(\hat{x}) = 0
$$

So:

$$
f_0(\hat{x}) - g(\lambda^*, \nu^*) = 0
$$

Thus:

$$
f_0(\hat{x}) = g(\lambda^*, \nu^*)
$$

## The Squeeze

We can finally complete the circuit. From weak duality:

$$
g(\lambda^*, \nu^*) \le p^*
$$

Substitute:

$$
f_0(\hat{x}) \le p^*
$$

From feasibility:

$$
p^* \le f_0(\hat{x})
$$

So:

$$
f_0(\hat{x}) \le p^* \le f_0(\hat{x})
$$

Therefore:

$$
f_0(\hat{x}) = p^*
$$

## Conclusion

So we have shown that: 

- $\hat{x}$ is feasible  
- $f_0(\hat{x}) = p^*$  

So:

$$
\hat{x} \text{ is primal optimal}
$$

We now rename:

$$
x^* = \hat{x}
$$


# Constrained vs Penalized Optimization

In general, for a fixed choice of $(\lambda, \nu)$, the penalized problem is generally **not equivalent** to the constrained problem.

- The solution may violate constraints  
- The solution may not be optimal  

So simply “adding penalties” does not automatically enforce constraints.


However, under standard convexity and regularity conditions, there is a precise **implicit correspondence at optimality**. Consider the constrained problem

$$
\min_x f_0(x)
\quad \text{s.t.} \quad
f_i(x) \le 0, \qquad h_j(x) = 0.
$$

Assume:

- the problem is convex  
- a solution $x^*$ exists  
- a regularity condition such as Slater’s condition holds (ensuring strong duality)

There exists a pair of multipliers $(\lambda^*, \nu^*)$ such that the triple $(x^*, \lambda^*, \nu^*)$ satisfies the KKT conditions, and
$$
x^* \in \arg\min_x \mathcal{L}(x, \lambda^*, \nu^*).
$$

This implies:

$$
\min_x f_0(x)
\quad \text{s.t. constraints}
\quad \equiv \quad
\min_x \mathcal{L}(x, \lambda^*, \nu^*).
$$

Consequently, 

- Each constraint has an associated multiplier, but the correct values $(\lambda^*, \nu^*)$ are determined **jointly**.
- There is an implicit relationship defined between constraints and multipliers through the KKT conditions.
- Only **active constraints** influence the solution: if $f_i(x^*) < 0$, then $\lambda_i^* = 0$.

Consider a problem with a single convex constraint:

$$
\min_x f_0(x)
\quad \text{s.t.} \quad
g(x) \le t,
$$

where $g(x)$ is convex.

The corresponding penalized form is:

$$
\min_x f_0(x) + \lambda g(x).
$$

Under convexity and regularity conditions (e.g. strong duality), the following holds:

> For each constraint level $t$, there exists a multiplier $\lambda^* \ge 0$ such that the solutions of the constrained and penalized problems coincide.

Moreover, if the solution is unique and the constraint is active (i.e. $g(x^*) = t$), then this correspondence is effectively one-to-one:

$$
t \;\longleftrightarrow\; \lambda^*.
$$


The KKT condition

$$
\lambda^* \big(g(x^*) - t\big) = 0
$$

links these two parameters through the solution $x^*$. Some Geometric intuition:

- The constrained problem restricts $x$ to a set $\{x : g(x) \le t\}$  
- The penalized problem tilts the objective by adding $\lambda g(x)$  

At the correct $\lambda^*$, the unconstrained minimizer lands exactly on the boundary $g(x^*) = t$.
