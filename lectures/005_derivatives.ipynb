# Matrix Calculus for Machine Learning

In single-variable calculus, the basic question is how one quantity changes as another quantity changes. If $f: \mathbb{R} \to \mathbb{R}$, then the **derivative** $f'(x)$ measures how the scalar output changes as the scalar input changes. The familiar rules describe common patterns: constants differentiate to zero, linear terms differentiate to their coefficients, and quadratic terms produce a factor of two.

In multivariable calculus, the same idea is extended to functions with several inputs. For a scalar-valued function such as $L(w_1,\ldots,w_d)$, the **gradient** collects the partial derivatives into one vector. For a function with several outputs, the **Jacobian** records how each output coordinate changes with respect to each input coordinate. For a scalar-valued function, the **Hessian** records second partial derivatives and describes curvature.

**Matrix calculus** systematizes this notation and extends it to expressions involving vectors, matrices, and tensors. The underlying calculus is not new. The main difference is that every object now has a shape, and the notation has to keep track of those shapes.

A term such as $y^\top y$ may look more complicated than a constant from single-variable calculus, but it is just compact notation for a real-valued object. Similarly, terms like $w^\top a$ and $w^\top A w$ are vector and matrix versions of familiar linear and quadratic expressions. Much of what we are doing is learning to recognize familiar calculus patterns after they have been written in vector and matrix notation.

## Taxonomy of derivative objects

Once functions can have scalar, vector, matrix, or **tensor** (basically, multi-dimensional matrices) inputs and outputs, the word "derivative" can refer to several different kinds of objects. The main task is to keep track of the input shape, the output shape, and the convention being used to arrange the derivative.

### Gradients: Scalar Output

#### Scalar input → scalar output

The simplest case is a function from one real number to another:

$$
f: \mathbb{R} \to \mathbb{R}.
$$

The derivative is the ordinary single-variable derivative:

$$
\frac{df}{dx}.
$$

#### Vector input → scalar output

A very common case in optimization and machine learning is a scalar-valued function of a vector:

$$
L: \mathbb{R}^d \to \mathbb{R}.
$$

Here the input is a vector and the output is a scalar. We usually write the input as:

$$
w =
\begin{bmatrix}
w_1 \\
w_2 \\
\vdots \\
w_d
\end{bmatrix}.
$$

The derivative is collected into the gradient:

$$
\nabla_w L =
\begin{bmatrix}
\frac{\partial L}{\partial w_1} \\
\frac{\partial L}{\partial w_2} \\
\vdots \\
\frac{\partial L}{\partial w_d}
\end{bmatrix}.
$$

If $w \in \mathbb{R}^d$, then:

$$
\nabla_w L \in \mathbb{R}^d.
$$

This is one of the most important shape rules in applied matrix calculus:

$$
\text{If } L \text{ is scalar-valued, then } \nabla_w L \text{ has the same shape as } w.
$$

For example, if:

$$
L(w) = w^\top w,
$$

then:

$$
\nabla_w L = 2w.
$$

**Why?**

The output $L(w)$ is a scalar, and the gradient is a vector with the same shape as $w$.

#### Matrix input → scalar output

Another common case is a scalar-valued function of a matrix:

$$
L: \mathbb{R}^{m \times n} \to \mathbb{R}.
$$

For example, $L(W)$ might depend on a matrix $W$ of parameters. If:

$$
W \in \mathbb{R}^{m \times n},
$$

then the gradient with respect to $W$ is arranged as a matrix of the same shape:

$$
\nabla_W L \in \mathbb{R}^{m \times n}.
$$

The entries of this gradient are:

$$
(\nabla_W L)_{ij} = \frac{\partial L}{\partial W_{ij}}.
$$

So:

$$
\nabla_W L =
\begin{bmatrix}
\frac{\partial L}{\partial W_{11}} & \cdots & \frac{\partial L}{\partial W_{1n}} \\
\vdots & \ddots & \vdots \\
\frac{\partial L}{\partial W_{m1}} & \cdots & \frac{\partial L}{\partial W_{mn}}
\end{bmatrix}.
$$

Again, for scalar-valued functions, the gradient has the same shape as the variable.

For example, if:

$$
L(W) = \|W\|_F^2 = \sum_{i=1}^m \sum_{j=1}^n W_{ij}^2,
$$

then:

$$
\nabla_W L = 2W.
$$

This is essentially the matrix version of:

$$
\frac{d}{dw} w^2 = 2w.
$$

#### Tensor input → scalar output

The same idea extends to tensor inputs. One may think of a tensor as just multidimensional array of numbers. Scalars are 0-dimensional tensors, vectors are 1-dimensional tensors, matrices are 2-dimensional tensors, and arrays with three or more axes are higher-order tensors. Suppose:

$$
L: \mathbb{R}^{d_1 \times d_2 \times \cdots \times d_k} \to \mathbb{R}.
$$

If the input tensor is $T$, then the gradient with respect to $T$ has the same shape as $T$:

$$
\nabla_T L \in \mathbb{R}^{d_1 \times d_2 \times \cdots \times d_k}.
$$

Each entry of the gradient tells us the partial derivative of $L$ with respect to the corresponding entry of $T$:

$$
(\nabla_T L)_{i_1,\ldots,i_k} = \frac{\partial L}{\partial T_{i_1,\ldots,i_k}}.
$$

This is how gradients are usually represented in modern array-based computation. A scalar objective depends on many arrays, and each array receives a gradient array of the same shape.

For scalar-valued functions, the pattern is therefore:

| Variable | Shape of variable | Shape of gradient |
|---|---:|---:|
| $w$ | $d$ | $d$ |
| $W$ | $m \times n$ | $m \times n$ |
| $T$ | $d_1 \times \cdots \times d_k$ | $d_1 \times \cdots \times d_k$ |

This same-shape rule is one of the main conventions used in machine learning notation and software.

### Jacobians: Vector Output

#### Vector input → vector output

Consider a vector-valued function of a vector:

$$
f: \mathbb{R}^n \to \mathbb{R}^m.
$$

The input is:

$$
x =
\begin{bmatrix}
x_1 \\
\vdots \\
x_n
\end{bmatrix},
$$

and the output is:

$$
f(x) =
\begin{bmatrix}
f_1(x) \\
\vdots \\
f_m(x)
\end{bmatrix}.
$$

The derivative is the **Jacobian** matrix:

$$
J_f(x) =
\begin{bmatrix}
\frac{\partial f_1}{\partial x_1} & \cdots & \frac{\partial f_1}{\partial x_n} \\
\vdots & \ddots & \vdots \\
\frac{\partial f_m}{\partial x_1} & \cdots & \frac{\partial f_m}{\partial x_n}
\end{bmatrix}.
$$

Under this convention:

$$
J_f(x) \in \mathbb{R}^{m \times n}.
$$

The rows correspond to **output** coordinates, and the columns correspond to **input** coordinates. The size of the Jacobian $J_f$ is always **output** $\times$ **input**. 

**Note:** This is different from the scalar-output gradient case. If the output is scalar, the gradient has the same shape as the input. If the output is vector-valued, the derivative needs to track how every output coordinate depends on every input coordinate, so it becomes a matrix.

#### Scalar input → vector output

A special case of the Jacobian is a vector-valued function of one scalar:

$$
f: \mathbb{R} \to \mathbb{R}^m.
$$

The output is:

$$
f(t) =
\begin{bmatrix}
f_1(t) \\
\vdots \\
f_m(t)
\end{bmatrix}.
$$

The Jacobian is:

$$
\frac{df}{dt} =
\begin{bmatrix}
\frac{df_1}{dt} \\
\vdots \\
\frac{df_m}{dt}
\end{bmatrix}.
$$

**Note:** This is a useful place to notice a mild notational tension. For scalar-valued functions, we usually write gradients as column vectors with the same shape as the input. For vector-valued functions, we usually write Jacobians with output coordinates first and input coordinates second.

So if $L:\mathbb{R}^n \to \mathbb{R}$, the Jacobian convention gives:

$$
J_L(x) \in \mathbb{R}^{1 \times n}.
$$

But the gradient convention gives:

$$
\nabla_x L \in \mathbb{R}^n.
$$

These are not different mathematical information. They are transposes of each other:

$$
J_L(x) = (\nabla_x L)^\top.
$$

This is one reason we usually say **"gradient"** for scalar-valued outputs and **"Jacobian"** for vector-valued outputs. 

### Matrix-valued Output

Matrix-output and tensor-output functions appear whenever an operation produces an array rather than a scalar or vector. These cases are common in applied work, but the full derivative is usually not written explicitly.

#### Scalar input to matrix output

A simple example is a matrix that depends on one scalar parameter:

$$
F:\mathbb{R} \to \mathbb{R}^{m \times n}.
$$

For instance, $F(t)$ might be a time-dependent matrix, a rotation matrix depending on an angle, or a covariance matrix depending on a scalar parameter. The derivative has the same shape as the output:

$$
\frac{dF}{dt} \in \mathbb{R}^{m \times n}.
$$

Entry by entry:

$$
\left(\frac{dF}{dt}\right)_{ij} = \frac{dF_{ij}}{dt}.
$$

A common example is a 2D rotation matrix depending on an angle $\theta$:

$$
R(\theta) =
\begin{bmatrix}
\cos \theta & -\sin \theta \\
\sin \theta & \cos \theta
\end{bmatrix}.
$$

Then:

$$
\frac{dR}{d\theta} =
\begin{bmatrix}
-\sin \theta & -\cos \theta \\
\cos \theta & -\sin \theta
\end{bmatrix}.
$$

This case is conceptually straightforward because there is only one input direction. The derivative is just the entrywise derivative of the matrix-valued function.

## Hessians

The Hessian is the derivative of the gradient with respect to $w$:

$$
H_L(w) = \nabla_w^2 L(w).
$$

In coordinates:

$$
H_L(w) =
\begin{bmatrix}
\frac{\partial^2 L}{\partial w_1 \partial w_1} & \cdots & \frac{\partial^2 L}{\partial w_1 \partial w_d} \\
\vdots & \ddots & \vdots \\
\frac{\partial^2 L}{\partial w_d \partial w_1} & \cdots & \frac{\partial^2 L}{\partial w_d \partial w_d}
\end{bmatrix}.
$$

So:

$$
H_L(w) \in \mathbb{R}^{d \times d}.
$$

The Hessian records second-order information. It describes how the gradient changes as $w$ changes.

For example, if:

$$
L(w) = w^\top A w,
$$

and $A$ is symmetric, then:

$$
\nabla_w L = 2Aw,
$$

and:

$$
\nabla_w^2 L = 2A.
$$
(verify these for yourself.) 

The Hessian is important because it describes curvature. In a second-order Taylor approximation:

$$
L(w + h) \approx L(w) + (\nabla_w L)^\top h + \frac{1}{2}h^\top H_L(w)h.
$$

**Note.** With the usual column-vector convention, the Hessian of a scalar-valued function is the Jacobian of its gradient. That is, for $L:\mathbb{R}^d \to \mathbb{R}$,

$$
\nabla^2 L(w) = J_{\nabla L}(w) \in \mathbb{R}^{d \times d}.
$$

Different derivative-layout conventions may transpose this object, but for twice continuously differentiable scalar functions the Hessian is symmetric.

## Numerator layout and denominator layout

There is one important notational warning. Different books arrange derivatives of vectors in different ways.

For a function:

$$
f: \mathbb{R}^n \to \mathbb{R}^m,
$$

this lecture uses the convention:

$$
J_f(x)_{ij} = \frac{\partial f_i}{\partial x_j}.
$$

So:

$$
J_f(x) \in \mathbb{R}^{m \times n}.
$$

This is often called the **numerator-layout** convention for Jacobians. It is common in applied mathematics, optimization, and much of machine learning because it gives the clean approximation:

$$
f(x + dx) \approx f(x) + J_f(x)dx.
$$

Some sources use the transpose convention, especially in parts of statistics or older matrix calculus references. Under that convention, the Jacobian of $f:\mathbb{R}^n \to \mathbb{R}^m$ may be written with shape $n \times m$.

This can make formulas look different across books. For example, one source may write a chain rule using $J^\top v$, while another may write an equivalent expression using $v^\top J$ or a transposed Jacobian definition.

The safest habit is to check shapes. If the formula is supposed to produce a vector in $\mathbb{R}^n$, then the multiplication must produce a vector in $\mathbb{R}^n$. Shape checking usually reveals which convention is being used.

### Summary table

The following table summarizes the most common derivative objects.

| Function type | Example notation | Derivative object | Typical shape |
|---|---|---|---|
| Scalar to scalar | $f:\mathbb{R}\to\mathbb{R}$ | ordinary derivative | scalar |
| Vector to scalar | $L:\mathbb{R}^d\to\mathbb{R}$ | gradient $\nabla_w L$ | $d$ |
| Matrix to scalar | $L:\mathbb{R}^{m\times n}\to\mathbb{R}$ | matrix gradient $\nabla_W L$ | $m \times n$ |
| Vector to vector | $f:\mathbb{R}^n\to\mathbb{R}^m$ | Jacobian $J_f$ | $m \times n$ |
| Scalar to vector | $f:\mathbb{R}\to\mathbb{R}^m$ | vector derivative $\frac{df}{dt}$ | $m$ |
| Scalar to matrix | $F:\mathbb{R}\to\mathbb{R}^{m\times n}$ | matrix derivative $\frac{dF}{dt}$ | $m \times n$ |


The most important practical distinction is between scalar-output derivatives and vector-output derivatives. For scalar-output functions, gradients are usually arranged to match the shape of the variable. For vector-output functions, the derivative needs to track how each output coordinate depends on each input coordinate, so extra axes appear.

**Practical rule**

When reading or deriving a matrix calculus expression, ask three questions:

1. What is the variable being differentiated?
2. What is the shape of the output?
3. What shape must the derivative object have?

## Basic derivative identities 

## Scalar output, scalar input

Let:

$$
f:\mathbb{R}\to\mathbb{R}.
$$

The basic rules are the familiar single-variable rules:

$$
\frac{d}{dx}c = 0.
$$

$$
\frac{d}{dx}(ax) = a.
$$

$$
\frac{d}{dx}(ax^2) = 2ax.
$$

$$
\frac{d}{dx}\frac{1}{2}ax^2 = ax.
$$

These patterns reappear in vector and matrix form as constant, linear, and quadratic terms.

The chain rule is:

$$
\frac{d}{dx}f(g(x)) = f'(g(x))g'(x).
$$

## Scalar output, vector input

Let:

$$
L:\mathbb{R}^d\to\mathbb{R}.
$$

If:

$$
w \in \mathbb{R}^d,
$$

then:

$$
\nabla_w L \in \mathbb{R}^d.
$$

So the gradient has the same shape as the vector being differentiated.

### Constant, linear, and quadratic terms

If $c$ does not depend on $w$, then:

$$
\nabla_w c = 0.
$$

If $a \in \mathbb{R}^d$ is constant with respect to $w$, then:

$$
\nabla_w(a^\top w) = a.
$$

$$
\nabla_w(w^\top a) = a.
$$

The squared norm is:

$$
\nabla_w(w^\top w) = 2w.
$$

Equivalently:

$$
\nabla_w \frac{1}{2}w^\top w = w.
$$

If $A \in \mathbb{R}^{d \times d}$ is constant with respect to $w$, then:

$$
\nabla_w(w^\top A w) = (A + A^\top)w.
$$

If $A$ is symmetric, then:

$$
\nabla_w(w^\top A w) = 2Aw.
$$

With a factor of $\frac{1}{2}$:

$$
\nabla_w \frac{1}{2}w^\top A w = Aw.
$$

when $A$ is symmetric.

### Chain rule through a vector intermediate

Suppose:

$$
z = f(w), \qquad L = g(z),
$$

where:

$$
w \in \mathbb{R}^d, \qquad z \in \mathbb{R}^m, \qquad L \in \mathbb{R}.
$$

If:

$$
J_f(w) \in \mathbb{R}^{m \times d},
$$

then:

$$
\nabla_w L = J_f(w)^\top \nabla_z L.
$$

This is the main chain rule for scalar-valued objectives. The gradient with respect to $z$ is pulled back to a gradient with respect to $w$ by multiplying by the transposed Jacobian.

## Scalar output, matrix input

Let:

$$
L:\mathbb{R}^{m \times n}\to\mathbb{R}.
$$

If:

$$
W \in \mathbb{R}^{m \times n},
$$

then:

$$
\nabla_W L \in \mathbb{R}^{m \times n}.
$$

The gradient with respect to a matrix has the same shape as the matrix.

### Constant and linear terms

If $c$ does not depend on $W$, then:

$$
\nabla_W c = 0.
$$

A linear function of a matrix is often written using the trace. If $A \in \mathbb{R}^{m \times n}$ is constant with respect to $W$, then:

$$
\nabla_W \operatorname{tr}(A^\top W) = A.
$$

Equivalently:

$$
\nabla_W \operatorname{tr}(W^\top A) = A.
$$

This is the matrix version of:

$$
\nabla_w(a^\top w) = a.
$$

### Frobenius norm

The Frobenius norm is the matrix version of the Euclidean norm:

$$
\|W\|_F^2 = \sum_{i=1}^m \sum_{j=1}^n W_{ij}^2 = \operatorname{tr}(W^\top W).
$$

Its gradient is:

$$
\nabla_W \|W\|_F^2 = 2W.
$$

Equivalently:

$$
\nabla_W \frac{1}{2}\|W\|_F^2 = W.
$$

## Vector output, vector input

Let:

$$
f:\mathbb{R}^n\to\mathbb{R}^m.
$$

The derivative is the Jacobian:

$$
J_f(x) \in \mathbb{R}^{m \times n}.
$$

### Linear and affine maps

If:

$$
f(x) = Ax,
$$

where:

$$
A \in \mathbb{R}^{m \times n},
$$

then:

$$
J_f(x) = A.
$$

If:

$$
f(x) = Ax + b,
$$

then:

$$
J_f(x) = A.
$$

The constant shift $b$ does not affect the Jacobian.

### Chain rule for vector-output functions

If:

$$
x \in \mathbb{R}^n, \qquad z = f(x) \in \mathbb{R}^m, \qquad y = g(z) \in \mathbb{R}^p,
$$

then:

$$
J_y(x) = J_{g\circ f}(x) = J_g(z)J_f(x).
$$

The shapes are:

$$
J_g(z) \in \mathbb{R}^{p \times m}, \qquad J_f(x) \in \mathbb{R}^{m \times n}.
$$

So:

$$
J_{g\circ f}(x) \in \mathbb{R}^{p \times n}.
$$

For scalar output, this becomes the gradient chain rule we saw earlier:

$$
\nabla_x L = J_f(x)^\top \nabla_z L.
$$

## Matrix output, scalar input

Let:

$$
F:\mathbb{R}\to\mathbb{R}^{m \times n}.
$$

This means a matrix depends on one scalar variable. The derivative has the same shape as the matrix output:

$$
\frac{dF}{dt} \in \mathbb{R}^{m \times n}.
$$

Entry by entry:

$$
\left(\frac{dF}{dt}\right)_{ij} = \frac{dF_{ij}}{dt}.
$$

### Constant, linear, and elementwise rules

If $A$ is constant with respect to $t$, then:

$$
\frac{d}{dt}A = 0.
$$

If:

$$
F(t) = tA,
$$

then:

$$
\frac{dF}{dt} = A.
$$

If:

$$
F(t) = t^2A,
$$

then:

$$
\frac{dF}{dt} = 2tA.
$$

If $\phi$ is applied entrywise to a matrix $F(t)$, then:

$$
\frac{d}{dt}\phi(F(t)) = \phi'(F(t)) \odot \frac{dF}{dt}.
$$
where $\odot$ is element-wise (Hadamard) multiplication.
