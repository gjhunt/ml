# Statistical Learning

These lecture notes introduce core ideas in statistical learning, with an emphasis on how models are formulated, fit, evaluated, and compared. The notes use empirical risk minimization as a recurring framework, then build toward model selection, regularization, neural networks, unsupervised learning, and tree-based methods.

The goal is not to catalog every model or algorithm. The goal is to develop a practical way of thinking about learning from data: what objective is being optimized, what assumptions are being made, how flexible the model is, and how well its performance is likely to generalize.

## Course Map

| Part | Main Question | Topics |
|---|---|---|
| Foundations and Linear Models | What does it mean to fit a model? | Supervised learning, ERM, derivatives, linear regression, logistic regression, multiclass classification |
| Evaluation and Model Selection | How do we know whether a model generalizes? | Train/test performance, validation, cross-validation, KNN, bias-variance tradeoff |
| Regularization and Penalized Models | How do we control model complexity? | Ridge, lasso, elastic net, penalized logistic regression |
| Neural Networks | How do models learn representations? | Feedforward networks, optimization, backpropagation, architecture choices |
| Unsupervised Learning | What structure can we find without labels? | Dimensionality reduction, PCA, clustering, k-means |
| Tree-Based Methods | How do adaptive rule-based models and ensembles work? | Decision trees, random forests, boosting |

## Prerequisites

These notes assume comfort with basic Python, plotting, and numerical computing. They also assume working familiarity with functions, derivatives, vectors, matrices, probability, and statistical estimation. The early lectures review some of the mathematical notation used later, but the notes are written for readers who are willing to move between equations, code, and visual interpretation.

## Practice Questions

The practice questions are paired with the lecture topics. They are meant to be attempted actively, not just read. A good first pass is to answer them without looking back at the notes, then revisit the relevant lecture sections to repair gaps in notation, interpretation, or implementation.
